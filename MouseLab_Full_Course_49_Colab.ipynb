# MouseLab — آموزش جامع ۴۹درسی از داده تا تصمیم طراحی

## هدف و شیوهٔ مطالعه

این فایل **نسخهٔ کاملِ همان ۴۹ درسِ وب‌سایت، در همان ترتیب هفت‌فصلی** است؛ نه خلاصهٔ ۳۹ تحلیل قبلی. در هر درس، هدف، تشریح مفهوم، فرمول، مثال حل‌شده، کد قابل اجرا، خروجی، تفسیر، کاربرد و تمرین با پاسخ وجود دارد. تفسیر «اجرای حاضر» از همان متغیرهای محاسبه‌شده تولید می‌شود؛ اعداد مثالِ مرجع در متن برای مقایسه حفظ شده‌اند. تغییر داده می‌تواند آن اعداد مرجع را از خروجی جدید متفاوت کند.

۱. در Google Colab از **File → Upload notebook** همین فایل `.ipynb` را انتخاب کنید. سپس **Runtime → Run all** را بزنید. CPU کافی است؛ GPU لازم نیست. نصب اولیهٔ کتابخانه‌ها به اینترنت نیاز دارد.

۲. از بالا اجرا کنید؛ درس‌ها آگاهانه به متغیرهای درس‌های قبل متکی‌اند. بعد از تغییر فیلتر کارگاه واقعی، همان سلول را دوباره اجرا کنید. بعد از تغییر دادهٔ پایه، از ابتدا اجرا کنید. آزمون‌های انتهایی برای نسخهٔ مرجع‌اند و در آزمایش با دادهٔ تازه باید متناسب بازنویسی شوند.

۳. سلول داده‌های فشرده را تغییر ندهید. فایل‌های دقیق CSV درون همین نوت‌بوک‌اند و با SHA-256 کنترل می‌شوند؛ هیچ مسیر محلی، سرور شخصی یا رمز لازم نیست. خروجی‌های ذخیره‌شده اجازه می‌دهند بدون اجرای مجدد هم مطالعه کنید.

۴. نمودارهای پایتون برای جلوگیری از به‌هم‌ریختگی حروف فارسی در محیط‌های مختلف، محور و عنوان انگلیسی دارند؛ توضیح و تفسیر کنارشان فارسی است. کد رسم همهٔ نمودارها قابل ویرایش است.

۵. ویوئر سایت خروجی‌های همین فایل را نمایش می‌دهد، اما موتور اجرای پایتون نیست. اجرا و ذخیرهٔ تغییرات در Colab/Jupyter انجام می‌شود؛ این فایل قبلاً روی محیط آزمون اجرا می‌شود، نه در حساب Google شما.

### سه نوع داده و مرز نتیجه‌گیری

- ده فرد آموزشی، خرابی‌ها و فروش ماهانهٔ ساختگی: برای یادگیری روش، نه تعمیم به بازار.
- ANSUR II: ۶٬۰۶۸ نظامی آمریکایی با ابعاد دست به میلی‌متر؛ نمایندهٔ خودکار کاربران ایرانی نیست.
- Bike Sharing: ۱۷٬۳۷۹ ساعت ثبت‌شده در ۲۰۱۱–۲۰۱۲؛ هر سطر ساعت است، نه فرد یکتا. ساعت غایب با صفر پر نشده است.

در فایل ساعتی، اختلاف مستندات README بسته با صفحهٔ UCI در تبدیل دما و نام فصل وجود دارد. پس دما و باد در مقیاس نرمال‌شدهٔ خام و فصل فقط با کد ۱–۴ تحلیل می‌شوند. رطوبت در نمودار به درصد تبدیل می‌شود. تبدیل خطی مثبت همبستگی را عوض نمی‌کند؛ اختلاف فراداده نباید به برچسب قطعیِ درجهٔ سلسیوس تبدیل شود.

## آماده‌سازی محیط

کد زیر فقط بسته‌های گمشده را نصب می‌کند و نسخه‌های واقعی اجرا را نشان می‌دهد. نسخه‌های الگوریتم و بهینه‌ساز ممکن است خروجی تصادفی یا بهینه‌سازی را اندکی تغییر دهند؛ بذر ۴۲ و محدودیت نخ‌ها برای بازتولیدپذیری تعیین شده‌اند. بسته‌ها: NumPy، pandas، SciPy، Matplotlib، scikit-learn، statsmodels. هیچ کدی از اینترنت دانلود و اجرا نمی‌شود؛ فقط نصب معمول کتابخانه‌های علمی در صورت فقدان آن‌هاست.


````python
import os
os.environ['OMP_NUM_THREADS']='1'
os.environ['OPENBLAS_NUM_THREADS']='1'
os.environ['MKL_NUM_THREADS']='1'
import importlib.util, subprocess, sys
required = {'numpy':'numpy', 'pandas':'pandas', 'matplotlib':'matplotlib',
            'scipy':'scipy', 'sklearn':'scikit-learn', 'statsmodels':'statsmodels'}
missing = [package for module, package in required.items() if importlib.util.find_spec(module) is None]
if missing:
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', *missing])
import numpy as np, pandas as pd, scipy, sklearn, statsmodels, matplotlib
import matplotlib.pyplot as plt
from scipy import stats
from IPython.display import display, Markdown, HTML
from sklearn.base import clone
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.metrics import mean_squared_error, r2_score, roc_auc_score, roc_curve, confusion_matrix

SEED = 42
rng = np.random.default_rng(SEED)
plt.rcParams.update({'figure.figsize': (8, 4.5), 'figure.dpi': 110,
                     'axes.spines.top': False, 'axes.spines.right': False,
                     'axes.grid': True, 'grid.alpha': .18, 'font.size': 10})
BLUE, GOLD, TEAL = '#2475a4', '#d09331', '#168a7c'
print('Python:', sys.version.split()[0], '| numpy:', np.__version__, '| scipy:', scipy.__version__,
      '| sklearn:', sklearn.__version__, '| matplotlib:', matplotlib.__version__)
plt.rcParams.update({'axes.prop_cycle':matplotlib.cycler(color=[BLUE,GOLD,TEAL,'#8d638a']), 'savefig.bbox':'tight'})
pd.set_option('display.max_columns',12)
pd.set_option('display.max_rows',20)
def say(text):
    display(Markdown('**تفسیر خروجی اجرای حاضر:** '+text))
def group_figure(groups,labels,title):
    fig,ax=plt.subplots(figsize=(8,4))
    for i,values in enumerate(groups):
        jitter=np.linspace(-.09,.09,len(values))
        ax.scatter(np.full(len(values),i)+jitter,values,color=BLUE,alpha=.8)
        ax.errorbar(i,np.mean(values),yerr=np.std(values,ddof=1),fmt='s',color=GOLD,capsize=6)
    ax.set(xticks=range(len(labels)),xticklabels=labels,ylabel='Hand length (cm)',title=title)
    ax.text(.02,.97,'Gold square ±1 sample SD (NOT a confidence interval)',transform=ax.transAxes,va='top',fontsize=9)
    plt.tight_layout(); plt.show()
def prediction_figure(actual,pred,title,unit):
    fig,ax=plt.subplots(); ax.scatter(actual,pred,color=BLUE,s=60)
    limits=[min(actual.min(),pred.min()),max(actual.max(),pred.max())]
    ax.plot(limits,limits,color='0.4',ls='--',label='Perfect prediction')
    for i,(a,b) in enumerate(zip(actual,pred)): ax.annotate(str(i+1),(a,b),xytext=(3,3),textcoords='offset points')
    ax.set(xlabel='Observed '+unit,ylabel='Held-out predicted '+unit,title=title); ax.legend(); plt.show()
print('pandas:',pd.__version__,'| statsmodels:',statsmodels.__version__)
````

## فصل: داده، توصیف و شکل توزیع

### درس 01. داده‌های خام و کیفیت اولیه

**هدف یادگیری:** از سؤال طراحی به دادهٔ قابل تحلیل برسید.

#### مفهوم و تشریح کامل

جامعهٔ هدف کسانی‌اند که می‌خواهیم تصمیم دربارهٔ آن‌ها معتبر باشد؛ نمونه فقط کسانی است که واقعاً اندازه‌گیری شده‌اند. واحد مشاهده را قبل از شروع مشخص کنید: فرد، محصول یا ساعت. طول و عرض دست متغیر کمّی پیوسته‌اند؛ شناسهٔ فرد عدد اندازه‌گیری نیست. ابتدا روش اندازه‌گیری، واحد، رضایت و معیار ورود را ثابت کنید، سپس سطرها را ثبت کنید.

#### فرمول و قرارداد

```text
هر سطر = یک فرد مستقل؛ هر ستون = یک ویژگی با تعریف و واحد مشخص
```

#### مثال حل‌شدهٔ مرجع

در جدول پایه ۱۰ فرد، چهار ویژگی و صفر خانهٔ گمشده داریم. مجموع طول‌ها ۲۱۵ سانتی‌متر است. این ده عدد برای تمرین در فایل‌های اولیه آمده‌اند؛ ادعای دو میلیون گیمر یا نمونه‌گیری تصادفیِ واقعی از آن‌ها قابل استخراج نیست. برای تبدیل واحد، ۱۸ سانتی‌متر برابر ۱۸۰ میلی‌متر است؛ ترکیب این دو واحد در یک ستون، همهٔ تحلیل‌های بعدی را خراب می‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
df = pd.DataFrame({
    'user_id': range(1, 11),
    'hand_length_cm': [18,19,20,20,21,21,22,23,24,27],
    'hand_width_cm': [8,8.5,9,9,9.5,9.5,10,10,11,12],
    'age_years': [22,25,19,28,24,30,21,26,33,27],
    'gaming_years': [3,6,2,8,5,10,4,7,12,9],
})
assert len(df) == 10 and df.user_id.is_unique and not df.isna().any().any()
assert df.hand_length_cm.between(10, 40).all()
display(df)
print('n =', len(df), '| missing cells =', int(df.isna().sum().sum()))
numeric_cols=['hand_length_cm','hand_width_cm','age_years','gaming_years']

say(f'این جدول {len(df)} فرد مستقل و مجموع طول {df.hand_length_cm.sum():.0f} سانتی‌متر دارد؛ فقدان مقدار خالی، نمایندگی جامعهٔ هدف را ثابت نمی‌کند.')
````

#### راهنمای کامل خواندن خروجی و نمودار

جدول را سطری و ستونی بخوانید: کاربر ۱۰ طول دست ۲۷، عرض ۱۲، سن ۲۷ و سابقهٔ ۹ سال دارد. ابتدا معقول‌بودن و تعریف اندازه‌گیری او را بررسی می‌کنیم؛ بزرگ‌بودن مقدار به‌تنهایی مجوز حذف نیست. در دادهٔ واقعی، شناسهٔ یکتا، تکرار اندازه‌گیری و علت مقدارهای خالی را هم ثبت کنید.

#### کاربرد و محدودیت

اگر دو بار از یک فرد اندازه‌گیری کرده‌اید، دو سطر مستقل به دست نیاورده‌اید. می‌توان برای کاهش خطای اندازه‌گیری میانگین تکرارها را گرفت، اما تعداد افراد همچنان همان است.

#### تمرین

اگر یک طول دست به‌جای ۲۱٫۵ با مقدار ۲۱۵ وارد شده باشد، اولین اقدام چیست؟

#### پاسخ تشریحی

واحد و منبع ثبت را بررسی کنید. پس از تأیید میلی‌متر بودن، آن را به سانتی‌متر تبدیل و تغییر را ثبت کنید. حذف یا تقسیم خودکار همهٔ مقدارهای بزرگ بدون بررسی، درست نیست.

### درس 02. نمودار پراکندگی کاربران

**هدف یادگیری:** قبل از هر خلاصه‌سازی، تک‌تک مشاهده‌ها را ببینید.

#### مفهوم و تشریح کامل

نمودار نقاط خام کمک می‌کند دامنه، تجمع و مشاهدهٔ متفاوت را ببینیم. در این شکل محور افقی شناسهٔ فرد است؛ فاصلهٔ شناسه‌ها معنای فیزیکی ندارد. بنابراین از بالا رفتن نقاط نمی‌توان رشد زمانی یا اثر تجربه را نتیجه گرفت. مرتب‌بودن طول‌ها در جدول، خودش ظاهر صعودی شکل را ایجاد کرده است.

#### فرمول و قرارداد

```text
نقطهٔ i = (شناسهٔ i، طول دست i)
```

#### مثال حل‌شدهٔ مرجع

نقاط از ۱۸ تا ۲۷ سانتی‌متر امتداد دارند؛ ۲۰ و ۲۱ هرکدام دو بار تکرار شده‌اند. فاصلهٔ ۲۴ تا ۲۷ از بسیاری فاصله‌های مجاور بزرگ‌تر است. مقدار ۲۷ را فعلاً مشاهدهٔ انتهایی می‌نامیم؛ در درس چارک‌ها خواهیم دید برچسب دورافتاده به قرارداد محاسبه هم وابسته است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
fig, ax = plt.subplots()
ax.scatter(df.user_id, df.hand_length_cm, s=70, color=BLUE)
ax.set(xlabel='User ID', ylabel='Hand length (cm)', title='Hand length across 10 sample users')
ax.set_xticks(df.user_id)
plt.show()

say(f'ده نقطهٔ مستقل نشان داده شده‌اند؛ بیشترین طول {df.hand_length_cm.max():.0f} سانتی‌متر است. محور شناسه فقط نام فرد است، نه روند زمانی.')
````

#### راهنمای کامل خواندن خروجی و نمودار

ارتفاع نقطه، اندازهٔ دست است. تعداد نقاط، تعداد افراد است. این شکل رابطهٔ دو ویژگیِ اندازه‌گیری‌شده را نمی‌سنجد؛ برای رابطهٔ طول و عرض باید همان دو ویژگی روی محورهای X و Y قرار گیرند. رسم خط اتصال بین کاربران می‌تواند ترتیبِ پیوسته‌ای القا کند که در داده وجود ندارد.

#### کاربرد و محدودیت

در آزمایش ارگونومی، یک نمودار نقاطِ جداگانه برای هر روش گرفتن ماوس یا گروه هدف می‌تواند پوشش نمونه را آشکار کند. بدون شناسه، یافتن خطای ثبت یا بازگشت به مشاهدهٔ غیرمعمول دشوار می‌شود.

#### تمرین

چرا افزودن خطی که کاربر ۱ را به ۲ و سپس به ۳ وصل می‌کند ممکن است گمراه‌کننده باشد؟

#### پاسخ تشریحی

افراد ایستگاه‌های یک مسیر یا زمان‌های متوالی نیستند. خط اتصال پیوستگی و روندی القا می‌کند که از شناسهٔ قراردادی به دست نمی‌آید.

### درس 03. میانگین طول دست

**هدف یادگیری:** مرکز داده و حساسیت میانگین را با عدد بفهمید.

#### مفهوم و تشریح کامل

میانگین نقطهٔ تعادل عددی داده است: مجموع انحراف‌های علامت‌دار از آن صفر می‌شود. میانه مرکزِ ترتیب است و به بزرگی دقیق مشاهدهٔ انتهایی کمتر حساس است. انتخاب شاخص به پرسش وابسته است؛ برای مجموع مصرف، میانگین مهم است و برای توصیف فرد میانی، میانه اغلب خواناتر است.

#### فرمول و قرارداد

```text
x̄ = Σxᵢ / n = 215 / 10 = 21.5 cm
```

#### مثال حل‌شدهٔ مرجع

دو مقدار میانیِ دادهٔ مرتب‌شده هر دو ۲۱ هستند، پس میانه ۲۱ است؛ میانگین ۲۱٫۵ است. اگر فقط برای تحلیل حساسیت ۲۷ را با ۲۴ جایگزین کنیم، مجموع به ۲۱۲ و میانگین به ۲۱٫۲ می‌رسد. اگر ۲۷ را کنار بگذاریم، میانگین نه‌نفره ۱۸۸÷۹≈۲۰٫۸۸۹ است. این‌ها سناریو هستند، نه پاک‌سازی داده.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
x = df.hand_length_cm.to_numpy(dtype=float)
mean_length = x.mean()
print(f'Mean = {mean_length:.2f} cm; median = {np.median(x):.2f} cm')
fig, ax = plt.subplots()
ax.scatter(df.user_id, x, color=BLUE, label='Observations')
ax.axhline(mean_length, color=GOLD, lw=2, label=f'Mean {mean_length:.2f} cm')
ax.set(xlabel='User ID', ylabel='Hand length (cm)', title='Observed lengths and arithmetic mean')
ax.legend(); plt.show()

say(f'میانگین {mean_length:.2f} و میانه {np.median(x):.2f} سانتی‌متر است. فاصلهٔ مثبت آن‌ها با انتهای راست سازگار است، اما اثبات شکل جامعه نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

خط افقی شکل میانگین است؛ نقاط بالاتر انحراف مثبت و نقاط پایین‌تر انحراف منفی دارند. فاصله‌ها را با واحد سانتی‌متر بخوانید. تفاوت میانگین و میانه سرنخ شکل توزیع است، اما به‌تنهایی جهت چولگی را در هر توزیع ممکن تعیین نمی‌کند.

#### کاربرد و محدودیت

از میانگین طول دست نمی‌توان طول بهینهٔ بدنه را با نسبت ثابتی مثل یک‌دوم نتیجه گرفت. هندسهٔ گرفتن، عرض، ارتفاع، ترجیح کاربر و معیار عملکرد باید در آزمون پروتوتایپ بررسی شوند.

#### تمرین

اگر به طول دست همهٔ افراد ۲ سانتی‌متر اضافه کنیم، میانگین و میانه چه می‌شوند؟

#### پاسخ تشریحی

هر دو دقیقاً ۲ واحد زیاد می‌شوند: میانگین ۲۳٫۵ و میانه ۲۳. پراکندگی پیرامون میانگین تغییری نمی‌کند.

### درس 04. هیستوگرام و منحنی نرمال

**هدف یادگیری:** شکل توزیع را با محور و مقیاس درست بخوانید.

#### مفهوم و تشریح کامل

هیستوگرام محور عددی را به بازه‌ها تقسیم و مشاهده‌های هر بازه را می‌شمارد. انتخاب عرض و مبدأ بازه بر شکل ظاهری اثر می‌گذارد؛ در ده مشاهده، تغییر یک مرز می‌تواند یک قله را ایجاد یا حذف کند. منحنی نرمالِ برازش‌شده مدلی برای مقایسهٔ بصری است و از روی آن نمی‌توان توزیع جامعه را اثبات کرد.

#### فرمول و قرارداد

```text
در هیستوگرام چگالی: ارتفاع = تعداد بازه / (n × عرض بازه)
```

#### مثال حل‌شدهٔ مرجع

اگر در بازهٔ عرض ۲ سانتی‌متر، ۳ نفر از ۱۰ نفر قرار بگیرند، فراوانی ۳، سهم ۳۰٪ و چگالی ۳÷(۱۰×۲)=۰٫۱۵ بر سانتی‌متر است. نمودار این درس چگالی دارد؛ بنابراین مساحت ستون‌ها جمعاً ۱ می‌شود. منحنی همراه، میانگین ۲۱٫۵ و انحراف معیار نمونه‌ای حدود ۲٫۶۳۵ دارد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
fig, ax = plt.subplots()
bins = np.arange(17.5, 28.6, 2)
ax.hist(x, bins=bins, color=BLUE, alpha=.75, edgecolor='white', density=True)
grid = np.linspace(17, 28, 300)
ax.plot(grid, stats.norm.pdf(grid, x.mean(), x.std(ddof=1)), color=GOLD, lw=2,
        label='Fitted normal density')
ax.set(xlabel='Hand length (cm)', ylabel='Density', title='Sample distribution and fitted normal curve')
ax.legend(); plt.show()

say(f'مساحت هیستوگرام چگالی تقریباً یک است، نه جمع ارتفاع ستون‌ها. منحنی نرمال با میانگین {x.mean():.2f} و SD نمونه‌ای {x.std(ddof=1):.3f} فقط یک مرجع برازش است.')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور افقی طول دست و محور عمودی چگالی است. ارتفاع ۰٫۱۵ به معنی ۱۵ نفر یا احتمال دقیق در یک نقطه نیست. ناحیهٔ زیر نمودار در یک فاصله به سهم آن فاصله مربوط می‌شود. شکل را با نقاط خام و Q–Q کنار هم ببینید، به‌ویژه برای انتهای راست.

#### کاربرد و محدودیت

برای مقایسهٔ گروه‌های با حجم متفاوت از چگالی یا درصد با تعریف یکسان استفاده کنید. مقایسهٔ تعداد خامِ ۴٬۰۸۲ مرد با ۱٬۹۸۶ زن می‌تواند فقط اختلاف تعداد سطرها را نمایش دهد.

#### تمرین

چرا منحنی چگالی نرمال را نمی‌توان بدون تبدیل روی محور «تعداد افراد» قرار داد؟

#### پاسخ تشریحی

واحد محور‌ها متفاوت است. برای محور تعداد، مقدار چگالی باید در n و عرض بازه ضرب شود؛ یا هر دو نمایش روی چگالی تنظیم شوند.

### درس 05. نمودار جعبه‌ای و چارک‌ها

**هدف یادگیری:** چارک، سبیل و قاعدهٔ دورافتاده را از هم جدا کنید.

#### مفهوم و تشریح کامل

چارک‌ها دادهٔ مرتب‌شده را در احتمال‌های ۰٫۲۵، ۰٫۵ و ۰٫۷۵ خلاصه می‌کنند. برای نمونهٔ محدود، روش‌های درون‌یابی متفاوت‌اند. در این آموزش روش خطی نوع ۷ به کار رفته است: جایگاه صفرمبنا h=(n−1)p و درون‌یابی بین دو همسایه. سبیلِ جعبه تا دورترین مشاهدهٔ داخل حصار می‌رود، نه الزاماً تا مقدار خود حصار.

#### فرمول و قرارداد

```text
Q(p)=x[j]+(h−j)(x[j+1]−x[j]); IQR=Q3−Q1; fences=Q1−1.5IQR, Q3+1.5IQR
```

#### مثال حل‌شدهٔ مرجع

برای Q3، h=۹×۰٫۷۵=۶٫۷۵ است؛ مقدارهای صفرمبنای ۶ و ۷ برابر ۲۲ و ۲۳ هستند، پس Q3=۲۲٫۷۵. Q1=۲۰، میانه=۲۱، IQR=۲٫۷۵ و حصار بالا=۲۶٫۸۷۵ است. بنابراین ۲۷ با این قرارداد علامت می‌خورد. روش میانهٔ دو نیمه Q3=۲۳ و حصار بالا=۲۷٫۵ می‌دهد و آن را علامت نمی‌زند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
q1, median, q3 = np.quantile(x, [.25, .5, .75])
iqr = q3-q1
fences = (q1-1.5*iqr, q3+1.5*iqr)
print({'min':x.min(), 'Q1':q1, 'median':median, 'Q3':q3, 'max':x.max(),
       'IQR':iqr, 'fences':fences, 'flagged':x[(x<fences[0]) | (x>fences[1])].tolist()})
fig, ax = plt.subplots(figsize=(7, 2.7))
ax.boxplot(x, orientation='horizontal')
ax.set(xlabel='Hand length (cm)', title='Five-number summary and potential outliers')
plt.show()

say(f'چارک خطی نوع ۷: Q1={q1:.2f}، Q3={q3:.2f}، IQR={iqr:.2f}؛ حصار بالا {fences[1]:.3f} سانتی‌متر است. مقدارهای علامت‌خورده {x[(x<fences[0])|(x>fences[1])].tolist()} نیازمند بررسی‌اند، نه حذف خودکار.')
````

#### راهنمای کامل خواندن خروجی و نمودار

هر دو نتیجه با قرارداد خود سازگارند؛ خطای متن قدیمی، مخلوط‌کردن قراردادها و نتیجه‌گیری قطعی بود. جعبه ناحیهٔ چارکی را نشان می‌دهد و با داده‌های گسسته لزوماً دقیقاً نصف نقاط داخل آن قرار نمی‌گیرد. مقدار علامت‌خورده را با عنوان «نیازمند بررسی» گزارش کنید.

#### کاربرد و محدودیت

در آزمایشگاه چارک‌ها همین روش را روی ANSUR II امتحان کنید. حصار، حد ارگونومیک یا مشخصهٔ تلرانس محصول نیست؛ صدک‌ها نیز به‌تنهایی طول بدنهٔ مناسب را تعیین نمی‌کنند.

#### تمرین

با Q1=۱۰ و Q3=۱۴، حصارها چیست و آیا ۲۰ بیرون است؟

#### پاسخ تشریحی

IQR=۴ و حصارها ۴ و ۲۰ هستند. قاعدهٔ معمول مشاهدهٔ کمتر از ۴ یا بیشتر از ۲۰ را علامت می‌زند؛ خود ۲۰ بیرون حصار نیست.

### درس 06. گشتاورهای خام

**هدف یادگیری:** چهار گشتاور خام را واقعاً محاسبه و واحدشان را تفسیر کنید.

#### مفهوم و تشریح کامل

گشتاور خام فاصله را از صفر می‌سنجد. توان اول میانگین است؛ توان دوم هم مرکز و هم پراکندگی را در خود دارد. توان سوم و چهارم به مقدارهای بزرگ وزن شدید می‌دهند. بزرگی این اعداد بین مرتبه‌ها قابل مقایسهٔ مستقیم نیست، چون واحد و توان متفاوت است.

#### فرمول و قرارداد

```text
m′r=(1/n)Σxᵢʳ; m′2=m2+x̄²
```

#### مثال حل‌شدهٔ مرجع

در دادهٔ دست: m′1=۲۱٫۵، m′2=۴۶۸٫۵، m′3=۱۰٬۳۵۳٫۵ و m′4=۲۳۲٬۱۵۷٫۳ است. مثلاً برای [۲،۴،۶،۸]، مجموع مربع‌ها ۱۲۰ و m′2=۳۰ است؛ واریانس این نمونه ۵ است، نه ۳۰. رابطهٔ ۳۰−۵²=۵ اختلاف مرکز و پراکندگی را روشن می‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
raw_moments = {order: float(np.mean(x**order)) for order in range(1, 5)}
display(pd.DataFrame({'order': list(raw_moments), 'raw_moment': list(raw_moments.values())}))
assert np.isclose(raw_moments[1], mean_length)

fig, axes = plt.subplots(2,2,figsize=(10,6),layout='constrained')
for order, ax in enumerate(axes.flat,1):
    ax.bar(df.user_id,x**order,color=BLUE)
    ax.axhline(raw_moments[order],color=GOLD,ls='--',label='Raw moment (mean)')
    ax.set(xlabel='User ID',ylabel=f'Length^{order} (cm^{order})',title=f'Raw moment order {order}')
axes[0,0].legend(); plt.show()

say(f'گشتاور خام اول {raw_moments[1]:.2f} همان میانگین است؛ گشتاور خام دوم {raw_moments[2]:.2f} واحد cm² دارد و واریانس نیست: {raw_moments[2]:.2f}−{mean_length**2:.2f}={x.var():.2f}. مقیاس هر پنل جداست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

جدول خروجی مرتبه و میانگین توان را نشان می‌دهد. واحد ستونِ مرتبهٔ چهار cm⁴ است. اگر سانتی‌متر را به میلی‌متر تبدیل کنیم، مرتبهٔ r در ۱۰ʳ ضرب می‌شود؛ پس افزایش عدد، تغییر شکل جامعه را ثابت نمی‌کند. نمودار لگاریتمی نیز فاصله‌ها را به نسبت نشان می‌دهد، نه اختلاف ساده.

#### کاربرد و محدودیت

برای بازیابی واریانس یا چولگی از گشتاورهای خام باید تبدیل‌های جبری را درست انجام داد. در کاربرد محاسباتی، تفریق دو عدد بسیار بزرگ و نزدیک ممکن است دقت عددی را کاهش دهد؛ محاسبهٔ مستقیم فاصله‌ها اغلب مطمئن‌تر است.

#### تمرین

برای [۱،۳] گشتاور خام دوم و واریانس توصیفی را پیدا کنید.

#### پاسخ تشریحی

میانگین ۲، m′2=(۱+۹)/۲=۵ و m2=۵−۴=۱ است. تفاوت این دو مقدار ناشی از مبدأ اندازه‌گیری است.

### درس 07. گشتاورهای مرکزی

**هدف یادگیری:** معنای هندسی و عددیِ توان‌های فاصله از میانگین را یاد بگیرید.

#### مفهوم و تشریح کامل

در گشتاور مرکزی، نخست میانگین را کم می‌کنیم. توان فرد علامت دو سمت را نگه می‌دارد و توان زوج آن را حذف می‌کند. مرکزیِ اول صفر یک هویت جبری است؛ حتی دادهٔ بسیار نامتقارن هم این ویژگی را دارد. مرکزیِ سوم و چهارم هنوز واحددارند و خودِ چولگی و کشیدگی استاندارد نیستند.

#### فرمول و قرارداد

```text
mr=(1/n)Σ(xᵢ−x̄)ʳ; m1=0; m2=variance with denominator n
```

#### مثال حل‌شدهٔ مرجع

در دادهٔ دست جمع مربع فاصله‌ها ۶۲٫۵ است و m2=۶٫۲۵. جمع مکعب‌ها ۱۲۰ است و m3=۱۲. جمع توان چهارم ۱٬۱۵۸٫۶۲۵ است و m4=۱۱۵٫۸۶۲۵. برای مشاهدهٔ ۲۷، فاصله ۵٫۵ و توان چهارم ۹۱۵٫۰۶۲۵ است؛ همین یک فرد حدود ۷۹٪ مجموع توان چهارم را می‌سازد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
central_moments = {order: float(np.mean((x-mean_length)**order)) for order in range(1, 5)}
display(pd.DataFrame({'order': list(central_moments), 'central_moment': list(central_moments.values())}))
assert abs(central_moments[1]) < 1e-12

fig, axes = plt.subplots(2,2,figsize=(10,6),layout='constrained')
for order, ax in enumerate(axes.flat,1):
    ax.bar(df.user_id,(x-mean_length)**order,color=BLUE)
    ax.axhline(0,color='0.3',lw=.8)
    ax.axhline(central_moments[order],color=GOLD,ls='--')
    ax.set(xlabel='User ID',ylabel=f'Deviation^{order} (cm^{order})',title=f'Central moment order {order}')
plt.show()

say(f'گشتاور مرکزی اول {central_moments[1]:.3g}، دوم {central_moments[2]:.3f}، سوم {central_moments[3]:.3f} و چهارم {central_moments[4]:.4f} است. توان زوج همیشه نامنفی و توان سوم جهت‌دار است؛ این اعداد واحدهای متفاوت دارند.')
````

#### راهنمای کامل خواندن خروجی و نمودار

جدول هر مرتبه میانگین سهم‌ها را نمایش می‌دهد. اگر همهٔ طول‌ها را ۱۰ واحد جابه‌جا کنید، میانگین تغییر می‌کند اما فاصله‌های مرکزی ثابت می‌مانند. اگر همه را دو برابر کنید، m2 چهار برابر و m4 شانزده برابر می‌شود. این همان دلیل نیاز به استانداردسازی برای مقایسهٔ شکل است.

#### کاربرد و محدودیت

در آزمایشگاه گشتاورها یک مقدار انتهایی را جابه‌جا کنید و سهم هر نقطه را ببینید. توان‌های بالا برای تشخیص تمرکز سهم مفیدند، اما یک مشاهدهٔ اشتباه می‌تواند آن‌ها را به‌شدت منحرف کند.

#### تمرین

چرا m1=۰، تقارن داده را ثابت نمی‌کند؟

#### پاسخ تشریحی

میانگین دقیقاً طوری تعریف می‌شود که مجموع انحراف‌ها صفر باشد. در [۱،۱،۱۰] نیز این جمع صفر است، با اینکه داده متقارن نیست.

### درس 08. انحراف معیار و بازهٔ طراحی

**هدف یادگیری:** پراکندگی افراد و عدم‌قطعیت میانگین را تفکیک کنید.

#### مفهوم و تشریح کامل

انحراف معیار جذر واریانس است و به واحد اصلی برمی‌گردد. برای توصیف همین داده تقسیم بر n و برای برآورد رایج واریانس جامعه از نمونهٔ مستقل تقسیم بر n−1 به کار می‌رود. خطای استاندارد میانگین s/√n است؛ دربارهٔ نوسان برآورد میانگین صحبت می‌کند، نه اختلاف بین افراد.

#### فرمول و قرارداد

```text
σdescriptive=√(62.5/10)=2.5; s=√(62.5/9)≈2.635; SE(x̄)=s/√10≈0.833
```

#### مثال حل‌شدهٔ مرجع

بازهٔ ۲۱٫۵±۲×۲٫۵ برابر [۱۶٫۵،۲۶٫۵] است؛ با s نمونه‌ای [۱۶٫۲۳،۲۶٫۷۷] می‌شود. هر دو در همین ده نفر فقط ۹ نفر را می‌پوشانند، یعنی ۹۰٪ مشاهده‌ها. در مقابل بازهٔ t برای میانگین با t≈۲٫۲۶۲ حدود [۱۹٫۶۲،۲۳٫۳۸] است، با فرض‌های نمونه‌گیری مناسب.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
sd_sample = x.std(ddof=1)
sd_population = x.std(ddof=0)
print(f'Sample SD = {sd_sample:.3f} cm; descriptive population SD = {sd_population:.3f} cm')
print(f'Mean ± 2 sample SD = [{mean_length-2*sd_sample:.2f}, {mean_length+2*sd_sample:.2f}] cm')

mean_ci = stats.t.interval(.95,len(x)-1,loc=mean_length,scale=sd_sample/np.sqrt(len(x)))
fig,ax=plt.subplots(figsize=(9,3.5))
bounds=[(mean_length-2*sd_sample,mean_length+2*sd_sample),mean_ci]
for row,(a,b) in enumerate(bounds):
    ax.plot([a,b],[row,row],lw=5,color=[BLUE,GOLD][row]); ax.scatter([mean_length],[row],color='black',zorder=3)
ax.set(yticks=[0,1],yticklabels=['Mean ± 2 sample SD (spread)','95% t CI of mean (uncertainty)'],xlabel='Hand length (cm)',title='Spread of observations is not uncertainty of the mean')
plt.tight_layout(); plt.show()

say(f'SD توصیفی={sd_population:.3f} و نمونه‌ای={sd_sample:.3f} سانتی‌متر است؛ SE میانگین={sd_sample/np.sqrt(len(x)):.3f}. بازهٔ t برای میانگین [{mean_ci[0]:.3f}, {mean_ci[1]:.3f}] است؛ این بازه با پراکندگی افراد فرق دارد.')
````

#### راهنمای کامل خواندن خروجی و نمودار

سه چیز متفاوت را جدا بخوانید: دامنهٔ توصیفیِ افراد، بازهٔ اطمینان پارامتر میانگین و بازهٔ پیش‌بینی یک فرد آینده. قاعدهٔ تقریبی ±۲SD به مدل توزیع و جامعه وابسته است؛ برچسب «پوشش قطعی ۹۵٪» روی آن نگذارید.

#### کاربرد و محدودیت

برای اندازهٔ محصول، دامنهٔ کاربران و پوشش چندمتغیره را بررسی کنید. برای مقایسهٔ دو نمونهٔ مستقل دربارهٔ میانگین، عدم‌قطعیت برآورد اهمیت دارد. افزایش n خطای استاندارد را کم می‌کند؛ لزوماً تنوع واقعی دست‌ها را کم نمی‌کند.

#### تمرین

اگر n چهار برابر شود و s ثابت بماند، SE چه تغییری می‌کند؟

#### پاسخ تشریحی

SE نصف می‌شود، چون √(۴n)=۲√n. خود انحراف معیار بین افراد ثابت فرض شده است.

### درس 09. بازهٔ اطمینان Bootstrap برای میانگین

**هدف یادگیری:** بوت‌استرپ را از بازنمونه تا تفسیر بازه دنبال کنید.

#### مفهوم و تشریح کامل

یک بازنمونهٔ بوت‌استرپ n مقدار از دادهٔ موجود با جایگذاری می‌گیرد؛ ممکن است فردی چند بار انتخاب و فردی اصلاً انتخاب نشود. میانگین هر بازنمونه را محاسبه و این کار را تکرار می‌کنیم. توزیع میانگین‌های بازنمونه تقریبِ نوسان برآوردگر است، مشروط به مناسب‌بودن داده و طرح نمونه‌گیری.

#### فرمول و قرارداد

```text
CIpercentile=[quantile(bootstrap means,0.025), quantile(bootstrap means,0.975)]
```

#### مثال حل‌شدهٔ مرجع

کد با بذر ۴۲، ده‌هزار بازنمونهٔ ده‌تایی می‌سازد؛ بازهٔ صدکی ثبت‌شده [۲۰٫۱۰،۲۳٫۱۰] سانتی‌متر است. مثلاً بازنمونهٔ [۱۸،۱۸،۲۰،۲۰،۲۱،۲۱،۲۲،۲۳،۲۷،۲۷] مجاز است و میانگینش ۲۱٫۷ است. این مثال نشان می‌دهد بازنمونه اندازهٔ جدیدی بیرون از داده اختراع نمی‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
boot_means = np.random.default_rng(SEED).choice(x, size=(10000, len(x)), replace=True).mean(axis=1)
boot_ci = np.quantile(boot_means, [.025, .975])
print(f'Bootstrap 95% percentile CI for mean: [{boot_ci[0]:.2f}, {boot_ci[1]:.2f}] cm')
fig, ax = plt.subplots()
ax.hist(boot_means, bins=35, color=TEAL, alpha=.8)
ax.axvline(boot_ci[0], color=GOLD, ls='--'); ax.axvline(boot_ci[1], color=GOLD, ls='--')
ax.set(xlabel='Bootstrap mean (cm)', ylabel='Resamples', title='Bootstrap distribution of the mean (10,000 resamples)')
plt.show()

say(f'بازهٔ صدکی ۹۵٪ از ۱۰٬۰۰۰ بازنمونه [{boot_ci[0]:.2f}, {boot_ci[1]:.2f}] سانتی‌متر است؛ عدم‌قطعیت میانگین همین نمونه را تحت فرض نمونه‌گیری مدل می‌کند، نه پوشش ۹۵٪ دست افراد.')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور افقی نمودار «میانگین بازنمونه» است و محور عمودی تعداد بازنمونه‌ها. دو خط، صدک‌های توزیع میانگین‌ها هستند. اینکه اکثر میانگین‌ها در فاصله‌ای قرار دارند، به معنی قرارداشتن اکثر دست‌های افراد در همان فاصله نیست. در تعبیر فراوانی‌گرایانه، روشِ ساخت بازه در تکرار نمونه‌گیری هدف پوشش پارامتر را دارد.

#### کاربرد و محدودیت

برای دادهٔ ساعتی متوالی، بوت‌استرپ سادهٔ تک‌سطرها وابستگی زمانی را نادیده می‌گیرد؛ باید طرحی مثل بلوکی متناسب با مسئله در نظر گرفت. افزایش بازنمونه از ۲هزار به ۱۰هزار، سوگیری انتخاب افراد را اصلاح نمی‌کند.

#### تمرین

آیا تکرار بوت‌استرپ یک میلیون بار، نمونهٔ ده‌نفره را به یک میلیون فرد تبدیل می‌کند؟

#### پاسخ تشریحی

خیر. اطلاعات مستقل همچنان از همان ده نفر می‌آید. فقط خطای محاسباتیِ تقریب بازنمونه‌ای کمتر می‌شود.

### درس 10. تابع توزیع تجمعی تجربی ECDF

**هدف یادگیری:** سهم پوشش را مستقیم از داده بخوانید.

#### مفهوم و تشریح کامل

تابع تجمعی تجربی در آستانهٔ t، تعداد مشاهده‌های ≤t را بر n تقسیم می‌کند. نمودار پله‌ای است: بین مشاهده‌ها تغییر نمی‌کند و در مقدارهای تکراری جهش بزرگ‌تر دارد. این ویژگی برای خواندن پوشش یک آستانه و مقایسهٔ توزیع‌ها مفید است.

#### فرمول و قرارداد

```text
F̂(t)=#{xᵢ≤t}/n
```

#### مثال حل‌شدهٔ مرجع

در دادهٔ دست F̂(۲۰)=۴/۱۰=۰٫۴، F̂(۲۱)=۶/۱۰=۰٫۶، F̂(۲۲)=۷/۱۰=۰٫۷ و F̂(۲۳)=۸/۱۰=۰٫۸ است. پس عبارت قدیمی «۵۰٪ کمتر از ۲۱٫۵» دقیق نیست؛ شش نفر یعنی ۶۰٪ کمتر یا مساوی ۲۱٫۵ هستند. میانهٔ ۲۱ با تعریف موقعیت مرکزی سازگار است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
sorted_x = np.sort(x)
ecdf_y = np.arange(1, len(x)+1)/len(x)
fig, ax = plt.subplots()
ax.step(sorted_x, ecdf_y, where='post', color=BLUE)
ax.scatter(sorted_x, ecdf_y, color=BLUE)
ax.set(xlabel='Hand length threshold (cm)', ylabel='Share ≤ threshold', ylim=(0,1.05),
       title='Empirical cumulative distribution')
plt.show()
print('Share with length ≤ 22 cm:', (x<=22).mean())

say(f'در آستانهٔ ۲۱ سهم تجمعی {(x<=21).mean():.0%} و در ۲۲ سهم {(x<=22).mean():.0%} است. پله‌ها دادهٔ دیده‌شده را توصیف می‌کنند؛ ECDF را از صفر تا یک دنبال کنید.')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور X یک آستانهٔ طول و محور Y سهم تجمعی از صفر تا یک است. شیب‌نداشتن پله بین ۲۴ و ۲۷ به معنی نبود مشاهده در آن فاصله است، نه نبودِ ممکن آن اندازه در جامعه. از روی این ده نفر سهم‌های بازار را با دقت درصدی گزارش نکنید.

#### کاربرد و محدودیت

اگر ابعاد یک ابزار تنها برای طول‌های حداکثر ۲۲ مناسب فرض شده باشد، ECDF پوششِ تک‌بعدی ۷۰٪ همین نمونه را نشان می‌دهد. سازگاری هم‌زمان طول و عرض، یک رویداد مشترک است و باید جدا محاسبه شود.

#### تمرین

اگر طول ۲۰ دو بار تکرار شود، پلهٔ ECDF در ۲۰ با n=۱۰ چقدر بالا می‌رود؟

#### پاسخ تشریحی

۰٫۲ یا ۲۰ واحد درصد. هر فرد یک‌دهم به توزیع تجربی اضافه می‌کند.

### درس 11. چولگی

**هدف یادگیری:** چولگی را از سهمِ مکعب فاصله‌ها بفهمید.

#### مفهوم و تشریح کامل

چولگی گشتاوری، سومین گشتاور مرکزی را به پراکندگی استاندارد می‌کند تا بی‌واحد شود. سمت راستِ میانگین سهم مکعبی مثبت و سمت چپ سهم منفی دارد. غالب‌شدن سهم راست، ضریب مثبت می‌دهد. نزدیک صفر می‌تواند از تقارن یا خنثی‌شدن الگوهای پیچیده‌تر حاصل شود.

#### فرمول و قرارداد

```text
g1=m3/m2^(3/2)=12/6.25^(3/2)=0.768
```

#### مثال حل‌شدهٔ مرجع

در دادهٔ ده‌نفره g1=۰٫۷۶۸ است؛ برآورد اصلاح‌شدهٔ نمونه‌ای حدود ۰٫۹۱۱ می‌شود. این اختلاف خطا نیست؛ قرارداد تصحیح فرق می‌کند. مثال [۲،۴،۶،۱۸] چولگی حدود +۰٫۹۷۹ دارد؛ بازتاب آن [۲،۱۴،۱۶،۱۸] مقدار −۰٫۹۷۹ می‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
moment_skew = central_moments[3] / central_moments[2]**1.5
adjusted_skew = stats.skew(x, bias=False)
print(f'Moment skewness = {moment_skew:.3f}; adjusted sample skewness = {adjusted_skew:.3f}')

z=(x-x.mean())/x.std(ddof=0)
fig,ax=plt.subplots(); ax.bar(df.user_id,z**3,color=BLUE); ax.axhline(0,color='0.3')
ax.set(xlabel='User ID',ylabel='Standardized deviation cubed',title=f'Signed contributions: mean(z³)={moment_skew:.3f}')
plt.show()

say(f'چولگی گشتاوری={moment_skew:.3f} و اصلاح‌شده={adjusted_skew:.3f}. علامت مثبت از سهم توان سوم فاصله‌های بزرگِ سمت راست می‌آید؛ اختلاف دو برآوردگر در n=10 محسوس است.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در هیستوگرام، به گستردگی انتهاها و محل تجمع نگاه کنید؛ جهت چولگی با محل بیشترین ستون یکی نیست. در دادهٔ دوچرخه هم ضریب کل مثبت است، اما باید با فیلتر هر زیرگروه دوباره محاسبه شود. مقدار مرکزی سومِ واحددار را با g1 مقایسه نکنید.

#### کاربرد و محدودیت

چولگی مثبت انگیزه‌ای برای بررسی کاربران انتهایی است، اما تعداد سایزهای محصول را تعیین نمی‌کند. ممکن است یک محصول قابل تنظیم یا چند اندازه مناسب باشد؛ این انتخاب به هزینه و آزمون کاربرد وابسته است.

#### تمرین

با دو برابرکردن همهٔ xها، m3 هشت برابر می‌شود. آیا چولگی هم هشت برابر می‌شود؟

#### پاسخ تشریحی

خیر. m2^(۳/۲) نیز هشت برابر می‌شود و نسبت ثابت می‌ماند. تغییر مقیاس مثبت، چولگی استاندارد را تغییر نمی‌دهد.

### درس 12. کشیدگی

**هدف یادگیری:** کشیدگی را با حساسیت به فاصله‌های انتهایی تفسیر کنید.

#### مفهوم و تشریح کامل

توان چهارم همهٔ فاصله‌ها را نامنفی و مشاهده‌های دور را بسیار پراثر می‌کند. کشیدگی استاندارد m4/m2² و کشیدگی اضافی آن منهای ۳ است. مقدار ۳ مرجع نرمال در این قرارداد است؛ برابری کشیدگی با نرمال، برابری کل توزیع‌ها را نتیجه نمی‌دهد.

#### فرمول و قرارداد

```text
β2=115.8625/6.25²≈2.96608; excess=β2−3≈−0.03392
```

#### مثال حل‌شدهٔ مرجع

دادهٔ دست کشیدگی گشتاوری نزدیک ۲٫۹۶۶ دارد؛ برآورد اصلاح‌شدهٔ کشیدگی اضافی حدود ۰٫۹۰۴ است. برای [۲،۴،۶،۸]، β2=۱٫۶۴ و excess=−۱٫۳۶ است. در [۱،۲،۳،۴،۵،۶،۷،۸،۹،۳۰]، β2≈۶٫۷۰ می‌شود. این نمونه‌ها مصنوعی و برای دیدن اثر فاصلهٔ دورند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
moment_excess = central_moments[4]/central_moments[2]**2 - 3
adjusted_excess = stats.kurtosis(x, fisher=True, bias=False)
print(f'Moment excess kurtosis = {moment_excess:.3f}; adjusted sample estimate = {adjusted_excess:.3f}')

fig,ax=plt.subplots(); ax.bar(df.user_id,z**4,color=GOLD)
ax.axhline(moment_excess+3,color='0.3',ls='--',label='Mean z^4')
ax.set(xlabel='User ID',ylabel='Standardized deviation to fourth power',title='Kurtosis: tail contributions, not just peak height')
ax.legend(); plt.show()

say(f'کشیدگی گشتاوری={moment_excess+3:.5f}، اضافی={moment_excess:.5f} و اضافی اصلاح‌شده={adjusted_excess:.3f}. نزدیک‌بودن اولین مقدار به ۳، اثبات نرمال‌بودن نیست؛ پنل سهم z⁴ حساسیت به انتهاها را نشان می‌دهد.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در جدول یا نمودار، اول مشخص کنید «کشیدگی» نمایش داده شده یا «اضافی». صفرِ اضافی با ۳ِ معمولی یک مرجع دارد. فقط از ارتفاع قله نمی‌توان این ضریب را فهمید؛ بهتر است سهم نواحی دور از میانگین در مجموع توان چهارم را هم دید، مانند آزمایشگاه گشتاورها.

#### کاربرد و محدودیت

در دادهٔ کم، یک مشاهدهٔ معتبر یا خطای ثبت می‌تواند ضریب را جابه‌جا کند. گزارش تحلیل حساسیت و تعریف دقیق ضریب از برچسب‌های قطعی «نرمال» یا «غیرنرمال» مفیدتر است.

#### تمرین

اگر β2=۴٫۴۱۶ باشد، کشیدگی اضافی چقدر است و چه چیزی را ثابت نمی‌کند؟

#### پاسخ تشریحی

۱٫۴۱۶ است. این عدد حساسیت بیشترِ گشتاور چهارم نسبت به مرجع را توصیف می‌کند؛ به‌تنهایی علت، شکل کامل توزیع یا نتیجهٔ آزمون نرمال‌بودن نیست.

### درس 13. نمودار Q–Q

**هدف یادگیری:** صدک مشاهده و صدک نظری را مقایسه کنید.

#### مفهوم و تشریح کامل

Q–Q دادهٔ مرتب‌شده را مقابل صدک‌های یک توزیع مرجع قرار می‌دهد. برای مرجع نرمال، اگر توزیع نمونه با تغییر مکان و مقیاس به آن شبیه باشد، نقاط تقریباً خطی می‌شوند. خمیدگی و انحراف انتهاها دربارهٔ ناسازگاری شکل سرنخ می‌دهند.

#### فرمول و قرارداد

```text
محور افقی: quantile of N(0,1)؛ محور عمودی: ordered observed x
```

#### مثال حل‌شدهٔ مرجع

این شکل فقط ده نقطه دارد و بالاترین مقدار آن ۲۷ است. یک انحراف انتهایی با این n می‌تواند ناشی از نوسان نمونه‌گیری یا الگوی واقعی باشد؛ برای تصمیم، دادهٔ بیشتر و منطق اندازه‌گیری لازم است. اگر تمام طول‌ها ۲ واحد زیاد شوند، شکل Q–Q عموداً جابه‌جا می‌شود ولی الگوی خمیدگی حفظ می‌شود.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
fig, ax = plt.subplots()
stats.probplot(x, dist='norm', plot=ax)
ax.set(xlabel='Theoretical normal quantiles', ylabel='Observed hand length (cm)',
       title='Normal Q–Q plot for hand length')
plt.show()

say('نزدیکی نقاط Q–Q به خط، سازگاری بصری با نرمال را بررسی می‌کند. انحراف انتهاها و تعداد فقط ده مشاهده را با هم ببینید؛ نمودار آزمون قطعی نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور X نمرهٔ نظریِ بی‌واحد و محور Y سانتی‌متر است. خط مرجع «خط برابری x=y» نیست، چون واحدها یکی نیستند. انحراف از خط را نسبت به پراکندگی و تعداد نقاط بسنجید؛ یک شکل شبیه خط، اثبات نرمال‌بودن جامعه نیست.

#### کاربرد و محدودیت

در استنباط رگرسیون، نرمال‌بودن خطاها موضوع اصلی است؛ برای آن Q–Q باقیمانده‌ها را بررسی کنید. نرمال‌بودنِ خود X شرط عمومیِ رگرسیون خطی نیست.

#### تمرین

آیا Q–Q طول دست می‌تواند جای Q–Q باقیماندهٔ مدل عرض بر طول را بگیرد؟

#### پاسخ تشریحی

خیر. این دو دربارهٔ دو متغیر متفاوت‌اند؛ فرض خطای مدل را با باقیمانده‌ها و طراحی داده‌برداری بررسی می‌کنیم.

### درس 14. نقشهٔ همبستگی

**هدف یادگیری:** ماتریس همبستگی و حدود نتیجه‌گیری آن را بخوانید.

#### مفهوم و تشریح کامل

هر خانه رابطهٔ خطیِ دو ستون را خلاصه می‌کند. ماتریس متقارن است و قطر برای ستونِ دارای پراکندگی مقدار ۱ دارد. رنگ فقط راهنمای الگو است؛ مقدار عددی، حجم نمونه و نمودار پراکندگی باید همراهش خوانده شوند. همبستگی صفر نیز استقلال کامل متغیرها را ثابت نمی‌کند.

#### فرمول و قرارداد

```text
rXY=Σ(x−x̄)(y−ȳ)/√[Σ(x−x̄)²Σ(y−ȳ)²]
```

#### مثال حل‌شدهٔ مرجع

در خروجی گرد‌شده، طول و عرض r≈۰٫۹۹، سن و سابقه r≈۰٫۹۸، طول و سن r≈۰٫۴۳ و طول و سابقه r≈۰٫۵۸ دارند. بنابراین ادعای اولیهٔ استقلال ابعاد دست از سن از همین جدول به دست نمی‌آید. سن و سابقهٔ بسیار همبسته می‌توانند در مدل چندگانه هم‌خطی ایجاد کنند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
numeric_cols = ['hand_length_cm','hand_width_cm','age_years','gaming_years']
corr = df[numeric_cols].corr()
display(corr.round(2))
fig, ax = plt.subplots(figsize=(6,5))
im = ax.imshow(corr, vmin=-1, vmax=1, cmap='BrBG')
ax.set_xticks(range(4), numeric_cols, rotation=30, ha='right')
ax.set_yticks(range(4), numeric_cols)
for i in range(4):
    for j in range(4): ax.text(j,i,f'{corr.iloc[i,j]:.2f}',ha='center',va='center',bbox=dict(facecolor='white',alpha=.85,edgecolor='none'))
fig.colorbar(im, ax=ax, label='Pearson r')
ax.set_title('Correlation among measured attributes')
plt.tight_layout(); plt.show()

say(f'پیرسون طول–عرض={corr.iloc[0,1]:.3f} و سن–سابقه={corr.iloc[2,3]:.3f} است. وابستگی شدید پیش‌بین‌ها می‌تواند ضرایب رگرسیون چندگانه را ناپایدار کند؛ رابطهٔ عددی علّیت نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در نقشه، قطر را برای کشف رابطهٔ تازه تفسیر نکنید؛ هر متغیر با خودش مقایسه شده است. خانهٔ طول–عرض را با نقاط ده کاربر کنترل کنید؛ یک مشاهدهٔ انتهایی می‌تواند r را بالا ببرد. برای رابطهٔ یکنواختِ غیرخطی، اسپیرمن را هم ببینید.

#### کاربرد و محدودیت

در پروژهٔ همبستگی، دما با اجاره روی ۱۷٬۳۷۹ ساعت بررسی می‌شود. وابستگی زمان و ترکیب فصل‌ها همچنان مهم است؛ زیادبودن تعداد سطر، همهٔ اشکال نمونه‌گیری را برطرف نمی‌کند.

#### تمرین

اگر طول و عرض r≈۰٫۹۹ داشته باشند، آیا اندازه‌گیری عرض برای طراحی قابل حذف است؟

#### پاسخ تشریحی

فقط با این شاهد خیر. برای پوشش موارد انتهایی، خطای اندازه‌گیری و کاربران جدید باید خطای پیش‌بینی و هزینهٔ حذف ویژگی مستقلاً ارزیابی شود.

## فصل: استنباط و مدل خطی

### درس 15. منطق استنباط، اندازهٔ اثر، توان و چندمقایسه‌ای

**هدف یادگیری:** قبل از خواندن p-value، پرسش، برآورد و هزینهٔ تصمیم را روشن کنید.

#### مفهوم و تشریح کامل

آمار توصیفی دربارهٔ دادهٔ دیده‌شده است و استنباط به جامعه‌ای فراتر از آن مربوط می‌شود. این گذار به نمونه‌گیری و مدل وابسته است. α آستانهٔ تصمیمِ از پیش تعیین‌شده و p سازگاری داده با فرض صفر در همان چارچوب است. اندازهٔ اثر می‌گوید اختلاف چقدر است؛ بازهٔ اطمینان دقت برآورد را نشان می‌دهد. توان، احتمال کشف اثرِ تعریف‌شده در یک طرح مشخص است و با حجم نمونه، پراکندگی و آستانه تغییر می‌کند.

#### فرمول و قرارداد

```text
گزارش کامل = برآورد اثر + واحد + بازه + n + روش + طرح داده‌برداری
```

#### مثال حل‌شدهٔ مرجع

فرض کنید راحتیِ دو ماوس در مقیاس۱۰ نمره، میانگین‌های۷٫۱ و۷٫۲ داشته باشد. حتی اگر با نمونهٔ بسیار بزرگ p کوچک شود، اختلاف۰٫۱ ممکن است ارزش عملی نداشته باشد. برعکس، اختلاف۲ نمره‌ای در نمونهٔ کم می‌تواند بازهٔ پهن و p بزرگ داشته باشد. برای مثال ده دست، اختلاف گروه تجربه≈۲٫۰۸۳ سانتی‌متر است؛ عدم رد آن را صفر نمی‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from statsmodels.stats.power import TTestIndPower
power_model=TTestIndPower(); sizes=np.arange(5,151,5)
powers={d:power_model.power(effect_size=d,nobs1=sizes,alpha=.05,ratio=1) for d in [.2,.5,.8]}
family_error=1-(1-.05)**20
fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained')
for d,curve in powers.items(): axes[0].plot(sizes,curve,label=f'Cohen d={d}',ls={.2:':',.5:'--',.8:'-'}[d])
axes[0].axhline(.8,color='0.3',ls='--'); axes[0].set(xlabel='n per group (independent, equal n)',ylabel='Power',ylim=(0,1),title='Two-sided t-test planning: model-based power'); axes[0].legend()
tests=np.arange(1,31); axes[1].plot(tests,1-.95**tests,color=GOLD); axes[1].set(xlabel='Number of independent true-null tests',ylabel='P(at least one false rejection)',ylim=(0,1),title='Family-wise error without correction')
plt.show()
say(f'با ۲۰ آزمون مستقلِ صفرِ درست، احتمال حداقل یک رد اشتباه {family_error:.1%} است. در مدل برنامه‌ریزی t با d=0.5 و n=30 در هر گروه، توان {power_model.power(.5,30,alpha=.05):.1%} است. این توان پیشینیِ مدل است؛ نه توانِ اثبات‌شدهٔ نمونهٔ دست. p کوچک اندازهٔ عملی اثر را تعیین نمی‌کند.')
````

#### راهنمای کامل خواندن خروجی و نمودار

ابتدا ستون برآورد و بازه را بخوانید، سپس p و سطح تصمیم را. «برابرند»، «اثر ندارد» و «فرض صفر ثابت شد» از عدم رد نتیجه نمی‌شوند. برای ادعای هم‌ارزی باید حاشیهٔ هم‌ارزیِ کاربردی از پیش تعیین و آزمون متناسب انجام شود. اگر۲۰ آزمون مستقل درα=۰٫۰۵ و با همهٔ فرض‌های صفر درست انجام شوند، احتمال حداقل یک رد حدود۱−۰٫۹۵²⁰≈۶۴٪ است؛ استقلال در این مثال یک فرض محاسباتی است.

#### کاربرد و محدودیت

پیش از مشاهدهٔ خروجی، پاسخ اصلی، مقایسه‌ها و حداقل اختلاف مهم را بنویسید. تحلیل‌های اکتشافی بعدی را با همین عنوان گزارش کنید. توافق چند آزمون روی همان ده نفر، شاهد مستقل چندباره ایجاد نمی‌کند.

#### تمرین

آیا p=۰٫۰۱ یعنی اثر حتماً بزرگ است؟

#### پاسخ تشریحی

خیر. p به n و پراکندگی و مدل وابسته است. اندازه و واحد اثر و بازهٔ برآورد باید جدا بررسی شوند.

### درس 16. ماتریس پراکندگی و بررسی رابطهٔ غیرخطی

**هدف یادگیری:** پشت هر خانهٔ همبستگی، نقاط واقعی را ببینید.

#### مفهوم و تشریح کامل

در ماتریس پراکندگی، هر خانه یک زوج ویژگی را رسم می‌کند؛ محورهای سطر و ستون تعیین می‌کنند چه چیزی مقایسه می‌شود. خانه‌های دو سوی قطر اطلاعات جفتی مشابه با محورهای جابه‌جا دارند. قطر را می‌توان با هیستوگرام همان متغیر پر کرد. این نمایش، رابطهٔ خمیده، تجمع، ties و نقاط اثرگذار را از یک عدد همبستگی بهتر آشکار می‌کند.

#### فرمول و قرارداد

```text
چهار ویژگی → ۴×۴ خانه؛ تعداد زوج‌های یکتا = 4×3/2 = 6
```

#### مثال حل‌شدهٔ مرجع

در دادهٔ نمونه، دو زوج طول–عرض و سن–سابقه همبستگی نزدیک۱ دارند. اما در زوج طول–سن r≈۰٫۴۳ است و تنها ده نقطه وجود دارد. در مثال جداگانه X=[−۲،−۱،۰،۱،۲] وY=[۴،۱،۰،۱،۴]، پیرسون صفر می‌شود ولی شکلU کاملاً مشخص است. بنابراین نقشهٔ رنگیِ نزدیک صفر ممکن است یک رابطهٔ مهم را پنهان کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from pandas.plotting import scatter_matrix
scatter_matrix(df[numeric_cols],diagonal='hist',figsize=(10,10),alpha=.8,color=BLUE)
plt.suptitle('Four measured attributes: pairwise observations and histograms',y=1.01); plt.show()
u=np.arange(-2,3); v=u**2
fig,ax=plt.subplots(); ax.scatter(u,v,color=GOLD,s=70); ax.set(xlabel='X (constructed)',ylabel='Y = X²',title=f'Nonlinear dependence despite Pearson r={stats.pearsonr(u,v).statistic:.2f}'); plt.show()
say(f'چهار ویژگی شش زوج یکتا می‌سازند. در مثال ساخته‌شدهٔ Uشکل، r={stats.pearsonr(u,v).statistic:.2f} ولی Y دقیقاً تابع X است؛ پس همبستگی صفر به معنی نبود هر رابطه نیست. مقیاس هر محور را جدا بخوانید.')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور هر خانه را از برچسب سطر و ستون بخوانید. مقیاس‌ها بین متغیرهای متفاوت یکی نیستند. در شکل طول–عرض، تراکم دو نقطهٔ برابر ممکن است یک نقطه دیده شود؛ جدول شناسه‌ها را هم بررسی کنید. خط اتصالِ بین افراد لازم نیست.

#### کاربرد و محدودیت

برای انتخاب مدل، از این بررسیِ اکتشافی سرنخ بگیرید ولی همان دادهٔ آزمون نهایی را بارها برای انتخاب ویژگی و شکل مدل مصرف نکنید. نمودار به تشخیص کمک می‌کند؛ به‌تنهایی مدل علّی نمی‌سازد.

#### تمرین

برای شش ویژگی چند زوج یکتا وجود دارد؟

#### پاسخ تشریحی

۶×۵÷۲=۱۵ زوج. تعداد خانه‌های کل۳۶ است، اما قطر و نیمهٔ تکراری رابطهٔ تازه‌ای اضافه نمی‌کنند.

### درس 17. رگرسیون خطی ساده و باقیمانده‌ها

**هدف یادگیری:** از خط برازش به پیش‌بینی و باقیمانده برسید.

#### مفهوم و تشریح کامل

کمترین مربعات خطی را انتخاب می‌کند که مجموع مربعِ فاصله‌های عمودی Y از خط کمینه شود. شیب در واحد پاسخ به ازای یک واحد X است. باقیمانده اختلاف مشاهده و برازش است؛ مثبت یعنی مدل Y را کمتر از مقدار واقعی پیش‌بینی کرده است.

#### فرمول و قرارداد

```text
ŷ=0.104+0.444x; e=y−ŷ; R²=1−SSE/SST
```

#### مثال حل‌شدهٔ مرجع

برای طول ۲۲ سانتی‌متر، عرض پیش‌بینی‌شده حدود ۹٫۸۷۲ است. عرض ثبت‌شدهٔ همان فرد ۱۰ است، پس با ضرایب گرد‌شده e≈۰٫۱۲۸ سانتی‌متر. R² آموزشی≈۰٫۹۸۴ است؛ یعنی خط نسبت به مدل میانگین بخش بزرگی از تغییر Yِ همین داده را بازسازی می‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
import statsmodels.api as sm
simple_fit = sm.OLS(df.hand_width_cm, sm.add_constant(df.hand_length_cm)).fit()
print(pd.DataFrame({'term':simple_fit.params.index, 'estimate':simple_fit.params.values}).to_string(index=False))
print(f'In-sample R² = {simple_fit.rsquared:.3f}')
fig, axes = plt.subplots(1, 2, figsize=(11,4))
axes[0].scatter(x, df.hand_width_cm, color=BLUE)
grid = np.linspace(x.min(),x.max(),100)
axes[0].plot(grid, simple_fit.params.iloc[0]+simple_fit.params.iloc[1]*grid, color=GOLD)
axes[0].set(xlabel='Hand length (cm)', ylabel='Hand width (cm)', title='Linear fit')
axes[1].scatter(simple_fit.fittedvalues, simple_fit.resid, color=TEAL)
axes[1].axhline(0, color='black', lw=1)
axes[1].set(xlabel='Fitted width (cm)', ylabel='Residual (cm)', title='Residual check')
plt.tight_layout(); plt.show()

say(f'عرضِ برازش‌شده={simple_fit.params.iloc[0]:.3f}+{simple_fit.params.iloc[1]:.3f}×طول؛ R² آموزشی={simple_fit.rsquared:.4f}. شیب واحد cm/cm دارد؛ باقیمانده‌ها باید جداگانه بررسی شوند و این R² ارزیابیِ فرد جدید نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در نمودار اول محور X طول و Y عرض است. در نمودار باقیمانده، محور افقی برازش و عمودی e است؛ به خمیدگی، قیف، نقاط اثرگذار و وابستگی زمانی نگاه کنید. نبودِ الگوی آشکار در ده نقطه به معنی اثبات همهٔ فرض‌ها نیست. R² بالا هم خطای کاربران جدید را گزارش نمی‌کند.

#### کاربرد و محدودیت

عرض دستِ پیش‌بینی‌شده، عرض بدنهٔ ماوس نیست. برای تبدیل آن به طراحی باید یک رابطهٔ ارگونومیک مجزا را با آزمون کاربری برقرار کنید. بیرون بازهٔ ۱۸ تا ۲۷ نیز برون‌یابیِ همین خط قابل اتکای خودکار نیست.

#### تمرین

اگر y=۹ و ŷ=۹٫۴ باشد، باقیمانده چیست و جهت خطا کدام است؟

#### پاسخ تشریحی

e=−۰٫۴ سانتی‌متر؛ مدل مقدار عرض را ۰٫۴ بیش‌برآورد کرده است.

### درس 18. آزمون Shapiro–Wilk

**هدف یادگیری:** p-value و عدم رد را درست تعبیر کنید.

#### مفهوم و تشریح کامل

Shapiro–Wilk ناسازگاری داده با مدل نرمال را با آمارهٔ W می‌سنجد. فرض صفر، مدل نرمال است. p احتمال درست‌بودن فرض صفر نیست؛ با فرضِ آن مدل و دیگر شروط، میزان غیرعادی‌بودن شواهد مشاهده‌شده را توصیف می‌کند. در نمونهٔ بسیار کوچک قدرت کشف انحراف محدود است.

#### فرمول و قرارداد

```text
H0: distribution is normal; تصمیم آموزشی با α=0.05 از پیش تعیین‌شده
```

#### مثال حل‌شدهٔ مرجع

طول دست W≈۰٫۹۴۸ و p≈۰٫۶۴۳۵ دارد. باقیمانده‌های رگرسیون W≈۰٫۹۵۶ و p≈۰٫۷۳۷۷ دارند. در سطح ۰٫۰۵ هیچ‌کدام رد نمی‌شوند. جملهٔ قابل دفاع: «شواهد کافی برای رد مدل نرمال در این نمونه نداریم»، نه «نرمال‌بودن ثابت شد».

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
for name, values in [('Hand length',x), ('Simple-regression residuals', simple_fit.resid)]:
    result = stats.shapiro(values)
    print(f'{name}: W={result.statistic:.3f}, p={result.pvalue:.4f}')

fig,axes=plt.subplots(1,2,figsize=(10,4),layout='constrained')
stats.probplot(x,dist='norm',plot=axes[0]); stats.probplot(simple_fit.resid,dist='norm',plot=axes[1])
axes[0].set(title='Q–Q: observed length',ylabel='Length (cm)'); axes[1].set(title='Q–Q: regression residuals',ylabel='Residual (cm)')
plt.show()

say(f'p شاپیرو برای طول={stats.shapiro(x).pvalue:.4f} و برای باقیمانده={stats.shapiro(simple_fit.resid).pvalue:.4f}. در α=0.05 رد نمی‌کنیم؛ توان کم ده مشاهده اجازهٔ اثبات نرمالیت نمی‌دهد. در رگرسیون فرض اصلی به خطا مربوط است، نه الزاماً X.')
````

#### راهنمای کامل خواندن خروجی و نمودار

W نزدیک ۱ معمولاً سازگاری بیشتر را نشان می‌دهد، اما مقدار بحرانی و p به n وابسته‌اند. شکل Q–Q و علل تولید داده را کنار عدد بخوانید. برای دادهٔ بسیار بزرگ، انحراف ناچیز می‌تواند p کوچک بدهد؛ برای دادهٔ کم، انحراف مهم ممکن است کشف نشود.

#### کاربرد و محدودیت

انتخاب روش را صرفاً به یک آزمون مقدماتی نسپارید. پرسش، استقلال، اندازهٔ اثر، مقاومت روش و تشخیص باقیمانده‌ها را هم در نظر بگیرید.

#### تمرین

اگر p=۰٫۶۴ باشد، آیا احتمال نرمال‌بودن ۶۴٪ است؟

#### پاسخ تشریحی

خیر. p احتمال فرضیه نیست. آن عدد تحت فرض صفر و شروط محاسبه تعریف شده و به‌تنهایی احتمال درستی مدل را نمی‌دهد.

### درس 19. بازهٔ اطمینان ۹۵٪ شیب

**هدف یادگیری:** اندازه و عدم‌قطعیت شیب را هم‌زمان گزارش کنید.

#### مفهوم و تشریح کامل

برآورد شیب با تغییر نمونه نوسان می‌کند. خطای استاندارد شیب از پراکندگی باقیمانده و گستردگی X به دست می‌آید. با همان خطای مدل، گسترده‌تر بودن X اطلاعات بیشتری دربارهٔ شیب فراهم می‌کند. بازهٔ t به فرض‌های مدل و استقلال وابسته است.

#### فرمول و قرارداد

```text
CI(b)=b±t(0.975,n−2)×SE(b); SE(b)=√[SSE/(n−2)] / √Σ(x−x̄)²
```

#### مثال حل‌شدهٔ مرجع

شیب ۰٫۴۴۴ و بازهٔ ۹۵٪ آن تقریباً [۰٫۳۹۷،۰٫۴۹۱] است. واحد: سانتی‌متر عرض به ازای یک سانتی‌متر طول. برای اختلاف ۴ سانتی‌متری X، اختلاف متوسط برازش‌شده ۱٫۷۷۶ سانتی‌متر است؛ این محاسبه دربارهٔ تغییر میانگین مدل است و دامنهٔ Y یک فرد را نمی‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
slope_name = 'hand_length_cm'
slope_ci = simple_fit.conf_int(alpha=.05).loc[slope_name]
print(f'Slope = {simple_fit.params[slope_name]:.3f} cm width / cm length')
print(f'95% CI = [{slope_ci.iloc[0]:.3f}, {slope_ci.iloc[1]:.3f}]')

fig,ax=plt.subplots(figsize=(8,2.8)); ax.errorbar([simple_fit.params[slope_name]],[0],xerr=[[simple_fit.params[slope_name]-slope_ci.iloc[0]],[slope_ci.iloc[1]-simple_fit.params[slope_name]]],fmt='o',capsize=8,color=BLUE)
ax.axvline(0,color='0.4',ls='--'); ax.set(xlabel='Slope (cm width / cm length)',yticks=[],title='Slope and model-based 95% confidence interval'); plt.show()

say(f'شیب {simple_fit.params[slope_name]:.3f} و بازهٔ ۹۵٪ [{slope_ci.iloc[0]:.3f}, {slope_ci.iloc[1]:.3f}] است؛ صفر بیرون این بازه است، تحت فرض‌های مدل. این بازه برای شیب است، نه پیش‌بینی عرض یک فرد.')
````

#### راهنمای کامل خواندن خروجی و نمودار

بازه صفر را شامل نمی‌کند و با جهت مثبت رابطه سازگار است. عرض بازه دقت برآوردِ مشروط به مدل را نشان می‌دهد. نباید آن را با بازهٔ پیش‌بینی عرضِ یک فرد اشتباه گرفت؛ برای پیش‌بینی فردی، خطای فرد نیز وارد می‌شود.

#### کاربرد و محدودیت

در گزارش طراحی جملهٔ «شیب مثبت با این بازه» را کنار نمودار، n و بازهٔ مشاهده‌شدهٔ X بنویسید. از آن برای اثبات رابطهٔ علّی میان ویژگی‌های فیزیکی یا تبدیل مستقیم به ابعاد ماوس استفاده نکنید.

#### تمرین

دو شیب برابرند ولی بازهٔ یکی بسیار پهن‌تر است. چه تفاوتی در دانش ما وجود دارد؟

#### پاسخ تشریحی

برآورد نقطه‌ای یکسان است، اما دقت برآورد دوم کمتر است. صرف مقایسهٔ شیب‌های گرد‌شده این عدم‌قطعیت را پنهان می‌کند.

### درس 20. ANOVA یک‌طرفه

**هدف یادگیری:** ANOVA را با نسبتِ تغییر بین و درون گروه بفهمید.

#### مفهوم و تشریح کامل

آزمون کلیِ ANOVA می‌پرسد آیا میانگین همهٔ گروه‌ها برابر است. تغییر کل به جزء بین گروه و درون گروه تجزیه می‌شود. اگر اختلاف مراکز نسبت به نوسان درون گروه‌ها بزرگ باشد، F افزایش می‌یابد. فرض استقلال را از طراحی می‌گیریم؛ آزمون نرمال‌بودن آن را بررسی نمی‌کند.

#### فرمول و قرارداد

```text
F=(SSbetween/(k−1))/(SSwithin/(N−k)); H0: μ1=μ2=μ3
```

#### مثال حل‌شدهٔ مرجع

گروه جوانِ سن≤۲۲ شامل طول‌های [۱۸،۲۰،۲۲] با میانگین ۲۰ است؛ میانیِ سن۲۳ تا۲۷ شامل [۱۹،۲۱،۲۳،۲۷] با میانگین۲۲٫۵؛ بالاترِ سن≥۲۸ شامل [۲۰،۲۱،۲۴] با میانگین≈۲۱٫۶۶۷ است. F≈۰٫۷۳۴ و p≈۰٫۵۱۳۶ با درجات آزادی۲ و۷ به دست آمده است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
age_group = np.select([df.user_id.isin([1,3,7]), df.user_id.isin([2,5,8,10])],
                      ['young','middle'], default='older')
df['age_group'] = age_group
groups_age = [df.loc[df.age_group==g,'hand_length_cm'].to_numpy() for g in ['young','middle','older']]
f_stat, p_anova = stats.f_oneway(*groups_age)
display(df.groupby('age_group').hand_length_cm.agg(['count','mean','std']).round(3))
print(f'F={f_stat:.3f}, p={p_anova:.4f}')

group_figure(groups_age,['young','middle','older'],'Length by age group (observations, mean and SD)')

say(f'ANOVA: F={f_stat:.3f} و p={p_anova:.4f}. با وجود اختلاف میانگین‌های نمونه، شواهد کافی برای رد برابری میانگین‌ها در α=0.05 نداریم؛ عدم رد به معنی هم‌ارزی گروه‌ها نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

اختلافِ میانگین‌های نمونه وجود دارد، ولی شواهد این نمونه برای رد برابری میانگین جامعه در سطح انتخابی کافی نیست. «عدم رد» به معنای یکسان‌بودن اندازهٔ دستِ گروه‌های سنی نیست. گروه‌های ۳ یا۴ نفره توان محدودی دارند و بازه‌ها می‌توانند پهن باشند.

#### کاربرد و محدودیت

اگر نتیجهٔ کلی رد شد، باید معلوم شود کدام مقایسه‌ها هدف پژوهش بوده‌اند و برای چندمقایسه‌ای چه روشی به کار می‌رود. انتخاب چند آزمون به امید p کوچک، نتیجه را مخدوش می‌کند.

#### تمرین

با سه گروه و ده مشاهده، درجات آزادی بین و درون گروه چیست؟

#### پاسخ تشریحی

بین گروه k−۱=۲ و درون گروه N−k=۷ است. آمارهٔ F روی همین دو درجهٔ آزادی تفسیر می‌شود.

### درس 21. آزمون t مستقل

**هدف یادگیری:** اختلاف دو میانگین را همراه حجم و پراکندگی مقایسه کنید.

#### مفهوم و تشریح کامل

آزمون t مستقل دربارهٔ اختلاف میانگینِ دو جمعیت است. نسخهٔ Welch برای هر گروه واریانس جداگانه می‌گیرد و درجهٔ آزادی را تصحیح می‌کند. استقلال افراد و تناسب مدل همچنان لازم است. اگر یک فرد قبل و بعد سنجیده شود، مسئله زوجی است و فرمول مستقل مناسب نیست.

#### فرمول و قرارداد

```text
t=(x̄A−x̄B)/√(sA²/nA+sB²/nB)
```

#### مثال حل‌شدهٔ مرجع

گروه سابقه≤۵ سال طول‌های [۱۸،۲۰،۲۱،۲۲] با n=۴ و میانگین۲۰٫۲۵ دارد؛ گروه سابقه>۵ سال n=۶ و میانگین≈۲۲٫۳۳۳ دارد. اختلاف A−B≈−۲٫۰۸۳ سانتی‌متر است؛ Welch t≈−۱٫۴۱۳ و p≈۰٫۱۹۵۶ می‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
df['experience_group'] = np.where(df.gaming_years<=5, 'low', 'high')
low_exp = df.loc[df.experience_group=='low','hand_length_cm'].to_numpy()
high_exp = df.loc[df.experience_group=='high','hand_length_cm'].to_numpy()
t_result = stats.ttest_ind(low_exp, high_exp, equal_var=False)
print(f'low n={len(low_exp)}, mean={low_exp.mean():.2f}; high n={len(high_exp)}, mean={high_exp.mean():.2f}')
print(f'Welch t={t_result.statistic:.3f}, p={t_result.pvalue:.4f}')

se_diff=np.sqrt(low_exp.var(ddof=1)/len(low_exp)+high_exp.var(ddof=1)/len(high_exp))
df_diff=(low_exp.var(ddof=1)/len(low_exp)+high_exp.var(ddof=1)/len(high_exp))**2/((low_exp.var(ddof=1)/len(low_exp))**2/(len(low_exp)-1)+(high_exp.var(ddof=1)/len(high_exp))**2/(len(high_exp)-1))
exp_ci=(low_exp.mean()-high_exp.mean())+np.array([-1,1])*stats.t.ppf(.975,df_diff)*se_diff
group_figure([low_exp,high_exp],['low experience','high experience'],'Experience groups: observations, mean and SD')

say(f'اختلاف میانگین کم‌تجربه منهای پرتجربه {low_exp.mean()-high_exp.mean():.3f} سانتی‌متر است؛ Welch t={t_result.statistic:.3f}، p={t_result.pvalue:.4f} و CI اختلاف [{exp_ci[0]:.3f}, {exp_ci[1]:.3f}]. گروه‌بندی مشاهده‌ای، اثر علّی تجربه را مشخص نمی‌کند.')
````

#### راهنمای کامل خواندن خروجی و نمودار

علامت منفی فقط به ترتیب تفریق گروه‌ها وابسته است. p بزرگ‌تر از۰٫۰۵، اختلاف ثبت‌شده را صفر نمی‌کند و ثابت نمی‌کند تفاوت عملی وجود ندارد. برای تصمیم، اختلاف، بازهٔ آن و حداقل تفاوت مهمِ کاربردی را هم گزارش کنید.

#### کاربرد و محدودیت

سابقهٔ بازی در این داده مشاهده‌ای است و با سن همبستگی زیادی دارد؛ تفاوت طول دست را به «اثر بازی‌کردن» نسبت ندهید. برای مقایسهٔ راحتی دو ماوس روی افراد یکسان از طرح زوجی و کنترل ترتیب استفاده کنید.

#### تمرین

اگر ترتیب گروه‌ها را عوض کنیم، t و p دوطرفه چه تغییری می‌کنند؟

#### پاسخ تشریحی

علامت t عوض می‌شود، اما قدرمطلق آن و p دوطرفه ثابت می‌ماند. جهت اختلاف باید با تعریف A−B گزارش شود.

### درس 22. رگرسیون چندگانه

**هدف یادگیری:** ضرایب شرطی، هم‌خطی و پیچیدگی مدل را درک کنید.

#### مفهوم و تشریح کامل

در رگرسیون چندگانه هر ضریب با ثابت نگه‌داشتن دیگر پیش‌بین‌های مدل تفسیر می‌شود. این یک توصیف شرطی است؛ در دادهٔ مشاهده‌ای الزاماً اثر مداخله نیست. پیش‌بین‌های شدیداً همبسته می‌توانند ضرایب را ناپایدار کنند، حتی زمانی که پیش‌بینی مجموع مدل تغییر کمی دارد.

#### فرمول و قرارداد

```text
width≈1.4355+0.4149×length−0.0516×age+0.0923×experience
```

#### مثال حل‌شدهٔ مرجع

مدل سه پیش‌بین و یک عرض از مبدأ روی ده نفر دارد؛ درجهٔ آزادی باقیمانده ۶ است. R²≈۰٫۹۸۶ و تعدیل‌شده≈۰٫۹۷۸ است. با ورودی طول ۲۲، سن ۲۶ و سابقه ۷، با ضرایب گرد‌شده ŷ≈۹٫۸۶۸ سانتی‌متر به دست می‌آید. p سن≈۰٫۶۷۶ و سابقه≈۰٫۶۱۱ در حضور بقیهٔ متغیرهاست.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
multi_X = df[['hand_length_cm','age_years','gaming_years']]
multi_fit = sm.OLS(df.hand_width_cm, sm.add_constant(multi_X)).fit()
display(pd.DataFrame({'coefficient':multi_fit.params, 'p_value':multi_fit.pvalues}).round(4))
print(f'R²={multi_fit.rsquared:.3f}; adjusted R²={multi_fit.rsquared_adj:.3f}; residual df={multi_fit.df_resid:.0f}')
print('Design matrix condition number:', round(np.linalg.cond(sm.add_constant(multi_X)), 1))

coef_ci=multi_fit.conf_int().drop('const'); coef_values=multi_fit.params.drop('const')
fig,ax=plt.subplots(figsize=(9,4)); ax.errorbar(coef_values,np.arange(3),xerr=[coef_values-coef_ci[0],coef_ci[1]-coef_values],fmt='o',capsize=6,color=BLUE)
ax.axvline(0,color='0.4',ls='--'); ax.set(yticks=np.arange(3),yticklabels=coef_values.index,xlabel='Conditional coefficient, 95% model-based CI (different input units)',title='Multiple-regression coefficients'); plt.tight_layout(); plt.show()

say(f'R² آموزشی={multi_fit.rsquared:.4f} و تعدیل‌شده={multi_fit.rsquared_adj:.4f} با df باقیمانده={multi_fit.df_resid:.0f}. ضریب سن به‌شرط ثابت‌بودن بقیه است؛ هم‌خطی و نمونهٔ کم تفسیر مستقل و تعمیم را محدود می‌کنند.')
````

#### راهنمای کامل خواندن خروجی و نمودار

بالارفتن جزئی R² نسبت به مدل ساده انتظار می‌رود؛ افزودن پیش‌بین معمولاً خطای آموزش را کم می‌کند. مقایسهٔ ضرایب خامِ سن و طول به‌عنوان اهمیت درست نیست، چون واحدها فرق دارند. عدد شرطِ ماتریسِ بدون مقیاس‌بندی نیز به مقیاس حساس است؛ هم‌خطی را فقط از آن تشخیص ندهید.

#### کاربرد و محدودیت

برای مقایسهٔ پیش‌بینی، یک pipeline با پیش‌پردازش درون هر fold و معیار مشترک بسازید. ضریب کوچک و p بزرگ، بی‌اهمیتی قطعیِ متغیر را ثابت نمی‌کند، به‌ویژه با ده سطر.

#### تمرین

آیا با افزودن ده پیش‌بین دیگر به ده فرد، کیفیت تصمیم بهتر می‌شود؟

#### پاسخ تشریحی

تعداد پارامترها نسبت به اطلاعات بسیار زیاد و مدل نامعین یا بیش‌برازش می‌شود. ابتدا هدف، دادهٔ بیشتر و ارزیابی خارج از نمونه لازم است.

## فصل: مقایسهٔ گروه‌ها و آزمون‌های مقاوم

### درس 23. آزمون تعقیبی Tukey HSD

**هدف یادگیری:** مقایسهٔ چند جفت را بدون افزایش کنترل‌نشدهٔ خطا انجام دهید.

#### مفهوم و تشریح کامل

با k گروه، k(k−1)/۲ جفت وجود دارد. اجرای t جداگانهٔ بدون تعدیل برای همهٔ جفت‌ها، احتمال حداقل یک رد تصادفی را افزایش می‌دهد. Tukey خانوادهٔ مقایسه‌های جفتیِ میانگین را کنترل می‌کند؛ برای حجم‌های نامساوی از صورت Tukey–Kramer استفاده می‌شود.

#### فرمول و قرارداد

```text
SEij=√[(MSE/2)(1/ni+1/nj)]; qij=|mean_i−mean_j|/SEij
```

#### مثال حل‌شدهٔ مرجع

سه گروه سنی سه جفت می‌سازند. در خروجی، تفاوت جوان نسبت به میانی −۲٫۵ سانتی‌متر و p تعدیل‌شده≈۰٫۴۸۷۴ است. بازهٔ هم‌زمان همین مقایسه حدود [−۸٫۶۱۱،۳٫۶۱۱] است و صفر را شامل می‌شود. ترتیب group1/group2 را از سرستون بخوانید، چون علامت اختلاف به آن وابسته است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from statsmodels.stats.multicomp import pairwise_tukeyhsd
tukey = pairwise_tukeyhsd(df.hand_length_cm, df.age_group, alpha=.05)
print(tukey.summary())

fig,ax=plt.subplots(figsize=(8,4));
for i,(delta,ci) in enumerate(zip(tukey.meandiffs,tukey.confint)):
    ax.plot(ci,[i,i],color=BLUE); ax.scatter(delta,i,color=GOLD)
ax.axvline(0,color='0.4',ls='--'); ax.set(yticks=range(3),yticklabels=['middle → older','middle → young','older → young'],xlabel='group2 − group1 length (cm)',title='Tukey simultaneous 95% confidence intervals'); plt.tight_layout(); plt.show()

say('در جدول Tukey ستون meandiff اختلاف group2−group1 است؛ p-adj برای خانوادهٔ مقایسه‌ها اصلاح شده است. بازه‌های همهٔ زوج‌ها صفر را شامل می‌شوند؛ هیچ زوجی در سطح ۰٫۰۵ رد نمی‌شود، اما برابری اثبات نشده است.')
````

#### راهنمای کامل خواندن خروجی و نمودار

ستون reject=False به معنی نبودِ رد در سطح تعیین‌شده است. بازهٔ پهن نشان می‌دهد چه اختلاف‌های منفی و مثبتی هنوز با داده سازگارند. به جای تنها انتخاب کوچک‌ترین p، همهٔ مقایسه‌های از پیش تعریف‌شده و مقیاس اختلاف را ببینید.

#### کاربرد و محدودیت

وقتی واریانس‌ها متفاوت‌اند، پرسش تعقیبی مشابه را می‌توان با Games–Howell بررسی کرد. انتخاب آزمون با فرض‌ها و طرح پژوهش تعیین می‌شود، نه اینکه کدام روش نتیجهٔ معنادار تولید کند.

#### تمرین

برای پنج گروه چند مقایسهٔ جفتی داریم؟

#### پاسخ تشریحی

۵×۴÷۲=۱۰ مقایسه. هرچه خانواده بزرگ‌تر شود، لزوم کنترل چندمقایسه‌ای بیشتر به چشم می‌آید.

### درس 24. آزمون Levene / Brown–Forsythe

**هدف یادگیری:** پرسش برابری پراکندگی را از برابری میانگین جدا کنید.

#### مفهوم و تشریح کامل

Levene روی فاصلهٔ مطلق هر مشاهده از مرکز گروه کار می‌کند و آن فاصله‌ها را بین گروه‌ها مقایسه می‌کند. مرکز می‌تواند میانگین یا میانه باشد؛ صورت با میانه اغلب Brown–Forsythe نامیده می‌شود. این آزمون دربارهٔ واریانس‌هاست و پاسخ ANOVAِ میانگین‌ها را تکرار نمی‌کند.

#### فرمول و قرارداد

```text
Zij=|Yij−median(group i)|; سپس ANOVA روی Zها
```

#### مثال حل‌شدهٔ مرجع

انحراف معیار گروه‌های جوان، میانی و بالاتر به‌ترتیب۲٫۰۰۰،۳٫۴۱۶ و۲٫۰۸۲ سانتی‌متر است. آماره≈۰٫۶۲۴ و p≈۰٫۵۶۳۳ شده است. نسبت بزرگ‌ترین و کوچک‌ترین SD حدود۱٫۷۱ است؛ نتیجهٔ p را کنار همین مقدارهای خام بخوانید.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
lev = stats.levene(*groups_age, center='median')
print(f'Brown–Forsythe/Levene statistic={lev.statistic:.3f}, p={lev.pvalue:.4f}')
print('Group SDs:', [round(np.std(g, ddof=1),3) for g in groups_age])

fig,ax=plt.subplots(); ax.bar(['young','middle','older'],[np.std(g,ddof=1) for g in groups_age],color=BLUE)
ax.set(ylabel='Sample SD of length (cm)',title='Within-group spread (not a mean comparison)'); plt.show()

say(f'آزمون Levene با مرکز میانه: آماره={lev.statistic:.3f}، p={lev.pvalue:.4f}. این آزمون دربارهٔ پراکندگی است، نه میانگین؛ عدم رد با این نمونهٔ کوچک تضمین یکسانی واریانس‌ها نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

عدم رد با نمونه‌های کوچک، برابر بودن واریانس‌ها را تضمین نمی‌کند. ممکن است آزمون توان کافی برای کشف اختلاف نداشته باشد. نمودار انحراف‌های مطلق و نقاط خام گروه‌ها کمک می‌کند یک مشاهدهٔ دور یا عدم توازن حجم را تشخیص دهید.

#### کاربرد و محدودیت

برای روش‌های مقاوم، لازم نیست ابتدا با یک آزمون کم‌توان «اجازه» بگیرید. می‌توانید با توجیه روش‌شناختی Welch را انتخاب و برآوردها و پراکندگی‌ها را شفاف گزارش کنید.

#### تمرین

دو گروه میانگین برابر دارند. آیا Levene الزاماً p بزرگ می‌دهد؟

#### پاسخ تشریحی

خیر. میانگین برابر دربارهٔ پراکندگی چیزی نمی‌گوید. یک گروه می‌تواند بسیار متمرکز و دیگری بسیار پراکنده باشد.

### درس 25. آزمون Mann–Whitney U

**هدف یادگیری:** آزمون رتبه‌ای را با ادعای درست استفاده کنید.

#### مفهوم و تشریح کامل

Mann–Whitney داده‌های دو گروه مستقل را در کنار هم رتبه‌بندی می‌کند. برای مقدارهای مساوی میانگین رتبه می‌گیرد. تفسیر عمومی مربوط به توزیع‌ها و برتری نسبی است؛ تفسیر صرفِ اختلاف میانه به فرض‌های بیشتر دربارهٔ شکل و جابه‌جایی توزیع‌ها نیاز دارد.

#### فرمول و قرارداد

```text
UA=RA−nA(nA+1)/2; UB=nA×nB−UA
```

#### مثال حل‌شدهٔ مرجع

در مقایسهٔ سابقه، nA=۴ و nB=۶ است؛ UA=۷ و UB=۲۴−۷=۱۷ به دست آمده است. p دوطرفهٔ ثبت‌شده با تقریبِ اصلاح‌شده برای ties≈۰٫۳۳۴۴ است. میانهٔ دو گروه۲۰٫۵ و۲۲ است؛ این تفاوت توصیفی را از تصمیم آزمون جدا گزارش کنید.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
mw = stats.mannwhitneyu(low_exp, high_exp, alternative='two-sided', method='auto')
print(f'U={mw.statistic:.3f}, p={mw.pvalue:.4f}')
print('Medians:', np.median(low_exp), np.median(high_exp))

ranked=stats.rankdata(np.r_[low_exp,high_exp]); print('Average-tie ranks:',ranked)
fig,ax=plt.subplots(); ax.scatter(ranked[:len(low_exp)],np.zeros(len(low_exp)),marker='o',s=80,label='low',color=BLUE); ax.scatter(ranked[len(low_exp):],np.ones(len(high_exp)),marker='s',s=80,label='high',color=GOLD)
ax.set(xlabel='Pooled average rank',yticks=[0,1],yticklabels=['Low experience','High experience'],title='Mann–Whitney compares ranks'); plt.show()

say(f'Mann–Whitney: U={mw.statistic:.3f}، p={mw.pvalue:.4f}؛ میانه‌ها {np.median(low_exp):.2f} و {np.median(high_exp):.2f}. تعبیر صرفاً اختلاف میانه به شکل مشابه توزیع‌ها نیاز دارد؛ این آزمون رتبه‌ای است.')
````

#### راهنمای کامل خواندن خروجی و نمودار

رتبه‌ها بزرگی دقیق فاصله‌ها را از بین می‌برند، ولی ترتیب را نگه می‌دارند. وجود ties روی روش محاسبهٔ p اثر می‌گذارد. برای گروه بسیار کوچک همراه ties، آزمون جایگشتی با تعریف روشن می‌تواند بررسی تکمیلی مناسبی باشد؛ عدد روش‌های مختلف ممکن است کمی فرق کند.

#### کاربرد و محدودیت

غیرپارامتری به معنی بی‌فرض نیست؛ استقلال و تعریف جامعه همچنان مهم‌اند. برای اندازه‌گیری قبل و بعدِ افراد یکسان، Mann–Whitney مستقل مناسبِ ساختار داده نیست.

#### تمرین

اگر UA=۷ و nA×nB=۲۴ باشد، UB چقدر است؟

#### پاسخ تشریحی

۱۷. این هویت یک کنترل سادهٔ محاسبه است؛ جهت U باید با ترتیب گروه‌ها خوانده شود.

### درس 26. Welch ANOVA

**هدف یادگیری:** میانگین چند گروه با واریانس‌های متفاوت را مقایسه کنید.

#### مفهوم و تشریح کامل

Welch ANOVA برای هر گروه از حجم و واریانس همان گروه وزن می‌سازد و درجهٔ آزادی مخرج را اصلاح می‌کند. فرضِ واریانس برابر را نیاز ندارد، اما استقلال، تناسب داده و برآورد معقول واریانس هر گروه لازم است. گروه با واریانس صفر یا حجم ناکافی نیاز به بررسی ویژه دارد.

#### فرمول و قرارداد

```text
wi=ni/si²; weighted mean=Σwi×mean_i /Σwi
```

#### مثال حل‌شدهٔ مرجع

برای سه گروه سنی، Welch F≈۰٫۷۸۷، df1=۲، df2≈۴٫۶۵۳ و p≈۰٫۵۰۷۵ است. ANOVA کلاسیک p≈۰٫۵۱۳۶ داده بود؛ هر دو در این نمونه به عدم رد می‌رسند، ولی این دو شاهد مستقل نیستند چون از همان افراد محاسبه شده‌اند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from statsmodels.stats.oneway import anova_oneway
welch = anova_oneway(groups_age, use_var='unequal', welch_correction=True)
print(f'Welch F={welch.statistic:.3f}, p={welch.pvalue:.4f}, df={welch.df}')

group_figure(groups_age,['young','middle','older'],'Welch ANOVA inputs: unequal-variance model')

say(f'Welch ANOVA: F={welch.statistic:.3f}، p={welch.pvalue:.4f} و df={welch.df}. نابرابری واریانس را مجاز می‌کند، اما کوچک‌بودن و استقلال نمونه‌ها همچنان مهم است؛ رتبه‌ای نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

درجهٔ آزادی کسری خطای محاسبه نیست؛ نتیجهٔ تصحیح Welch است. مقدار F را با درجهٔ آزادی خودش بخوانید. توافق دو روش، محدودیت گروه‌های سه‌نفره و عدم قطعیتِ تخمین اختلاف را حذف نمی‌کند.

#### کاربرد و محدودیت

برای گزارش، حجم، میانگین و SD هر گروه را کنار آزمون بنویسید. در دادهٔ واقعی با ناهمگنی و اندازه‌های نامتوازن، Welch اغلب گزینه‌ای مناسب برای پرسش میانگین‌هاست.

#### تمرین

آیا Welch همان Kruskal–Wallis است؟

#### پاسخ تشریحی

خیر. Welch همچنان میانگین‌ها را با وزن‌دهی واریانسی مقایسه می‌کند؛ Kruskal–Wallis رتبه‌ها و تفاوت توزیع‌ها را بررسی می‌کند.

### درس 27. مقایسهٔ Games–Howell

**هدف یادگیری:** مقایسهٔ جفتیِ میانگین‌ها را با واریانس جدا انجام دهید.

#### مفهوم و تشریح کامل

Games–Howell برای هر جفت، خطای استاندارد و درجهٔ آزادی جدا می‌سازد و تعدیل چندمقایسه‌ای را با توزیع دامنهٔ دانشجوشده انجام می‌دهد. این روش پارامتری است و نام‌گذاریِ «تعقیبی غیرپارامتری» در فایل اولیه درست نبود.

#### فرمول و قرارداد

```text
SEij=√(si²/ni+sj²/nj); q=√2×|mean_i−mean_j|/SEij
```

#### مثال حل‌شدهٔ مرجع

اختلاف جوان−میانی −۲٫۵ با df≈۴٫۸۵ و p تعدیل‌شده≈۰٫۴۹۷۸ است. جوان−بالاتر −۱٫۶۶۷ و p≈۰٫۶۱۵۲؛ میانی−بالاتر +۰٫۸۳۳ و p≈۰٫۹۱۷۴ است. هیچ‌کدام در سطح۰٫۰۵ رد نمی‌شوند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from itertools import combinations
rows = []
group_names = ['young','middle','older']
for i,j in combinations(range(3),2):
    a,b = groups_age[i],groups_age[j]
    va,vb = np.var(a,ddof=1),np.var(b,ddof=1)
    se = np.sqrt(va/len(a)+vb/len(b))
    dof = (va/len(a)+vb/len(b))**2 / ((va/len(a))**2/(len(a)-1)+(vb/len(b))**2/(len(b)-1))
    q = np.sqrt(2)*abs(a.mean()-b.mean())/se
    p = stats.studentized_range.sf(q, len(groups_age), dof)
    rows.append([group_names[i],group_names[j],a.mean()-b.mean(),dof,p])
display(pd.DataFrame(rows,columns=['group A','group B','mean difference (cm)','df','adjusted p']).round(4))

gh_table=pd.DataFrame(rows,columns=['group A','group B','mean difference (cm)','df','adjusted p'])
fig,ax=plt.subplots(); ax.barh([r[0]+' − '+r[1] for r in rows],[r[2] for r in rows],color=BLUE); ax.axvline(0,color='0.3'); ax.set(xlabel='Mean difference (cm)',title='Games–Howell pairwise effect estimates (see adjusted p table)'); plt.tight_layout(); plt.show()

say(f'در Games–Howell سه مقایسهٔ میانگین با واریانس جدا و p اصلاح‌شده انجام شد. کوچک‌ترین p={gh_table["adjusted p"].min():.4f} است؛ در سطح ۰٫۰۵ هیچ زوج رد نمی‌شود. ستون اختلاف این جدول group A−group B است.')
````

#### راهنمای کامل خواندن خروجی و نمودار

هر سطر پرسش جدا با برآورد اختلافِ دارای جهت است. بزرگ بودن p به معنی احتمال برابری گروه‌ها نیست. با سه یا چهار نفر در گروه، واریانس تخمینی خودش ناپایدار است و نمی‌توان با عدم رد، محصول مشترک برای همهٔ گروه‌ها را اثبات کرد.

#### کاربرد و محدودیت

در صورت نگرانی دربارهٔ ناهمگنی، Welch و Games–Howell چارچوب سازگارتری برای پرسش میانگین‌ها فراهم می‌کنند. برای پرسش رتبه‌ای یا دادهٔ زوجی باید روش متناسب با همان ساختار انتخاب شود.

#### تمرین

چرا همهٔ جفت‌های Games–Howell لزوماً df یکسان ندارند؟

#### پاسخ تشریحی

چون n و s² هر دو گروه در تصحیح درجهٔ آزادی وارد می‌شوند و ترکیب این مقدارها برای هر جفت متفاوت است.

### درس 28. ANOVA عاملی: کنترل شناسایی‌پذیری

**هدف یادگیری:** پیش از آزمون تعامل، امکان شناسایی آن را بررسی کنید.

#### مفهوم و تشریح کامل

در طرح دو عاملی، اثر عامل A، عامل B و تعامل یعنی تغییر اثر A در سطوح B را می‌سنجیم. برای جداسازی این اثرها باید ترکیب‌های عامل‌ها مشاهده شده باشند. خانهٔ خالی یا هم‌پوشانی کامل دو عامل می‌تواند ماتریس طرح را کم‌رتبه کند.

#### فرمول و قرارداد

```text
Y=μ+α(age)+β(experience)+αβ+ε; نخست جدول شمارشِ ترکیب عامل‌ها
```

#### مثال حل‌شدهٔ مرجع

با سن≤۲۴ در برابر سن>۲۴ و سابقه≤۵ در برابر>۵، جدول فقط دو خانهٔ پر دارد: جوان‌تر/کم‌سابقه۴ نفر و بالاتر/پرسابقه۶ نفر؛ دو خانهٔ دیگر صفرند. بنابراین اثر سن و سابقه در این تقسیم از هم جدا نمی‌شوند و جدول p تعامل معتبر تولید نشده است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
df['age_binary'] = np.where(df.age_years<=24,'younger','older')
design = pd.crosstab(df.age_binary, df.experience_group)
display(design)
full_design = bool((design>0).all().all())
print('Full 2×2 design:', full_design)
if not full_design:
    print('Interaction ANOVA is not estimable from these data: at least one cell is empty.')
else:
    import statsmodels.formula.api as smf
    from statsmodels.stats.anova import anova_lm
    fit = smf.ols('hand_length_cm ~ C(age_binary)*C(experience_group)', data=df).fit()
    display(anova_lm(fit, typ=2))

from patsy import dmatrix
design_matrix=dmatrix('C(age_binary)*C(experience_group)',data=df)
design_rank=np.linalg.matrix_rank(design_matrix); design_columns=design_matrix.shape[1]
fig,ax=plt.subplots(figsize=(5,4)); ax.imshow(design,cmap='Blues',vmin=0,vmax=design.to_numpy().max())
ax.set(xticks=range(2),xticklabels=design.columns,yticks=range(2),yticklabels=design.index,title='Observed 2×2 cell counts (zeros are structural gaps)')
for i in range(2):
    for j in range(2): ax.text(j,i,str(design.iloc[i,j]),ha='center',va='center',color='black',bbox=dict(facecolor='white',alpha=.8,edgecolor='none'))
plt.show()

say(f'طرح کامل ۲×۲؟ {full_design}. دو ترکیب صفر دارند؛ ماتریس طرح رتبهٔ {design_rank} از {design_columns} ستون دارد. در این داده اثر سن و تجربه از هم جدا نمی‌شود؛ خروجی عددیِ اثر متقابل گزارش نمی‌کنیم.')
````

#### راهنمای کامل خواندن خروجی و نمودار

به جدول شمارش نگاه کنید، نه فقط یک interaction plot که نرم‌افزار ممکن است با خانه‌های حذف‌شده رسم کند. برای تصور تعامل در یک مثال مستقل، پاسخ روش A در مبتدی/حرفه‌ای۶ و۸ و روش B در همان گروه‌ها۷ و۷ باشد: اثر A نسبت به B از−۱ به+۱ عوض می‌شود.

#### کاربرد و محدودیت

برای پژوهش جدید از هر ترکیب عامل نمونه جمع کنید و تعداد کافی در سلول‌ها داشته باشید. اگر تخصیص مشاهده‌ای است، حتی طرحِ قابل برازش نیز علت را بدون کنترل لازم ثابت نمی‌کند.

#### تمرین

آیا نرم‌افزاری که با خانه‌های خالی یک جدول ANOVA می‌دهد، مشکل طراحی را حل کرده است؟

#### پاسخ تشریحی

خیر. ممکن است مدل کاهش‌یافته، قرارداد محدودیت یا نتیجهٔ تکین گزارش شود. اول رتبهٔ طرح و فرض‌های قابل شناسایی را بررسی و صریحاً بیان کنید.

### درس 29. Kruskal–Wallis

**هدف یادگیری:** مقایسهٔ رتبه‌ایِ بیش از دو گروه را تفسیر کنید.

#### مفهوم و تشریح کامل

Kruskal–Wallis رتبه‌های کل مشاهده‌ها را بین گروه‌ها مقایسه می‌کند و ties را اصلاح می‌کند. فرض عمومی دربارهٔ یکسانی توزیع‌هاست. تنها با شکل‌های قابل مقایسه می‌توان آن را به جابه‌جایی مکان ربط داد؛ هر اختلاف پراکندگی هم ممکن است بر نتیجه اثر بگذارد.

#### فرمول و قرارداد

```text
H=12/[N(N+1)] × Σ(Ri²/ni) −3(N+1), با تصحیحِ ties
```

#### مثال حل‌شدهٔ مرجع

سه گروه سنی در این نمونه H≈۱٫۳۴۱ و p≈۰٫۵۱۱۴ دارند. مجموع رتبه‌ها از همان طول‌های ده نفر به دست آمده است. اعداد۲۰ و۲۱ تکرار دارند و باید رتبهٔ میانگین بگیرند؛ رتبه‌دادنِ متفاوت به دو مقدار برابر نتیجه را تغییر می‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
kw = stats.kruskal(*groups_age)
print(f'H={kw.statistic:.3f}, p={kw.pvalue:.4f}')

rank_all=stats.rankdata(np.concatenate(groups_age)); starts=np.cumsum([0]+[len(g) for g in groups_age])
fig,ax=plt.subplots(); ax.boxplot([rank_all[starts[i]:starts[i+1]] for i in range(3)],tick_labels=['young','middle','older']); ax.set(ylabel='Pooled average rank',title='Rank distributions used in Kruskal–Wallis'); plt.show()

say(f'Kruskal–Wallis: H={kw.statistic:.3f} و p={kw.pvalue:.4f}. در سطح ۰٫۰۵ شواهد کافی برای اختلاف توزیع‌های رتبه‌ای نیست؛ این نتیجه اثبات برابر بودن همهٔ میانه‌های جامعه نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

نتیجهٔ کلی رد نشده است؛ این به معنی یکسان‌بودن همهٔ توزیع‌ها نیست. حتی در صورت رد کلی، آزمون تعیین نمی‌کند کدام جفت متفاوت است؛ پس‌آزمون رتبه‌ای و کنترل چندمقایسه‌ای لازم می‌شود. با گروه‌های بسیار کوچک، تقریب مجانبی را با احتیاط بخوانید.

#### کاربرد و محدودیت

استفاده از روش رتبه‌ای، سوگیری نمونه‌گیری یا وابستگی افراد را حل نمی‌کند. یک داده را با چند آزمون بررسی‌کردن به معنی چند منبع شاهد مستقل نیست.

#### تمرین

وقتی دو فرد مقدار۲۰ دارند، چگونه رتبه می‌گیرند؟

#### پاسخ تشریحی

اگر جایگاه‌های سوم وچهارم باشند، هر دو رتبه۳٫۵ می‌گیرند. مجموع رتبه‌ها همچنان باید N(N+۱)/۲ باشد.

## فصل: کاهش بُعد و گروه‌بندی

### درس 30. PCA: تحلیل مؤلفه‌های اصلی

**هدف یادگیری:** فشرده‌سازی خطی را از طبقه‌بندی و کشف علت جدا کنید.

#### مفهوم و تشریح کامل

PCA جهت‌هایی متعامد می‌سازد که واریانس دادهٔ مرکز‌شده را به‌ترتیب بیشینه کنند. چون سانتی‌متر و سال مقیاس‌های متفاوت دارند، در این مثال ابتدا هر ستون استاندارد می‌شود. مؤلفه‌ها از همهٔ ستون‌ها ترکیب می‌شوند؛ «مستقل بودن آماری» از متعامد یا ناهمبسته بودن نتیجه نمی‌شود.

#### فرمول و قرارداد

```text
Z=(X−mean)/SD; scores=ZV; explained ratio=λj/Σλ
```

#### مثال حل‌شدهٔ مرجع

سهم PC1≈۷۵٫۷٪ و PC2≈۲۴٫۰٪ است؛ مجموع حدود۹۹٫۷٪ واریانسِ همین ویژگی‌های استانداردشده را نگه می‌دارند. ضرایب PC1 تقریباً[۰٫۴۹۸،۰٫۵۰۷،۰٫۴۷۳،۰٫۵۲۲] و PC2≈[−۰٫۵۰۷،−۰٫۴۷۸،۰٫۵۷۸،۰٫۴۲۴] است. مؤلفهٔ دوم اندازه‌های دست را در برابر سن/سابقه قرار می‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from sklearn.decomposition import PCA
X_all = df[numeric_cols].to_numpy(dtype=float)
X_scaled = StandardScaler().fit_transform(X_all)
pca = PCA(n_components=2).fit(X_scaled)
scores_pca = pca.transform(X_scaled)
print('Explained variance ratio:', np.round(pca.explained_variance_ratio_,3))
display(pd.DataFrame(pca.components_.T, index=numeric_cols, columns=['PC1','PC2']).round(3))
fig, ax = plt.subplots()
ax.scatter(scores_pca[:,0], scores_pca[:,1], color=BLUE)
for i, (a,b) in enumerate(scores_pca): ax.annotate(str(i+1),(a,b),xytext=(4,4),textcoords='offset points')
ax.set(xlabel='PC1 (standardized units)', ylabel='PC2 (standardized units)', title='PCA projection of 10 users')
plt.show()

say(f'PC1={100*pca.explained_variance_ratio_[0]:.2f}٪ و PC2={100*pca.explained_variance_ratio_[1]:.2f}٪، مجموع {100*pca.explained_variance_ratio_.sum():.2f}٪ واریانس استانداردشده را حفظ می‌کنند؛ محور‌ها واحد سانتی‌متر ندارند.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در شکل هر نقطه یک فرد با مختصات PC1 وPC2 است. این محورها سانتی‌متر نیستند و درصد واریانس «دقت پیش‌بینی» نیست. علامت هر مؤلفه قراردادی است؛ وارونگی هم‌زمان محور و ضرایب، همان هندسه را می‌دهد.

#### کاربرد و محدودیت

برای استفادهٔ پیش‌بینی، استانداردسازی و PCA را داخل fold آموزش برازش کنید. روی همهٔ افراد fit کردن و سپس ارزیابی روی افراد کنارگذاشته‌شده، اطلاعات آزمون را به تبدیل وارد می‌کند.

#### تمرین

آیا دو مؤلفه با ۹۹٫۷٪ واریانس، ۹۹٫۷٪ دقتِ تعیین سایز می‌دهند؟

#### پاسخ تشریحی

خیر. واریانسِ ویژگی‌ها با صحتِ یک هدف برچسب‌دار متفاوت است. برای دقت باید هدف واقعی و ارزیابی جدا تعریف شود.

### درس 31. Scree، بارگذاری، Biplot و بازسازی PCA

**هدف یادگیری:** سه خروجی متفاوت PCA را به یکدیگر وصل کنید.

#### مفهوم و تشریح کامل

Scree سهم یا مقدار ویژهٔ مؤلفه‌ها را به‌ترتیب نشان می‌دهد؛ افت شدید می‌تواند برای انتخاب بُعد سرنخ باشد. بارگذاری‌ها نشان می‌دهند هر ویژگی چگونه در مؤلفه شرکت می‌کند. score مختصاتِ افراد روی مؤلفه‌هاست. در Biplot افراد و بردار ویژگی‌ها با یک قرارداد مقیاس‌بندی کنار هم قرار می‌گیرند؛ بدون اعلام قرارداد، طول پیکان‌ها قابل مقایسهٔ دقیق نیست.

#### فرمول و قرارداد

```text
scores=ZV; Ẑ=scores Vᵀ; X̂=scaler.inverse_transform(Ẑ)
```

#### مثال حل‌شدهٔ مرجع

دو مؤلفهٔ اول حدود۷۵٫۷٪ و۲۴٫۰٪ را نگه می‌دارند. برای فردی با بردار استاندارد z، score اول با ضرب هر چهار مقدار در[۰٫۴۹۸،۰٫۵۰۷،۰٫۴۷۳،۰٫۵۲۲] و جمع آن‌ها محاسبه می‌شود. PC2 برای طول/عرض علامت منفی و برای سن/سابقه مثبت دارد. در این تعریف components_ ضرایبِ بردارهای مؤلفه‌اند؛ برخی کتاب‌ها loading را با ضرب در√λ تعریف می‌کنند و عددهایشان متفاوت است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
scaler_detail=StandardScaler().fit(df[numeric_cols]); Z_detail=scaler_detail.transform(df[numeric_cols])
full_pca=PCA().fit(Z_detail); detail_pca=PCA(n_components=2).fit(Z_detail); detail_scores=detail_pca.transform(Z_detail)
detail_reconstruction=detail_pca.inverse_transform(detail_scores)
display(pd.DataFrame(full_pca.components_.T,index=numeric_cols,columns=['PC1','PC2','PC3','PC4']).round(4))
fig,axes=plt.subplots(1,2,figsize=(12,5),layout='constrained')
axes[0].bar(np.arange(1,5),100*full_pca.explained_variance_ratio_,color=BLUE); axes[0].plot(np.arange(1,5),100*full_pca.explained_variance_ratio_.cumsum(),color=GOLD,marker='o'); axes[0].set(xlabel='Component',ylabel='Explained variance (%)',xticks=[1,2,3,4],ylim=(0,105),title='Scree bars and cumulative proportion')
axes[1].scatter(detail_scores[:,0],detail_scores[:,1],color=BLUE)
for i,(a,b) in enumerate(detail_scores): axes[1].annotate(str(i+1),(a,b),xytext=(4,4),textcoords='offset points')
short_names=['Length','Width','Age','Experience']
label_offsets=[(-18,-24),(8,-5),(-25,18),(8,-12)]
for label,(a,b),offset in zip(short_names,detail_pca.components_.T,label_offsets):
    axes[1].annotate('',xy=(2*a,2*b),xytext=(0,0),arrowprops=dict(arrowstyle='->',color=GOLD,lw=1.5)); axes[1].annotate(label,(2*a,2*b),xytext=offset,textcoords='offset points')
axes[1].set(xlabel='PC1 score',ylabel='PC2 score',title='Scores + component vectors ×2 (declared scaling)',xlim=(-4,5),ylim=(-3,3)); plt.show()
say(f'دو مؤلفه {detail_pca.explained_variance_ratio_.sum():.3%} واریانس را حفظ می‌کنند؛ MSE بازسازی آموزشی در فضای استاندارد {mean_squared_error(Z_detail,detail_reconstruction):.5f} است. پیکان‌ها components_×۲ هستند، نه همبستگی‌های دقیقِ بارگذاری. علامت محور دلخواه و خطای فوق آموزشی است.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در Scree جمع نسبت‌ها باید۱ شود. در نمودار score، دوربودن نقطه ازمرکز می‌تواند ترکیبِ اندازه‌های استانداردِ بزرگ باشد. زاویهٔ پیکان‌ها فقط با مقیاس‌بندی مناسب و بازنمایی خوبِ ویژگی‌ها سرنخ همبستگی است؛ در صفحهٔ دوبعدیِ ناقص آن را دقیق نخوانید.

#### کاربرد و محدودیت

اگر هدف فشرده‌سازی است، MSE بازسازیِ دادهٔ کنارگذاشته‌شده را هم بسنجید. اگر هدف پیش‌بینی است، تعداد مؤلفه را برای همان معیار پیش‌بینی انتخاب کنید؛ بیشترین واریانس الزاماً بیشترین اطلاعاتِ هدف نیست.

#### تمرین

با وارونه‌کردن علامت score اول و بردار اول، بازسازی تغییر می‌کند؟

#### پاسخ تشریحی

خیر. حاصل‌ضرب دو علامت منفی مثبت است و همان مؤلفهٔ بازسازی به دست می‌آید. تفسیر فیزیکی را به جهت ظاهری محور گره نزنید.

### درس 32. خوشه‌بندی K-means

**هدف یادگیری:** خوشه‌بندی را به‌عنوان سناریوی اکتشافی ارزیابی کنید.

#### مفهوم و تشریح کامل

K-means فاصلهٔ مربع‌شدهٔ نقاط تا مرکز خوشه‌شان را کمینه می‌کند. مقیاس ویژگی‌ها تعیین‌کننده است؛ اینجا طول و عرض استاندارد شده‌اند. الگوریتم همیشه برای k درخواستی خوشه می‌سازد، پس دیدن سه رنگ به معنی کشفِ قطعی سه گروه طبیعی نیست.

#### فرمول و قرارداد

```text
J=Σk Σi∈Ck ||zi−μk||²; silhouette compares within/between cluster distance
```

#### مثال حل‌شدهٔ مرجع

برای k=۲، SSE≈۶٫۷۶۹ و silhouette≈۰٫۵۵۹ است؛ k=۳:۲٫۷۴۸ و۰٫۴۶۵؛ k=۴:۱٫۳۹۲ و۰٫۴۱۳. SSE با افزایش k پایین می‌آید، اما در این سه سناریو silhouette برای۲ بهتر است. بنابراین ادعای خودکار «بهترین تعداد سایز=۳» با این جدول پشتیبانی نمی‌شود.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
X_size = StandardScaler().fit_transform(df[['hand_length_cm','hand_width_cm']])
cluster_checks = []
for k in [2,3,4]:
    model = KMeans(n_clusters=k, random_state=SEED, n_init=10).fit(X_size)
    cluster_checks.append([k,model.inertia_,silhouette_score(X_size,model.labels_)])
display(pd.DataFrame(cluster_checks,columns=['k','within-cluster SSE','silhouette']).round(3))
k3 = KMeans(n_clusters=3,random_state=SEED,n_init=10).fit(X_size)
fig, ax = plt.subplots()
ax.scatter(df.hand_length_cm, df.hand_width_cm, c=np.array([BLUE,GOLD,TEAL])[k3.labels_], s=80)
for i,row in df.iterrows(): ax.annotate(str(row.user_id),(row.hand_length_cm,row.hand_width_cm),xytext=(4,4),textcoords='offset points')
ax.set(xlabel='Hand length (cm)',ylabel='Hand width (cm)',title='Illustrative k=3 size clusters')
plt.show()

say(f'برای k=2,3,4 امتیاز silhouette برابر {[round(row[2],3) for row in cluster_checks]} است. در این مقایسه k={max(cluster_checks,key=lambda row:row[2])[0]} بیشترین امتیاز را دارد؛ شکل k=3 فقط یک سناریوست، نه اثبات وجود سه سایز طبیعی.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در شکل، نقاط کاربر و رنگ‌ها انتساب خوشه‌اند. مرکز خوشه میانگین ویژگی‌هاست و الزاماً شکل یا ابعاد محصول قابل تولید نیست. فرد انتهایی می‌تواند یک خوشهٔ کوچک بسازد؛ پایداری با نمونه‌ها و شروع‌های مختلف را بسنجید.

#### کاربرد و محدودیت

k=۳ می‌تواند فرضیهٔ طراحی S/M/L باشد. انتخاب نهایی به هزینهٔ قالب، پوشش کاربران، موجودی و نتایج آزمون راحتی وابسته است. خوشه‌های ناپایدار را به بخش بازار تبدیل نکنید.

#### تمرین

چرا کم‌ترین SSE به‌تنهایی k مناسب را مشخص نمی‌کند؟

#### پاسخ تشریحی

اگر هر فرد خوشهٔ خودش باشد، SSE صفر می‌شود. باید پیچیدگی، جدایی، پایداری و هدف کاربردی را هم در نظر گرفت.

### درس 33. PPCA: تحلیل مؤلفه‌های اصلی احتمالاتی

**هدف یادگیری:** تفسیر احتمالاتیِ کاهش بُعد را یاد بگیرید.

#### مفهوم و تشریح کامل

PPCA فرض می‌کند داده از یک متغیر نهفتهٔ کم‌بُعد با توزیع گاوسی و نویز همسانگرد ساخته می‌شود. در کنار تصویر دوبعدی، یک مدل احتمالاتی برای کوواریانس و درست‌نمایی فراهم می‌کند. برآوردِ نویز کوچک فقط در چارچوب همین فرض‌ها معنا دارد.

#### فرمول و قرارداد

```text
x=Wz+μ+ε; z~N(0,I); ε~N(0,σ²I)
```

#### مثال حل‌شدهٔ مرجع

برای دو مؤلفه روی چهار ستونِ استانداردشده، noise_variance_≈۰٫۰۰۷۵ و میانگین log-likelihood آموزش≈−۱٫۲۱۷۴ ثبت شده است. این اعداد در واحدِ فضای استانداردند، نه cm² مستقیم. سهم واریانسِ دو مؤلفه همان حدود۷۵٫۷٪ و۲۴٫۰٪ است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
ppca = PCA(n_components=2, svd_solver='full').fit(X_scaled)
print('Noise variance estimate:', round(ppca.noise_variance_,4))
print('Training mean log-likelihood per sample:', round(ppca.score(X_scaled),4))
print('Explained variance:', np.round(ppca.explained_variance_ratio_,3))

display(pd.DataFrame({'user_id':df.user_id,'training_log_density':ppca.score_samples(X_scaled)}))
fig,ax=plt.subplots(); ax.bar(df.user_id,ppca.score_samples(X_scaled),color=TEAL); ax.set(xlabel='User ID',ylabel='Training log density',title='PPCA density diagnostic (not held-out performance)'); plt.show()

say(f'واریانس نویز PPCA={ppca.noise_variance_:.5f} و log-likelihood میانگین آموزش={ppca.score(X_scaled):.4f}. این کیفیت روی همان دادهٔ برازش‌شده است، نه احتمال موفقیت محصول یا نتیجهٔ اعتبارسنجی مستقل.')
````

#### راهنمای کامل خواندن خروجی و نمودار

درست‌نمایی عددی برای مقایسهٔ مدل‌ها با تعریف یکسانِ داده و مقیاس به کار می‌رود؛ بزرگ‌بودن مقدارِ آموزش به‌تنهایی مدل بهترِ کاربران آینده را اثبات نمی‌کند. تغییر استانداردسازی یا تعداد ویژگی‌ها مقیاس آن را عوض می‌کند.

#### کاربرد و محدودیت

برای انتخاب تعداد مؤلفه، امتیاز دادهٔ اعتبارسنجی و هدف بازسازی/پیش‌بینی را بررسی کنید. PPCA خوشه‌بند نیست و محورها برچسب مستقیمِ اندازهٔ محصول ندارند.

#### تمرین

آیا σ² کوچک، نبود خطای اندازه‌گیری واقعی را ثابت می‌کند؟

#### پاسخ تشریحی

خیر. این پارامترِ نویزِ مدل گاوسی در فضای انتخابی است و با خطای دستگاه یا همهٔ انواع خطای داده برابر نیست.

### درس 34. t-SNE: نمایش غیرخطی

**هدف یادگیری:** نقشهٔ t-SNE را بدون داستان‌سازی بخوانید.

#### مفهوم و تشریح کامل

t-SNE شباهت همسایگی‌های فضای اصلی را با شباهت در نقشهٔ کم‌بُعد تقریب می‌زند. محورهای نقشه واحد فیزیکی ندارند. فاصلهٔ خوشه‌های دور، مساحت جزیره‌ها و جهت چرخش را نباید مثل فاصله‌های اصلی تفسیر کرد.

#### فرمول و قرارداد

```text
هدف: کمینه‌سازی KL(P||Q) برای شباهت‌های جفتی
```

#### مثال حل‌شدهٔ مرجع

در این اجرا چهار ویژگی استاندارد، n=۱۰، perplexity=۳، شروع PCA و بذر۴۲ استفاده شده است. perplexity باید با حجم داده سازگار باشد. اگر آن را به۲ یا۴ تغییر دهید، ممکن است چیدمان عوض شود؛ این آزمایشِ حساسیت است و دادهٔ جدید تولید نمی‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from sklearn.manifold import TSNE
tsne_xy = TSNE(n_components=2, perplexity=3, init='pca', learning_rate='auto',
               random_state=SEED).fit_transform(X_scaled)
fig, ax = plt.subplots()
ax.scatter(tsne_xy[:,0], tsne_xy[:,1], color=TEAL)
for i,(a,b) in enumerate(tsne_xy): ax.annotate(str(i+1),(a,b),xytext=(4,4),textcoords='offset points')
ax.set(xlabel='t-SNE 1 (arbitrary)',ylabel='t-SNE 2 (arbitrary)',title='Exploratory t-SNE map, perplexity=3')
plt.show()

say('نقشهٔ t-SNE با perplexity=3 و بذر ثابت ساخته شد. همسایگی‌های محلی سرنخ اکتشافی‌اند؛ جهت، فاصلهٔ خوشه‌های دور و اندازهٔ فیزیکی محور قابل تفسیر نیست. گروه‌بندی تجاری از این شکل به‌تنهایی به دست نمی‌آید.')
````

#### راهنمای کامل خواندن خروجی و نمودار

هر نقطه همان یک کاربر است. نزدیکی چند کاربر می‌تواند همسایگی محلی را منعکس کند، ولی فاصلهٔ زیاد دو جزیره دلیل دو گروه تجاری نیست. در نمودارهای بدون برچسب مستقل، نام «دست کوچک» و«بزرگ» را از روی جایگاه بصری تحمیل نکنید.

#### کاربرد و محدودیت

برای اکتشاف دادهٔ بزرگ‌تر، نقشه را کنار ویژگی‌های اصلی، برچسب‌های مستقل و اجرای چند تنظیم بخوانید. برای این ده نفر، جدول و پراکندگی طول–عرض اغلب از نقشهٔ پیچیده قابل تفسیرتر است.

#### تمرین

آیا کاربری که دو برابر دورتر روی t-SNE است، دو برابر تفاوت فیزیکی دارد؟

#### پاسخ تشریحی

خیر. محور و فاصلهٔ جهانیِ نقشه چنین مقیاس مستقیمی ندارند؛ هدف روش تقریبِ ساختار محلی است.

## فصل: مدل‌های پیش‌بینی

### درس 35. ارزیابی مدل، OOB، خط پایه و اهمیت جایگشتی

**هدف یادگیری:** کیفیت مدل‌ها را با سؤال و معیار مشترک مقایسه کنید.

#### مفهوم و تشریح کامل

آموزش برای یادگیری پارامترها، اعتبارسنجی برای انتخاب تنظیمات و آزمون برای برآورد نهاییِ تعمیم است. اگر یک فرد چند سطر دارد، سطرهای او باید در یک گروهِ جداسازی بمانند. برای زمان، گذشته باید قبل از آینده باشد. درpipeline، scaler وPCA و هر انتخاب ویژگی فقط روی بخش آموزش fit می‌شوند.

#### فرمول و قرارداد

```text
MAE=Σ|e|/n; RMSE=√(Σe²/n); R²=1−SSE/SST
```

#### مثال حل‌شدهٔ مرجع

برای خطاهای[−۱،۰،۲]، MAE=۱، MSE=۵/۳ وRMSE≈۱٫۲۹۱ است. RMSE به خطای بزرگ حساس‌تر است. در مدل‌های طولِ این دوره، جنگلMAE≈۱٫۲۰۱ وBoosting≈۱٫۲۳۵ دارند؛ خط پایهٔ میانگین≈۲٫۲۲۲. MLP باMAE≈۰٫۲۳۴ هدفِ دیگری یعنی عرض دارد و در همین رتبه‌بندی قرار نمی‌گیرد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from sklearn.dummy import DummyRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.inspection import permutation_importance
eval_X=df[['age_years','gaming_years','hand_width_cm']].to_numpy(); eval_y=df.hand_length_cm.to_numpy()
eval_cv=LeaveOneOut(); eval_models={'Mean baseline':DummyRegressor(),'Random forest':RandomForestRegressor(n_estimators=100,min_samples_leaf=2,random_state=SEED)}
eval_rows=[]
for name,estimator in eval_models.items():
    pred=cross_val_predict(estimator,eval_X,eval_y,cv=eval_cv)
    eval_rows.append([name,np.mean(abs(eval_y-pred)),np.sqrt(mean_squared_error(eval_y,pred))])
eval_table=pd.DataFrame(eval_rows,columns=['model','LOOCV MAE (cm length)','LOOCV RMSE (cm length)']); display(eval_table.round(4))
oob_demo=RandomForestRegressor(n_estimators=500,min_samples_leaf=2,oob_score=True,random_state=SEED).fit(eval_X,eval_y)
print('Illustrative OOB R²:',oob_demo.oob_score_)
# A fixed 7/3 random split, not a reliable benchmark or an extra independent population.
from sklearn.model_selection import train_test_split
perm_train,perm_test=train_test_split(np.arange(len(eval_y)),test_size=.3,random_state=SEED)
perm_model=RandomForestRegressor(n_estimators=100,min_samples_leaf=2,random_state=SEED).fit(eval_X[perm_train],eval_y[perm_train])
perm=permutation_importance(perm_model,eval_X[perm_test],eval_y[perm_test],scoring='neg_mean_absolute_error',n_repeats=50,random_state=SEED)
print('Permutation test user IDs:',df.user_id.iloc[perm_test].tolist())
fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained')
axes[0].bar(eval_table.model,eval_table.iloc[:,1],color=BLUE); axes[0].set(ylabel='LOOCV MAE (cm length)',title='Same target, same held-out scheme')
axes[1].barh(['Age','Experience','Width'],perm.importances_mean,xerr=perm.importances_std,color=GOLD,capsize=4); axes[1].axvline(0,color='0.4'); axes[1].set(xlabel='Increase in test MAE (cm length)',title='Permutation importance: only 3 test users'); plt.show()
say(f'خط پایهٔ LOOCV برابر {eval_rows[0][1]:.3f} و RF برابر {eval_rows[1][1]:.3f} سانتی‌متر طول خطا دارد. OOB R²={oob_demo.oob_score_:.3f} یک برآورد داخلی جنگل است. اهمیت جایگشتی روی فقط سه فرد کنارگذاشته‌شده بسیار ناپایدار است؛ میلهٔ خطا SD تکرار جایگشت است، نه CI جمعیتی. انتخاب مدل با این ده فرد به اعتبارسنجی مستقل نیاز دارد.')
````

#### راهنمای کامل خواندن خروجی و نمودار

دقت آموزش را از دقت کنارگذاشته‌شده جدا بخوانید. در bootstrapِ جنگل، احتمال کنارماندن یک مشاهده از یک بازنمونهٔnتایی (۱−۱/n)ⁿ و برایn بزرگ حدود۳۶٫۸٪ است. OOB از همین درخت‌ها برای آن مشاهده استفاده می‌کند. در اهمیت جایگشتی، یک ستونِ آزمون را جابه‌جا و افزایش خطا را می‌سنجیم؛ همبستگی ویژگی‌ها روی این تفسیر هم اثر دارد.

#### کاربرد و محدودیت

با جست‌وجوی مکرر تنظیمات روی یک مجموعهٔ آزمون، آن مجموعه دیگر آزمونِ دست‌نخورده نیست. با دادهٔ کافی از ارزیابی تودرتو یا آزمون مستقل استفاده کنید. R² آزمون حتی می‌تواند منفی شود؛ یعنی از میانگینِ مرجعِ آن ارزیابی ضعیف‌تر است.

#### تمرین

اگر هر فرد در پنج روز اندازه‌گیری شده، جداسازی تصادفیِ سطرها چه مشکلی دارد؟

#### پاسخ تشریحی

اطلاعات همان فرد می‌تواند هم درآموزش و هم درآزمون باشد و تعمیم به فرد جدید خوش‌بینانه شود. GroupKFold یا طرح متناسب با فرد و زمان را بررسی کنید.

## فصل: مدل‌های پیش‌بینی و ارزیابی

### درس 36. طبقه‌بندی LDA

**هدف یادگیری:** طبقه‌بندی، برچسب و ارزیابی کنارگذاشته‌شده را بشناسید.

#### مفهوم و تشریح کامل

LDA از میانگین کلاس‌ها و کوواریانس مشترک، تابع تصمیم خطی می‌سازد. در این تمرین برچسبِ B برابر طول دست>۲۱ است و پیش‌بین‌ها عرض و سن‌اند. این برچسب ترجیح واقعی سایز ماوس نیست. در LOOCV هر بار یک فرد کنار گذاشته و تبدیل و مدل روی نه فرد دیگر آموزش داده می‌شود.

#### فرمول و قرارداد

```text
δk(x)=xᵀΣ⁻¹μk−0.5μkᵀΣ⁻¹μk+log πk
```

#### مثال حل‌شدهٔ مرجع

ماتریس خطای کنارگذاشته‌شده برای ترتیب A/B برابر [[۶،۰]،[۲،۲]] است. ازده فرد۸ درست‌اند و دقت۸۰٪ می‌شود. برای B، حساسیت۲÷۴=۵۰٪ و برای A نرخ تشخیص۶÷۶=۱۰۰٪ است؛ پس دقت کلی بخشی از ضعف تشخیص B را پنهان می‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
y_class = (df.hand_length_cm>21).astype(int).to_numpy()
X_class = df[['hand_width_cm','age_years']].to_numpy()
lda_pipe = make_pipeline(StandardScaler(), LinearDiscriminantAnalysis())
loo = LeaveOneOut()
lda_cv_pred = cross_val_predict(lda_pipe, X_class, y_class, cv=loo, method='predict')
print('LOOCV confusion matrix [A,B]:')
display(pd.DataFrame(confusion_matrix(y_class, lda_cv_pred), index=['actual A','actual B'],
                     columns=['pred A','pred B']))
print('LOOCV accuracy:', round((lda_cv_pred==y_class).mean(),3))

cm=confusion_matrix(y_class,lda_cv_pred); class_sensitivity=cm[1,1]/cm[1].sum(); class_precision=cm[1,1]/cm[:,1].sum()
fig,ax=plt.subplots(figsize=(5,4)); ax.imshow(cm,cmap='Blues',vmin=0)
ax.set(xticks=[0,1],xticklabels=['≤21 cm','>21 cm'],yticks=[0,1],yticklabels=['≤21 cm','>21 cm'],xlabel='Predicted class',ylabel='Actual class',title='LDA LOOCV confusion counts')
for i in range(2):
    for j in range(2): ax.text(j,i,str(cm[i,j]),ha='center',va='center',bbox=dict(facecolor='white',edgecolor='none',alpha=.8))
plt.show()

say(f'در LOOCV دقت={(lda_cv_pred==y_class).mean():.0%}، حساسیت کلاس مثبت={class_sensitivity:.0%} و precision={class_precision:.0%} است. کلاس مثبت طول بیشتر از ۲۱ سانتی‌متر است؛ تنها چهار مثبت واقعی داریم، پس دقت ۸۰٪ به‌تنهایی کافی نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

سطر ماتریس کلاس واقعی و ستون کلاس پیش‌بینی است. جای سطر و ستون را با قرارداد نمودار کنترل کنید. دو خطای B→A دربارهٔ از دست رفتن کاربران کلاس B است و هزینهٔ کاربردیِ آن می‌تواند از یک خطای نوع دیگر متفاوت باشد.

#### کاربرد و محدودیت

برای توصیهٔ اندازه، هدف واقعی را با آزمون تناسب و راحتی ثبت کنید. برچسب‌سازی از طول و سپس استفاده از همان طول به‌عنوان ورودی، مسئله را تقریباً به بازسازی یک آستانه تبدیل می‌کند.

#### تمرین

در این خروجی تعداد مثبت کاذب برای کلاسB چقدر است؟

#### پاسخ تشریحی

صفر؛ هیچ فردA بهB پیش‌بینی نشده است. منفی کاذب برایB برابر۲ است و با وجود دقت۸۰٪، حساسیتB فقط۵۰٪ است.

### درس 37. ROC/AUC برای LDA

**هدف یادگیری:** دقت، آستانه و توان رتبه‌بندی را از هم تشخیص دهید.

#### مفهوم و تشریح کامل

ROC با تغییر آستانهٔ امتیاز، TPR را مقابل FPR رسم می‌کند. AUC توان رتبه‌بندی یک مثبت در برابر یک منفی را خلاصه می‌کند؛ دقت در یک آستانه یا کالیبراسیون احتمال نیست. امتیاز باید از فردی به دست آمده باشد که در برازش آن مدل استفاده نشده است.

#### فرمول و قرارداد

```text
TPR=TP/(TP+FN); FPR=FP/(FP+TN)
```

#### مثال حل‌شدهٔ مرجع

برای آستانهٔ طبقه‌بندی LDAِ این نمونه، TP=۲، FN=۲، TN=۶ وFP=۰ است؛ بنابراین TPR=۰٫۵، FPR=۰، precision=۱ و accuracy=۰٫۸ است. precision یک‌بودن، از دست‌رفتن نصف مثبت‌ها را پنهان نمی‌کند. منحنی از امتیازهای احتمالِ LOOCV ساخته شده است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
lda_cv_prob = cross_val_predict(lda_pipe, X_class, y_class, cv=loo, method='predict_proba')[:,1]
auc = roc_auc_score(y_class, lda_cv_prob)
fpr,tpr,thresholds = roc_curve(y_class, lda_cv_prob)
fig, ax = plt.subplots()
ax.plot(fpr,tpr,marker='o',color=BLUE,label=f'LOOCV AUC = {auc:.2f}')
ax.plot([0,1],[0,1],ls='--',color='gray',label='Chance line')
ax.set(xlabel='False positive rate',ylabel='True positive rate',xlim=(0,1),ylim=(0,1.05),
       title='LDA ROC from held-out predictions')
ax.legend(); plt.show()

say(f'AUC کنارگذاشته‌شده={auc:.4f}؛ خط چین مرجع رتبه‌بندی تصادفی است. AUC توان رتبه‌بندی دو کلاس را توصیف می‌کند و دقت در یک آستانه یا کالیبراسیون احتمال نیست؛ n=10 عدم‌قطعیت زیادی دارد.')
````

#### راهنمای کامل خواندن خروجی و نمودار

خط قطری مرجعِ رتبه‌بندی تصادفی است؛ بالا–چپ معمولاً مطلوب‌تر است، اما هزینهٔ خطاها و کاربرد انتخاب آستانه را تعیین می‌کنند. با فقط۴ مثبت و۶ منفی، پله‌های منحنی درشت‌اند و عدد AUC ناپایدار است. حتی تجمیع امتیازهای foldهای مختلف محدودیتِ مقایسهٔ امتیاز دارد.

#### کاربرد و محدودیت

اگر کلاس مثبت نادر باشد، precision–recall و تعداد واقعی خطاها را هم گزارش کنید. انتخاب آستانه روی دادهٔ آزمون نهایی، آزمون را به بخشی از تنظیم مدل تبدیل می‌کند.

#### تمرین

اگر پیش‌بینی همهٔ افراد را منفی کنیم، حساسیتB وFPR چیست؟

#### پاسخ تشریحی

هر دو صفر می‌شوند. FPR کم به‌تنهایی کافی نیست، چون هیچ مثبت واقعی کشف نشده است.

### درس 38. جنگل تصادفی و اهمیت متغیرها

**هدف یادگیری:** جنگل تصادفی و اهمیت ویژگی را در چارچوب پیش‌بینی بخوانید.

#### مفهوم و تشریح کامل

جنگل از درخت‌های متعدد روی بازنمونه‌ها و زیرمجموعه‌های تصادفی ویژگی‌ها استفاده می‌کند. میانگین‌گیری پیش‌بینی‌ها نوسان درخت منفرد را کاهش می‌دهد، اما دادهٔ کم را پرحجم نمی‌کند. اهمیت ناخالصی سهم کاهش خطای تقسیم‌های آموزشی است و علت یا ارزش طراحی را نمی‌سنجد.

#### فرمول و قرارداد

```text
ŷ(x)=(1/B)Σb tree_b(x); MAE=mean(|y−ŷ|)
```

#### مثال حل‌شدهٔ مرجع

پاسخ این مدل طول دست است و ورودی‌ها سن، سابقه و عرض‌اند. خطای آموزش≈۰٫۶۸۷ سانتی‌متر و LOOCV≈۱٫۲۰۱ است. اهمیت آموزشی عرض≈۰٫۸۶۱، سابقه≈۰٫۰۸۲ و سن≈۰٫۰۵۷ است. افزایش خطا در فرد کنارگذاشته‌شده، تفاوت یادگیری و حفظ‌کردن را نشان می‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from sklearn.ensemble import RandomForestRegressor
X_reg = df[['age_years','gaming_years','hand_width_cm']].to_numpy()
y_reg = df.hand_length_cm.to_numpy()
rf = RandomForestRegressor(n_estimators=100, min_samples_leaf=2, random_state=SEED)
rf_cv = cross_val_predict(rf, X_reg, y_reg, cv=loo)
rf.fit(X_reg,y_reg)
print(f'Train MAE={np.mean(abs(y_reg-rf.predict(X_reg))):.3f} cm; LOOCV MAE={np.mean(abs(y_reg-rf_cv)):.3f} cm')
display(pd.DataFrame({'feature':['age','experience','width'], 'training impurity importance':rf.feature_importances_}).round(3))

prediction_figure(y_reg,rf_cv,'RF LOOCV: hand length','Hand length (cm)')

say(f'RF: MAE آموزش={np.mean(abs(y_reg-rf.predict(X_reg))):.3f} و LOOCV={np.mean(abs(y_reg-rf_cv)):.3f} سانتی‌متر طول. اختلاف دو خطا خوش‌بینیِ آموزش را نشان می‌دهد؛ اهمیتِ ناخالصی سنجهٔ اثر علّی نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

اهمیت‌ها در این تعریف جمعاً۱ می‌شوند؛ مقدار۰٫۸۶۱ به معنی «۸۶٫۱٪ اثر علّی عرض» نیست. همبستگی ویژگی‌ها می‌تواند اهمیت را بین آن‌ها تقسیم کند. مقایسهٔ مدل‌ها باید با پاسخ، افراد آزمون و واحد خطای یکسان انجام شود.

#### کاربرد و محدودیت

برای دادهٔ بزرگ‌تر، اهمیت جایگشتی را روی دادهٔ کنارگذاشته‌شده بسنجید. OOB نیز برای هر نمونه از درخت‌هایی استفاده می‌کند که آن نمونه در بازنمونهٔ آموزشی‌شان نبوده است؛ در نمونهٔ خیلی کوچک خودِ این برآورد ناپایدار است.

#### تمرین

آیا MAE≈۱٫۲۰۱ یعنی خطای هر فرد دقیقاً۱٫۲۰۱ است؟

#### پاسخ تشریحی

خیر؛ این میانگین قدرمطلق خطاهای افراد است. خطای فردی می‌تواند کوچک‌تر یا بزرگ‌تر باشد و باید توزیع خطا را هم دید.

### درس 39. شبکهٔ عصبی کوچک MLP

**هدف یادگیری:** شبکهٔ عصبی را با داده و خطای مناسب ارزیابی کنید.

#### مفهوم و تشریح کامل

MLP ترکیب‌های خطی و تبدیل غیرخطی را پشت سر هم می‌گذارد. تعداد پارامترها می‌تواند نسبت به ده مشاهده زیاد شود، بنابراین استانداردسازی، منظم‌سازی و ارزیابیِ درست مهم‌اند. در این اجرا یک لایهٔ پنهان چهارواحدی و solver=lbfgs به کار رفته است.

#### فرمول و قرارداد

```text
h=f(W1z+b1); ŷ=W2h+b2; سه ورودی → چهار واحد پنهان → یک پاسخ
```

#### مثال حل‌شدهٔ مرجع

هدف این مدل عرض دست است. MAE کنارگذاشته‌شده≈۰٫۲۳۴ سانتی‌متر شده است. برای کاربر۱، مشاهده۸ و پیش‌بینیِ کنارگذاشته‌شده≈۸٫۲۶ است، یعنی خطا≈+۰٫۲۶؛ برای فرد۹، مشاهده۱۱ و پیش‌بینی≈۱۰٫۶۳ است، یعنی خطا≈−۰٫۳۷.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from sklearn.neural_network import MLPRegressor
X_mlp = df[['hand_length_cm','age_years','gaming_years']].to_numpy()
y_mlp = df.hand_width_cm.to_numpy()
mlp = make_pipeline(StandardScaler(), MLPRegressor(hidden_layer_sizes=(4,), solver='lbfgs',
                        alpha=1.0, max_iter=1000, random_state=SEED))
mlp_cv = cross_val_predict(mlp, X_mlp, y_mlp, cv=loo)
print(f'MLP LOOCV MAE={np.mean(abs(y_mlp-mlp_cv)):.3f} cm')
display(pd.DataFrame({'observed_width_cm':y_mlp,'held_out_prediction_cm':mlp_cv}).round(2))

prediction_figure(y_mlp,mlp_cv,'MLP LOOCV: hand width','Hand width (cm)')

say(f'MLP: MAE کنارگذاشته‌شده={np.mean(abs(y_mlp-mlp_cv)):.3f} سانتی‌متر عرض دست. هدف این مدل با RF که طول را پیش‌بینی می‌کند فرق دارد؛ مقایسهٔ مستقیم دو MAE برای انتخاب الگوریتم معتبر نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

این خطا را با MAE جنگلِ درس قبل مستقیم رتبه‌بندی نکنید؛ آن مدل طول را پیش‌بینی می‌کرد و این یکی عرض را. کاهش loss آموزش نیز با خطای افراد آینده یکسان نیست. lbfgs الزاماً یک منحنی loss در هر epoch مانند SGD ارائه نمی‌کند.

#### کاربرد و محدودیت

برای مقایسهٔ عادلانه، یک رگرسیون ساده و یک شبکه را روی همان پاسخ، همان foldها و همان معیار بسنجید. تنظیم مدل با دیدن نتایج همهٔ foldها می‌تواند برآورد را خوش‌بینانه کند؛ برای تنظیم جدی ارزیابیِ تودرتو یا آزمون مستقل لازم است.

#### تمرین

چرا scaler باید داخل pipeline و داخل fold fit شود؟

#### پاسخ تشریحی

میانگین وSD افراد آزمون نباید در پیش‌پردازش آموزشی دخالت کند. pipeline از ورود این اطلاعات به مدلِ آن fold جلوگیری می‌کند.

### درس 40. رگرسیون Gradient Boosting

**هدف یادگیری:** یادگیری ترتیبی و مقایسه با خط پایه را ببینید.

#### مفهوم و تشریح کامل

Gradient Boosting با مدل ساده شروع و در هر گام یک یادگیرنده برای اصلاح خطا اضافه می‌کند. در زیان مربع، هدفِ گام بعدی باقیمانده‌های فعلی است. نرخ یادگیری کوچک، سهم هر درخت را محدود می‌کند؛ تعداد گام و عمق باید با ارزیابی مناسب تنظیم شوند.

#### فرمول و قرارداد

```text
Fm(x)=Fm−1(x)+νhm(x); برای square loss: residual=y−Fm−1(x)
```

#### مثال حل‌شدهٔ مرجع

مدل۳۰ درختِ عمق۱ با نرخ۰٫۰۵ برای پیش‌بینی طول دارد. LOOCV MAE≈۱٫۲۳۵ سانتی‌متر است؛ خط پایهٔ میانگینِ نه فرد آموزش≈۲٫۲۲۲ است. نسبت کاهشِ این خطا در همین ارزیابی حدود۴۴٫۴٪ است؛ این به معنی افزایش۴۴٫۴٪ فروش یا اثر طراحی نیست.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from sklearn.ensemble import GradientBoostingRegressor
gb = GradientBoostingRegressor(n_estimators=30, learning_rate=.05, max_depth=1,
                               min_samples_leaf=2, random_state=SEED)
gb_cv = cross_val_predict(gb, X_reg, y_reg, cv=loo)
baseline_cv = np.array([y_reg[np.arange(len(y_reg))!=i].mean() for i in range(len(y_reg))])
print(f'Boosting LOOCV MAE={np.mean(abs(y_reg-gb_cv)):.3f} cm')
print(f'Mean baseline LOOCV MAE={np.mean(abs(y_reg-baseline_cv)):.3f} cm')
gb.fit(X_reg,y_reg)
fig, ax = plt.subplots()
ax.plot(gb.train_score_,color=BLUE)
ax.set(xlabel='Boosting stage',ylabel='Training loss',title='Training loss (not a validation curve)')
plt.show()

say(f'Boosting LOOCV MAE={np.mean(abs(y_reg-gb_cv)):.3f} در برابر خط پایه={np.mean(abs(y_reg-baseline_cv)):.3f} سانتی‌متر طول. نمودار کاهش زیان فقط آموزش را نشان می‌دهد؛ مقایسهٔ مناسب با RF روی همین هدف ممکن است، با MLP عرض خیر.')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور افقی منحنی loss تعداد مراحل boosting و عمودی خطای آموزش است. نزول آن نشان می‌دهد مدل دادهٔ آموزش را بهتر بازسازی می‌کند. برای تعیین نقطهٔ توقف باید منحنیِ اعتبارسنجی را داشته باشیم؛ کمینهٔ آموزش معمولاً مدل مناسبِ آینده را تضمین نمی‌کند.

#### کاربرد و محدودیت

جنگل وBoosting در این مثال پاسخ وfoldهای مشترک دارند و MAE آن‌ها نزدیک است؛ با این n اختلاف کوچک را برتری پایدار اعلام نکنید. پیچیدگی، پایداری و قابلیت توضیح را هم مقایسه کنید.

#### تمرین

اگر loss آموزش کم شود ولی خطای اعتبارسنجی زیاد شود، نشانهٔ چه وضعیتی است؟

#### پاسخ تشریحی

بیش‌برازش محتمل است؛ مدل جزئیات خاص آموزش را یاد می‌گیرد. توقف، منظم‌سازی یا مدل ساده‌تر را با طرح ارزیابی مناسب بررسی کنید.

### درس 41. Autoencoder و فضای نهفته

**هدف یادگیری:** بازسازیِ خوب را از کشفِ ساختار واقعی جدا کنید.

#### مفهوم و تشریح کامل

Autoencoder ورودی را به گلوگاه کم‌بُعد می‌برد و دوباره بازسازی می‌کند. در این مثال ورودی و هدف هر دو چهار ویژگی استانداردشده‌اند؛ پس مسئله نظارت‌شدهٔ پیش‌بینی عرض از بقیه نیست. مدل کم‌ظرفیت می‌تواند ساختار مشترک را فشرده کند، ولی بازسازی آموزشی به‌تنهایی تعمیم را نمی‌سنجد.

#### فرمول و قرارداد

```text
z=tanh(Xscaled We+be); X̂=decoder(z); loss=mean((Xscaled−X̂)²)
```

#### مثال حل‌شدهٔ مرجع

معماری۴→۲→۴ است و MSE بازسازیِ آموزش≈۰٫۰۶۰۹ در واحد استانداردشده ثبت شده است. اجرای ذخیره‌شده هشدار همگراییِ lbfgs دارد؛ بنابراین همین خروجی را با قیدِ همگرا نشدن گزارش می‌کنیم و از آن ادعای مدل نهایی نمی‌سازیم.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
auto = MLPRegressor(hidden_layer_sizes=(2,), activation='tanh', solver='adam',
                    alpha=.01, max_iter=20000, tol=1e-7, n_iter_no_change=100, learning_rate_init=.01, random_state=SEED).fit(X_scaled, X_scaled)
latent = np.tanh(X_scaled @ auto.coefs_[0] + auto.intercepts_[0])
reconstruction = auto.predict(X_scaled)
print(f'Training reconstruction MSE in standardized units={mean_squared_error(X_scaled,reconstruction):.4f}')
fig, ax = plt.subplots()
ax.scatter(latent[:,0],latent[:,1],color=TEAL)
for i,(a,b) in enumerate(latent): ax.annotate(str(i+1),(a,b),xytext=(4,4),textcoords='offset points')
ax.set(xlabel='Latent unit 1',ylabel='Latent unit 2',title='Two-neuron autoencoder bottleneck (training data)')
plt.show()

say(f'MSE بازسازی آموزشی={mean_squared_error(X_scaled,reconstruction):.5f} در واحد استانداردشده است. دو نورون گلوگاه کد فشرده‌اند، نه دو ویژگی فیزیکی. نسخهٔ حاضر با Adam، alpha=0.01، نرخ یادگیری 0.01 و توقف پس از {auto.n_iter_} تکرار اجرا شد؛ عدد اجرای قدیمیِ lbfgs در مثال مرجع حفظ شده ولی نتیجهٔ این اجرا نیست. توقف بهینه‌ساز و اعتبار خارج از نمونه دو موضوع جدا هستند.')
````

#### راهنمای کامل خواندن خروجی و نمودار

هر نقطهٔ فضای نهفته یک فرد است؛ مختصات آن طول یا سن نیست. جهت محور و فاصله‌ها به وزن‌های آموزش وابسته‌اند. برای بازسازی در واحد اصلی باید تبدیل معکوس scaler انجام شود؛ خطای عددیِ استاندارد را مستقیم سانتی‌متر نخوانید.

#### کاربرد و محدودیت

مدل را باPCA، دادهٔ کنارگذاشته‌شده و چند بذر مقایسه کنید. با ده نفر، یک شبکهٔ پیچیده می‌تواند تقریباً حفظ کند؛ وجود جزیره در فضای نهفته دلیل تعیین سه سایز نیست.

#### تمرین

چرا MSE=۰٫۰۶۰۹ را نباید «خطای۰٫۰۶۰۹ سانتی‌متر» نوشت؟

#### پاسخ تشریحی

داده‌ها استاندارد شده‌اند و خروجی میانگین مربع خطا روی چهار ویژگی با واحدهای اولیهٔ متفاوت است. برای تفسیر فیزیکی باید خطای هر ستون را پس از تبدیل معکوس سنجید.

## فصل: بقا و پیش‌بینی زمانی

### درس 42. تحلیل بقا Kaplan–Meier

**هدف یادگیری:** خرابی و سانسور را در منحنی بقا تفکیک کنید.

#### مفهوم و تشریح کامل

تابع بقا S(t) احتمالِ باقی‌ماندن بدون رخداد تا بعد از t را توصیف می‌کند. Kaplan–Meier در زمان رخداد با ضریب۱−d/nr افت می‌کند. سانسور راست یعنی می‌دانیم محصول تا زمان پیگیری کار کرده ولی زمان خرابی بعدی‌اش را ندیده‌ایم؛ سانسور، خرابی نیست.

#### فرمول و قرارداد

```text
Ŝ(t)=∏(1−di/ni), برای زمان‌های رخداد ti≤t
```

#### مثال حل‌شدهٔ مرجع

مثال موجود ساختگی است و سه گروه۵تاییِ بدون سانسور دارد. در گروه سبک، خرابی‌ها۲۴،۲۶،۲۸،۳۰،۳۲ ماه‌اند؛ بعد از نخستین خرابی S=۴/۵=۰٫۸ و بعد از دومین S=۰٫۶ است. میانهٔ بقا نخستین زمان S≤۰٫۵ است: سبک۲۸، متوسط۲۲ و سنگین۱۴ ماه.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
survival = {'light':[24,26,28,30,32], 'medium':[18,20,22,24,26],
            'heavy':[10,12,14,16,18]}
def km_all_events(times):
    values, counts = np.unique(times, return_counts=True)
    at_risk, probability, steps = len(times), 1.0, [(0,1.0)]
    for t,d in zip(values, counts):
        probability *= 1-d/at_risk
        steps.append((t,probability))
        at_risk -= d
    return np.asarray(steps)
fig, ax = plt.subplots()
for name, times in survival.items():
    steps = km_all_events(times)
    ax.step(steps[:,0],steps[:,1],where='post',label=f'{name} (median {np.median(times):.0f} mo)')
ax.set(xlabel='Months to simulated failure',ylabel='Survival probability',ylim=(0,1.05),
       title='Kaplan–Meier curves: simulated failures only')
ax.legend(); plt.show()

say(f'میانه‌های بقای ساختگی: سبک={np.median(survival["light"]):.0f}، متوسط={np.median(survival["medium"]):.0f}، سنگین={np.median(survival["heavy"]):.0f} ماه. هر ۱۵ مشاهده رخداد دارد؛ این مثال شاهد گارانتی واقعی نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور افقی ماه و عمودی احتمال بقا ازصفر تا۱ است. پلهٔ افقی به معنی نبودِ رخداد ثبت‌شده در آن فاصله است. اگر سانسور وجود داشته باشد، تعداد در معرض خطر کاهش می‌یابد ولی خود سانسور افتِ احتمال ایجاد نمی‌کند. نقاط انتهایی با افراد در معرض خطرِ کم، عدم‌قطعیت بیشتری دارند.

#### کاربرد و محدودیت

برای گارانتی باید دادهٔ واقعیِ تست، وضعیت رخداد، شرایط استفاده، علت خروج و هزینهٔ خدمات ثبت شود. میانه‌های دادهٔ ساختگی هیچ ضمانتی دربارهٔ عمر محصول تولیدی ایجاد نمی‌کنند.

#### تمرین

محصولی تا ماه۱۸ کار کرده و سپس مطالعه تمام شده است. status آن چیست؟

#### پاسخ تشریحی

در تعریف رایج، status=۰ یعنی سانسور در۱۸؛ status=۱ برای رخداد خرابی است. ثبت آن به‌عنوان خرابی۱۸ماهه منحنی را غلط می‌کند.

### درس 43. سانسور و آزمون Log-rank، قدم‌به‌قدم

**هدف یادگیری:** افت منحنی بقا و مقایسهٔ دو منحنی را با شمارشِ در معرض خطر یاد بگیرید.

#### مفهوم و تشریح کامل

برای مقایسهٔ دو منحنی، Log-rank در هر زمان رخداد تعداد رخدادِ مشاهده‌شدهٔ یک گروه را با تعداد مورد انتظار تحت فرض صفر مقایسه می‌کند. تعداد در معرض خطر باید شامل افرادِ زنده و هنوز تحت پیگیری باشد. سانسور مستقل از خطر رخداد، مشروط به اطلاعات مناسب، از فرض‌های مهم تفسیر است.

#### فرمول و قرارداد

```text
E1j=dj×n1j/nj; U=Σ(O1j−E1j); χ²=U²/Var(U), df=1
```

#### مثال حل‌شدهٔ مرجع

یک مثال مستقلِ چهارتایی در زمان‌های ۲، ۳، ۴، ۵ داریم: در ۲ یک خرابی، در ۳ یک سانسور، در ۴ یک خرابی و در ۵ یک سانسور. S پس از ۲ برابر ۳/۴=۰٫۷۵ است؛ سانسور ۳ آن را کم نمی‌کند. قبل از ۴ دو محصول در معرض خطرند، پس S پس از ۴ برابر ۰٫۷۵×(۱−۱/۲)=۰٫۳۷۵ است. این همان میانهٔ بقای برآوردی ۴ است. در مقایسهٔ جداگانهٔ گروه‌های ساختگیِ سبک و سنگین، χ²≈۹٫۷۰۰۷ با یک درجهٔ آزادی و p≈۰٫۰۰۱۸۴۲ به دست می‌آید؛ این عدد توصیفِ سناریوی ساختگی است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from statsmodels.duration.survfunc import SurvfuncRight, survdiff
light=np.array([24,26,28,30,32]); heavy=np.array([10,12,14,16,18])
lr_time=np.r_[light,heavy]; lr_status=np.ones(10,dtype=int); lr_group=np.r_[np.zeros(5,dtype=int),np.ones(5,dtype=int)]
lr_chi,lr_p=survdiff(lr_time,lr_status,lr_group)
example=SurvfuncRight(np.array([2,3,4,5]),np.array([1,0,1,0]))
display(pd.DataFrame({'event_time':example.surv_times,'estimated_survival':example.surv_prob}))
lr_rows=[]
for t in np.unique(lr_time):
    n1=(light>=t).sum(); n2=(heavy>=t).sum(); d1=(light==t).sum(); d2=(heavy==t).sum(); n=n1+n2; d=d1+d2
    variance=n1*n2*d*(n-d)/(n*n*(n-1)) if n>1 else 0
    lr_rows.append([t,n1,n2,d1,d2,d*n1/n,variance])
lr_table=pd.DataFrame(lr_rows,columns=['time','risk light','risk heavy','events light','events heavy','expected light','variance'])
display(lr_table.round(4)); lr_u=(lr_table['events light']-lr_table['expected light']).sum(); lr_manual=lr_u**2/lr_table.variance.sum()
assert np.isclose(lr_manual,lr_chi)
fig,ax=plt.subplots(figsize=(8,4)); ax.step(np.r_[0,example.surv_times,5],np.r_[1,example.surv_prob,example.surv_prob[-1]],where='post',color=BLUE)
ax.scatter([3,5],[.75,.375],marker='+',s=140,color=GOLD,label='Censoring (no drop)'); ax.axhline(.5,color='0.4',ls='--'); ax.set(xlabel='Time (constructed units)',ylabel='Estimated survival probability',ylim=(0,1.05),title='Censoring removes risk time, not survival probability'); ax.legend(); plt.show()
say(f'مثال سانسور: S(2)=0.75 و S(4)=0.375؛ میانهٔ بقا ۴ است. مقایسهٔ جداگانهٔ سبک/سنگینِ ساختگی χ²={lr_chi:.4f} و p={lr_p:.6f} می‌دهد؛ محاسبهٔ دستی جدول دقیقاً با کتابخانه تطبیق دارد. سانسور ۳ و ۵ منحنی را پایین نمی‌آورد. این شواهد از دادهٔ خرابی صنعتی نیست.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در مثال ساختگیِ سه گروه قبلی، هر۱۵ محصول خرابی دارند؛ status همه۱ است. اگر دو گروه سبک و سنگین را باکد زیر مقایسه کنیم، p صرفاً خروجیِ همان سناریوی ساختگی است. در دادهٔ واقعی، هم‌پوشانی و عبور منحنی‌ها مهم است و یکp همهٔ شکل اختلاف را خلاصه نمی‌کند.

#### کاربرد و محدودیت

میانهٔ بقا، نرخ خرابی و هزینهٔ گارانتی پرسش‌های متفاوت‌اند. برای اعتبار صنعتی، شرایط آزمایش و پیگیریِ خروج‌کرده‌ها باید ثبت شوند. متغیرهای استفاده یا گروه‌های تصادفی‌نشده، استنباط علّی را محدود می‌کنند.

#### تمرین

چرا حذفِ کامل محصولاتِ هنوز سالم از جدول، منحنی را عوض می‌کند؟

#### پاسخ تشریحی

آن‌ها تا زمان سانسور در معرض خطر بوده‌اند. حذفشان مخرجِ رخدادهای قبلی را کاهش می‌دهد و معمولاً بقا را کمتر از واقع نشان می‌دهد.

### درس 44. سری زمانی و پیش‌بینی فروش

**هدف یادگیری:** پیش‌بینی زمانی را با گذشتهٔ مجاز ارزیابی کنید.

#### مفهوم و تشریح کامل

در سری زمانی ترتیب داده بخشی از مسئله است. ماه‌های آینده نباید در برازش یا انتخاب تنظیماتِ پیش‌بینیِ گذشته استفاده شوند. خط پایهٔ فصلی مقدار همان ماهِ سال قبل را پیش‌بینی می‌کند؛ خط روند فقط تغییر خطی متوسط را می‌بیند. دادهٔ۲۴ماههٔ این درس نمایشی است.

#### فرمول و قرارداد

```text
seasonal naive: ŷt=yt−12; MAEtest=mean(|yt−ŷt|), t=19,…,24
```

#### مثال حل‌شدهٔ مرجع

ماه‌های۱ تا۱۸ آموزش و۱۹ تا۲۴ آزمون‌اند. برای ماه۱۹ مقدار واقعی۲۴هزار و فصلیِ ساده۲۲هزار است؛ خطای مطلق۲هزار. MAE شش ماهِ آزمون برای فصلی۲ و برای روند≈۲٫۴۷۸هزار است. پیش‌بینی فصلیِ ماه‌های۲۵ تا۳۰ برابر[۱۴،۱۶،۱۷،۱۹،۲۰،۲۲]هزار است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
sales = np.array([12,14,15,17,18,20,22,21,19,18,16,15,
                  14,16,17,19,20,22,24,23,21,20,18,17], dtype=float)
months = np.arange(1,25)
train_end = 18
seasonal_test = sales[np.arange(train_end,24)-12]
trend_coef = np.polyfit(months[:train_end],sales[:train_end],deg=1)
trend_test = np.polyval(trend_coef,months[train_end:])
test_actual = sales[train_end:]
comparison = pd.DataFrame({'month':months[train_end:], 'actual_demo_sales_k':test_actual,
                           'seasonal_naive_k':seasonal_test, 'linear_trend_k':trend_test})
display(comparison.round(2))
for name,pred in [('seasonal naive',seasonal_test),('linear trend',trend_test)]:
    print(name, 'holdout MAE (thousand units)=',round(np.mean(abs(test_actual-pred)),3))
future_months = np.arange(25,31)
future_forecast = sales[future_months-13]
display(pd.DataFrame({'future_month':future_months,'seasonal_naive_forecast_k':future_forecast}))
fig, ax = plt.subplots()
ax.plot(months,sales,marker='o',color=BLUE,label='Illustrative history')
ax.plot(future_months,future_forecast,marker='o',ls='--',color=GOLD,label='Seasonal-naive forecast')
ax.axvline(train_end+.5,color='gray',ls=':',label='Holdout begins')
ax.set(xlabel='Month index',ylabel='Illustrative sales (thousand units)',
       title='24-month demo series and six-month seasonal baseline')
ax.legend(); plt.show()

say(f'MAE شش ماه آخر: فصلی ساده={np.mean(abs(test_actual-seasonal_test)):.3f} و روند خطی={np.mean(abs(test_actual-trend_test)):.3f} هزار واحد. پیش‌بینی آیندهٔ ماه‌های ۲۵–۳۰={future_forecast.tolist()}. سری ساختگی است و بازهٔ عدم‌قطعیت آینده در این خط پایه برآورد نشده است.')
````

#### راهنمای کامل خواندن خروجی و نمودار

بخش تاریخی مشاهدهٔ سریِ نمایشی و بخش خط‌چین ادامهٔ خط پایه است. محل شروع آزمون را ببینید؛ به‌هم‌زدن تصادفی ماه‌ها می‌تواند آینده را به آموزش ببرد. بدون مدل عدم‌قطعیت و دادهٔ کافی، نوار اطمینانِ ساختگی روی شکل قرار نمی‌دهیم.

#### کاربرد و محدودیت

برای برنامه‌ریزی تولید واقعی، موجودی، کمبود، کمپین، قیمت، مرجوعی و تغییر بازار را ثبت کنید. پیش‌بینی تقاضا با فروشِ محدودشده به موجودی یکی نیست. روش‌های پیچیده باید از خط پایهٔ ساده در ارزیابیِ زمانی بهتر باشند.

#### تمرین

برای پیش‌بینیِ ماه۱۹ به روش فصلی ساده، از کدام ماه استفاده می‌شود؟

#### پاسخ تشریحی

ماه۷، چون۱۹−۱۲=۷. برای پیش‌بینی ماهِ آینده باید فقط اطلاعاتی به کار رود که در لحظهٔ پیش‌بینی واقعاً موجود بوده است.

### درس 45. میانگین متحرک، تجزیه و Holt–Winters

**هدف یادگیری:** هموارسازیِ تاریخ، تجزیهٔ الگو و پیش‌بینیِ آینده را جدا بخوانید.

#### مفهوم و تشریح کامل

میانگین متحرک trailing از زمان جاری و گذشته استفاده می‌کند؛ میانگین centered برای توصیف می‌تواند از آیندهٔ همان نقطه هم استفاده کند و برای پیش‌بینیِ آنلاین بدون تغییر مناسب نیست. تجزیهٔ جمعی، سری را به روند، فصل و باقیمانده می‌شکند. Holt–Winters سطح و روند و الگوی فصلی را به‌صورت بازگشتی به‌روزرسانی می‌کند.

#### فرمول و قرارداد

```text
MA3(t)=(y[t−2]+y[t−1]+y[t])/3; y[t]=T[t]+S[t]+R[t]
ℓ[t]=α(y[t]−s[t−m])+(1−α)(ℓ[t−1]+b[t−1])
b[t]=β(ℓ[t]−ℓ[t−1])+(1−β)b[t−1]
s[t]=γ(y[t]−ℓ[t])+(1−γ)s[t−m]
forecast at t+h = ℓ[t]+h×b[t]+appropriate seasonal index
```

#### مثال حل‌شدهٔ مرجع

سه فروش آخرِ سری نمایشی ۲۰، ۱۸، ۱۷ هزار است، پس MA3 در ماه ۲۴ برابر ۱۸٫۳۳۳ هزار می‌شود. این هموارسازیِ تاریخ است. برای یک گامِ نمونهٔ Holt–Winters جمعی، سطح قبلی ۱۸، روند قبلی ۱، فصلِ متناظرِ قبلی ۲ و مشاهدهٔ تازه ۲۳ را بگیرید. با α=۰٫۳، β=۰٫۲ و γ=۰٫۴، سطح جدید ۰٫۳×(۲۳−۲)+۰٫۷×(۱۸+۱)=۱۹٫۶ است؛ روند جدید ۰٫۲×(۱۹٫۶−۱۸)+۰٫۸×۱=۱٫۱۲ و جزء فصلی جدید ۰٫۴×(۲۳−۱۹٫۶)+۰٫۶×۲=۲٫۵۶ است. این مثالِ یک گام، مستقل از برازش خودکارِ سری ۲۴ماهه است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.seasonal import seasonal_decompose
series=pd.Series(sales,index=pd.date_range('2020-01-01',periods=24,freq='MS'))
ma3=series.rolling(3).mean(); decomp=seasonal_decompose(series,model='additive',period=12)
fig,ax=plt.subplots(figsize=(10,4)); ax.plot(series,marker='o',color=BLUE,label='Constructed history'); ax.plot(ma3,color=GOLD,ls='--',label='Trailing MA3'); ax.set(ylabel='Demo sales (thousand units)',xlabel='Month',title='Trailing moving average: smoothing, not validated forecasting'); ax.legend(); plt.show()
fig=decomp.plot(); fig.set_size_inches(10,8); fig.suptitle('Additive decomposition: only two constructed seasonal cycles',y=1.01); fig.tight_layout(); plt.show()
hw=ExponentialSmoothing(series,trend='add',seasonal='add',seasonal_periods=12,initialization_method='estimated').fit()
hw_future=hw.forecast(6); display(pd.DataFrame({'forecast_demo_sales_k':hw_future}).round(3))
fig,ax=plt.subplots(figsize=(10,4)); ax.plot(series,color=BLUE,label='History used to fit'); ax.plot(hw.fittedvalues,color=TEAL,ls=':',label='In-sample fit'); ax.plot(hw_future,color=GOLD,marker='o',ls='--',label='Illustrative forecast (no interval)'); ax.set(xlabel='Month',ylabel='Demo sales (thousand units)',title='Holt–Winters fitted on all 24 months: not a holdout evaluation'); ax.legend(); plt.show()
level=.3*(23-2)+.7*(18+1); trend=.2*(level-18)+.8*1; seasonal=.4*(23-level)+.6*2
say(f'آخرین MA3={ma3.iloc[-1]:.3f} هزار واحد است. مثال یک گام مستقل: سطح={level:.2f}، روند={trend:.2f}، فصل={seasonal:.2f}. پیش‌بینی خودکار HW شش ماه بعد {np.round(hw_future.to_numpy(),3).tolist()} است. همهٔ ۲۴ ماه برای برازش مصرف شده‌اند؛ این نمودار ارزیابی آزمون یا بازهٔ تضمین‌شده نیست.')
````

### درس 46. مطالعهٔ واقعی ۱: چارک‌های ANSUR II و پوشش اندازه

**هدف یادگیری:** روش چارک و تفسیر را روی هزاران اندازه‌گیری واقعی اجرا کنید.

#### مفهوم و تشریح کامل

نسخهٔ منتخب ANSUR II شامل ۶٬۰۶۸ فرد، ۴٬۰۸۲ مرد و ۱٬۹۸۶ زن از نظامیان آمریکاست. طول، عرض و طول کف دست در فایل اصلی میلی‌مترند. این جامعه با کاربران ایرانیِ ماوس یکی نیست؛ تعداد زیاد سطر به‌تنهایی نمایندگی بازار هدف را ایجاد نمی‌کند. گروه و متغیر را پیش از مقایسه مشخص کنید.

#### فرمول و قرارداد

```text
type 7 quantile: h=(n−1)p; cm=mm/10
```

#### مثال حل‌شدهٔ مرجع

برای طول دست مردان Q1=۱۸۷، میانه=۱۹۳ و Q3=۲۰۰ میلی‌متر است؛ IQR=۱۳ و حصارها ۱۶۷٫۵ و ۲۱۹٫۵ میلی‌مترند. برای زنان Q1=۱۷۴، میانه=۱۸۰ و Q3=۱۸۷ است؛ IQR نیز ۱۳ میلی‌متر است. میانه‌ها ۱۳ میلی‌متر فرق دارند؛ برابر بودن IQR به معنی توزیع یا بازهٔ پوششِ برابر نیست.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
# Editable analysis parameters; run this cell again after changing them.
ANSUR_SEX='male'  # 'male', 'female', or 'all'; do not infer a target-market sex mix.
ANSUR_VARIABLE='handlength'  # handlength / handbreadth / palmlength (mm)
ansur_view=ansur if ANSUR_SEX=='all' else ansur.loc[ansur.sex.eq(ANSUR_SEX)]
av=ansur_view[ANSUR_VARIABLE].to_numpy(float); aq=np.quantile(av,[.05,.25,.5,.75,.95],method='linear'); aiqr=aq[3]-aq[1]; afences=(aq[1]-1.5*aiqr,aq[3]+1.5*aiqr)
summary_ansur=ansur.groupby('sex')[ANSUR_VARIABLE].agg(['count','min','mean','median','max'])
for p in [.05,.25,.75,.95]: summary_ansur[f'P{100*p:g}']=ansur.groupby('sex')[ANSUR_VARIABLE].quantile(p,interpolation='linear')
display(summary_ansur.round(3))
fig,axes=plt.subplots(1,2,figsize=(12,4.5),layout='constrained')
axes[0].hist(av,bins=35,color=BLUE,edgecolor='white');
for q,label,ls in zip(aq[1:4],['Q1','Median','Q3'],['--','-',':']): axes[0].axvline(q,color=GOLD,ls=ls,label=f'{label}={q:g} mm')
axes[0].set(xlabel=f'{ANSUR_VARIABLE} (mm)',ylabel='People',title=f'ANSUR II: {ANSUR_SEX}, n={len(av):,}'); axes[0].legend()
axes[1].boxplot([ansur.loc[ansur.sex.eq(s),ANSUR_VARIABLE] for s in ['female','male']],tick_labels=['Female (n=1,986)','Male (n=4,082)'],showfliers=True); axes[1].set(ylabel=f'{ANSUR_VARIABLE} (mm)',title='Separate populations, same measurement scale'); plt.show()
az=np.sort(av); fig,ax=plt.subplots(); ax.step(az,np.arange(1,len(az)+1)/len(az),where='post',color=BLUE); ax.axhline(.95,color=GOLD,ls='--'); ax.axvline(aq[-1],color=GOLD,ls=':'); ax.set(xlabel=f'{ANSUR_VARIABLE} (mm)',ylabel='Empirical cumulative share',ylim=(0,1.03),title='Percentile coverage in the selected ANSUR population'); plt.show()
say(f'انتخاب فعلی: {ANSUR_SEX}، متغیر {ANSUR_VARIABLE}، n={len(av):,}. چارک‌ها {aq[1]:.2f}، {aq[2]:.2f} و {aq[3]:.2f} میلی‌متر؛ IQR={aiqr:.2f} و حصارها [{afences[0]:.2f},{afences[1]:.2f}] هستند. صدک ۹۵={aq[-1]:.2f} میلی‌متر توصیف همین گروه است؛ طول بدنهٔ ماوس از آن به‌طور مستقیم به دست نمی‌آید. نمایندگی کاربران ایرانی یا بازار هدف ادعا نمی‌شود.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در آزمایشگاه چارک‌ها ابتدا «طول دست/مردان» را انتخاب کنید؛ جای جعبه و میانه را با همین اعداد تطبیق دهید. سپس گروه را عوض و حرکت جعبه را ببینید. تغییر متغیر به عرض دست، واحد را ثابت نگه می‌دارد اما معنای اندازه و دامنه را تغییر می‌دهد. علامت‌های بیرون حصار، خطای قطعی نیستند.

#### کاربرد و محدودیت

برای فرضیهٔ سایزبندی، ابتدا جامعهٔ هدفِ واقعی و پوشش مطلوب را تعریف کنید، سپس نسبت‌های جمعیتی را با طرح نمونه‌گیری و وزن‌دهی مناسب وارد کنید. تقسیم دلخواهِ طول به سه بخش، تناسب چندبعدی یا تعداد مدل تجاری را تضمین نمی‌کند.

#### تمرین

اگر Q3=۲۰۰ میلی‌متر باشد، آیا طول بدنهٔ ماوس باید ۲۰ سانتی‌متر شود؟

#### پاسخ تشریحی

خیر. این صدکِ اندازهٔ دست است و رابطهٔ مستقیم با طول بدنه از آن استخراج نشده است. ابعاد محصول باید با هندسهٔ گرفتن و آزمون تناسب تعیین شوند.

### درس 47. مطالعهٔ واقعی ۲: گشتاورهای تقاضای ساعتی

**هدف یادگیری:** توان‌های فاصله و حساسیت به انتهاها را در ۱۷٬۳۷۹ ساعت ببینید.

#### مفهوم و تشریح کامل

در Bike Sharing هر سطر یک تاریخ×ساعت است؛ cnt تعداد اجاره و نه تعداد افراد یکتاست. این فایل همهٔ ساعت‌های تقویمی را ندارد و ساعت غایب را نباید خودکار صفر کرد. دادهٔ دو سال، فصل و زمان روز دارد؛ مخلوط این وضعیت‌ها بر شکل توزیع اثر می‌گذارد.

#### فرمول و قرارداد

```text
m2=Σ(x−x̄)²/n; g1=m3/m2^(3/2); β2=m4/m2²
```

#### مثال حل‌شدهٔ مرجع

برای کل فایل، میانگین≈۱۸۹٫۴۶۳، میانه ۱۴۲، SD توصیفی≈۱۸۱٫۳۸۲ و بیشینه ۹۷۷ اجاره در ساعت است. چولگی≈۱٫۲۷۷، کشیدگی≈۴٫۴۱۶ و اضافی≈۱٫۴۱۶ است. تعداد ۹۳۸ ساعت، یعنی حدود ۵٫۴٪، حداقل ۲SD از میانگین دورند و حدود ۸۱٫۷٪ مجموع توان چهارم فاصله‌ها را می‌سازند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
BIKE_VARIABLE='cnt'  # cnt / casual / registered
BIKE_YEAR='all'  # 'all', 2011, 2012
bv_frame=bike if BIKE_YEAR=='all' else bike.loc[bike.dteday.dt.year.eq(int(BIKE_YEAR))]
bv=bv_frame[BIKE_VARIABLE].to_numpy(float); bm=bv.mean(); bc=bv-bm; bsd=bv.std(ddof=0); bz=bc/bsd
braw=np.array([np.mean(bv**o) for o in range(1,5)]); bcentral=np.array([np.mean(bc**o) for o in range(1,5)])
bskew=np.mean(bz**3); bkurt=np.mean(bz**4); tail=np.abs(bz)>=2; tailshare=np.sum(bz[tail]**4)/np.sum(bz**4)
display(pd.DataFrame({'order':[1,2,3,4],'raw moment (rental^order)':braw,'central moment (rental^order)':bcentral}))
display(pd.DataFrame({'n':[len(bv)],'mean':[bm],'median':[np.median(bv)],'SD descriptive':[bsd],'skew moment':[bskew],'excess moment':[bkurt-3],'skew adjusted':[stats.skew(bv,bias=False)],'excess adjusted':[stats.kurtosis(bv,bias=False)]}).round(5))
fig,axes=plt.subplots(1,2,figsize=(12,4.5),layout='constrained'); axes[0].hist(bv,bins=45,color=BLUE,edgecolor='white'); axes[0].axvline(bm,color=GOLD,label=f'Mean={bm:.1f}'); axes[0].axvline(np.median(bv),color='0.3',ls='--',label=f'Median={np.median(bv):.1f}'); axes[0].set(xlabel=f'{BIKE_VARIABLE}: rentals per observed hour',ylabel='Observed hours',title=f'Bike Sharing {BIKE_YEAR}: n={len(bv):,}'); axes[0].legend()
axes[1].bar([0,1],[100*tail.mean(),100*tailshare],color=[BLUE,GOLD]); axes[1].set(xticks=[0,1],xticklabels=['Share of hours','Share of total z⁴'],ylabel='Percent (different denominators)',ylim=(0,100),title='Hours with |z| ≥ 2: frequency vs fourth-moment influence');
for i,v in enumerate([100*tail.mean(),100*tailshare]): axes[1].text(i,v+2,f'{v:.1f}%',ha='center')
plt.show()
fig,axes=plt.subplots(1,2,figsize=(12,4),layout='constrained'); axes[0].hist(bz,bins=50,density=True,color=BLUE,alpha=.65,label='Standardized observed rentals'); ng=np.linspace(-3,5,250); axes[0].plot(ng,stats.norm.pdf(ng),color=GOLD,label='Standard normal reference'); axes[0].set(xlabel='z = (rentals − mean)/descriptive SD',ylabel='Density',title='Shape comparison, not a normality conclusion'); axes[0].legend(); stats.probplot(bv,dist='norm',plot=axes[1]); axes[1].set(title='Normal Q–Q of observed hourly rentals',ylabel='Rentals per observed hour'); plt.show()
say(f'فیلتر فعلی {BIKE_VARIABLE}/{BIKE_YEAR}: n={len(bv):,}، میانگین={bm:.3f}، میانه={np.median(bv):.1f} و SD={bsd:.3f}. چولگی گشتاوری={bskew:.3f} و کشیدگی={bkurt:.3f} (اضافی={bkurt-3:.3f}) است. {tail.sum():,} ساعت ({tail.mean():.2%}) با |z|≥۲، سهم {tailshare:.2%} از Σz⁴ دارند. میلهٔ دوم تعداد ساعت نیست. این فایل ساعت‌های ثبت‌شدهٔ ۲۰۱۱–۲۰۱۲ است، نه افراد یکتا یا همهٔ ساعات تقویم.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در هیستوگرام، بیشتر ساعت‌ها نسبت به بیشینه کم‌تقاضاترند و انتهای راست گسترده است. در نمودار سهمِ چهارم، درصد ساعت‌ها و درصد Σz⁴ مخرج‌های متفاوت دارند. ستون طلایی بزرگِ گروه انتهایی به معنی فراوانی بیشترِ آن گروه نیست؛ به معنی وزن بسیار زیادِ فاصله‌های آن در توان چهار است.

#### کاربرد و محدودیت

متغیر را به casual یا registered و سال را به ۲۰۱۲ تغییر دهید. در cnt سال ۲۰۱۲ چولگی≈۰٫۹۸۷ و کشیدگی≈۳٫۳۷ است؛ پس تفسیر کل داده را برای هر زیرگروه کپی نکنید. برازش توزیع تقاضا نیازمند توجه به ساعت روز و فصل است.

#### تمرین

چرا درصد ۵٫۴٪ ساعت با سهم ۸۱٫۷٪ توان چهار تناقض ندارد؟

#### پاسخ تشریحی

یکی بر تعداد ساعت‌ها تقسیم شده، دیگری بر مجموع توان چهارم فاصله‌ها. چند فاصلهٔ بزرگ می‌توانند سهم بیشترِ گشتاور چهارم را بسازند.

### درس 48. مطالعهٔ واقعی ۳: همبستگی، رتبه و ترکیب زیرگروه‌ها

**هدف یادگیری:** یک رابطه را با جهت، شدت، شکل و محدودهٔ معتبر گزارش کنید.

#### مفهوم و تشریح کامل

آزمایشگاه همبستگی دما، رطوبت یا شاخص باد را مقابل تعداد کل، گذری یا عضو قرار می‌دهد. تمام ضرایب از همهٔ زوج‌های همان فیلتر حساب می‌شوند. برای دما و نام فصل بین README بسته و صفحهٔ فعلی UCI اختلاف وجود دارد؛ بنابراین دما را به‌صورت شاخص نرمال‌شدهٔ فایل و فصل‌ها را با کد ۱ تا ۴ گزارش می‌کنیم. رطوبت ۱۰۰×hum درصد است.

#### فرمول و قرارداد

```text
Pearson: linear association; Spearman: Pearson of average-tie ranks
```

#### مثال حل‌شدهٔ مرجع

در کل ساعت‌ها، شاخص دما با cnt پیرسون≈۰٫۴۰۵ و اسپیرمن≈۰٫۴۲۳ دارد و r²≈۰٫۱۶۴ است. برای رطوبت با cnt، پیرسون≈−۰٫۳۲۳ و اسپیرمن≈−۰٫۳۶۰ است. در فصل کد ۱ سال ۲۰۱۲، پیرسونِ رطوبت≈−۰٫۲۵۸ می‌شود. تبدیل خطیِ مثبتِ شاخص دما ضرایب را عوض نمی‌کند. تغییر با فیلتر، توصیفِ جامعهٔ انتخاب‌شده است، نه اثبات اینکه فیلتر علت رابطه بوده است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
CORR_X='temp'  # normalized temp / hum / windspeed
CORR_Y='cnt'  # cnt / casual / registered
CORR_YEAR='all'  # all / 2011 / 2012
CORR_SEASON='all'  # all / 1 / 2 / 3 / 4 (codes only: conflicting metadata)
cf=bike.copy()
if CORR_YEAR!='all': cf=cf.loc[cf.dteday.dt.year.eq(int(CORR_YEAR))]
if CORR_SEASON!='all': cf=cf.loc[cf.season.eq(int(CORR_SEASON))]
assert len(cf)>3
cx=cf[CORR_X].to_numpy(float)*(100 if CORR_X=='hum' else 1); cy=cf[CORR_Y].to_numpy(float)
cr=stats.pearsonr(cx,cy).statistic; crho=stats.spearmanr(cx,cy).statistic; cfit=np.polyfit(cx,cy,1)
cx_label={'temp':'Normalized temperature index (original file)','hum':'Relative humidity (%)','windspeed':'Normalized windspeed index'}[CORR_X]
fig,axes=plt.subplots(1,2,figsize=(12,4.5),layout='constrained'); axes[0].scatter(cx,cy,s=9,alpha=.12,color=BLUE,rasterized=True); cgrid=np.linspace(cx.min(),cx.max(),100); axes[0].plot(cgrid,np.polyval(cfit,cgrid),color=GOLD,lw=2); axes[0].set(xlabel=cx_label,ylabel=f'{CORR_Y}: rentals per hour',title=f'All {len(cf):,} selected hours; descriptive linear fit')
bins=pd.cut(cx,bins=10); cbin=pd.DataFrame({'band':bins,'x':cx,'rentals':cy}).groupby('band',observed=True).agg(x=('x','mean'),mean_rentals=('rentals','mean'),n=('rentals','size'))
axes[1].plot(cbin.x,cbin.mean_rentals,marker='o',color=BLUE); axes[1].set(xlabel=cx_label,ylabel='Mean rentals within band',title='Binned averages hide within-band spread')
for row in cbin.itertuples(): axes[1].annotate(f'n={row.n}',(row.x,row.mean_rentals),xytext=(0,8),textcoords='offset points',ha='center',fontsize=8)
plt.show(); display(cbin.round(3))
correlation_rows=[]
for (year,season_code),part in cf.groupby([cf.dteday.dt.year,'season']):
    correlation_rows.append([year,season_code,len(part),part[CORR_X].corr(part[CORR_Y]),part[CORR_X].corr(part[CORR_Y],method='spearman')])
display(pd.DataFrame(correlation_rows,columns=['year','season code','observed hours','Pearson r','Spearman rho']).round(4))
check_mask=(bike.dteday.dt.year.eq(2012)&bike.season.eq(1)); humidity_sub_r=bike.loc[check_mask,'hum'].corr(bike.loc[check_mask,'cnt'])
say(f'انتخاب {CORR_X} در برابر {CORR_Y}، سال={CORR_YEAR}، فصل کد={CORR_SEASON}: n={len(cf):,}، r={cr:.4f}، ρ={crho:.4f} و r²={cr**2:.3f}. r² سهم توضیح توصیفیِ خط ساده با عرض از مبدأ است، نه سهم اثر علّی. مثال کنترل رطوبت/کل در فصل کد ۱ سال ۲۰۱۲: r={humidity_sub_r:.4f}. همهٔ زوج‌های انتخاب‌شده برای ضرایب و نقاط مصرف شده‌اند؛ ساعت روز و روند رشد می‌توانند مخدوش‌کننده باشند.')
````

#### راهنمای کامل خواندن خروجی و نمودار

در ابر نقاط هر نقطه یک ساعت و شفافیّت نشان‌دهندهٔ تراکم است. خط طلایی برازش خطیِ توصیفی است. در نمودار بازه‌ای، ارتفاع میانگین اجاره و n تعداد ساعت در بازه است؛ میانگین‌گیری می‌تواند پراکندگیِ درون بازه را پنهان کند. r²=۰٫۱۶۴ را «۱۶٫۴٪ اثر علّی دما» ننویسید.

#### کاربرد و محدودیت

سال، فصل و نوع روز را تغییر دهید و جدول گروه‌ها را بخوانید. هوا و تقاضا با زمان روز، تعطیلی، فصل، اندازهٔ شبکه و رشد سیستم درهم‌آمیخته‌اند. cnt از جمع casual و registered ساخته شده، پس همبستگی اجزا با کل بخشی از وابستگیِ تعریف‌شدهٔ جمع را منعکس می‌کند.

#### تمرین

اگر ρ بزرگ‌تر از r باشد، آیا حتماً رابطه علّی است؟

#### پاسخ تشریحی

خیر. ممکن است همراهی رتبه‌ای از نزدیکی به خط قوی‌تر باشد یا مقدارهای انتهایی بر پیرسون اثر گذاشته باشند. شکل نمودار و طرح داده‌برداری را بررسی کنید.

### درس 49. از همهٔ تحلیل‌ها تا یک تصمیم طراحیِ قابل آزمون

**هدف یادگیری:** یک آموزش آماری را به برنامهٔ پژوهش و پروتوتایپ تبدیل کنید.

#### مفهوم و تشریح کامل

زنجیرهٔ تصمیم با هدف کاربر شروع می‌شود: راحتی، دقت، خستگی یا سرعت. ابعاد بدن فقط بخشی از توضیح عملکردند. سپس جامعهٔ هدف و تنوع گرفتن را تعریف، اندازه‌گیری را کالیبره و چند پروتوتایپ می‌سازیم. روش آماری باید به پرسشِ از پیش تعریف‌شده پاسخ دهد؛ زیادکردن تعداد مدل جای طرح پژوهش را نمی‌گیرد.

#### فرمول و قرارداد

```text
پرسش → معیار موفقیت → دادهٔ مناسب → برآورد و عدم‌قطعیت → پروتوتایپ → آزمون مستقل
```

#### مثال حل‌شدهٔ مرجع

برای ده فرد آموزشی، تقسیم ≤۲۰، بیش از ۲۰ تا ۲۳ و >۲۳ به‌ترتیب ۴، ۴ و ۲ نفر می‌دهد. این یک سناریوی S/M/L است. برای آزمون، سه پروتوتایپ را با ترتیب متوازن روی افراد جدید امتحان کنید، دقت وظیفه و امتیاز راحتی را ثبت و اثر تکرارِ هر شخص را در تحلیل لحاظ کنید. آستانه‌های قبول باید از معنای عملی بیایند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
size_scenario=pd.cut(df.hand_length_cm,bins=[-np.inf,20,23,np.inf],labels=['S (≤20)','M (>20 to 23)','L (>23)'],right=True)
size_counts=size_scenario.value_counts(sort=False); display(size_counts.rename('people').to_frame())
# Explicitly constructed engineering tolerance example; these are NOT measurements of manufactured parts.
demo_parts=np.array([113.8,114.2,114.7,114.9,115.0,115.1,115.3,115.8,116.0,116.2]); specification=(114,116)
in_spec=(demo_parts>=specification[0])&(demo_parts<=specification[1])
fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained'); axes[0].bar(size_counts.index.astype(str),size_counts,color=BLUE); axes[0].set(ylabel='People in the 10-person teaching sample',title='An arbitrary size scenario: not market segmentation')
axes[1].scatter(np.arange(1,11),demo_parts,color=BLUE); axes[1].axhspan(*specification,color=GOLD,alpha=.18,label='Engineering specification 114–116 mm'); axes[1].axhline(115,color='0.3',ls='--'); axes[1].set(xlabel='Illustrative part ID',ylabel='Constructed part length (mm)',title='Specification limits are not statistical control limits'); axes[1].legend(); plt.show()
say(f'سناریوی قراردادی سایزها {size_counts.tolist()} فرد است. در مثال مستقلِ کاملاً ساختگیِ قطعه، {in_spec.sum()} از {len(demo_parts)} مقدار در محدودهٔ ۱۱۴–۱۱۶ میلی‌مترند؛ این عدد کیفیت خط تولید واقعی یا حدود کنترل را بیان نمی‌کند. تصمیم نهایی باید با افراد جدید، معیار راحتی/دقت/خستگی، ترتیب متوازن و تحلیل اندازه‌گیری تکراری ارزیابی شود.')
````

#### راهنمای کامل خواندن خروجی و نمودار

جدول تصمیم باید برای هر ادعا شاهد و محدودیت داشته باشد: ANSUR برای اندازهٔ بدنِ جامعهٔ خود، دادهٔ دوچرخه برای آموزش تقاضای ساعتی، نمونهٔ ده‌نفره برای محاسبه، زمان خرابی و فروش ساختگی برای روش. هیچ‌یک به‌تنهایی طول ۱۱٫۵ سانتی‌متر یا گارانتی ۱۸ ماه را اثبات نمی‌کنند.

#### کاربرد و محدودیت

کنترل کیفیت تولید نیز جداست: اگر طول طراحی ۱۱۵ میلی‌متر و تلرانس ±۱ باشد، پایش میانگین و SD قطعه‌های واقعی به کنترل فرایند مربوط است؛ پراکندگی دست کاربران تلرانس ماشین‌کاری نیست. حدود کنترل از رفتار فرایند و حدود مشخصات از نیاز مهندسی می‌آیند.

#### تمرین

پروژهٔ پایانی: برای انتخاب بین ۲ و ۳ سایز، سه خروجی قابل اندازه‌گیری و یک طرح ارزیابی بنویسید.

#### پاسخ تشریحی

مثلاً سهم کاربران دارای راحتی بالاتر از آستانه، خطای دقت در کار استاندارد، و تغییر خستگی پس از زمان معین. طرحِ افراد جدید با اندازه‌های گوناگون، ترتیب متوازن پروتوتایپ‌ها، ثبت اندازه‌گیری تکراری و معیارِ از پیش تعریف‌شده بسازید؛ تصمیم را همراه هزینهٔ تولید و موجودی بسنجید.

## کنترل‌های نهایی و جمع‌بندی

سلول زیر اعداد کلیدی را با محاسبات مستقل کنترل می‌کند: مجموع و واریانس دستی، همبستگی حاصل از حاصل‌ضرب مرکزی، PCA از تجزیهٔ SVD، U رتبه‌ای، Log-rank دستی، تعداد سطر داده‌های واقعی و قرارداد چارک‌ها. این‌ها کنترل نسخهٔ مرجع‌اند؛ اگر عمداً داده‌ها را تغییر داده‌اید، انتظار تطابق با اعداد مرجع نداشته باشید.


````python
assert np.isclose(sum(x)/len(x),21.5)
assert np.isclose(sum((value-21.5)**2 for value in x)/len(x),6.25)
assert np.allclose(np.quantile(x,[.25,.5,.75],method='linear'),[20,21,22.75])
direct_r=np.sum((x-x.mean())*(df.hand_width_cm-df.hand_width_cm.mean()))/np.sqrt(np.sum((x-x.mean())**2)*np.sum((df.hand_width_cm-df.hand_width_cm.mean())**2))
assert np.isclose(direct_r,corr.loc['hand_length_cm','hand_width_cm'])
_,sv,_=np.linalg.svd(X_scaled,full_matrices=False)
assert np.allclose(sv[:2]**2/np.sum(sv**2),pca.explained_variance_ratio_)
manual_u=sum((a>b)+.5*(a==b) for a in low_exp for b in high_exp)
assert np.isclose(manual_u,mw.statistic)
assert np.isclose(lr_chi,lr_manual)
assert not full_design and design_rank<design_columns
assert np.allclose(ansur.loc[ansur.sex.eq('male'),'handlength'].quantile([.25,.5,.75]),[187,193,200])
assert len(bike)==17379 and np.isclose(bike.cnt.mean(),189.46308763450142)
assert np.isclose(bike.temp.corr(bike.cnt),.404772275778658)
assert len(future_forecast)==6 and np.allclose(future_forecast,[14,16,17,19,20,22])
say('همهٔ کنترل‌های نسخهٔ مرجع گذشتند؛ ۴۹ درس اجرا شد. داده، واحد، روش، عدم‌قطعیت و مرز تعمیم باید همراه هر نتیجه گزارش شوند. مدل پیچیده جای نمونه‌گیری و آزمایش مستقل را نمی‌گیرد.')
````

## واژه‌نامهٔ کامل دوره

**جامعه و نمونه:** جامعه هدفِ تعمیم است؛ نمونه افراد یا واحدهای واقعاً دیده‌شده‌اند.

**پارامتر و آماره:** پارامتر ویژگی جامعه و آماره تابعی از نمونه است.

**برآورد نقطه‌ای و بازه‌ای:** یک عدد برای پارامتر، در برابر مجموعه‌ای از مقادیر با روش و پوشش مشخص.

**SD و SE:** SD پراکندگی مشاهده‌ها و SE نوسان برآوردگر در تکرار نمونه‌گیری است.

**چگالی و احتمال:** احتمالِ یک بازه از مساحت چگالی به دست می‌آید؛ ارتفاع یک نقطه احتمال آن نقطه نیست.

**چارک و صدک:** آستانه‌های موقعیتیِ توزیع؛ تعریف نرم‌افزاری در نمونهٔ محدود باید مشخص شود.

**مقدار انتهایی و خطای ثبت:** انتهایی‌بودن یک مشاهده لزوماً خطا نیست؛ اصالت با منبع و اندازه‌گیری بررسی می‌شود.

**p-value و α:** p از داده و مدل محاسبه می‌شود؛ α آستانهٔ تصمیمِ از پیش تعیین‌شده است.

**اندازهٔ اثر:** مقدارِ اختلاف یا رابطه با واحد و تعریفِ متناسب با پرسش.

**عدم رد و هم‌ارزی:** عدم رد، نبودِ شواهد کافی است؛ هم‌ارزی نیازمند حاشیه و آزمون مشخص است.

**باقیمانده:** اختلاف مشاهده و مقدار برازش‌شده؛ نشانه‌ای برای بررسیِ ساختار خطای مدل.

**هم‌خطی:** وابستگی شدید پیش‌بین‌ها که می‌تواند ضرایبِ شرطی را ناپایدار کند.

**بارگذاری و score:** ضرایبِ ویژگی‌ها در مؤلفه، در برابر مختصاتِ هر مشاهده روی مؤلفه.

**آموزش و آزمون:** دادهٔ یادگیری مدل، در برابر دادهٔ کنارگذاشته‌شده برای سنجش تعمیم.

**نشت داده:** ورود اطلاعاتی به فرایند آموزش/انتخاب که در لحظهٔ استفاده در دسترس نیست.

**سانسور:** اطلاع ناقص اما مفید دربارهٔ زمان رخداد؛ از جمله سالم‌بودن تا پایان پیگیری.

**اثر متقابل:** تغییر رابطهٔ یک عامل با پاسخ در سطح‌های عامل دیگر.

**خط پایه:** پیش‌بینی ساده‌ای که مدل پیچیده باید ارزش افزوده‌اش را نسبت به آن نشان دهد.

## گام‌های بعدی و پروژهٔ پایانی

۱. فیلترهای سه کارگاه واقعی را تغییر دهید و برای هر انتخاب، n، واحد، برآورد و محدودیت را دوباره گزارش کنید. تفسیر خودکار کنار خروجی تغییر می‌کند؛ مثال مرجع متن ثابت می‌ماند.

۲. یک نتیجهٔ نامعتبر رایج را اصلاح کنید: SD به‌جای CI، p بزرگ به‌جای هم‌ارزی، r² به‌جای اثر علّی، یا خطای آموزش به‌جای آزمون. پاسخ باید به عدد و طراحی مطالعه ارجاع بدهد.

۳. برای تصمیم بین دو و سه سایز ماوس، جمعیت هدف، روش اندازه‌گیری، معیار حداقل اثر مهم، نمونهٔ مستقل و ثبت خستگی/دقت/راحتی را از پیش تعریف کنید. ANSUR و ده فرد آموزشی جای آن آزمایش را نمی‌گیرند.
