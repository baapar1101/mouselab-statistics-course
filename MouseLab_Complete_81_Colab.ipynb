# MouseLab — آموزش یکپارچهٔ ۸۱ درس آمار و دیتا

## هدف و شیوهٔ مطالعه

این فایل دقیقاً با ترتیب ۸۱ درس و ۱۳ فصل سایت هماهنگ است؛ از تعریف سؤال و کیفیت داده تا احتمال، استنباط، علّیت، مدل و تصمیم. تمام متن آموزشی، فرمول‌ها، مثال‌های حل‌شده، کدهای تحلیل، راهنمای نمودار، تمرین‌ها و پاسخ‌های نسخهٔ قبلی حفظ شده‌اند. فقط شماره و فصل تغییر کرده و ورودی‌های لازم برای اجرای مستقل هر مثال به کد اضافه شده‌اند.

۱. از پیوند مستقیم Google Colab باز کنید و **Runtime → Run all** را بزنید؛ آپلود دستی لازم نیست. CPU کافی است و نصب کتابخانه‌های گمشده به اینترنت نیاز دارد. برای ذخیرهٔ تغییرات **File → Save a copy in Drive** را بزنید. اجرا و ذخیره در Colab/Jupyter انجام می‌شود؛ ویوئر سایت موتور اجرای پایتون نیست. اجرای کنترل‌شدهٔ این نسخه در محیط آزمون انجام شده، نه حساب Google شما.

۲. سه سلول آماده‌سازی آغاز فایل را اجرا کنید. سپس هر سلول درس تابعی با ورودی‌های صریح است و می‌تواند دوباره اجرا شود، بدون اینکه `x`، `df` یا مدل‌های درس دیگری را خراب کند. نام تابع با شمارهٔ درس سایت یکی است؛ کد اصلی مثال داخل تابع همچنان قابل ویرایش است. نتیجه‌ها در `lesson_results['شناسهٔ درس']` قابل بررسی‌اند. دادهٔ دست از `make_hand_data()` می‌آید؛ برای تغییر پایه آن تابع را ویرایش و همهٔ سلول‌ها را دوباره اجرا کنید. فیلترهای کارگاه واقعی در تابع همان درس قرار دارند. آزمون‌های نهایی برای نسخهٔ مرجع‌اند و با تغییر عمدی داده باید بازبینی شوند.

۳. تفسیر «اجرای حاضر» از محاسبات همان سلول ساخته می‌شود؛ اعداد مثالِ مرجع در متن ثابت نگه داشته شده‌اند. تغییر داده می‌تواند مثال مرجع را از خروجی تازه متفاوت کند. پیش از بازکردن پاسخ، تمرین را خودتان حل کنید و یک جملهٔ تفسیر بنویسید.

۴. نمودارها محور و عنوان انگلیسی و توضیح فارسی دارند تا نمایش آن‌ها در Colab و محیط‌های مختلف پایدار بماند. کد رسم قابل ویرایش است. CSVهای لازم با SHA-256 داخل فایل فشرده‌اند؛ مسیر محلی، سرور شخصی، رمز و GPU لازم نیست. هش هویت ورودی را کنترل می‌کند، نه صحت علمی هر اندازه‌گیری.

### داده‌ها و مرز نتیجه‌گیری

- ده فرد آموزشی، خرابی‌ها و فروش ماهانهٔ ساختگی: برای یادگیری روش، نه تعمیم به بازار.
- ANSUR II: ۶٬۰۶۸ نظامی آمریکایی با ابعاد دست به میلی‌متر؛ نمایندهٔ خودکار کاربران ایرانی نیست.
- Bike Sharing: ۱۷٬۳۷۹ ساعت ثبت‌شده در ۲۰۱۱–۲۰۱۲؛ هر سطر ساعت است، نه فرد یکتا. ساعت غایب با صفر پر نشده است.

در فایل ساعتی، اختلاف مستندات README بسته با صفحهٔ UCI در تبدیل دما و نام فصل وجود دارد. پس دما و باد در مقیاس نرمال‌شدهٔ خام و فصل فقط با کد ۱–۴ تحلیل می‌شوند. رطوبت در نمودار به درصد تبدیل می‌شود. تبدیل خطی مثبت همبستگی را عوض نمی‌کند؛ اختلاف فراداده نباید به برچسب قطعیِ درجهٔ سلسیوس تبدیل شود.

۳۲ سناریوی مصنوعیِ افزوده اکنون میان فصل‌های مربوط همین مسیر قرار دارند؛ توصیف زیر فقط دربارهٔ آن‌هاست، نه کارگاه‌های دادهٔ واقعی.

همهٔ داده‌های این ۳۲ درس **ساختگی و با سازوکار صریح** هستند؛ عددهای «حقیقت» فقط پارامترهای معلوم شبیه‌سازی‌اند، نه آمار بازار یا پژوهش واقعی. استفادهٔ آموزشی از دادهٔ مصنوعی کمک می‌کند علت خطای روش را ببینیم؛ برای تصمیم عملی باید داده و طرحِ مسئلهٔ خود را بررسی کنید. هیچ توصیهٔ پزشکی، حقوقی یا مالیِ فردی از مثال‌ها استخراج نمی‌شود.

## آماده‌سازی محیط

کد زیر فقط بسته‌های گمشده را نصب می‌کند و نسخه‌های واقعی اجرا را نشان می‌دهد. نسخه‌های الگوریتم و بهینه‌ساز ممکن است خروجی تصادفی یا بهینه‌سازی را اندکی تغییر دهند؛ بذر ۴۲ و محدودیت نخ‌ها برای بازتولیدپذیری تعیین شده‌اند. بسته‌ها: NumPy، pandas، SciPy، Matplotlib، scikit-learn، statsmodels. هیچ کدی از اینترنت دانلود و اجرا نمی‌شود؛ فقط نصب معمول کتابخانه‌های علمی در صورت فقدان آن‌هاست.

قرارداد اجرا: هر درس محیط متغیر محلی و بذر ثابت خودش را دارد. آماده‌سازی بعضی درس‌ها یک برازش مرجع را بی‌صدا بازسازی می‌کند؛ مثلاً برای بررسی باقیمانده در Shapiro، خط OLS همان‌جا محاسبه می‌شود، حتی اگر آموزش کامل رگرسیون در فصل بعد باشد. این آماده‌سازی خروجی آموزشی اضافه تولید نمی‌کند و از نتیجهٔ پنهان درس قبلی استفاده نمی‌کند.

````python
import os
os.environ['OMP_NUM_THREADS']='1'
os.environ['OPENBLAS_NUM_THREADS']='1'
os.environ['MKL_NUM_THREADS']='1'
import importlib.util, subprocess, sys
required = {'numpy':'numpy', 'pandas':'pandas', 'matplotlib':'matplotlib',
            'scipy':'scipy', 'sklearn':'scikit-learn', 'statsmodels':'statsmodels'}
missing = [package for module, package in required.items() if importlib.util.find_spec(module) is None]
if missing:
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-q', *missing])
import numpy as np, pandas as pd, scipy, sklearn, statsmodels, matplotlib
import matplotlib.pyplot as plt
from scipy import stats
from IPython.display import display, Markdown, HTML
from sklearn.base import clone
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import LeaveOneOut, cross_val_predict
from sklearn.metrics import mean_squared_error, r2_score, roc_auc_score, roc_curve, confusion_matrix

SEED = 42
rng = np.random.default_rng(SEED)
plt.rcParams.update({'figure.figsize': (8, 4.5), 'figure.dpi': 110,
                     'axes.spines.top': False, 'axes.spines.right': False,
                     'axes.grid': True, 'grid.alpha': .18, 'font.size': 10})
BLUE, GOLD, TEAL = '#2475a4', '#d09331', '#168a7c'
print('Python:', sys.version.split()[0], '| numpy:', np.__version__, '| scipy:', scipy.__version__,
      '| sklearn:', sklearn.__version__, '| matplotlib:', matplotlib.__version__)
plt.rcParams.update({'axes.prop_cycle':matplotlib.cycler(color=[BLUE,GOLD,TEAL,'#8d638a']), 'savefig.bbox':'tight'})
pd.set_option('display.max_columns',12)
pd.set_option('display.max_rows',20)
def say(text):
    display(Markdown('**تفسیر خروجی اجرای حاضر:** '+text))
def group_figure(groups,labels,title):
    fig,ax=plt.subplots(figsize=(8,4))
    for i,values in enumerate(groups):
        jitter=np.linspace(-.09,.09,len(values))
        ax.scatter(np.full(len(values),i)+jitter,values,color=BLUE,alpha=.8)
        ax.errorbar(i,np.mean(values),yerr=np.std(values,ddof=1),fmt='s',color=GOLD,capsize=6)
    ax.set(xticks=range(len(labels)),xticklabels=labels,ylabel='Hand length (cm)',title=title)
    ax.text(.02,.97,'Gold square ±1 sample SD (NOT a confidence interval)',transform=ax.transAxes,va='top',fontsize=9)
    plt.tight_layout(); plt.show()
def prediction_figure(actual,pred,title,unit):
    fig,ax=plt.subplots(); ax.scatter(actual,pred,color=BLUE,s=60)
    limits=[min(actual.min(),pred.min()),max(actual.max(),pred.max())]
    ax.plot(limits,limits,color='0.4',ls='--',label='Perfect prediction')
    for i,(a,b) in enumerate(zip(actual,pred)): ax.annotate(str(i+1),(a,b),xytext=(3,3),textcoords='offset points')
    ax.set(xlabel='Observed '+unit,ylabel='Held-out predicted '+unit,title=title); ax.legend(); plt.show()
print('pandas:',pd.__version__,'| statsmodels:',statsmodels.__version__)
import statsmodels.api as sm
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neural_network import MLPRegressor

def show():
    plt.show()
    plt.close('all')
````

````python
def make_hand_data():
    df = pd.DataFrame({
        'user_id': range(1, 11),
        'hand_length_cm': [18,19,20,20,21,21,22,23,24,27],
        'hand_width_cm': [8,8.5,9,9,9.5,9.5,10,10,11,12],
        'age_years': [22,25,19,28,24,30,21,26,33,27],
        'gaming_years': [3,6,2,8,5,10,4,7,12,9],
    })
    return df

# Local result dictionaries retain inspectable intermediates for final checks.
lesson_results = {}
STYLE_HAND = {'figure.figsize':(8,4.5),'figure.dpi':110,'font.size':10,
    'axes.spines.top':False,'axes.spines.right':False,'axes.grid':True,
    'grid.alpha':.18,'axes.prop_cycle':matplotlib.cycler(color=[BLUE,GOLD,TEAL,'#8d638a']),
    'savefig.bbox':'tight'}
STYLE_SIM = {**STYLE_HAND,'grid.alpha':.15,
    'axes.prop_cycle':matplotlib.cycler(color=['#277f9f','#c08b2c','#278b80'])}
assert make_hand_data().shape == (10,5)
````

ابتدا کمیت هدف، واحد مشاهده، کیفیت اندازه‌گیری و جامعهٔ هدف را مشخص می‌کنیم؛ هنوز انتخاب آزمون مطرح نیست.

### درس 01. قبل از عدد: سؤال، نوع داده و کمیتی که می‌خواهیم بدانیم

**پرسش شروع:** دو فروشگاه هر دو «۹۰٪ موفقیت» اعلام می‌کنند؛ آیا واقعاً برابرند؟

#### مفهوم، دقیق اما ساده

آمار با انتخاب آزمون شروع نمی‌شود؛ با تعریف کمیت هدف یا estimand شروع می‌شود. می‌خواهیم میانگین زمان انتظار هر فرد را بدانیم، نسبت سفارش‌های موفق را، یا اثر تغییر طراحی را؟ واحد مشاهده، جامعه، بازهٔ زمانی، صورت و مخرج باید روشن باشند. شناسه عددی است اما اندازهٔ کمّی نیست؛ شهر متغیر اسمی است؛ رتبهٔ رضایت ترتیبی است و فاصلهٔ بین رتبه‌ها لزوماً برابر نیست. تعداد خرید گسسته و زمان پاسخ پیوسته است. انتخاب میانگین، میانه یا نسبت باید با این معنا سازگار باشد. اگر یک فرد ده سفارش داشته باشد، میانگین سفارش‌ها همان میانگین افراد نیست. هر دو ممکن است مفید باشند، اما پرسش متفاوتی را پاسخ می‌دهند.

#### فرمول و قرارداد

```text
Overall rate = sum(successes) / sum(eligible trials)
Unweighted average of rates answers a different question.
```

#### مثال حل‌شده

در فروشگاه کوچک، ۹ موفقیت از ۱۰ تلاش و در فروشگاه بزرگ ۵۰ موفقیت از ۱۰۰ تلاش داریم. میانگین سادهٔ نرخ‌ها ۷۰٪ می‌شود؛ ولی نرخ هر تلاش در مجموع ۵۹÷۱۱۰≈۵۳٫۶۴٪ است. عدد اول هر فروشگاه را یک واحد با وزن برابر می‌بیند، عدد دوم هر تلاش را. مثال دوم: درآمدهای ۱۰، ۱۰ و ۱۰۰ میانگین ۴۰ و میانهٔ ۱۰ دارند؛ «کاربر معمولی» با «درآمد به ازای هر کاربر» یک سؤال نیست.

````python
def run_lesson_01(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: estimand
    rng=np.random.default_rng(202601)
    success=np.array([9,50]); total=np.array([10,100]); rates=success/total
    pooled=success.sum()/total.sum(); unweighted=rates.mean()
    display(pd.DataFrame({'shop':['Small','Large'],'success':success,'trials':total,'rate':rates}))
    fig,ax=plt.subplots(); ax.bar(['Small n=10','Large n=100','Pooled n=110'],100*np.r_[rates,pooled],color=[BLUE,BLUE,GOLD]); ax.set(ylabel='Success rate (%)',ylim=(0,100),title='The denominator defines the question'); show()
    say(f'میانگین بدون وزن={unweighted:.2%}؛ نرخ هر تلاش در کل={pooled:.2%}. تفاوت، خطای گردکردن نیست؛ کمیت هدف و وزن‌ها متفاوت‌اند.')
    # END PRESERVED ANALYSIS: estimand
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['estimand'] = run_lesson_01()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

در نمودار، ارتفاع ستون نرخ و برچسب n حجم مخرج را نشان می‌دهد. نرخ تجمیعی بین دو نرخ قرار می‌گیرد و به گروه بزرگ‌تر نزدیک‌تر است. در جدول به جای نگاه‌کردن صرف به درصد، موفقیت و تعداد تلاش را با هم بخوانید. هیچ‌کدام از این اعداد علت عملکرد دو فروشگاه را نمی‌گویند.

#### دام رایج

میانگین درصدها بدون وزن، ترکیب جمعیت را پنهان می‌کند. نسبتِ مجموع‌ها را هم وقتی مخرج‌ها هم‌پوشان‌اند یا تعریف متفاوت دارند، کورکورانه نسازید.

#### تمرین

برای «رضایت کاربران»، هر سفارش باید وزن برابر بگیرد یا هر شخص؟

#### پاسخ تشریحی

به هدف بستگی دارد. برای تجربهٔ فرد معمولی، ابتدا خلاصهٔ هر فرد را بسازید و سپس افراد را وزن دهید. برای کیفیت هر تعامل، سفارش واحد تحلیل است؛ وابستگی سفارش‌های یک فرد را در عدم‌قطعیت لحاظ کنید.

### درس 02. داده‌های خام و کیفیت اولیه

**هدف یادگیری:** از سؤال طراحی به دادهٔ قابل تحلیل برسید.

#### مفهوم و تشریح کامل

جامعهٔ هدف کسانی‌اند که می‌خواهیم تصمیم دربارهٔ آن‌ها معتبر باشد؛ نمونه فقط کسانی است که واقعاً اندازه‌گیری شده‌اند. واحد مشاهده را قبل از شروع مشخص کنید: فرد، محصول یا ساعت. طول و عرض دست متغیر کمّی پیوسته‌اند؛ شناسهٔ فرد عدد اندازه‌گیری نیست. ابتدا روش اندازه‌گیری، واحد، رضایت و معیار ورود را ثابت کنید، سپس سطرها را ثبت کنید.

#### فرمول و قرارداد

```text
هر سطر = یک فرد مستقل؛ هر ستون = یک ویژگی با تعریف و واحد مشخص
```

#### مثال حل‌شدهٔ مرجع

در جدول پایه ۱۰ فرد، چهار ویژگی و صفر خانهٔ گمشده داریم. مجموع طول‌ها ۲۱۵ سانتی‌متر است. این ده عدد برای تمرین در فایل‌های اولیه آمده‌اند؛ ادعای دو میلیون گیمر یا نمونه‌گیری تصادفیِ واقعی از آن‌ها قابل استخراج نیست. برای تبدیل واحد، ۱۸ سانتی‌متر برابر ۱۸۰ میلی‌متر است؛ ترکیب این دو واحد در یک ستون، همهٔ تحلیل‌های بعدی را خراب می‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_02(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # BEGIN PRESERVED ANALYSIS: lesson-1
    df = pd.DataFrame({
        'user_id': range(1, 11),
        'hand_length_cm': [18,19,20,20,21,21,22,23,24,27],
        'hand_width_cm': [8,8.5,9,9,9.5,9.5,10,10,11,12],
        'age_years': [22,25,19,28,24,30,21,26,33,27],
        'gaming_years': [3,6,2,8,5,10,4,7,12,9],
    })
    assert len(df) == 10 and df.user_id.is_unique and not df.isna().any().any()
    assert df.hand_length_cm.between(10, 40).all()
    display(df)
    print('n =', len(df), '| missing cells =', int(df.isna().sum().sum()))
    numeric_cols=['hand_length_cm','hand_width_cm','age_years','gaming_years']

    say(f'این جدول {len(df)} فرد مستقل و مجموع طول {df.hand_length_cm.sum():.0f} سانتی‌متر دارد؛ فقدان مقدار خالی، نمایندگی جامعهٔ هدف را ثابت نمی‌کند.')
    # END PRESERVED ANALYSIS: lesson-1
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-1'] = run_lesson_02()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

جدول را سطری و ستونی بخوانید: کاربر ۱۰ طول دست ۲۷، عرض ۱۲، سن ۲۷ و سابقهٔ ۹ سال دارد. ابتدا معقول‌بودن و تعریف اندازه‌گیری او را بررسی می‌کنیم؛ بزرگ‌بودن مقدار به‌تنهایی مجوز حذف نیست. در دادهٔ واقعی، شناسهٔ یکتا، تکرار اندازه‌گیری و علت مقدارهای خالی را هم ثبت کنید.

#### کاربرد و محدودیت

اگر دو بار از یک فرد اندازه‌گیری کرده‌اید، دو سطر مستقل به دست نیاورده‌اید. می‌توان برای کاهش خطای اندازه‌گیری میانگین تکرارها را گرفت، اما تعداد افراد همچنان همان است.

#### تمرین

اگر یک طول دست به‌جای ۲۱٫۵ با مقدار ۲۱۵ وارد شده باشد، اولین اقدام چیست؟

#### پاسخ تشریحی

واحد و منبع ثبت را بررسی کنید. پس از تأیید میلی‌متر بودن، آن را به سانتی‌متر تبدیل و تغییر را ثبت کنید. حذف یا تقسیم خودکار همهٔ مقدارهای بزرگ بدون بررسی، درست نیست.

### درس 03. خطای اندازه‌گیری، Reliability و توافق ابزارها

**پرسش شروع:** دو خط‌کش همیشه یک سانتی‌متر اختلاف دارند ولی همبستگی‌شان نزدیک یک است؛ آیا توافق دارند؟

#### مفهوم، دقیق اما ساده

همبستگی شدت همراهی را می‌سنجد، نه توافق عددی دو ابزار. یک خطای ثابت می‌تواند همبستگی را دست‌نخورده بگذارد و همهٔ اندازه‌ها را جابه‌جا کند. در Bland–Altman، اختلاف دو ابزار در برابر میانگین آن‌ها رسم می‌شود تا سوگیری ثابت، وابستگی اختلاف به اندازه و پراکندگی دیده شود. حدود توافقِ ساده میانگین اختلاف±۱٫۹۶SD اختلاف‌اند و CI برای میانگین اختلاف نیستند؛ این تعبیر به فرض‌های توزیع و استقلال وابسته است و در دادهٔ تکراری باید اصلاح شود. خطای تصادفی کلاسیکِ مستقل در پیش‌بین، شیب رگرسیون ساده را معمولاً به صفر می‌کشد؛ در خطاهای افتراقی یا پیچیده چنین قاعدهٔ ساده‌ای نداریم.

#### فرمول و قرارداد

```text
Agreement bias=mean(A−B)
Approx. limits of agreement=mean difference±1.96 SD(differences)
Correlation is invariant to an additive constant.
```

#### مثال حل‌شده

۲۰۰ طول واقعیِ ساخته‌شده را حول۲۰ بسازید. ابزار A نویز۰٫۲ و ابزار B علاوه بر نویز۰٫۲، سوگیری۱ واحد دارد. همبستگی می‌تواند بسیار بالا بماند اما اختلاف A−B نزدیک−۱ است. اگر تلرانسِ کاربرد ±۰٫۳ باشد، همبستگی بالا کافی نیست. در مثال مستقلِ خطای پیش‌بین، افزودن نویز به X ممکن است شیب پیش‌بینی y را کم کند؛ آن را با خطای ثبت y یا خطای واحد یکی ندانید.

````python
def run_lesson_03(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: measurement
    rng=np.random.default_rng(202630)
    latent=rng.normal(20,2,200); tool_a=latent+rng.normal(scale=.2,size=200); tool_b=latent+1+rng.normal(scale=.2,size=200); differences=tool_a-tool_b; center=differences.mean(); sd=differences.std(ddof=1)
    fig,ax=plt.subplots(); ax.scatter((tool_a+tool_b)/2,differences,color=BLUE,alpha=.55); ax.axhline(center,color=GOLD,label='Mean difference'); ax.axhline(center-1.96*sd,color='0.4',ls='--',label='Approx. 95% agreement limits'); ax.axhline(center+1.96*sd,color='0.4',ls='--'); ax.set(xlabel='Average of tools A and B (demo length units)',ylabel='A − B (same length units)',title='High correlation does not ensure measurement agreement'); ax.legend(); show()
    say(f'r={np.corrcoef(tool_a,tool_b)[0,1]:.4f} ولی سوگیری A−B={center:.3f} است؛ حدود توافق تقریبی [{center-1.96*sd:.3f},{center+1.96*sd:.3f}]. این‌ها CI میانگین یا حدود مشخصات ابزار نیستند.')
    # END PRESERVED ANALYSIS: measurement
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['measurement'] = run_lesson_03()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

محور افقی میانگین دو اندازه و عمودی اختلاف A−B در همان واحد است. خط وسط میانگین اختلاف و دو خط بیرونی حدود توافق تقریبی‌اند. این خط‌ها حدود مجاز مهندسی نیستند و برای قضاوت کفایت ابزار باید با نیاز عملیِ ازپیش‌تعریف‌شده مقایسه شوند. تعداد محدود مشاهده، برای خود حدود توافق نیز عدم‌قطعیت ایجاد می‌کند که در این نمایش برآورد نشده است.

#### دام رایج

Pearson r بالا یا R² بالا را به‌جای توافق و کالیبراسیون ابزار گزارش نکنید. تغییر واحد یا سوگیری ثابت را با «پرت» رفع نکنید.

#### تمرین

اگر B=A+۱ دقیقاً باشد، همبستگی چند و میانگین اختلاف A−B چند است؟

#### پاسخ تشریحی

با واریانس غیرصفر A، همبستگی دقیقاً۱ و میانگین اختلاف−۱ است. همراهی کامل می‌تواند با عدم توافق ثابت همراه باشد.

### درس 04. مهندسی داده برای تحلیل: Grain، Join، Schema و آزمون

**پرسش شروع:** چرا فقط با اضافه‌کردن جدول کاربران، جمع فروش ناگهان زیاد شد؟

#### مفهوم، دقیق اما ساده

Grain یعنی هر سطر دقیقاً نمایندهٔ چیست. جدول سفارش، سفارش‌قلم و شخص، grain یکسان ندارند. در join، تکرار کلید در سمت بُعد می‌تواند هر سطر واقعیت را چند بار تکثیر کند؛ جمع‌ها دیگر به سؤال اولیه پاسخ نمی‌دهند. پیش از ادغام، یکتایی، نوع رابطهٔ یک‌به‌یک/چندبه‌یک، کلیدهای بی‌تطابق و زمان اعتبار رکورد را بررسی کنید. حذف تکرار با keep=first بدون منطق کسب‌وکار فقط خطا را پنهان می‌کند. Schema شامل نوع، کران، مجازبودن null، واحد و قرارداد معنایی است. آزمون داده فقط syntax نیست: جمع مبلغ، تعداد سفارش یکتا و مقدارهای شناخته‌شده باید قبل و بعد reconcile شوند.

#### فرمول و قرارداد

```text
Fact grain: one row per order
Many-to-one join requires unique dimension keys.
Assertions: unique keys + reconciliation + unmatched-key audit.
```

#### مثال حل‌شده

سه سفارش با user_idهای۱،۲،۲ و مبلغ‌های۱۰۰،۲۰۰،۳۰۰ داریم؛ جمع۶۰۰. در بُعد کاربر، شناسه۲ به اشتباه دو سطر دارد. join خام دو سفارش مربوط به۲ را دو بار می‌شمارد و جمع۱٬۱۰۰ می‌شود. validate="many_to_one" ادغام را متوقف می‌کند. نسخهٔ معتبرِ بُعد با قاعدهٔ اعلام‌شدهٔ «یک رکورد جاری تأییدشده برای هر شخص» جدا ساخته می‌شود. در مثال واقعی، رکورد تاریخی باید با تاریخ مؤثر join شود، نه حذف دلخواه یک سطر.

````python
def run_lesson_04(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: joins
    rng=np.random.default_rng(202629)
    orders=pd.DataFrame({'order_id':[1,2,3],'user_id':[1,2,2],'amount':[100,200,300]}); bad_users=pd.DataFrame({'user_id':[1,2,2],'group':['A','B','B']})
    inflated=orders.merge(bad_users,on='user_id')
    try:
        orders.merge(bad_users,on='user_id',validate='many_to_one')
    except pd.errors.MergeError as exc:
        print('Expected quality guard caught duplicate dimension keys:',str(exc))
    confirmed_users=pd.DataFrame({'user_id':[1,2],'group':['A','B']}); valid=orders.merge(confirmed_users,on='user_id',validate='many_to_one',indicator=True)
    assert valid.order_id.is_unique and valid.amount.sum()==orders.amount.sum() and valid['_merge'].eq('both').all()
    display(valid); fig,ax=plt.subplots(); ax.bar(['Original orders','Unchecked join','Validated join'],[orders.amount.sum(),inflated.amount.sum(),valid.amount.sum()],color=[BLUE,GOLD,TEAL]); ax.set(ylabel='Total amount (constructed units)',title='Duplicate dimension keys silently inflate an aggregate'); show()
    say(f'جمع پایه={orders.amount.sum()}، join غلط={inflated.amount.sum()} و join معتبر={valid.amount.sum()}. کنترل یکتایی، شمارش کلید و تطبیق جمع جلوی یک نتیجهٔ عددیِ ظاهراً معتبر را گرفت.')
    # END PRESERVED ANALYSIS: joins
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['joins'] = run_lesson_04()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

جدول قبل و بعد تعداد سطر و جمع را با هم نشان می‌دهد. یک ردیف بیشتر ممکن است مشروع باشد، مثلاً تبدیل سفارش به قلم؛ اما در آن صورت جمع مبلغ سطح سفارش دیگر بدون اصلاح قابل جمع‌زدن نیست. پیام خطای validate در این درس خروجیِ مورد انتظارِ کنترل کیفیت است و با try/except محدود ثبت می‌شود، نه خطای پنهان‌شدهٔ تحلیل.

#### دام رایج

COUNT(*) و COUNT(DISTINCT order_id) یک چیز نیستند. DISTINCT پس از join همیشه درمان نیست؛ ممکن است رکوردهای واقعیِ هم‌مقدار را حذف کند.

#### تمرین

اگر دو کاربر نام یکسان داشته باشند، آیا نام می‌تواند کلید یکتای join باشد؟

#### پاسخ تشریحی

معمولاً نه؛ کلید پایدار و تعریف‌شده لازم است. یکتایی اتفاقی در فایل کوچک، قرارداد پایدار داده محسوب نمی‌شود. کیفیت کلید و رکوردهای بی‌تطابق را کنترل کنید.

### درس 05. دادهٔ گمشده: MCAR، MAR، MNAR و جایگذاری

**پرسش شروع:** افرادی که امتیاز پایین‌تری دارند، کمتر فرم رضایت را پر می‌کنند؛ میانگین پاسخ‌ها چه معنایی دارد؟

#### مفهوم، دقیق اما ساده

MCAR یعنی احتمال گم‌شدن مستقل از دادهٔ کامل است. MAR یعنی پس از شرطی‌کردن بر اطلاعات مشاهده‌شدهٔ مناسب، گم‌شدن به مقدارِ مشاهده‌نشده وابسته نیست. MNAR وابستگیِ باقی‌مانده به مقدارِ ندیده یا سازوکار نامشاهده را توصیف می‌کند. MAR یک فرض دربارهٔ فرایند است و عموماً فقط از جدولِ ناقص قابل اثبات نیست. حذف سطر، پرکردن با میانگین، مدل‌سازی و multiple imputation اهداف و فرض‌های متفاوت دارند. میانگین‌گذاری، پراکندگی را مصنوعی کم و رابطه‌ها را خراب می‌کند؛ «دیگر خانهٔ خالی نداریم» به معنی بازگرداندن اطلاعات نیست. برای پیش‌بینی، جایگذاری باید داخل هر fold و با آموزش همان fold انجام شود.

#### فرمول و قرارداد

```text
Observed data + missingness mechanism + assumptions → valid analysis
Single mean imputation treats guessed values as if known unless uncertainty is handled separately.
```

#### مثال حل‌شده

متغیر x را برای همه ببینید و y=۵۰+۱۰x+ε بسازید. هرچه x بزرگ‌تر باشد، احتمال ثبت y کمتر شود؛ چون x مشاهده شده و سازوکار فقط به آن وابسته است، این سناریو MAR است. میانگین yهای ثبت‌شده با میانگین کاملِ معلومِ شبیه‌سازی فرق می‌کند. پرکردن همهٔ yهای غایب با میانگین مشاهده‌شده، همان میانگین سوگیر را حفظ و SD را کم می‌کند. اگر ناپاسخ‌دهی به خود y حتی بعد از دانستن x وابسته بود، MNAR و حساسیت‌سنجی مهم می‌شد.

````python
def run_lesson_05(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: missing
    rng=np.random.default_rng(202620)
    xx=rng.normal(size=3000); complete=50+10*xx+rng.normal(scale=4,size=len(xx)); observed=rng.random(len(xx))<1/(1+np.exp(xx)); seen=complete[observed]; filled=np.where(observed,complete,seen.mean())
    fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained'); labels=['Complete simulation','Observed only','Mean-filled']
    axes[0].bar(labels,[complete.mean(),seen.mean(),filled.mean()],color=[GOLD,BLUE,TEAL]); axes[0].set(ylabel='Mean constructed response',title='Mean-filling does not fix selection bias'); axes[1].bar(labels,[complete.std(ddof=1),seen.std(ddof=1),filled.std(ddof=1)],color=[GOLD,BLUE,TEAL]); axes[1].set(ylabel='SD of values',title='Imputation artificially contracts spread');
    for ax in axes: ax.tick_params(axis='x',labelrotation=12)
    show(); say(f'{(~observed).mean():.1%} مقدار y غایب است. میانگین کامل={complete.mean():.2f}، مشاهده‌شده={seen.mean():.2f}؛ پرکردن با میانگین آن را به {filled.mean():.2f} می‌رساند، نه لزوماً حقیقت. SD از {seen.std(ddof=1):.2f} به {filled.std(ddof=1):.2f} می‌رسد.')
    # END PRESERVED ANALYSIS: missing
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['missing'] = run_lesson_05()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

پنل اول میانگین کاملِ قابل‌دانستن فقط در شبیه‌سازی را با میانگین مشاهده‌شده و پرشده مقایسه می‌کند. پنل دوم نشان می‌دهد SD دادهٔ پرشده کوچک شده است. این کاهش، بهبود دقت واقعیِ برآورد نیست؛ یک اثر مکانیکی از تکرار عدد میانگین است. در کار واقعی y کامل را نداریم تا به این سادگی خطا را بسنجیم.

#### دام رایج

پرکردن مقدار گمشده با صفر بدون معنای واقعیِ صفر، یا fitکردن imputer روی کل داده قبل از CV، از خطاهای پرتکرارند.

#### تمرین

اگر همهٔ مقادیر خالی را با میانگین پر کنیم، آیا می‌توان CI را با n کامل و SD جدید محاسبه کرد؟

#### پاسخ تشریحی

آن CI معمولاً بیش از حد مطمئن است؛ مقدارهای جایگذاری‌شده مشاهدهٔ مستقل تازه نیستند و عدم‌قطعیتِ گم‌شدن و جایگذاری باید در روش مناسب لحاظ شود.

### درس 06. سوگیری نمونه‌گیری، انتخاب و وزن‌دهی

**پرسش شروع:** اگر نظرسنجی را فقط در انجمن حرفه‌ای گیمرها بگذاریم، آیا یک میلیون پاسخ مشکل را حل می‌کند؟

#### مفهوم، دقیق اما ساده

خطای تصادفی از نمونه‌ای به نمونهٔ دیگر عوض می‌شود؛ سوگیری یک انحراف نظام‌مند نسبت به کمیت هدف است. چارچوب نمونه‌گیری، عدم پاسخ، نمونهٔ در دسترس و بازماندگان می‌توانند ترکیب داده را تغییر دهند. وزن‌دهی پس‌طبقه‌ای وقتی مفید است که سهم طبقه‌ها در جامعه معلوم باشد، همهٔ طبقه‌های مهم پوشش داشته باشند و پاسخ‌دهندگان درون طبقه‌ها برای کمیت موردنظر مناسب باشند. وزن‌دهی جادو نیست: ویژگیِ مشاهده‌نشده یا گروهی با احتمال انتخاب صفر را از هیچ نمی‌سازد. وزن‌های بسیار بزرگ هم واریانس را بالا می‌برند. باید هم برآورد وزنی و هم محدودیتِ فرایند انتخاب را گزارش کرد.

#### فرمول و قرارداد

```text
Post-stratified mean = Σ population_share[g] × sample_mean[g]
Large n reduces noise, not necessarily selection bias.
```

#### مثال حل‌شده

فرض ساخته‌شده: ۸۰٪ جامعه کاربر عادی با میانگین راحتی ۶ و ۲۰٪ حرفه‌ای با میانگین ۹ هستند؛ میانگین هدف ۶٫۶ است. در نظرسنجی، سهم‌ها برعکس ۲۰٪ و ۸۰٪ شده‌اند و میانگین خام ۸٫۴ به دست می‌آید. اگر میانگین درون هر گروه نماینده باشد و سهم واقعی را بدانیم، وزن‌دهی به ۶٫۶ برمی‌گردد. اگر حرفه‌ای‌های ناراضی اصلاً پاسخ نداده باشند، دانستن سهم حرفه‌ای‌ها به‌تنهایی آن سوگیری درون‌گروهی را رفع نمی‌کند.

````python
def run_lesson_06(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: sampling
    rng=np.random.default_rng(202606)
    population_share=np.array([.8,.2]); survey_share=np.array([.2,.8]); group_mean=np.array([6.,9.])
    target=population_share@group_mean; raw=survey_share@group_mean; weighted=population_share@group_mean
    fig,ax=plt.subplots(); ax.bar(['Target population','Unweighted survey','Post-stratified'],[target,raw,weighted],color=[GOLD,BLUE,TEAL]); ax.set(ylabel='Mean comfort score (constructed)',ylim=(0,10),title='Selection bias survives a large sample'); show()
    say(f'میانگین جامعهٔ ساختگی={target:.1f}، نمونهٔ خام={raw:.1f} و وزن‌دهی با فرض نمایندگی درون طبقه={weighted:.1f}. نتیجهٔ آخر به فرض‌ها وابسته است؛ نمونهٔ بزرگ به‌خودی‌خود پاسخ را معتبر نمی‌کند.')
    # END PRESERVED ANALYSIS: sampling
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['sampling'] = run_lesson_06()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

نمودار، سه برآورد از ترکیب‌های مختلفِ یک سناریوی معلوم است؛ اختلاف اعداد ناشی از تغییر ترکیب است. مقدار هدف را فقط در شبیه‌سازی می‌دانیم؛ در مطالعهٔ واقعی معمولاً باید جامعه و اطلاعات بیرونیِ وزن‌ها را مستند کنیم. فاصلهٔ میانگین خام از هدف با افزایش تکرار همین طرح انتخاب لزوماً کم نمی‌شود.

#### دام رایج

وزن‌های ساخته‌شده از خود پیامد یا اصلاح‌های انتخاب‌شده بعد از دیدن نتیجه، ممکن است سوگیری تازه بسازند.

#### تمرین

اگر هیچ کاربر چپ‌دست وارد نمونه نشده، آیا می‌توان با افزایش وزن راست‌دست‌ها دربارهٔ چپ‌دست‌ها نتیجه گرفت؟

#### پاسخ تشریحی

خیر. پوشش و مثبت‌بودن احتمال ورود لازم است. برای گروه غایب باید داده یا مدلِ انتقال‌پذیری با فرض روشن داشت؛ وزن‌دهی صرف اطلاعات مشاهده‌نشده را ایجاد نمی‌کند.

## فصل 02: توصیف داده و خلاصه‌های مقاوم

مشاهده‌ها را می‌بینیم، مرکز و صدک را محاسبه می‌کنیم و می‌پرسیم مقدار غیرعادی چه معنایی دارد.

### درس 07. نمودار پراکندگی کاربران

**هدف یادگیری:** قبل از هر خلاصه‌سازی، تک‌تک مشاهده‌ها را ببینید.

#### مفهوم و تشریح کامل

نمودار نقاط خام کمک می‌کند دامنه، تجمع و مشاهدهٔ متفاوت را ببینیم. در این شکل محور افقی شناسهٔ فرد است؛ فاصلهٔ شناسه‌ها معنای فیزیکی ندارد. بنابراین از بالا رفتن نقاط نمی‌توان رشد زمانی یا اثر تجربه را نتیجه گرفت. مرتب‌بودن طول‌ها در جدول، خودش ظاهر صعودی شکل را ایجاد کرده است.

```text
نقطهٔ i = (شناسهٔ i، طول دست i)
```

### درس 08. میانگین طول دست

**هدف یادگیری:** مرکز داده و حساسیت میانگین را با عدد بفهمید.

#### مفهوم و تشریح کامل

میانگین نقطهٔ تعادل عددی داده است: مجموع انحراف‌های علامت‌دار از آن صفر می‌شود. میانه مرکزِ ترتیب است و به بزرگی دقیق مشاهدهٔ انتهایی کمتر حساس است. انتخاب شاخص به پرسش وابسته است؛ برای مجموع مصرف، میانگین مهم است و برای توصیف فرد میانی، میانه اغلب خواناتر است.

#### فرمول و قرارداد

```text
x̄ = Σxᵢ / n = 215 / 10 = 21.5 cm
```

#### مثال حل‌شدهٔ مرجع

دو مقدار میانیِ دادهٔ مرتب‌شده هر دو ۲۱ هستند، پس میانه ۲۱ است؛ میانگین ۲۱٫۵ است. اگر فقط برای تحلیل حساسیت ۲۷ را با ۲۴ جایگزین کنیم، مجموع به ۲۱۲ و میانگین به ۲۱٫۲ می‌رسد. اگر ۲۷ را کنار بگذاریم، میانگین نه‌نفره ۱۸۸÷۹≈۲۰٫۸۸۹ است. این‌ها سناریو هستند، نه پاک‌سازی داده.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_08(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()

    # BEGIN PRESERVED ANALYSIS: lesson-3
    x = df.hand_length_cm.to_numpy(dtype=float)
    mean_length = x.mean()
    print(f'Mean = {mean_length:.2f} cm; median = {np.median(x):.2f} cm')
    fig, ax = plt.subplots()
    ax.scatter(df.user_id, x, color=BLUE, label='Observations')
    ax.axhline(mean_length, color=GOLD, lw=2, label=f'Mean {mean_length:.2f} cm')
    ax.set(xlabel='User ID', ylabel='Hand length (cm)', title='Observed lengths and arithmetic mean')
    ax.legend(); plt.show()

    say(f'میانگین {mean_length:.2f} و میانه {np.median(x):.2f} سانتی‌متر است. فاصلهٔ مثبت آن‌ها با انتهای راست سازگار است، اما اثبات شکل جامعه نیست.')
    # END PRESERVED ANALYSIS: lesson-3
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-3'] = run_lesson_08()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

خط افقی شکل میانگین است؛ نقاط بالاتر انحراف مثبت و نقاط پایین‌تر انحراف منفی دارند. فاصله‌ها را با واحد سانتی‌متر بخوانید. تفاوت میانگین و میانه سرنخ شکل توزیع است، اما به‌تنهایی جهت چولگی را در هر توزیع ممکن تعیین نمی‌کند.

#### کاربرد و محدودیت

از میانگین طول دست نمی‌توان طول بهینهٔ بدنه را با نسبت ثابتی مثل یک‌دوم نتیجه گرفت. هندسهٔ گرفتن، عرض، ارتفاع، ترجیح کاربر و معیار عملکرد باید در آزمون پروتوتایپ بررسی شوند.

#### تمرین

اگر به طول دست همهٔ افراد ۲ سانتی‌متر اضافه کنیم، میانگین و میانه چه می‌شوند؟

#### پاسخ تشریحی

هر دو دقیقاً ۲ واحد زیاد می‌شوند: میانگین ۲۳٫۵ و میانه ۲۳. پراکندگی پیرامون میانگین تغییری نمی‌کند.

### درس 09. نمودار جعبه‌ای و چارک‌ها

**هدف یادگیری:** چارک، سبیل و قاعدهٔ دورافتاده را از هم جدا کنید.

#### مفهوم و تشریح کامل

چارک‌ها دادهٔ مرتب‌شده را در احتمال‌های ۰٫۲۵، ۰٫۵ و ۰٫۷۵ خلاصه می‌کنند. برای نمونهٔ محدود، روش‌های درون‌یابی متفاوت‌اند. در این آموزش روش خطی نوع ۷ به کار رفته است: جایگاه صفرمبنا h=(n−1)p و درون‌یابی بین دو همسایه. سبیلِ جعبه تا دورترین مشاهدهٔ داخل حصار می‌رود، نه الزاماً تا مقدار خود حصار.

#### فرمول و قرارداد

```text
Q(p)=x[j]+(h−j)(x[j+1]−x[j]); IQR=Q3−Q1; fences=Q1−1.5IQR, Q3+1.5IQR
```

#### مثال حل‌شدهٔ مرجع

برای Q3، h=۹×۰٫۷۵=۶٫۷۵ است؛ مقدارهای صفرمبنای ۶ و ۷ برابر ۲۲ و ۲۳ هستند، پس Q3=۲۲٫۷۵. Q1=۲۰، میانه=۲۱، IQR=۲٫۷۵ و حصار بالا=۲۶٫۸۷۵ است. بنابراین ۲۷ با این قرارداد علامت می‌خورد. روش میانهٔ دو نیمه Q3=۲۳ و حصار بالا=۲۷٫۵ می‌دهد و آن را علامت نمی‌زند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_09(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)

    # BEGIN PRESERVED ANALYSIS: lesson-5
    q1, median, q3 = np.quantile(x, [.25, .5, .75])
    iqr = q3-q1
    fences = (q1-1.5*iqr, q3+1.5*iqr)
    print({'min':x.min(), 'Q1':q1, 'median':median, 'Q3':q3, 'max':x.max(),
           'IQR':iqr, 'fences':fences, 'flagged':x[(x<fences[0]) | (x>fences[1])].tolist()})
    fig, ax = plt.subplots(figsize=(7, 2.7))
    ax.boxplot(x, orientation='horizontal')
    ax.set(xlabel='Hand length (cm)', title='Five-number summary and potential outliers')
    plt.show()

    say(f'چارک خطی نوع ۷: Q1={q1:.2f}، Q3={q3:.2f}، IQR={iqr:.2f}؛ حصار بالا {fences[1]:.3f} سانتی‌متر است. مقدارهای علامت‌خورده {x[(x<fences[0])|(x>fences[1])].tolist()} نیازمند بررسی‌اند، نه حذف خودکار.')
    # END PRESERVED ANALYSIS: lesson-5
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-5'] = run_lesson_09()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

هر دو نتیجه با قرارداد خود سازگارند؛ خطای متن قدیمی، مخلوط‌کردن قراردادها و نتیجه‌گیری قطعی بود. جعبه ناحیهٔ چارکی را نشان می‌دهد و با داده‌های گسسته لزوماً دقیقاً نصف نقاط داخل آن قرار نمی‌گیرد. مقدار علامت‌خورده را با عنوان «نیازمند بررسی» گزارش کنید.

#### کاربرد و محدودیت

در آزمایشگاه چارک‌ها همین روش را روی ANSUR II امتحان کنید. حصار، حد ارگونومیک یا مشخصهٔ تلرانس محصول نیست؛ صدک‌ها نیز به‌تنهایی طول بدنهٔ مناسب را تعیین نمی‌کنند.

#### تمرین

با Q1=۱۰ و Q3=۱۴، حصارها چیست و آیا ۲۰ بیرون است؟

### درس 10. تابع توزیع تجمعی تجربی ECDF

**هدف یادگیری:** سهم پوشش را مستقیم از داده بخوانید.

#### مفهوم و تشریح کامل

تابع تجمعی تجربی در آستانهٔ t، تعداد مشاهده‌های ≤t را بر n تقسیم می‌کند. نمودار پله‌ای است: بین مشاهده‌ها تغییر نمی‌کند و در مقدارهای تکراری جهش بزرگ‌تر دارد. این ویژگی برای خواندن پوشش یک آستانه و مقایسهٔ توزیع‌ها مفید است.

#### فرمول و قرارداد

```text
F̂(t)=#{xᵢ≤t}/n
```

#### مثال حل‌شدهٔ مرجع

در دادهٔ دست F̂(۲۰)=۴/۱۰=۰٫۴، F̂(۲۱)=۶/۱۰=۰٫۶، F̂(۲۲)=۷/۱۰=۰٫۷ و F̂(۲۳)=۸/۱۰=۰٫۸ است. پس عبارت قدیمی «۵۰٪ کمتر از ۲۱٫۵» دقیق نیست؛ شش نفر یعنی ۶۰٪ کمتر یا مساوی ۲۱٫۵ هستند. میانهٔ ۲۱ با تعریف موقعیت مرکزی سازگار است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_10(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)

    # BEGIN PRESERVED ANALYSIS: lesson-10
    sorted_x = np.sort(x)
    ecdf_y = np.arange(1, len(x)+1)/len(x)
    fig, ax = plt.subplots()
    ax.step(sorted_x, ecdf_y, where='post', color=BLUE)
    ax.scatter(sorted_x, ecdf_y, color=BLUE)
    ax.set(xlabel='Hand length threshold (cm)', ylabel='Share ≤ threshold', ylim=(0,1.05),
           title='Empirical cumulative distribution')
    plt.show()
    print('Share with length ≤ 22 cm:', (x<=22).mean())

    say(f'در آستانهٔ ۲۱ سهم تجمعی {(x<=21).mean():.0%} و در ۲۲ سهم {(x<=22).mean():.0%} است. پله‌ها دادهٔ دیده‌شده را توصیف می‌کنند؛ ECDF را از صفر تا یک دنبال کنید.')
    # END PRESERVED ANALYSIS: lesson-10
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-10'] = run_lesson_10()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور X یک آستانهٔ طول و محور Y سهم تجمعی از صفر تا یک است. شیب‌نداشتن پله بین ۲۴ و ۲۷ به معنی نبود مشاهده در آن فاصله است، نه نبودِ ممکن آن اندازه در جامعه. از روی این ده نفر سهم‌های بازار را با دقت درصدی گزارش نکنید.

#### کاربرد و محدودیت

اگر ابعاد یک ابزار تنها برای طول‌های حداکثر ۲۲ مناسب فرض شده باشد، ECDF پوششِ تک‌بعدی ۷۰٪ همین نمونه را نشان می‌دهد. سازگاری هم‌زمان طول و عرض، یک رویداد مشترک است و باید جدا محاسبه شود.

#### تمرین

اگر طول ۲۰ دو بار تکرار شود، پلهٔ ECDF در ۲۰ با n=۱۰ چقدر بالا می‌رود؟

#### پاسخ تشریحی

۰٫۲ یا ۲۰ واحد درصد. هر فرد یک‌دهم به توزیع تجربی اضافه می‌کند.

### درس 11. آمار مقاوم، MAD، میانگین بریده و نفوذ مشاهده

**پرسش شروع:** درآمد یا زمان پاسخ یک نفر هزار برابر دیگران است؛ حذفش کنیم یا مدل را عوض کنیم؟

#### مفهوم، دقیق اما ساده

مقاومت یعنی حساسیت محدودترِ یک برآوردگر به آلودگی یا مقدارهای بسیار دور، نه نادیده‌گرفتن هر مشاهدهٔ نامطلوب. میانه با جابه‌جایی یک مقدار دور معمولاً کمتر از میانگین تغییر می‌کند. MAD میانهٔ قدر مطلق فاصله از میانه است؛ ضرب در حدود۱٫۴۸۲۶ آن را تحت نرمال به مقیاس SD سازگار می‌کند، نه اینکه برای هر توزیع همان SD شود. میانگین بریده بخشی از دو انتها را حذف می‌کند و عملاً کمیت خلاصهٔ متفاوتی هدف می‌گیرد. برآوردهای Huber برای خطاهای کوچک شبیه مربع و برای خطاهای بزرگ شبیه قدرمطلق‌اند. انتخاب روش به هدف، اصالت مقدار و هزینهٔ حساسیت بستگی دارد؛ پرت بودن آماری با خطای اندازه‌گیری یکی نیست.

#### فرمول و قرارداد

```text
MAD=median(|x−median(x)|)
Normal-consistent scale ≈1.4826×MAD
Mean has unbounded sensitivity to a single value.
```

#### مثال حل‌شده

۹ زمان ۹،۱۰،۱۰،۱۱،۱۱،۱۲،۱۲،۱۳،۱۴ بسازید و مقدار دهم را از۱۴ تا۲۰۰ تغییر دهید. میانگین رشد می‌کند، میانه به یک دامنهٔ محدود می‌چسبد و میانگین بریدهٔ۱۰٪ اثر بزرگ‌ترین و کوچک‌ترین مقدار را حذف می‌کند. اگر سؤال هزینهٔ کلِ انتظار کاربران باشد، همان مقدار خیلی بزرگ ممکن است واقعی و از نظر کسب‌وکار بسیار مهم باشد؛ میانه به‌تنهایی آن هزینه را پنهان می‌کند. پس «روش مقاوم» را همراه نمودار انتهاها و پرسش گزارش کنید.

````python
def run_lesson_11(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: robust
    rng=np.random.default_rng(202621)
    base=np.array([9,10,10,11,11,12,12,13,14]); extremes=np.linspace(14,200,100); methods=[]
    for value in extremes:
        vals=np.r_[base,value]; methods.append([vals.mean(),np.median(vals),stats.trim_mean(vals,.1)])
    fig,ax=plt.subplots(); arr=np.array(methods)
    for j,name,style in [(0,'Mean','-'),(1,'Median','--'),(2,'10% trimmed mean',':')]: ax.plot(extremes,arr[:,j],label=name,ls=style)
    ax.set(xlabel='Value of the tenth observation (demo time units)',ylabel='Estimated center',title='One observation, different sensitivity'); ax.legend(); show()
    vals=np.r_[base,200]; mad=stats.median_abs_deviation(vals,scale='normal')
    say(f'با مقدار دهم=۲۰۰، میانگین={vals.mean():.2f}، میانه={np.median(vals):.2f} و MAD مقیاس‌شدهٔ نرمال={mad:.2f} است. میانه و میانگین کمیت‌های متفاوتی‌اند؛ مقاوم‌تر بودن، اعتبار خودکارِ حذف انتها را نمی‌دهد.')
    # END PRESERVED ANALYSIS: robust
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['robust'] = run_lesson_11()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

محور افقی مقدار دهم و عمودی برآورد مرکز است؛ این نمودار تغییرِ برآورد به ازای یک مشاهده است، نه توزیع نمونه‌گیری. خطِ میانه صاف‌تر است اما الزاماً برای همهٔ تصمیم‌ها بهتر نیست. مقدار MAD را همراه قرارداد مقیاس‌بندی و وجود ties بخوانید؛ اگر بیش از نیمی از مقادیر یکسان باشند، MAD می‌تواند صفر شود.

#### دام رایج

حذف خودکار هر مقدار بیرون ۱٫۵IQR می‌تواند گروه‌های واقعی را حذف و واریانس/اثر را منحرف کند. ابتدا منبع و واحد را بررسی کنید.

#### تمرین

برای پایش زمان پاسخ، چرا هم میانه و هم صدک۹۹ مفیدند؟

#### پاسخ تشریحی

میانه تجربهٔ معمول و صدک۹۹ بخش کند انتهایی را توصیف می‌کند. یکی جای دیگری نیست؛ تعریف صدک، حجم نمونه و عدم‌قطعیت انتهاها نیز باید گزارش شود.

## فصل 03: احتمال و زبان توزیع‌ها

احتمال شرطی و نرخ پایه پیش‌نیاز بیزند؛ بعد از توزیع‌های پایه به هیستوگرام و چگالی برمی‌گردیم.

### درس 12. احتمال شرطی، استقلال و وابستگی

**پرسش شروع:** احتمال دیررسیدن بسته در روز بارانی با احتمال باران در روزِ دیررسیدن یکی نیست.

#### مفهوم، دقیق اما ساده

احتمال شرطی فضای مرجع را کوچک می‌کند: P(A|B) فقط بین مواردی حساب می‌شود که B برقرار است. استقلال یعنی دانستن B احتمال A را تغییر ندهد؛ صفر بودن همبستگی خطی به‌طور کلی استقلال نیست. رویدادهای ناسازگار با احتمال مثبت مستقل نیستند: اگر رخداد یکی دیگری را ناممکن کند، اطلاعات مهمی منتقل شده است. استقلال شرطی نیز با استقلال حاشیه‌ای فرق دارد؛ دو رفتار ممکن است پس از ثابت‌کردن ساعت روز مستقل شوند یا برعکس با انتخاب یک زیرگروه وابسته شوند. در یک جدول فراوانی، هر بار دقیقاً مشخص کنید کدام سطر، ستون یا کل جدول مخرج است.

#### فرمول و قرارداد

```text
P(A|B)=P(A∩B)/P(B), P(B)>0
Independence: P(A∩B)=P(A)P(B)
```

#### مثال حل‌شده

از ۱۰۰ روز ساخته‌شده، ۲۰ روز بارانی است. در ۱۲ روز بارانی و ۸ روز بدون باران بسته دیر رسیده است. بنابراین P(تأخیر|باران)=۱۲÷۲۰=۶۰٪ و P(تأخیر|بدون باران)=۸÷۸۰=۱۰٪. احتمال تأخیر در کل ۲۰٪ است. در این مثال خاص P(باران|تأخیر) هم ۶۰٪ می‌شود، فقط چون دو مخرج اتفاقاً برابرند؛ چنین تقارنی یک قانون نیست.

````python
def run_lesson_12(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: conditional
    rng=np.random.default_rng(202602)
    table=np.array([[12,8],[8,72]])
    conditional=table[:,0]/table.sum(axis=1)
    display(pd.DataFrame(table,index=['Rain','No rain'],columns=['Late','On time']))
    fig,ax=plt.subplots(); ax.bar(['Given rain','Given no rain'],100*conditional,color=[BLUE,GOLD]); ax.axhline(20,color='0.3',ls='--',label='Overall 20%'); ax.set(ylabel='P(late | condition), %',ylim=(0,100),title='Conditioning changes the denominator'); ax.legend(); show()
    say(f'تأخیر مشروط به باران={conditional[0]:.0%}، بدون باران={conditional[1]:.0%}؛ چون با هم برابر نیستند، تأخیر و باران در این جدول مستقل نیستند.')
    # END PRESERVED ANALYSIS: conditional
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['conditional'] = run_lesson_12()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

دو ستونِ شرطی هر کدام بر تعداد روزهای همان وضعیت تقسیم شده‌اند. اگر جدول شمارش را به نرخ تبدیل می‌کنید، جمع هر سطر باید با تعریف آن سازگار باشد. اختلاف ۶۰٪ و ۱۰٪ نشان‌دهندهٔ وابستگی در سناریوست، نه اندازهٔ اثر علّی باران؛ مسیرها و زمان ارسال هم می‌توانند تغییر کرده باشند.

#### دام رایج

جابه‌جاکردن P(A|B) و P(B|A)، یا شمردن موارد خالی به‌عنوان «خیر»، نتیجه را عوض می‌کند.

#### تمرین

اگر ۴۰ روز بارانی بود ولی همچنان ۱۲ تأخیر بارانی داشتیم، P(تأخیر|باران) چقدر می‌شد؟

#### پاسخ تشریحی

۱۲÷۴۰=۳۰٪. تعداد تأخیر ثابت مانده، ولی جامعهٔ شرطی و مخرج دو برابر شده است. احتمال جهت معکوس را باید با تعداد همهٔ تأخیرها جدا حساب کرد.

### درس 13. قضیهٔ بیز و تلهٔ نرخ پایه

**پرسش شروع:** هشدار یک سامانهٔ کشف تقلب با دقت حساسیت ۹۰٪؛ آیا هر هشدار تقریباً تقلب است؟

#### مفهوم، دقیق اما ساده

بیز احتمال پیشین را با اطلاعات تازه به‌روزرسانی می‌کند. حساسیت P(هشدار|تقلب) است، ولی چیزی که تصمیم‌گیر می‌خواهد P(تقلب|هشدار) است. برای رسیدن از اولی به دومی، نرخ پایه و هشدار اشتباه لازم‌اند. جدول تعدادهای طبیعی اغلب از فرمول قابل‌فهم‌تر است. اگر رویداد بسیار نادر باشد، حتی درصد کمی هشدار کاذب می‌تواند از هشدارهای درست بیشتر شود. این مسئله به معنی بی‌فایده‌بودن سامانه نیست؛ آستانهٔ اقدام و هزینهٔ بررسی باید متناسب انتخاب شود. نرخ پایه در جمعیت آموزش و استفاده ممکن است متفاوت باشد، پس یک precision را برای همهٔ بازارها کپی نکنید.

#### فرمول و قرارداد

```text
P(Fraud|Alert)=sensitivity×prevalence / [sensitivity×prevalence + false_positive_rate×(1−prevalence)]
```

#### مثال حل‌شده

در ۱۰٬۰۰۰ تراکنش با نرخ تقلب ۱٪، ۱۰۰ تقلب داریم. حساسیت ۹۰٪ یعنی ۹۰ هشدار درست. از ۹٬۹۰۰ تراکنش سالم، نرخ هشدار کاذب ۵٪ یعنی ۴۹۵ هشدار اشتباه. پس فقط ۹۰ از ۵۸۵ هشدار، حدود ۱۵٫۳۸٪، تقلب واقعی‌اند. این تعدادها انتظار مدل‌اند؛ در یک نمونهٔ تصادفی دقیقاً همین شمارش تضمین نمی‌شود. با نرخ پایهٔ ۱۰٪ و همان کیفیت آشکارساز، precision به حدود ۶۶٫۷٪ می‌رسد.

````python
def run_lesson_13(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: bayes
    rng=np.random.default_rng(202603)
    prev=.01; sensitivity=.9; fpr=.05; population=10000
    tp=population*prev*sensitivity; fp=population*(1-prev)*fpr; ppv=tp/(tp+fp)
    fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained'); axes[0].bar(['True alerts','False alerts'],[tp,fp],color=[BLUE,GOLD]); axes[0].set(ylabel='Expected alerts per 10,000 transactions',title='Constructed fraud-screening example')
    grid=np.linspace(.001,.3,300); axes[1].plot(100*grid,100*sensitivity*grid/(sensitivity*grid+fpr*(1-grid)),color=BLUE); axes[1].set(xlabel='Fraud prevalence (%)',ylabel='P(fraud | alert), %',ylim=(0,100),title='Same detector, changing base rate'); show()
    say(f'از {tp+fp:.0f} هشدارِ مورد انتظار، {tp:.0f} درست‌اند؛ precision={ppv:.2%}. حساسیت ۹۰٪ به معنای درست‌بودن ۹۰٪ هشدارها نیست.')
    # END PRESERVED ANALYSIS: bayes
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['bayes'] = run_lesson_13()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

دو بخش ستونِ هشدار را با هم بخوانید: بخش بزرگ هشدار اشتباه حاصل تعداد بسیار زیاد تراکنش سالم است. منحنی دوم نشان می‌دهد ثابت نگه‌داشتن حساسیت و نرخ خطای مثبت، precision را ثابت نگه نمی‌دارد. در این درس تصمیم اتهام‌زدن خودکار به فرد مطرح نیست؛ هشدار باید به بازبینی مناسب منتهی شود.

#### دام رایج

گزارش حساسیت به جای احتمال صحت هشدار، نرخ پایه را نادیده می‌گیرد. Accuracy هم در دادهٔ نادر می‌تواند فریبنده باشد.

#### تمرین

اگر همهٔ تراکنش‌ها را سالم اعلام کنیم، در نرخ تقلب ۱٪ دقت چند است؟ آیا سامانه خوب است؟

#### پاسخ تشریحی

دقت ۹۹٪ ولی حساسیت صفر است؛ هیچ تقلبی کشف نمی‌شود. معیار باید با هزینهٔ خطا، حجم بررسی و نرخ پایه سازگار باشد.

### درس 14. توزیع‌ها: برنولی، دوجمله‌ای، پواسون و نرمال

**پرسش شروع:** برای تعداد کلیک یک مدل می‌خواهیم؛ برای زمان پاسخ مدلی دیگر. چرا یک منحنی نرمال کافی نیست؟

#### مفهوم، دقیق اما ساده

توزیع، احتمال مقدارهای ممکن را مشخص می‌کند. برنولی یک موفقیت/شکست، دوجمله‌ای تعداد موفقیت در n تلاش مستقل با p ثابت، پواسون تعداد رخداد در یک بازه با ساختار نرخ مناسب و نرمال یک توزیع پیوستهٔ متقارن است. در مدل پواسون ساده میانگین و واریانس برابرند؛ در دوجمله‌ای واریانس np(1−p) است. برای متغیر پیوسته، احتمال یک نقطهٔ دقیق معمولاً صفر است و مساحت بازه معنا دارد؛ برای متغیر گسسته ارتفاع هر میله خود احتمال است. تابع توزیع تجمعی در هر دو حالت احتمال کمتر یا مساوی آستانه را می‌دهد. مدل را به دامنه و فرایند تولید داده وصل کنید؛ شباهت ظاهری یک هیستوگرام به‌تنهایی کافی نیست.

#### فرمول و قرارداد

```text
Bernoulli: E[X]=p, Var[X]=p(1−p)
Binomial: E[X]=np, Var[X]=np(1−p)
Poisson: E[X]=Var[X]=λ
```

#### مثال حل‌شده

در ۱۰ نمایش مستقل با احتمال کلیک ۲۰٪، تعداد کلیک دوجمله‌ای با میانگین ۲ و واریانس ۱٫۶ است. در مدل پواسون با λ=۲، میانگین همان ۲ ولی واریانس ۲ است و برخلاف دوجمله‌ای سقف ۱۰ ندارد. برای ساعت‌های شلوغ و خلوتِ مخلوط، یک پواسون با نرخ ثابت معمولاً توصیف کاملی نیست. مثال نرمالِ وزن قطعه هم مقدارهای منفی را از نظر ریاضی مجاز می‌داند؛ اگر آن احتمال مهم باشد باید مدل دیگری بررسی کرد.

````python
def run_lesson_14(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: distributions
    rng=np.random.default_rng(202604)
    k=np.arange(0,11); binprob=stats.binom.pmf(k,10,.2); poisprob=stats.poisson.pmf(k,2)
    fig,axes=plt.subplots(1,3,figsize=(12,3.8),layout='constrained'); axes[0].bar(k,binprob,color=BLUE); axes[1].bar(k,poisprob,color=GOLD)
    axes[0].set(xlabel='Clicks in 10 trials',ylabel='Probability mass',title='Binomial n=10, p=0.2'); axes[1].set(xlabel='Events in a fixed interval',ylabel='Probability mass',title='Poisson λ=2')
    g=np.linspace(-4,4,250); axes[2].plot(g,stats.norm.pdf(g),color=TEAL); axes[2].set(xlabel='Continuous standardized value',ylabel='Density',title='Normal: area, not point probability'); show()
    say(f'دوجمله‌ای: میانگین=۲، واریانس=۱٫۶ و احتمال حداقل یک کلیک={1-stats.binom.pmf(0,10,.2):.2%}. پواسون با میانگین ۲ واریانس ۲ دارد؛ برابر بودن میانگین‌ها به معنی برابر بودن توزیع‌ها نیست.')
    # END PRESERVED ANALYSIS: distributions
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['distributions'] = run_lesson_14()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

محور عمودی دو پنل اول احتمال جرم و در پنل نرمال چگالی است؛ ارتفاع‌ها را میان این نوع‌ها یکسان تفسیر نکنید. احتمال حداقل یک کلیک از ۱−P(X=0) محاسبه می‌شود. مجموع احتمالات دوجمله‌ای دقیقاً یک است؛ نمایش محدود پواسون ممکن است بخش بسیار کوچکی از انتها را بیرون محور بگذارد.

#### دام رایج

قرار دادن توزیع نرمال روی هر داده، محدودیت شمارش، صفرها، چولگی و کران‌ها را نادیده می‌گیرد.

#### تمرین

اگر نتیجهٔ تلاش‌ها مستقل نباشد، آیا فرمول واریانس np(1−p) همچنان کافی است؟

#### پاسخ تشریحی

نه لزوماً؛ کوواریانس تلاش‌ها هم وارد واریانس مجموع می‌شود. خوشه‌بندی رفتار کاربران می‌تواند پراکندگی تعداد کلیک‌ها را بیشتر کند.

### درس 15. هیستوگرام و منحنی نرمال

**هدف یادگیری:** شکل توزیع را با محور و مقیاس درست بخوانید.

#### مفهوم و تشریح کامل

هیستوگرام محور عددی را به بازه‌ها تقسیم و مشاهده‌های هر بازه را می‌شمارد. انتخاب عرض و مبدأ بازه بر شکل ظاهری اثر می‌گذارد؛ در ده مشاهده، تغییر یک مرز می‌تواند یک قله را ایجاد یا حذف کند. منحنی نرمالِ برازش‌شده مدلی برای مقایسهٔ بصری است و از روی آن نمی‌توان توزیع جامعه را اثبات کرد.

#### فرمول و قرارداد

```text
در هیستوگرام چگالی: ارتفاع = تعداد بازه / (n × عرض بازه)
```

#### مثال حل‌شدهٔ مرجع

اگر در بازهٔ عرض ۲ سانتی‌متر، ۳ نفر از ۱۰ نفر قرار بگیرند، فراوانی ۳، سهم ۳۰٪ و چگالی ۳÷(۱۰×۲)=۰٫۱۵ بر سانتی‌متر است. نمودار این درس چگالی دارد؛ بنابراین مساحت ستون‌ها جمعاً ۱ می‌شود. منحنی همراه، میانگین ۲۱٫۵ و انحراف معیار نمونه‌ای حدود ۲٫۶۳۵ دارد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_15(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)

    # BEGIN PRESERVED ANALYSIS: lesson-4
    fig, ax = plt.subplots()
    bins = np.arange(17.5, 28.6, 2)
    ax.hist(x, bins=bins, color=BLUE, alpha=.75, edgecolor='white', density=True)
    grid = np.linspace(17, 28, 300)
    ax.plot(grid, stats.norm.pdf(grid, x.mean(), x.std(ddof=1)), color=GOLD, lw=2,
            label='Fitted normal density')
    ax.set(xlabel='Hand length (cm)', ylabel='Density', title='Sample distribution and fitted normal curve')
    ax.legend(); plt.show()

    say(f'مساحت هیستوگرام چگالی تقریباً یک است، نه جمع ارتفاع ستون‌ها. منحنی نرمال با میانگین {x.mean():.2f} و SD نمونه‌ای {x.std(ddof=1):.3f} فقط یک مرجع برازش است.')
    # END PRESERVED ANALYSIS: lesson-4
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-4'] = run_lesson_15()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور افقی طول دست و محور عمودی چگالی است. ارتفاع ۰٫۱۵ به معنی ۱۵ نفر یا احتمال دقیق در یک نقطه نیست. ناحیهٔ زیر نمودار در یک فاصله به سهم آن فاصله مربوط می‌شود. شکل را با نقاط خام و Q–Q کنار هم ببینید، به‌ویژه برای انتهای راست.

#### کاربرد و محدودیت

برای مقایسهٔ گروه‌های با حجم متفاوت از چگالی یا درصد با تعریف یکسان استفاده کنید. مقایسهٔ تعداد خامِ ۴٬۰۸۲ مرد با ۱٬۹۸۶ زن می‌تواند فقط اختلاف تعداد سطرها را نمایش دهد.

#### تمرین

چرا منحنی چگالی نرمال را نمی‌توان بدون تبدیل روی محور «تعداد افراد» قرار داد؟

واحد محور‌ها متفاوت است. برای محور تعداد، مقدار چگالی باید در n و عرض بازه ضرب شود؛ یا هر دو نمایش روی چگالی تنظیم شوند.

## فصل 04: گشتاورها، شکل توزیع و تبدیل

گشتاور خام و مرکزی را به چولگی و کشیدگی وصل می‌کنیم؛ Q–Q، آزمون نرمال‌بودن و تبدیل لگاریتمی محدودیتِ خلاصه‌ها را روشن می‌کنند.

### درس 16. گشتاورهای خام

**هدف یادگیری:** چهار گشتاور خام را واقعاً محاسبه و واحدشان را تفسیر کنید.

#### مفهوم و تشریح کامل

گشتاور خام فاصله را از صفر می‌سنجد. توان اول میانگین است؛ توان دوم هم مرکز و هم پراکندگی را در خود دارد. توان سوم و چهارم به مقدارهای بزرگ وزن شدید می‌دهند. بزرگی این اعداد بین مرتبه‌ها قابل مقایسهٔ مستقیم نیست، چون واحد و توان متفاوت است.

#### فرمول و قرارداد

```text
m′r=(1/n)Σxᵢʳ; m′2=m2+x̄²
```

#### مثال حل‌شدهٔ مرجع

در دادهٔ دست: m′1=۲۱٫۵، m′2=۴۶۸٫۵، m′3=۱۰٬۳۵۳٫۵ و m′4=۲۳۲٬۱۵۷٫۳ است. مثلاً برای [۲،۴،۶،۸]، مجموع مربع‌ها ۱۲۰ و m′2=۳۰ است؛ واریانس این نمونه ۵ است، نه ۳۰. رابطهٔ ۳۰−۵²=۵ اختلاف مرکز و پراکندگی را روشن می‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_16(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)
    mean_length = x.mean()

    # BEGIN PRESERVED ANALYSIS: lesson-6
    raw_moments = {order: float(np.mean(x**order)) for order in range(1, 5)}
    display(pd.DataFrame({'order': list(raw_moments), 'raw_moment': list(raw_moments.values())}))
    assert np.isclose(raw_moments[1], mean_length)

    fig, axes = plt.subplots(2,2,figsize=(10,6),layout='constrained')
    for order, ax in enumerate(axes.flat,1):
        ax.bar(df.user_id,x**order,color=BLUE)
        ax.axhline(raw_moments[order],color=GOLD,ls='--',label='Raw moment (mean)')
        ax.set(xlabel='User ID',ylabel=f'Length^{order} (cm^{order})',title=f'Raw moment order {order}')
    axes[0,0].legend(); plt.show()

    say(f'گشتاور خام اول {raw_moments[1]:.2f} همان میانگین است؛ گشتاور خام دوم {raw_moments[2]:.2f} واحد cm² دارد و واریانس نیست: {raw_moments[2]:.2f}−{mean_length**2:.2f}={x.var():.2f}. مقیاس هر پنل جداست.')
    # END PRESERVED ANALYSIS: lesson-6
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-6'] = run_lesson_16()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

جدول خروجی مرتبه و میانگین توان را نشان می‌دهد. واحد ستونِ مرتبهٔ چهار cm⁴ است. اگر سانتی‌متر را به میلی‌متر تبدیل کنیم، مرتبهٔ r در ۱۰ʳ ضرب می‌شود؛ پس افزایش عدد، تغییر شکل جامعه را ثابت نمی‌کند. نمودار لگاریتمی نیز فاصله‌ها را به نسبت نشان می‌دهد، نه اختلاف ساده.

#### کاربرد و محدودیت

برای بازیابی واریانس یا چولگی از گشتاورهای خام باید تبدیل‌های جبری را درست انجام داد. در کاربرد محاسباتی، تفریق دو عدد بسیار بزرگ و نزدیک ممکن است دقت عددی را کاهش دهد؛ محاسبهٔ مستقیم فاصله‌ها اغلب مطمئن‌تر است.

#### تمرین

برای [۱،۳] گشتاور خام دوم و واریانس توصیفی را پیدا کنید.

#### پاسخ تشریحی

میانگین ۲، m′2=(۱+۹)/۲=۵ و m2=۵−۴=۱ است. تفاوت این دو مقدار ناشی از مبدأ اندازه‌گیری است.

### درس 17. گشتاورهای مرکزی

**هدف یادگیری:** معنای هندسی و عددیِ توان‌های فاصله از میانگین را یاد بگیرید.

#### مفهوم و تشریح کامل

در گشتاور مرکزی، نخست میانگین را کم می‌کنیم. توان فرد علامت دو سمت را نگه می‌دارد و توان زوج آن را حذف می‌کند. مرکزیِ اول صفر یک هویت جبری است؛ حتی دادهٔ بسیار نامتقارن هم این ویژگی را دارد. مرکزیِ سوم و چهارم هنوز واحددارند و خودِ چولگی و کشیدگی استاندارد نیستند.

#### فرمول و قرارداد

```text
mr=(1/n)Σ(xᵢ−x̄)ʳ; m1=0; m2=variance with denominator n
```

#### مثال حل‌شدهٔ مرجع

در دادهٔ دست جمع مربع فاصله‌ها ۶۲٫۵ است و m2=۶٫۲۵. جمع مکعب‌ها ۱۲۰ است و m3=۱۲. جمع توان چهارم ۱٬۱۵۸٫۶۲۵ است و m4=۱۱۵٫۸۶۲۵. برای مشاهدهٔ ۲۷، فاصله ۵٫۵ و توان چهارم ۹۱۵٫۰۶۲۵ است؛ همین یک فرد حدود ۷۹٪ مجموع توان چهارم را می‌سازد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_17(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)
    mean_length = x.mean()

    # BEGIN PRESERVED ANALYSIS: lesson-7
    central_moments = {order: float(np.mean((x-mean_length)**order)) for order in range(1, 5)}
    display(pd.DataFrame({'order': list(central_moments), 'central_moment': list(central_moments.values())}))
    assert abs(central_moments[1]) < 1e-12

    fig, axes = plt.subplots(2,2,figsize=(10,6),layout='constrained')
    for order, ax in enumerate(axes.flat,1):
        ax.bar(df.user_id,(x-mean_length)**order,color=BLUE)
        ax.axhline(0,color='0.3',lw=.8)
        ax.axhline(central_moments[order],color=GOLD,ls='--')
        ax.set(xlabel='User ID',ylabel=f'Deviation^{order} (cm^{order})',title=f'Central moment order {order}')
    plt.show()

    say(f'گشتاور مرکزی اول {central_moments[1]:.3g}، دوم {central_moments[2]:.3f}، سوم {central_moments[3]:.3f} و چهارم {central_moments[4]:.4f} است. توان زوج همیشه نامنفی و توان سوم جهت‌دار است؛ این اعداد واحدهای متفاوت دارند.')
    # END PRESERVED ANALYSIS: lesson-7
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-7'] = run_lesson_17()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

جدول هر مرتبه میانگین سهم‌ها را نمایش می‌دهد. اگر همهٔ طول‌ها را ۱۰ واحد جابه‌جا کنید، میانگین تغییر می‌کند اما فاصله‌های مرکزی ثابت می‌مانند. اگر همه را دو برابر کنید، m2 چهار برابر و m4 شانزده برابر می‌شود. این همان دلیل نیاز به استانداردسازی برای مقایسهٔ شکل است.

#### کاربرد و محدودیت

در آزمایشگاه گشتاورها یک مقدار انتهایی را جابه‌جا کنید و سهم هر نقطه را ببینید. توان‌های بالا برای تشخیص تمرکز سهم مفیدند، اما یک مشاهدهٔ اشتباه می‌تواند آن‌ها را به‌شدت منحرف کند.

#### تمرین

چرا m1=۰، تقارن داده را ثابت نمی‌کند؟

#### پاسخ تشریحی

میانگین دقیقاً طوری تعریف می‌شود که مجموع انحراف‌ها صفر باشد. در [۱،۱،۱۰] نیز این جمع صفر است، با اینکه داده متقارن نیست.

### درس 18. چولگی

**هدف یادگیری:** چولگی را از سهمِ مکعب فاصله‌ها بفهمید.

#### مفهوم و تشریح کامل

چولگی گشتاوری، سومین گشتاور مرکزی را به پراکندگی استاندارد می‌کند تا بی‌واحد شود. سمت راستِ میانگین سهم مکعبی مثبت و سمت چپ سهم منفی دارد. غالب‌شدن سهم راست، ضریب مثبت می‌دهد. نزدیک صفر می‌تواند از تقارن یا خنثی‌شدن الگوهای پیچیده‌تر حاصل شود.

#### فرمول و قرارداد

```text
g1=m3/m2^(3/2)=12/6.25^(3/2)=0.768
```

#### مثال حل‌شدهٔ مرجع

در دادهٔ ده‌نفره g1=۰٫۷۶۸ است؛ برآورد اصلاح‌شدهٔ نمونه‌ای حدود ۰٫۹۱۱ می‌شود. این اختلاف خطا نیست؛ قرارداد تصحیح فرق می‌کند. مثال [۲،۴،۶،۱۸] چولگی حدود +۰٫۹۷۹ دارد؛ بازتاب آن [۲،۱۴،۱۶،۱۸] مقدار −۰٫۹۷۹ می‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_18(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)
    mean_length = x.mean()
    central_moments = {order:float(np.mean((x-mean_length)**order)) for order in range(1,5)}

    # BEGIN PRESERVED ANALYSIS: lesson-11
    moment_skew = central_moments[3] / central_moments[2]**1.5
    adjusted_skew = stats.skew(x, bias=False)
    print(f'Moment skewness = {moment_skew:.3f}; adjusted sample skewness = {adjusted_skew:.3f}')

    z=(x-x.mean())/x.std(ddof=0)
    fig,ax=plt.subplots(); ax.bar(df.user_id,z**3,color=BLUE); ax.axhline(0,color='0.3')
    ax.set(xlabel='User ID',ylabel='Standardized deviation cubed',title=f'Signed contributions: mean(z³)={moment_skew:.3f}')
    plt.show()

    say(f'چولگی گشتاوری={moment_skew:.3f} و اصلاح‌شده={adjusted_skew:.3f}. علامت مثبت از سهم توان سوم فاصله‌های بزرگِ سمت راست می‌آید؛ اختلاف دو برآوردگر در n=10 محسوس است.')
    # END PRESERVED ANALYSIS: lesson-11
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-11'] = run_lesson_18()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

در هیستوگرام، به گستردگی انتهاها و محل تجمع نگاه کنید؛ جهت چولگی با محل بیشترین ستون یکی نیست. در دادهٔ دوچرخه هم ضریب کل مثبت است، اما باید با فیلتر هر زیرگروه دوباره محاسبه شود. مقدار مرکزی سومِ واحددار را با g1 مقایسه نکنید.

#### کاربرد و محدودیت

چولگی مثبت انگیزه‌ای برای بررسی کاربران انتهایی است، اما تعداد سایزهای محصول را تعیین نمی‌کند. ممکن است یک محصول قابل تنظیم یا چند اندازه مناسب باشد؛ این انتخاب به هزینه و آزمون کاربرد وابسته است.

#### تمرین

با دو برابرکردن همهٔ xها، m3 هشت برابر می‌شود. آیا چولگی هم هشت برابر می‌شود؟

### درس 19. کشیدگی

**هدف یادگیری:** کشیدگی را با حساسیت به فاصله‌های انتهایی تفسیر کنید.

#### مفهوم و تشریح کامل

توان چهارم همهٔ فاصله‌ها را نامنفی و مشاهده‌های دور را بسیار پراثر می‌کند. کشیدگی استاندارد m4/m2² و کشیدگی اضافی آن منهای ۳ است. مقدار ۳ مرجع نرمال در این قرارداد است؛ برابری کشیدگی با نرمال، برابری کل توزیع‌ها را نتیجه نمی‌دهد.

#### فرمول و قرارداد

```text
β2=115.8625/6.25²≈2.96608; excess=β2−3≈−0.03392
```

#### مثال حل‌شدهٔ مرجع

دادهٔ دست کشیدگی گشتاوری نزدیک ۲٫۹۶۶ دارد؛ برآورد اصلاح‌شدهٔ کشیدگی اضافی حدود ۰٫۹۰۴ است. برای [۲،۴،۶،۸]، β2=۱٫۶۴ و excess=−۱٫۳۶ است. در [۱،۲،۳،۴،۵،۶،۷،۸،۹،۳۰]، β2≈۶٫۷۰ می‌شود. این نمونه‌ها مصنوعی و برای دیدن اثر فاصلهٔ دورند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_19(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)
    mean_length = x.mean()
    central_moments = {order:float(np.mean((x-mean_length)**order)) for order in range(1,5)}
    z = (x-x.mean())/x.std(ddof=0)

    # BEGIN PRESERVED ANALYSIS: lesson-12
    moment_excess = central_moments[4]/central_moments[2]**2 - 3
    adjusted_excess = stats.kurtosis(x, fisher=True, bias=False)
    print(f'Moment excess kurtosis = {moment_excess:.3f}; adjusted sample estimate = {adjusted_excess:.3f}')

    fig,ax=plt.subplots(); ax.bar(df.user_id,z**4,color=GOLD)
    ax.axhline(moment_excess+3,color='0.3',ls='--',label='Mean z^4')
    ax.set(xlabel='User ID',ylabel='Standardized deviation to fourth power',title='Kurtosis: tail contributions, not just peak height')
    ax.legend(); plt.show()

    say(f'کشیدگی گشتاوری={moment_excess+3:.5f}، اضافی={moment_excess:.5f} و اضافی اصلاح‌شده={adjusted_excess:.3f}. نزدیک‌بودن اولین مقدار به ۳، اثبات نرمال‌بودن نیست؛ پنل سهم z⁴ حساسیت به انتهاها را نشان می‌دهد.')
    # END PRESERVED ANALYSIS: lesson-12
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-12'] = run_lesson_19()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

در جدول یا نمودار، اول مشخص کنید «کشیدگی» نمایش داده شده یا «اضافی». صفرِ اضافی با ۳ِ معمولی یک مرجع دارد. فقط از ارتفاع قله نمی‌توان این ضریب را فهمید؛ بهتر است سهم نواحی دور از میانگین در مجموع توان چهارم را هم دید، مانند آزمایشگاه گشتاورها.

#### کاربرد و محدودیت

در دادهٔ کم، یک مشاهدهٔ معتبر یا خطای ثبت می‌تواند ضریب را جابه‌جا کند. گزارش تحلیل حساسیت و تعریف دقیق ضریب از برچسب‌های قطعی «نرمال» یا «غیرنرمال» مفیدتر است.

#### تمرین

اگر β2=۴٫۴۱۶ باشد، کشیدگی اضافی چقدر است و چه چیزی را ثابت نمی‌کند؟

#### پاسخ تشریحی

۱٫۴۱۶ است. این عدد حساسیت بیشترِ گشتاور چهارم نسبت به مرجع را توصیف می‌کند؛ به‌تنهایی علت، شکل کامل توزیع یا نتیجهٔ آزمون نرمال‌بودن نیست.

### درس 20. نمودار Q–Q

**هدف یادگیری:** صدک مشاهده و صدک نظری را مقایسه کنید.

#### مفهوم و تشریح کامل

Q–Q دادهٔ مرتب‌شده را مقابل صدک‌های یک توزیع مرجع قرار می‌دهد. برای مرجع نرمال، اگر توزیع نمونه با تغییر مکان و مقیاس به آن شبیه باشد، نقاط تقریباً خطی می‌شوند. خمیدگی و انحراف انتهاها دربارهٔ ناسازگاری شکل سرنخ می‌دهند.

#### فرمول و قرارداد

```text
محور افقی: quantile of N(0,1)؛ محور عمودی: ordered observed x
```

#### مثال حل‌شدهٔ مرجع

این شکل فقط ده نقطه دارد و بالاترین مقدار آن ۲۷ است. یک انحراف انتهایی با این n می‌تواند ناشی از نوسان نمونه‌گیری یا الگوی واقعی باشد؛ برای تصمیم، دادهٔ بیشتر و منطق اندازه‌گیری لازم است. اگر تمام طول‌ها ۲ واحد زیاد شوند، شکل Q–Q عموداً جابه‌جا می‌شود ولی الگوی خمیدگی حفظ می‌شود.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_20(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)

    # BEGIN PRESERVED ANALYSIS: lesson-13
    fig, ax = plt.subplots()
    stats.probplot(x, dist='norm', plot=ax)
    ax.set(xlabel='Theoretical normal quantiles', ylabel='Observed hand length (cm)',
           title='Normal Q–Q plot for hand length')
    plt.show()

    say('نزدیکی نقاط Q–Q به خط، سازگاری بصری با نرمال را بررسی می‌کند. انحراف انتهاها و تعداد فقط ده مشاهده را با هم ببینید؛ نمودار آزمون قطعی نیست.')
    # END PRESERVED ANALYSIS: lesson-13
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-13'] = run_lesson_20()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور X نمرهٔ نظریِ بی‌واحد و محور Y سانتی‌متر است. خط مرجع «خط برابری x=y» نیست، چون واحدها یکی نیستند. انحراف از خط را نسبت به پراکندگی و تعداد نقاط بسنجید؛ یک شکل شبیه خط، اثبات نرمال‌بودن جامعه نیست.

#### کاربرد و محدودیت

در استنباط رگرسیون، نرمال‌بودن خطاها موضوع اصلی است؛ برای آن Q–Q باقیمانده‌ها را بررسی کنید. نرمال‌بودنِ خود X شرط عمومیِ رگرسیون خطی نیست.

#### تمرین

آیا Q–Q طول دست می‌تواند جای Q–Q باقیماندهٔ مدل عرض بر طول را بگیرد؟

#### پاسخ تشریحی

خیر. این دو دربارهٔ دو متغیر متفاوت‌اند؛ فرض خطای مدل را با باقیمانده‌ها و طراحی داده‌برداری بررسی می‌کنیم.

### درس 21. آزمون Shapiro–Wilk

**هدف یادگیری:** p-value و عدم رد را درست تعبیر کنید.

#### مفهوم و تشریح کامل

Shapiro–Wilk ناسازگاری داده با مدل نرمال را با آمارهٔ W می‌سنجد. فرض صفر، مدل نرمال است. p احتمال درست‌بودن فرض صفر نیست؛ با فرضِ آن مدل و دیگر شروط، میزان غیرعادی‌بودن شواهد مشاهده‌شده را توصیف می‌کند. در نمونهٔ بسیار کوچک قدرت کشف انحراف محدود است.

#### فرمول و قرارداد

```text
H0: distribution is normal; تصمیم آموزشی با α=0.05 از پیش تعیین‌شده
```

#### مثال حل‌شدهٔ مرجع

طول دست W≈۰٫۹۴۸ و p≈۰٫۶۴۳۵ دارد. باقیمانده‌های رگرسیون W≈۰٫۹۵۶ و p≈۰٫۷۳۷۷ دارند. در سطح ۰٫۰۵ هیچ‌کدام رد نمی‌شوند. جملهٔ قابل دفاع: «شواهد کافی برای رد مدل نرمال در این نمونه نداریم»، نه «نرمال‌بودن ثابت شد».

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_21(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    simple_fit = sm.OLS(df.hand_width_cm,sm.add_constant(df.hand_length_cm)).fit()
    x = df.hand_length_cm.to_numpy(dtype=float)

    # BEGIN PRESERVED ANALYSIS: lesson-16
    for name, values in [('Hand length',x), ('Simple-regression residuals', simple_fit.resid)]:
        result = stats.shapiro(values)
        print(f'{name}: W={result.statistic:.3f}, p={result.pvalue:.4f}')

    fig,axes=plt.subplots(1,2,figsize=(10,4),layout='constrained')
    stats.probplot(x,dist='norm',plot=axes[0]); stats.probplot(simple_fit.resid,dist='norm',plot=axes[1])
    axes[0].set(title='Q–Q: observed length',ylabel='Length (cm)'); axes[1].set(title='Q–Q: regression residuals',ylabel='Residual (cm)')
    plt.show()

    say(f'p شاپیرو برای طول={stats.shapiro(x).pvalue:.4f} و برای باقیمانده={stats.shapiro(simple_fit.resid).pvalue:.4f}. در α=0.05 رد نمی‌کنیم؛ توان کم ده مشاهده اجازهٔ اثبات نرمالیت نمی‌دهد. در رگرسیون فرض اصلی به خطا مربوط است، نه الزاماً X.')
    # END PRESERVED ANALYSIS: lesson-16
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-16'] = run_lesson_21()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

W نزدیک ۱ معمولاً سازگاری بیشتر را نشان می‌دهد، اما مقدار بحرانی و p به n وابسته‌اند. شکل Q–Q و علل تولید داده را کنار عدد بخوانید. برای دادهٔ بسیار بزرگ، انحراف ناچیز می‌تواند p کوچک بدهد؛ برای دادهٔ کم، انحراف مهم ممکن است کشف نشود.

#### کاربرد و محدودیت

انتخاب روش را صرفاً به یک آزمون مقدماتی نسپارید. پرسش، استقلال، اندازهٔ اثر، مقاومت روش و تشخیص باقیمانده‌ها را هم در نظر بگیرید.

#### تمرین

اگر p=۰٫۶۴ باشد، آیا احتمال نرمال‌بودن ۶۴٪ است؟

#### پاسخ تشریحی

خیر. p احتمال فرضیه نیست. آن عدد تحت فرض صفر و شروط محاسبه تعریف شده و به‌تنهایی احتمال درستی مدل را نمی‌دهد.

### درس 22. تبدیل لگاریتم، نابرابری Jensen و بازگشت به واحد اصلی

**پرسش شروع:** میانگین لگاریتم قیمت را exp می‌کنیم؛ چرا میانگین قیمت واقعی به دست نمی‌آید؟

#### مفهوم، دقیق اما ساده

تبدیل لگاریتم فاصله‌های ضربی را به جمعی می‌برد و می‌تواند پراکندگی وابسته به سطح را کمتر کند. اما هدف آماری هم تغییر می‌کند: exp(mean(log X)) میانگین هندسی است، نه میانگین حسابی. طبق Jensen برای متغیر مثبت، میانگین هندسی از حسابی بزرگ‌تر نیست. در مدل لگ‌نرمال با log X نرمالِ میانگین μ و واریانس σ²، میانه exp(μ) و میانگین exp(μ+σ²/2) است. این تصحیح فقط تحت همان مدل معتبر است؛ در مدل‌های دیگر تصحیح بازتبدیل، smearing و ناهمسانی واریانس باید متناسب بررسی شوند. log برای صفر و منفی تعریف حقیقی ندارد؛ log1p هم معنای هدف را متفاوت می‌کند.

#### فرمول و قرارداد

```text
Geometric mean=exp(mean(log X))≤mean(X)
If log X~Normal(μ,σ²): median(X)=exp(μ); mean(X)=exp(μ+σ²/2).
```

#### مثال حل‌شده

اعداد۱،۱۰،۱۰۰ میانگین حسابی۳۷ دارند؛ میانگین هندسی۱۰ است. در سناریوی لگ‌نرمال μ=۲ وσ=۱، میانهٔ نظری حدود۷٫۳۹ و میانگین حدود۱۲٫۱۸ می‌شود. ۵٬۰۰۰ داده می‌سازیم، هیستوگرام در مقیاس خام و لگ را کنار هم می‌بینیم و تخمین بازتبدیل ساده و تصحیح‌شده را مقایسه می‌کنیم. واحد پول یا زمان در این مثال قراردادی است؛ توصیهٔ قیمت‌گذاری یا سرمایه‌گذاری ارائه نمی‌شود.

````python
def run_lesson_22(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: transform
    rng=np.random.default_rng(202623)
    values=rng.lognormal(mean=2,sigma=1,size=5000); logs=np.log(values); naive_back=np.exp(logs.mean()); corrected=np.exp(logs.mean()+logs.var(ddof=1)/2)
    fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained'); axes[0].hist(values,bins=80,color=BLUE,edgecolor='white'); axes[0].axvline(values.mean(),color=GOLD,label='Arithmetic mean'); axes[0].set(xlabel='Positive value (demo units)',ylabel='Count',title='Raw lognormal observations'); axes[0].legend(); axes[1].hist(logs,bins=40,color=TEAL,edgecolor='white'); axes[1].set(xlabel='Natural logarithm of value',ylabel='Count',title='Transformation changes the target scale'); show()
    say(f'میانگین خام={values.mean():.3f}؛ exp میانگین لگ={naive_back:.3f}؛ تصحیح لگ‌نرمال={corrected:.3f}. مقدار آخر به فرض نرمال‌بودن لگ و ساختار واریانس وابسته است؛ فرمول عمومیِ بی‌فرض نیست.')
    # END PRESERVED ANALYSIS: transform
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['transform'] = run_lesson_22()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

پنل خام، انتهای راست را نشان می‌دهد؛ پنل لگ، شکل داده در واحدِ log را. خط میانگین روی نمودار لگ را نباید با میانگین در واحد خام یکی دانست. اگر پیش‌بینی هزینهٔ کل هدف باشد، پیش‌بینی میانه ممکن است هزینه را کم‌برآورد کند؛ اگر تجربهٔ معمول هدف باشد، میانه می‌تواند مناسب‌تر باشد.

#### دام رایج

برای زیباترشدن نمودار، log بگیرید ولی محور و واحد تازه را پنهان نکنید. exp کردن CI ضرایب نیز نوع پارامتر را تغییر می‌دهد؛ عنوان دقیق لازم است.

#### تمرین

در مدل log(y)=a+bx، افزایش یک واحد x چه تفسیر تقریبی دارد؟

#### پاسخ تشریحی

با فرض ساختار همان مدل، exp(b) نسبت میانه‌های شرطی در مدل لگ‌نرمالِ هم‌واریانس است؛ تقریب ۱۰۰b٪ برای b کوچک. نسبت میانگین‌ها با ناهمسانی واریانس ممکن است متفاوت باشد.

## فصل 05: از نمونه به عدم‌قطعیت و شواهد

استقلال و حجم نمونهٔ مؤثر، سپس CLT، SE، پوشش بازه، بوت‌استرپ و منطق استنباط؛ در پایان درست‌نمایی و پسین بیزی.

### درس 23. استقلال، اندازهٔ نمونهٔ مؤثر و شبه‌تکرار

**پرسش شروع:** ۱۰ نفر را هر کدام ۱۰۰ بار اندازه می‌گیریم؛ آیا ۱٬۰۰۰ نفر داده داریم؟

#### مفهوم، دقیق اما ساده

تعداد سطر با تعداد واحد مستقل فرق دارد. تکرارهای یک فرد معمولاً ویژگی مشترک دارند. اگر آن‌ها را مستقل فرض کنیم، SE بیش از حد کوچک و نتیجه بیش از حد مطمئن می‌شود؛ این شبه‌تکرار است. در طرح سادهٔ خوشه‌های هم‌اندازه با همبستگی درون‌خوشه‌ای ρ، اثر طرح تقریباً ۱+(m−۱)ρ و اندازهٔ نمونهٔ مؤثر برای میانگین کلی تقریباً n/اثرطرح است. این فرمول یک تقریب برای طرح و ساختار خاص است، نه قانون عمومی برای هر مدل. مدل چندسطحی، GEE، SE خوشه‌ای و بازنمونه‌گیری در سطح فرد از گزینه‌های متناسب‌اند؛ انتخاب به سؤال و تعداد خوشه‌ها هم بستگی دارد.

#### فرمول و قرارداد

```text
DEFF ≈ 1+(m−1)ρ; n_eff ≈ n/DEFF
Resample independent clusters, not repeated rows.
```

#### مثال حل‌شده

۲۰ فرد، هر کدام ۱۰ تکرار، در مجموع ۲۰۰ سطر داریم. اگر ρ=۰٫۶ باشد، اثر طرح ۶٫۴ و n مؤثرِ تقریبی ۳۱٫۲۵ می‌شود. این به معنی وجود ۳۱ فرد واقعی نیست؛ فقط دقت میانگینِ این طرح را با یک نمونهٔ مستقل مقایسه می‌کند. در شبیه‌سازی، اثر مشترک فرد و نویز تکرار جدا تولید می‌شوند. میانگین هر فرد را حساب کرده و SE میانگینِ ۲۰ میانگین فردی را با SE سادهٔ ۲۰۰ سطر مقایسه می‌کنیم.

````python
def run_lesson_23(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: dependence
    rng=np.random.default_rng(202607)
    rho=.6; people=20; repeats=10
    measure=rng.normal(0,np.sqrt(rho),size=(people,1))+rng.normal(0,np.sqrt(1-rho),size=(people,repeats))
    naive=measure.std(ddof=1)/np.sqrt(measure.size); clustered=measure.mean(axis=1).std(ddof=1)/np.sqrt(people)
    fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained'); axes[0].scatter(np.repeat(np.arange(1,people+1),repeats),measure.ravel(),color=BLUE,alpha=.6); axes[0].set(xlabel='Independent person ID',ylabel='Repeated constructed measurement',title='200 rows, only 20 people')
    axes[1].bar(['Naive row SE','Person-level SE'],[naive,clustered],color=[BLUE,GOLD]); axes[1].set(ylabel='Estimated SE of grand mean',title='Dependence changes uncertainty'); show()
    say(f'با ρ={rho} و m={repeats}، اثر طرح تقریبی={1+(repeats-1)*rho:.2f} و n مؤثر≈{measure.size/(1+(repeats-1)*rho):.2f}. در اجرای حاضر SE سطری={naive:.3f} و SE بر مبنای فرد={clustered:.3f} است.')
    # END PRESERVED ANALYSIS: dependence
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['dependence'] = run_lesson_23()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

نقاط ستون اول تکرارهای نزدیک یک شخص را نشان می‌دهند؛ تغییرات بین اشخاص مهم‌اند. پنل دوم SE ساده و SE در سطح افراد را مقایسه می‌کند، نه دو نوع پراکندگی دست. نمونهٔ محدود ممکن است دقیقاً نسبت نظری را نسازد. اگر اثر یادگیری یا زمان هم وجود داشته باشد، مدل تبادلیِ ساده کافی نیست.

#### دام رایج

دادن همان شخص به آموزش و آزمون، علاوه بر SE نادرست، ارزیابی مدل روی شخص جدید را خوش‌بینانه می‌کند.

#### تمرین

برای Bootstrap دادهٔ ۲۰ فرد با ۱۰ تکرار، کدام واحد را بازنمونه‌گیری می‌کنید؟

#### پاسخ تشریحی

برای تعمیم به افراد جدید، اشخاص را با همهٔ تکرارهایشان بازنمونه‌گیری می‌کنیم. اگر سؤال دربارهٔ تکرار در یک فرد مشخص باشد، طرح و واحد بازنمونه‌گیری باید دوباره تعریف شوند.

### درس 24. قانون اعداد بزرگ و قضیهٔ حد مرکزی

**پرسش شروع:** داده‌ها بسیار چوله‌اند؛ پس چگونه میانگین آن‌ها می‌تواند تقریباً نرمال شود؟

#### مفهوم، دقیق اما ساده

قانون اعداد بزرگ دربارهٔ نزدیک‌شدن میانگین نمونه به امیدریاضی در شرایط مناسب است؛ قضیهٔ حد مرکزی دربارهٔ شکل توزیع میانگین‌های استانداردشده در تکرار نمونه‌گیری است. هیچ‌کدام نمی‌گویند با زیادشدن n، خود داده‌های افراد نرمال می‌شوند. در نسخهٔ کلاسیکِ مستقل و هم‌توزیع با واریانس متناهی، SD میانگین برابر σ/√n است. برای وابستگی زمانی، دنباله‌های بسیار سنگین یا نمونه‌گیری سوگیرانه، این نتیجهٔ ساده را مستقیم به کار نبرید. عدد ۳۰ مرز جادویی نیست؛ شکل توزیع، انتهاها و هدف تقریب مهم‌اند. شبیه‌سازی با یک توزیع معلوم کمک می‌کند سطح «افراد» و «میانگینِ نمونه‌ها» را جدا ببینیم.

#### فرمول و قرارداد

```text
SE(mean)=σ/√n under i.i.d. finite variance
CLT standardizes the mean, not each observation.
```

#### مثال حل‌شده

زمان انتظار را از نمایی با میانگین و SD برابر ۲ واحد بسازید. انتظار فرد بسیار راست‌چوله است. اکنون ۴٬۰۰۰ بار نمونه‌های ۱، ۵، ۳۰ و ۱۰۰تایی بگیرید. میانگین‌های نمونهٔ بزرگ‌تر نزدیک ۲ جمع می‌شوند و شکلشان متقارن‌تر می‌شود؛ اما انتظارهای تکی هنوز همان توزیع چوله را دارند. دو برابرکردن دقت یعنی نصف‌کردن SE، تقریباً چهار برابر حجم نمونه می‌خواهد، نه دو برابر.

````python
def run_lesson_24(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: clt
    rng=np.random.default_rng(202605)
    fig,axes=plt.subplots(2,2,figsize=(10,6),layout='constrained'); rows=[]
    for ax,n in zip(axes.flat,[1,5,30,100]):
        means=rng.exponential(2,size=(4000,n)).mean(axis=1); rows.append([n,means.mean(),means.std(ddof=1),2/np.sqrt(n)])
        ax.hist(means,bins=45,density=True,color=BLUE,edgecolor='white'); ax.axvline(2,color=GOLD); ax.set(xlim=(0,10),xlabel='Sample mean waiting time (demo units)',ylabel='Density',title=f'4,000 independent samples, n={n}')
    show(); display(pd.DataFrame(rows,columns=['n','mean of means','simulation SE','theoretical SE']))
    say('هر پنل توزیع ۴٬۰۰۰ میانگین است، نه توزیع یک نمونهٔ فردی. کاهش SD میانگین‌ها با ۱/√n سازگار است؛ نرمال‌شدن تقریبیِ میانگین، چولگی زمان انتظار افراد را حذف نکرده است.')
    # END PRESERVED ANALYSIS: clt
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['clt'] = run_lesson_24()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

همهٔ پنل‌ها محور افقی مشترک دارند تا کاهش پراکندگی دیده شود. خط عمودی ۲ امیدریاضیِ سناریوی ساخته‌شده است، نه عددی که از دادهٔ واقعی کشف شده باشد. در خروجی SD میانگین‌های شبیه‌سازی را با ۲/√n مقایسه کنید؛ اختلاف کوچک حاصل شبیه‌سازی محدود است.

#### دام رایج

نمونهٔ بزرگ، سوگیری انتخاب را درمان نمی‌کند. میانگین هزاران داوطلبِ نامرتبط ممکن است با دقت بالا کمیتِ اشتباه را تخمین بزند.

#### تمرین

اگر n از ۲۵ به ۱۰۰ برسد، SE در شرایط همین مدل چه تغییری می‌کند؟

#### پاسخ تشریحی

σ/√۲۵=σ/۵ و σ/√۱۰۰=σ/۱۰؛ SE نصف می‌شود. این رابطه به فرض استقلال و واریانس متناهی وابسته است.

### درس 25. انحراف معیار و بازهٔ طراحی

**هدف یادگیری:** پراکندگی افراد و عدم‌قطعیت میانگین را تفکیک کنید.

#### مفهوم و تشریح کامل

انحراف معیار جذر واریانس است و به واحد اصلی برمی‌گردد. برای توصیف همین داده تقسیم بر n و برای برآورد رایج واریانس جامعه از نمونهٔ مستقل تقسیم بر n−1 به کار می‌رود. خطای استاندارد میانگین s/√n است؛ دربارهٔ نوسان برآورد میانگین صحبت می‌کند، نه اختلاف بین افراد.

#### فرمول و قرارداد

```text
σdescriptive=√(62.5/10)=2.5; s=√(62.5/9)≈2.635; SE(x̄)=s/√10≈0.833
```

#### مثال حل‌شدهٔ مرجع

بازهٔ ۲۱٫۵±۲×۲٫۵ برابر [۱۶٫۵،۲۶٫۵] است؛ با s نمونه‌ای [۱۶٫۲۳،۲۶٫۷۷] می‌شود. هر دو در همین ده نفر فقط ۹ نفر را می‌پوشانند، یعنی ۹۰٪ مشاهده‌ها. در مقابل بازهٔ t برای میانگین با t≈۲٫۲۶۲ حدود [۱۹٫۶۲،۲۳٫۳۸] است، با فرض‌های نمونه‌گیری مناسب.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_25(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)
    mean_length = x.mean()

    # BEGIN PRESERVED ANALYSIS: lesson-8
    sd_sample = x.std(ddof=1)
    sd_population = x.std(ddof=0)
    print(f'Sample SD = {sd_sample:.3f} cm; descriptive population SD = {sd_population:.3f} cm')
    print(f'Mean ± 2 sample SD = [{mean_length-2*sd_sample:.2f}, {mean_length+2*sd_sample:.2f}] cm')

    mean_ci = stats.t.interval(.95,len(x)-1,loc=mean_length,scale=sd_sample/np.sqrt(len(x)))
    fig,ax=plt.subplots(figsize=(9,3.5))
    bounds=[(mean_length-2*sd_sample,mean_length+2*sd_sample),mean_ci]
    for row,(a,b) in enumerate(bounds):
        ax.plot([a,b],[row,row],lw=5,color=[BLUE,GOLD][row]); ax.scatter([mean_length],[row],color='black',zorder=3)
    ax.set(yticks=[0,1],yticklabels=['Mean ± 2 sample SD (spread)','95% t CI of mean (uncertainty)'],xlabel='Hand length (cm)',title='Spread of observations is not uncertainty of the mean')
    plt.tight_layout(); plt.show()

    say(f'SD توصیفی={sd_population:.3f} و نمونه‌ای={sd_sample:.3f} سانتی‌متر است؛ SE میانگین={sd_sample/np.sqrt(len(x)):.3f}. بازهٔ t برای میانگین [{mean_ci[0]:.3f}, {mean_ci[1]:.3f}] است؛ این بازه با پراکندگی افراد فرق دارد.')
    # END PRESERVED ANALYSIS: lesson-8
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-8'] = run_lesson_25()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

سه چیز متفاوت را جدا بخوانید: دامنهٔ توصیفیِ افراد، بازهٔ اطمینان پارامتر میانگین و بازهٔ پیش‌بینی یک فرد آینده. قاعدهٔ تقریبی ±۲SD به مدل توزیع و جامعه وابسته است؛ برچسب «پوشش قطعی ۹۵٪» روی آن نگذارید.

#### کاربرد و محدودیت

برای اندازهٔ محصول، دامنهٔ کاربران و پوشش چندمتغیره را بررسی کنید. برای مقایسهٔ دو نمونهٔ مستقل دربارهٔ میانگین، عدم‌قطعیت برآورد اهمیت دارد. افزایش n خطای استاندارد را کم می‌کند؛ لزوماً تنوع واقعی دست‌ها را کم نمی‌کند.

#### تمرین

اگر n چهار برابر شود و s ثابت بماند، SE چه تغییری می‌کند؟

#### پاسخ تشریحی

SE نصف می‌شود، چون √(۴n)=۲√n. خود انحراف معیار بین افراد ثابت فرض شده است.

### درس 26. بازهٔ اطمینان، بازهٔ پیش‌بینی و پوشش در تکرار

**پرسش شروع:** «۹۵٪ اطمینان» یعنی ۹۵٪ داده‌ها داخل بازه‌اند یا ۹۵٪ احتمال برای میانگین؟

#### مفهوم، دقیق اما ساده

بازهٔ اطمینان فراوانی‌گرا یک روش تصادفیِ ساخت بازه است. اگر طرح و فرض‌ها درست باشند، در تکرارهای طولانی مثلاً ۹۵٪ بازه‌های ساخته‌شده پارامتر ثابت را می‌پوشانند. بعد از دیدن یک بازه، این تفسیر به معنای اختصاص احتمال بیزی ۹۵٪ به پارامتر نیست. بازهٔ پیش‌بینی برای مشاهدهٔ تازه است و پراکندگی افراد را هم در خود دارد؛ پس معمولاً از CI میانگین پهن‌تر است. یک بازهٔ جمعیتی هم موضوع دیگری است. وابستگی، انتخاب پس از نگاه به داده یا مدل نامناسب می‌تواند پوشش اسمی را خراب کند. سطح ۹۵٪ را از روش و فرض‌ها جدا گزارش نکنید.

#### فرمول و قرارداد

```text
Normal known σ: mean CI = x̄ ± 1.96σ/√n
New observation prediction interval = x̄ ± 1.96σ√(1+1/n)
```

#### مثال حل‌شده

در سناریوی معلوم μ=۱۰ و σ=۲، ۱۰۰ نمونهٔ مستقل ۲۵تایی می‌سازیم و برای هر کدام CI میانگین می‌کشیم. بعضی بازه‌ها ۱۰ را نمی‌پوشانند؛ این «اشتباه بودن آن نمونه» نیست، بلکه رفتار روش است. در اولین نمونه، بازهٔ پیش‌بینی یک مشاهدهٔ تازه حدود √۲۶ برابر CI میانگین پهنا دارد. برای σ نامعلوم باید از SE برآوردی و ضریب t متناسب استفاده کرد؛ این مثال عمداً σ را معلوم فرض می‌کند.

````python
def run_lesson_26(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: coverage
    rng=np.random.default_rng(202608)
    mu=10; sigma=2; n=25; samples=rng.normal(mu,sigma,size=(100,n)); means=samples.mean(axis=1); half=stats.norm.ppf(.975)*sigma/np.sqrt(n); covered=np.abs(means-mu)<=half
    fig,ax=plt.subplots(figsize=(9,6))
    for i,(m,hit) in enumerate(zip(means,covered)): ax.plot([m-half,m+half],[i,i],color=BLUE if hit else GOLD,lw=1.2,marker=None if hit else 'x')
    ax.axvline(mu,color='0.3',ls='--'); ax.set(xlabel='Mean and 95% known-σ confidence interval',ylabel='Independent sample index',title='Coverage is a repeated-sampling property'); show()
    prediction_half=stats.norm.ppf(.975)*sigma*np.sqrt(1+1/n)
    say(f'{covered.sum()} از ۱۰۰ CI این اجرا μ=۱۰ را پوشاندند. نیم‌عرض CI={half:.3f} و نیم‌عرض پیش‌بینی فرد تازه={prediction_half:.3f} است؛ این دو دربارهٔ کمیت‌های متفاوت‌اند.')
    # END PRESERVED ANALYSIS: coverage
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['coverage'] = run_lesson_26()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

هر خط افقی متعلق به یک نمونهٔ جداست؛ خط عمودی پارامتر واقعیِ شبیه‌سازی است. رنگ و علامت متفاوت، بازه‌های ناپوشاننده را مشخص می‌کنند. نرخ پوششِ ۱۰۰ بار ممکن است دقیقاً ۹۵٪ نباشد؛ خود این نرخ نیز نوسان دارد. پوشش هر زیرگروه یا پس از انتخاب مدل، خودکار از پوشش اسمی نتیجه نمی‌شود.

#### دام رایج

CI میانگین را به‌جای دامنهٔ قابل‌قبول افراد یا تضمین پیش‌بینی یک کاربر به کار نبرید.

#### تمرین

اگر n چهار برابر شود، CI میانگین و بازهٔ پیش‌بینی هر دو نصف می‌شوند؟

#### پاسخ تشریحی

CI میانگین با σ معلوم نصف می‌شود؛ اما عرض پیش‌بینی به کف پراکندگی فردی σ نزدیک می‌شود و نصف‌شدن آن تضمین نیست.

### درس 27. بازهٔ اطمینان Bootstrap برای میانگین

**هدف یادگیری:** بوت‌استرپ را از بازنمونه تا تفسیر بازه دنبال کنید.

#### مفهوم و تشریح کامل

یک بازنمونهٔ بوت‌استرپ n مقدار از دادهٔ موجود با جایگذاری می‌گیرد؛ ممکن است فردی چند بار انتخاب و فردی اصلاً انتخاب نشود. میانگین هر بازنمونه را محاسبه و این کار را تکرار می‌کنیم. توزیع میانگین‌های بازنمونه تقریبِ نوسان برآوردگر است، مشروط به مناسب‌بودن داده و طرح نمونه‌گیری.

#### فرمول و قرارداد

```text
CIpercentile=[quantile(bootstrap means,0.025), quantile(bootstrap means,0.975)]
```

#### مثال حل‌شدهٔ مرجع

کد با بذر ۴۲، ده‌هزار بازنمونهٔ ده‌تایی می‌سازد؛ بازهٔ صدکی ثبت‌شده [۲۰٫۱۰،۲۳٫۱۰] سانتی‌متر است. مثلاً بازنمونهٔ [۱۸،۱۸،۲۰،۲۰،۲۱،۲۱،۲۲،۲۳،۲۷،۲۷] مجاز است و میانگینش ۲۱٫۷ است. این مثال نشان می‌دهد بازنمونه اندازهٔ جدیدی بیرون از داده اختراع نمی‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_27(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)

    # BEGIN PRESERVED ANALYSIS: lesson-9
    boot_means = np.random.default_rng(SEED).choice(x, size=(10000, len(x)), replace=True).mean(axis=1)
    boot_ci = np.quantile(boot_means, [.025, .975])
    print(f'Bootstrap 95% percentile CI for mean: [{boot_ci[0]:.2f}, {boot_ci[1]:.2f}] cm')
    fig, ax = plt.subplots()
    ax.hist(boot_means, bins=35, color=TEAL, alpha=.8)
    ax.axvline(boot_ci[0], color=GOLD, ls='--'); ax.axvline(boot_ci[1], color=GOLD, ls='--')
    ax.set(xlabel='Bootstrap mean (cm)', ylabel='Resamples', title='Bootstrap distribution of the mean (10,000 resamples)')
    plt.show()

    say(f'بازهٔ صدکی ۹۵٪ از ۱۰٬۰۰۰ بازنمونه [{boot_ci[0]:.2f}, {boot_ci[1]:.2f}] سانتی‌متر است؛ عدم‌قطعیت میانگین همین نمونه را تحت فرض نمونه‌گیری مدل می‌کند، نه پوشش ۹۵٪ دست افراد.')
    # END PRESERVED ANALYSIS: lesson-9
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-9'] = run_lesson_27()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور افقی نمودار «میانگین بازنمونه» است و محور عمودی تعداد بازنمونه‌ها. دو خط، صدک‌های توزیع میانگین‌ها هستند. اینکه اکثر میانگین‌ها در فاصله‌ای قرار دارند، به معنی قرارداشتن اکثر دست‌های افراد در همان فاصله نیست. در تعبیر فراوانی‌گرایانه، روشِ ساخت بازه در تکرار نمونه‌گیری هدف پوشش پارامتر را دارد.

#### کاربرد و محدودیت

برای دادهٔ ساعتی متوالی، بوت‌استرپ سادهٔ تک‌سطرها وابستگی زمانی را نادیده می‌گیرد؛ باید طرحی مثل بلوکی متناسب با مسئله در نظر گرفت. افزایش بازنمونه از ۲هزار به ۱۰هزار، سوگیری انتخاب افراد را اصلاح نمی‌کند.

#### تمرین

آیا تکرار بوت‌استرپ یک میلیون بار، نمونهٔ ده‌نفره را به یک میلیون فرد تبدیل می‌کند؟

#### پاسخ تشریحی

خیر. اطلاعات مستقل همچنان از همان ده نفر می‌آید. فقط خطای محاسباتیِ تقریب بازنمونه‌ای کمتر می‌شود.

### درس 28. منطق استنباط، اندازهٔ اثر، توان و چندمقایسه‌ای

**هدف یادگیری:** قبل از خواندن p-value، پرسش، برآورد و هزینهٔ تصمیم را روشن کنید.

#### مفهوم و تشریح کامل

آمار توصیفی دربارهٔ دادهٔ دیده‌شده است و استنباط به جامعه‌ای فراتر از آن مربوط می‌شود. این گذار به نمونه‌گیری و مدل وابسته است. α آستانهٔ تصمیمِ از پیش تعیین‌شده و p سازگاری داده با فرض صفر در همان چارچوب است. اندازهٔ اثر می‌گوید اختلاف چقدر است؛ بازهٔ اطمینان دقت برآورد را نشان می‌دهد. توان، احتمال کشف اثرِ تعریف‌شده در یک طرح مشخص است و با حجم نمونه، پراکندگی و آستانه تغییر می‌کند.

#### فرمول و قرارداد

```text
گزارش کامل = برآورد اثر + واحد + بازه + n + روش + طرح داده‌برداری
```

#### مثال حل‌شدهٔ مرجع

فرض کنید راحتیِ دو ماوس در مقیاس۱۰ نمره، میانگین‌های۷٫۱ و۷٫۲ داشته باشد. حتی اگر با نمونهٔ بسیار بزرگ p کوچک شود، اختلاف۰٫۱ ممکن است ارزش عملی نداشته باشد. برعکس، اختلاف۲ نمره‌ای در نمونهٔ کم می‌تواند بازهٔ پهن و p بزرگ داشته باشد. برای مثال ده دست، اختلاف گروه تجربه≈۲٫۰۸۳ سانتی‌متر است؛ عدم رد آن را صفر نمی‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_28(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # BEGIN PRESERVED ANALYSIS: inference
    from statsmodels.stats.power import TTestIndPower
    power_model=TTestIndPower(); sizes=np.arange(5,151,5)
    powers={d:power_model.power(effect_size=d,nobs1=sizes,alpha=.05,ratio=1) for d in [.2,.5,.8]}
    family_error=1-(1-.05)**20
    fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained')
    for d,curve in powers.items(): axes[0].plot(sizes,curve,label=f'Cohen d={d}',ls={.2:':',.5:'--',.8:'-'}[d])
    axes[0].axhline(.8,color='0.3',ls='--'); axes[0].set(xlabel='n per group (independent, equal n)',ylabel='Power',ylim=(0,1),title='Two-sided t-test planning: model-based power'); axes[0].legend()
    tests=np.arange(1,31); axes[1].plot(tests,1-.95**tests,color=GOLD); axes[1].set(xlabel='Number of independent true-null tests',ylabel='P(at least one false rejection)',ylim=(0,1),title='Family-wise error without correction')
    plt.show()
    say(f'با ۲۰ آزمون مستقلِ صفرِ درست، احتمال حداقل یک رد اشتباه {family_error:.1%} است. در مدل برنامه‌ریزی t با d=0.5 و n=30 در هر گروه، توان {power_model.power(.5,30,alpha=.05):.1%} است. این توان پیشینیِ مدل است؛ نه توانِ اثبات‌شدهٔ نمونهٔ دست. p کوچک اندازهٔ عملی اثر را تعیین نمی‌کند.')
    # END PRESERVED ANALYSIS: inference
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['inference'] = run_lesson_28()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

ابتدا ستون برآورد و بازه را بخوانید، سپس p و سطح تصمیم را. «برابرند»، «اثر ندارد» و «فرض صفر ثابت شد» از عدم رد نتیجه نمی‌شوند. برای ادعای هم‌ارزی باید حاشیهٔ هم‌ارزیِ کاربردی از پیش تعیین و آزمون متناسب انجام شود. اگر۲۰ آزمون مستقل درα=۰٫۰۵ و با همهٔ فرض‌های صفر درست انجام شوند، احتمال حداقل یک رد حدود۱−۰٫۹۵²⁰≈۶۴٪ است؛ استقلال در این مثال یک فرض محاسباتی است.

#### کاربرد و محدودیت

پیش از مشاهدهٔ خروجی، پاسخ اصلی، مقایسه‌ها و حداقل اختلاف مهم را بنویسید. تحلیل‌های اکتشافی بعدی را با همین عنوان گزارش کنید. توافق چند آزمون روی همان ده نفر، شاهد مستقل چندباره ایجاد نمی‌کند.

#### تمرین

آیا p=۰٫۰۱ یعنی اثر حتماً بزرگ است؟

#### پاسخ تشریحی

خیر. p به n و پراکندگی و مدل وابسته است. اندازه و واحد اثر و بازهٔ برآورد باید جدا بررسی شوند.

### درس 29. درست‌نمایی، برآورد بیشینه و تابع زیان

**پرسش شروع:** کدام مقدار احتمال کلیک، دادهٔ دیده‌شده را بهتر توضیح می‌دهد؟

#### مفهوم، دقیق اما ساده

احتمال، با پارامتر معلوم دربارهٔ دادهٔ ممکن حرف می‌زند؛ درست‌نمایی، دادهٔ دیده‌شده را ثابت می‌گیرد و پارامترهای پیشنهادی را مقایسه می‌کند. بیشینهٔ درست‌نمایی یا MLE مقدار پارامتر با بیشترین سازگاری در همان مدل است. درست‌نمایی به‌تنهایی توزیع احتمال پارامتر نیست و لازم نیست مساحتش بر محور پارامتر یک باشد. لگاریتم حاصل‌ضرب احتمالات را به جمع تبدیل می‌کند و برای محاسبه پایدارتر است. تابع زیان نیز فقط یک انتخاب مهندسی نیست: کمترین مربعات با فرض‌های خطای گاوسی مرتبط است و خطای مطلق مرکز میانه‌ای را هدف می‌گیرد. مدل نادرست ممکن است با دقت بسیار خوب یک پارامتر نامناسب را برآورد کند.

#### فرمول و قرارداد

```text
L(p|k,n) ∝ p^k(1−p)^(n−k); p_MLE=k/n
Negative log-likelihood is a loss to minimize.
```

#### مثال حل‌شده

از ۲۰ تلاش، ۱۴ کلیک دیده‌ایم؛ MLE دوجمله‌ای ۰٫۷ است. حالا ۱۴۰ کلیک از ۲۰۰ تلاش را با همان نسبت تصور کنید. قله در هر دو ۰٫۷ است ولی منحنی درست‌نمایی نسبیِ نمونهٔ بزرگ‌تر باریک‌تر می‌شود. یک مثال دیگر: برای داده‌های ۱،۲،۳،۲۰، میانگین ۶٫۵ کمترین مربعات و هر عددی بین ۲ و۳ کمترین مجموع فاصلهٔ مطلق را می‌دهد؛ انتخاب زیان، هدف بهینه‌سازی را تغییر می‌دهد.

````python
def run_lesson_29(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: likelihood
    rng=np.random.default_rng(202609)
    pgrid=np.linspace(.01,.99,400); fig,ax=plt.subplots()
    for k,n,style in [(14,20,'-'),(140,200,'--')]:
        ll=stats.binom.logpmf(k,n,pgrid); ax.plot(pgrid,np.exp(ll-ll.max()),ls=style,label=f'{k}/{n} clicks')
    ax.axvline(.7,color='0.3',ls=':'); ax.set(xlabel='Candidate click probability p',ylabel='Likelihood / maximum likelihood',title='Same MLE, different concentration'); ax.legend(); show()
    say('هر دو نمونه MLE=۰٫۷ دارند؛ منحنیِ ۲۰۰ تلاش باریک‌تر است. ارتفاع نسبیِ یک را نباید احتمال یک‌بودن برآورد یا posterior قطعی تعبیر کرد.')
    # END PRESERVED ANALYSIS: likelihood
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['likelihood'] = run_lesson_29()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

منحنی‌ها بر بیشینهٔ خود تقسیم شده‌اند تا قلهٔ هر دو یک باشد؛ این نرمال‌سازی، آن‌ها را posterior یا چگالی احتمال پارامتر نمی‌کند. عرض قله دربارهٔ حساسیت برازش به پارامتر سرنخ می‌دهد؛ برای CI باید روش و فرض مربوط جدا به کار رود. در مرزهای p=۰ یا۱، تقریب‌های معمول می‌توانند نامناسب باشند.

#### دام رایج

Likelihood را به‌عنوان P(پارامتر|داده) گزارش نکنید؛ برای posterior، پیشین و نرمال‌سازی بیزی لازم است.

#### تمرین

چرا با زیادشدن تعداد داده ولی ثابت‌ماندن نسبت کلیک، برآورد نقطه‌ای تغییر نمی‌کند اما عدم‌قطعیت کم می‌شود؟

#### پاسخ تشریحی

نسبت k/n همان است، ولی دادهٔ بیشتری دربارهٔ همان فرایند جمع شده است. انحنای لگاریتم درست‌نمایی بیشتر و دامنهٔ پارامترهای تقریباً هم‌ساز محدودتر می‌شود، با فرض صحت مدل و طرح نمونه‌گیری.

### درس 30. بیزِ عملی: پیشین، پسین و پیش‌بینی آینده

**پرسش شروع:** یک فروشگاه تازه از دو سفارش، دو موفقیت دارد؛ آیا نرخ موفقیتش حتماً ۱۰۰٪ است؟

#### مفهوم، دقیق اما ساده

در استنباط بیزی، عدم‌قطعیت پارامتر با توزیع پیشین توصیف و با درست‌نمایی به پسین تبدیل می‌شود. در مدل برنولی با پیشین Beta(a,b)، پس از s موفقیت و f شکست، پسین Beta(a+s,b+f) است. میانگین پسین، ترکیبی از میانگین پیشین و نسبت نمونه است. فاصلهٔ اعتباری بیزی با فرض همین مدل و پیشین، احتمال پسین مشخصی را در خود دارد؛ با تفسیر تکرار نمونه‌گیریِ CI یکی نیست. پیش‌بینی آینده علاوه بر پارامتر نامعلوم، تصادفی‌بودن مشاهدهٔ آینده را هم در نظر می‌گیرد. پیشین باید با دانش و حساسیت‌سنجی توجیه شود، نه برای رساندن نتیجه به پاسخ دلخواه.

#### فرمول و قرارداد

```text
Prior: p~Beta(a,b)
Posterior: p|data~Beta(a+s,b+f)
Predictive P(next success)=E[p|data]=(a+s)/(a+b+s+f)
```

#### مثال حل‌شده

با پیشین یکنواخت Beta(1,1) و دو موفقیت از دو تلاش، پسین Beta(3,1) و میانگین ۰٫۷۵ است، نه یک. با پیشین متمرکزتر Beta(10,10)، میانگین پسین ۱۲÷۲۲≈۰٫۵۴۵ می‌شود. این اختلاف بزرگ نشان می‌دهد دو مشاهده برای غلبه بر اطلاعات پیشین کافی نیست. برای ۱۴ موفقیت از۲۰ تلاش، پیشین یکنواخت به Beta(15,7) و میانگین حدود ۰٫۶۸۲ می‌رسد؛ با افزایش داده، سهم نسبی پیشین معمولاً کمتر می‌شود، مشروط به اینکه مدل مناسب باشد.

````python
def run_lesson_30(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: posterior
    rng=np.random.default_rng(202610)
    a,b=1,1; successes,failures=14,6; pa,pb=a+successes,b+failures; g=np.linspace(.001,.999,500); ci=stats.beta.ppf([.025,.975],pa,pb)
    fig,ax=plt.subplots(); ax.plot(g,stats.beta.pdf(g,a,b),ls='--',color=GOLD,label='Prior Beta(1,1)'); ax.plot(g,stats.beta.pdf(g,pa,pb),color=BLUE,label='Posterior Beta(15,7)'); mask=(g>=ci[0])&(g<=ci[1]); ax.fill_between(g,0,stats.beta.pdf(g,pa,pb),where=mask,color=BLUE,alpha=.2); ax.set(xlabel='Success probability p',ylabel='Density over p',title='Bayesian learning from 14 successes in 20 trials'); ax.legend(); show()
    say(f'میانگین پسین و احتمال پیش‌بینی موفقیت بعدی={pa/(pa+pb):.3f}؛ بازهٔ اعتباری صدکی ۹۵٪=[{ci[0]:.3f},{ci[1]:.3f}]. این تفسیر به پیشین Beta(1,1) و مدل تلاش‌های مستقل با p ثابت مشروط است.')
    # END PRESERVED ANALYSIS: posterior
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['posterior'] = run_lesson_30()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

منحنی پیشین و پسین از نظر مساحت هر کدام یک‌اند؛ برخلاف درست‌نمایی نسبی درس قبل، این‌ها چگالی p هستند. بازهٔ سایه‌خورده صدکیِ ۹۵٪ پسین است. این بازه، دامنهٔ ۹۵٪ پیامدهای صفر و یک آینده نیست؛ دربارهٔ خود احتمال موفقیت است. احتمال موفقیت بعدی فقط یک عدد است، ولی تعداد موفقیت در چند تلاش آینده توزیع پیش‌بینی جدا دارد.

#### دام رایج

پیشین قوی را پنهان نکنید. «احتمال بیزی» فقط به اندازهٔ داده، مدل و پیشین قابل دفاع است.

#### تمرین

چرا مشاهدهٔ دو موفقیت، احتمال شکست آینده را صفر نمی‌کند؟

#### پاسخ تشریحی

دادهٔ محدود همهٔ pهای کمتر از یک را ناممکن نکرده است. پسین هنوز عدم‌قطعیت دارد و احتمال پیش‌بینی با میانگین‌گیری روی این عدم‌قطعیت به دست می‌آید.

## فصل 06: مقایسهٔ گروه‌ها و طراحی آزمون

از دو گروه به چند گروه و طرح عاملی می‌رویم؛ سپس هم‌ارزی، چندآزمونی، تصادفی‌سازی و توقف زودهنگام را بررسی می‌کنیم.

### درس 31. آزمون t مستقل

**هدف یادگیری:** اختلاف دو میانگین را همراه حجم و پراکندگی مقایسه کنید.

#### مفهوم و تشریح کامل

آزمون t مستقل دربارهٔ اختلاف میانگینِ دو جمعیت است. نسخهٔ Welch برای هر گروه واریانس جداگانه می‌گیرد و درجهٔ آزادی را تصحیح می‌کند. استقلال افراد و تناسب مدل همچنان لازم است. اگر یک فرد قبل و بعد سنجیده شود، مسئله زوجی است و فرمول مستقل مناسب نیست.

#### فرمول و قرارداد

```text
t=(x̄A−x̄B)/√(sA²/nA+sB²/nB)
```

#### مثال حل‌شدهٔ مرجع

گروه سابقه≤۵ سال طول‌های [۱۸،۲۰،۲۱،۲۲] با n=۴ و میانگین۲۰٫۲۵ دارد؛ گروه سابقه>۵ سال n=۶ و میانگین≈۲۲٫۳۳۳ دارد. اختلاف A−B≈−۲٫۰۸۳ سانتی‌متر است؛ Welch t≈−۱٫۴۱۳ و p≈۰٫۱۹۵۶ می‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_31(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()

    # BEGIN PRESERVED ANALYSIS: lesson-19
    df['experience_group'] = np.where(df.gaming_years<=5, 'low', 'high')
    low_exp = df.loc[df.experience_group=='low','hand_length_cm'].to_numpy()
    high_exp = df.loc[df.experience_group=='high','hand_length_cm'].to_numpy()
    t_result = stats.ttest_ind(low_exp, high_exp, equal_var=False)
    print(f'low n={len(low_exp)}, mean={low_exp.mean():.2f}; high n={len(high_exp)}, mean={high_exp.mean():.2f}')
    print(f'Welch t={t_result.statistic:.3f}, p={t_result.pvalue:.4f}')

    se_diff=np.sqrt(low_exp.var(ddof=1)/len(low_exp)+high_exp.var(ddof=1)/len(high_exp))
    df_diff=(low_exp.var(ddof=1)/len(low_exp)+high_exp.var(ddof=1)/len(high_exp))**2/((low_exp.var(ddof=1)/len(low_exp))**2/(len(low_exp)-1)+(high_exp.var(ddof=1)/len(high_exp))**2/(len(high_exp)-1))
    exp_ci=(low_exp.mean()-high_exp.mean())+np.array([-1,1])*stats.t.ppf(.975,df_diff)*se_diff
    group_figure([low_exp,high_exp],['low experience','high experience'],'Experience groups: observations, mean and SD')

    say(f'اختلاف میانگین کم‌تجربه منهای پرتجربه {low_exp.mean()-high_exp.mean():.3f} سانتی‌متر است؛ Welch t={t_result.statistic:.3f}، p={t_result.pvalue:.4f} و CI اختلاف [{exp_ci[0]:.3f}, {exp_ci[1]:.3f}]. گروه‌بندی مشاهده‌ای، اثر علّی تجربه را مشخص نمی‌کند.')
    # END PRESERVED ANALYSIS: lesson-19
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-19'] = run_lesson_31()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

علامت منفی فقط به ترتیب تفریق گروه‌ها وابسته است. p بزرگ‌تر از۰٫۰۵، اختلاف ثبت‌شده را صفر نمی‌کند و ثابت نمی‌کند تفاوت عملی وجود ندارد. برای تصمیم، اختلاف، بازهٔ آن و حداقل تفاوت مهمِ کاربردی را هم گزارش کنید.

#### کاربرد و محدودیت

سابقهٔ بازی در این داده مشاهده‌ای است و با سن همبستگی زیادی دارد؛ تفاوت طول دست را به «اثر بازی‌کردن» نسبت ندهید. برای مقایسهٔ راحتی دو ماوس روی افراد یکسان از طرح زوجی و کنترل ترتیب استفاده کنید.

#### تمرین

اگر ترتیب گروه‌ها را عوض کنیم، t و p دوطرفه چه تغییری می‌کنند؟

#### پاسخ تشریحی

علامت t عوض می‌شود، اما قدرمطلق آن و p دوطرفه ثابت می‌ماند. جهت اختلاف باید با تعریف A−B گزارش شود.

### درس 32. آزمون Levene / Brown–Forsythe

**هدف یادگیری:** پرسش برابری پراکندگی را از برابری میانگین جدا کنید.

#### مفهوم و تشریح کامل

Levene روی فاصلهٔ مطلق هر مشاهده از مرکز گروه کار می‌کند و آن فاصله‌ها را بین گروه‌ها مقایسه می‌کند. مرکز می‌تواند میانگین یا میانه باشد؛ صورت با میانه اغلب Brown–Forsythe نامیده می‌شود. این آزمون دربارهٔ واریانس‌هاست و پاسخ ANOVAِ میانگین‌ها را تکرار نمی‌کند.

#### فرمول و قرارداد

```text
Zij=|Yij−median(group i)|; سپس ANOVA روی Zها
```

#### مثال حل‌شدهٔ مرجع

انحراف معیار گروه‌های جوان، میانی و بالاتر به‌ترتیب۲٫۰۰۰،۳٫۴۱۶ و۲٫۰۸۲ سانتی‌متر است. آماره≈۰٫۶۲۴ و p≈۰٫۵۶۳۳ شده است. نسبت بزرگ‌ترین و کوچک‌ترین SD حدود۱٫۷۱ است؛ نتیجهٔ p را کنار همین مقدارهای خام بخوانید.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_32(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    age_group = np.select([df.user_id.isin([1,3,7]),df.user_id.isin([2,5,8,10])],['young','middle'],default='older')
    df['age_group'] = age_group
    groups_age = [df.loc[df.age_group==g,'hand_length_cm'].to_numpy() for g in ['young','middle','older']]

    # BEGIN PRESERVED ANALYSIS: lesson-22
    lev = stats.levene(*groups_age, center='median')
    print(f'Brown–Forsythe/Levene statistic={lev.statistic:.3f}, p={lev.pvalue:.4f}')
    print('Group SDs:', [round(np.std(g, ddof=1),3) for g in groups_age])

    fig,ax=plt.subplots(); ax.bar(['young','middle','older'],[np.std(g,ddof=1) for g in groups_age],color=BLUE)
    ax.set(ylabel='Sample SD of length (cm)',title='Within-group spread (not a mean comparison)'); plt.show()

    say(f'آزمون Levene با مرکز میانه: آماره={lev.statistic:.3f}، p={lev.pvalue:.4f}. این آزمون دربارهٔ پراکندگی است، نه میانگین؛ عدم رد با این نمونهٔ کوچک تضمین یکسانی واریانس‌ها نیست.')
    # END PRESERVED ANALYSIS: lesson-22
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-22'] = run_lesson_32()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

عدم رد با نمونه‌های کوچک، برابر بودن واریانس‌ها را تضمین نمی‌کند. ممکن است آزمون توان کافی برای کشف اختلاف نداشته باشد. نمودار انحراف‌های مطلق و نقاط خام گروه‌ها کمک می‌کند یک مشاهدهٔ دور یا عدم توازن حجم را تشخیص دهید.

#### کاربرد و محدودیت

برای روش‌های مقاوم، لازم نیست ابتدا با یک آزمون کم‌توان «اجازه» بگیرید. می‌توانید با توجیه روش‌شناختی Welch را انتخاب و برآوردها و پراکندگی‌ها را شفاف گزارش کنید.

#### تمرین

دو گروه میانگین برابر دارند. آیا Levene الزاماً p بزرگ می‌دهد؟

#### پاسخ تشریحی

خیر. میانگین برابر دربارهٔ پراکندگی چیزی نمی‌گوید. یک گروه می‌تواند بسیار متمرکز و دیگری بسیار پراکنده باشد.

### درس 33. آزمون Mann–Whitney U

**هدف یادگیری:** آزمون رتبه‌ای را با ادعای درست استفاده کنید.

#### مفهوم و تشریح کامل

Mann–Whitney داده‌های دو گروه مستقل را در کنار هم رتبه‌بندی می‌کند. برای مقدارهای مساوی میانگین رتبه می‌گیرد. تفسیر عمومی مربوط به توزیع‌ها و برتری نسبی است؛ تفسیر صرفِ اختلاف میانه به فرض‌های بیشتر دربارهٔ شکل و جابه‌جایی توزیع‌ها نیاز دارد.

#### فرمول و قرارداد

```text
UA=RA−nA(nA+1)/2; UB=nA×nB−UA
```

#### مثال حل‌شدهٔ مرجع

در مقایسهٔ سابقه، nA=۴ و nB=۶ است؛ UA=۷ و UB=۲۴−۷=۱۷ به دست آمده است. p دوطرفهٔ ثبت‌شده با تقریبِ اصلاح‌شده برای ties≈۰٫۳۳۴۴ است. میانهٔ دو گروه۲۰٫۵ و۲۲ است؛ این تفاوت توصیفی را از تصمیم آزمون جدا گزارش کنید.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_33(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    df['experience_group'] = np.where(df.gaming_years<=5,'low','high')
    low_exp = df.loc[df.experience_group=='low','hand_length_cm'].to_numpy()
    high_exp = df.loc[df.experience_group=='high','hand_length_cm'].to_numpy()

    # BEGIN PRESERVED ANALYSIS: lesson-23
    mw = stats.mannwhitneyu(low_exp, high_exp, alternative='two-sided', method='auto')
    print(f'U={mw.statistic:.3f}, p={mw.pvalue:.4f}')
    print('Medians:', np.median(low_exp), np.median(high_exp))

    ranked=stats.rankdata(np.r_[low_exp,high_exp]); print('Average-tie ranks:',ranked)
    fig,ax=plt.subplots(); ax.scatter(ranked[:len(low_exp)],np.zeros(len(low_exp)),marker='o',s=80,label='low',color=BLUE); ax.scatter(ranked[len(low_exp):],np.ones(len(high_exp)),marker='s',s=80,label='high',color=GOLD)
    ax.set(xlabel='Pooled average rank',yticks=[0,1],yticklabels=['Low experience','High experience'],title='Mann–Whitney compares ranks'); plt.show()

    say(f'Mann–Whitney: U={mw.statistic:.3f}، p={mw.pvalue:.4f}؛ میانه‌ها {np.median(low_exp):.2f} و {np.median(high_exp):.2f}. تعبیر صرفاً اختلاف میانه به شکل مشابه توزیع‌ها نیاز دارد؛ این آزمون رتبه‌ای است.')
    # END PRESERVED ANALYSIS: lesson-23
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-23'] = run_lesson_33()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

رتبه‌ها بزرگی دقیق فاصله‌ها را از بین می‌برند، ولی ترتیب را نگه می‌دارند. وجود ties روی روش محاسبهٔ p اثر می‌گذارد. برای گروه بسیار کوچک همراه ties، آزمون جایگشتی با تعریف روشن می‌تواند بررسی تکمیلی مناسبی باشد؛ عدد روش‌های مختلف ممکن است کمی فرق کند.

#### کاربرد و محدودیت

غیرپارامتری به معنی بی‌فرض نیست؛ استقلال و تعریف جامعه همچنان مهم‌اند. برای اندازه‌گیری قبل و بعدِ افراد یکسان، Mann–Whitney مستقل مناسبِ ساختار داده نیست.

#### تمرین

اگر UA=۷ و nA×nB=۲۴ باشد، UB چقدر است؟

#### پاسخ تشریحی

۱۷. این هویت یک کنترل سادهٔ محاسبه است؛ جهت U باید با ترتیب گروه‌ها خوانده شود.

### درس 34. ANOVA یک‌طرفه

**هدف یادگیری:** ANOVA را با نسبتِ تغییر بین و درون گروه بفهمید.

#### مفهوم و تشریح کامل

آزمون کلیِ ANOVA می‌پرسد آیا میانگین همهٔ گروه‌ها برابر است. تغییر کل به جزء بین گروه و درون گروه تجزیه می‌شود. اگر اختلاف مراکز نسبت به نوسان درون گروه‌ها بزرگ باشد، F افزایش می‌یابد. فرض استقلال را از طراحی می‌گیریم؛ آزمون نرمال‌بودن آن را بررسی نمی‌کند.

#### فرمول و قرارداد

```text
F=(SSbetween/(k−1))/(SSwithin/(N−k)); H0: μ1=μ2=μ3
```

#### مثال حل‌شدهٔ مرجع

گروه جوانِ سن≤۲۲ شامل طول‌های [۱۸،۲۰،۲۲] با میانگین ۲۰ است؛ میانیِ سن۲۳ تا۲۷ شامل [۱۹،۲۱،۲۳،۲۷] با میانگین۲۲٫۵؛ بالاترِ سن≥۲۸ شامل [۲۰،۲۱،۲۴] با میانگین≈۲۱٫۶۶۷ است. F≈۰٫۷۳۴ و p≈۰٫۵۱۳۶ با درجات آزادی۲ و۷ به دست آمده است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_34(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()

    # BEGIN PRESERVED ANALYSIS: lesson-18
    age_group = np.select([df.user_id.isin([1,3,7]), df.user_id.isin([2,5,8,10])],
                          ['young','middle'], default='older')
    df['age_group'] = age_group
    groups_age = [df.loc[df.age_group==g,'hand_length_cm'].to_numpy() for g in ['young','middle','older']]
    f_stat, p_anova = stats.f_oneway(*groups_age)
    display(df.groupby('age_group').hand_length_cm.agg(['count','mean','std']).round(3))
    print(f'F={f_stat:.3f}, p={p_anova:.4f}')

    group_figure(groups_age,['young','middle','older'],'Length by age group (observations, mean and SD)')

    say(f'ANOVA: F={f_stat:.3f} و p={p_anova:.4f}. با وجود اختلاف میانگین‌های نمونه، شواهد کافی برای رد برابری میانگین‌ها در α=0.05 نداریم؛ عدم رد به معنی هم‌ارزی گروه‌ها نیست.')
    # END PRESERVED ANALYSIS: lesson-18
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-18'] = run_lesson_34()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

اختلافِ میانگین‌های نمونه وجود دارد، ولی شواهد این نمونه برای رد برابری میانگین جامعه در سطح انتخابی کافی نیست. «عدم رد» به معنای یکسان‌بودن اندازهٔ دستِ گروه‌های سنی نیست. گروه‌های ۳ یا۴ نفره توان محدودی دارند و بازه‌ها می‌توانند پهن باشند.

#### کاربرد و محدودیت

اگر نتیجهٔ کلی رد شد، باید معلوم شود کدام مقایسه‌ها هدف پژوهش بوده‌اند و برای چندمقایسه‌ای چه روشی به کار می‌رود. انتخاب چند آزمون به امید p کوچک، نتیجه را مخدوش می‌کند.

#### تمرین

با سه گروه و ده مشاهده، درجات آزادی بین و درون گروه چیست؟

#### پاسخ تشریحی

بین گروه k−۱=۲ و درون گروه N−k=۷ است. آمارهٔ F روی همین دو درجهٔ آزادی تفسیر می‌شود.

### درس 35. Welch ANOVA

**هدف یادگیری:** میانگین چند گروه با واریانس‌های متفاوت را مقایسه کنید.

#### مفهوم و تشریح کامل

Welch ANOVA برای هر گروه از حجم و واریانس همان گروه وزن می‌سازد و درجهٔ آزادی مخرج را اصلاح می‌کند. فرضِ واریانس برابر را نیاز ندارد، اما استقلال، تناسب داده و برآورد معقول واریانس هر گروه لازم است. گروه با واریانس صفر یا حجم ناکافی نیاز به بررسی ویژه دارد.

#### فرمول و قرارداد

```text
wi=ni/si²; weighted mean=Σwi×mean_i /Σwi
```

#### مثال حل‌شدهٔ مرجع

برای سه گروه سنی، Welch F≈۰٫۷۸۷، df1=۲، df2≈۴٫۶۵۳ و p≈۰٫۵۰۷۵ است. ANOVA کلاسیک p≈۰٫۵۱۳۶ داده بود؛ هر دو در این نمونه به عدم رد می‌رسند، ولی این دو شاهد مستقل نیستند چون از همان افراد محاسبه شده‌اند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_35(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    age_group = np.select([df.user_id.isin([1,3,7]),df.user_id.isin([2,5,8,10])],['young','middle'],default='older')
    df['age_group'] = age_group
    groups_age = [df.loc[df.age_group==g,'hand_length_cm'].to_numpy() for g in ['young','middle','older']]

    # BEGIN PRESERVED ANALYSIS: lesson-25
    from statsmodels.stats.oneway import anova_oneway
    welch = anova_oneway(groups_age, use_var='unequal', welch_correction=True)
    print(f'Welch F={welch.statistic:.3f}, p={welch.pvalue:.4f}, df={welch.df}')

    group_figure(groups_age,['young','middle','older'],'Welch ANOVA inputs: unequal-variance model')

    say(f'Welch ANOVA: F={welch.statistic:.3f}، p={welch.pvalue:.4f} و df={welch.df}. نابرابری واریانس را مجاز می‌کند، اما کوچک‌بودن و استقلال نمونه‌ها همچنان مهم است؛ رتبه‌ای نیست.')
    # END PRESERVED ANALYSIS: lesson-25
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-25'] = run_lesson_35()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

درجهٔ آزادی کسری خطای محاسبه نیست؛ نتیجهٔ تصحیح Welch است. مقدار F را با درجهٔ آزادی خودش بخوانید. توافق دو روش، محدودیت گروه‌های سه‌نفره و عدم قطعیتِ تخمین اختلاف را حذف نمی‌کند.

#### کاربرد و محدودیت

برای گزارش، حجم، میانگین و SD هر گروه را کنار آزمون بنویسید. در دادهٔ واقعی با ناهمگنی و اندازه‌های نامتوازن، Welch اغلب گزینه‌ای مناسب برای پرسش میانگین‌هاست.

#### تمرین

آیا Welch همان Kruskal–Wallis است؟

#### پاسخ تشریحی

خیر. Welch همچنان میانگین‌ها را با وزن‌دهی واریانسی مقایسه می‌کند؛ Kruskal–Wallis رتبه‌ها و تفاوت توزیع‌ها را بررسی می‌کند.

### درس 36. آزمون تعقیبی Tukey HSD

**هدف یادگیری:** مقایسهٔ چند جفت را بدون افزایش کنترل‌نشدهٔ خطا انجام دهید.

#### مفهوم و تشریح کامل

با k گروه، k(k−1)/۲ جفت وجود دارد. اجرای t جداگانهٔ بدون تعدیل برای همهٔ جفت‌ها، احتمال حداقل یک رد تصادفی را افزایش می‌دهد. Tukey خانوادهٔ مقایسه‌های جفتیِ میانگین را کنترل می‌کند؛ برای حجم‌های نامساوی از صورت Tukey–Kramer استفاده می‌شود.

#### فرمول و قرارداد

```text
SEij=√[(MSE/2)(1/ni+1/nj)]; qij=|mean_i−mean_j|/SEij
```

#### مثال حل‌شدهٔ مرجع

سه گروه سنی سه جفت می‌سازند. در خروجی، تفاوت جوان نسبت به میانی −۲٫۵ سانتی‌متر و p تعدیل‌شده≈۰٫۴۸۷۴ است. بازهٔ هم‌زمان همین مقایسه حدود [−۸٫۶۱۱،۳٫۶۱۱] است و صفر را شامل می‌شود. ترتیب group1/group2 را از سرستون بخوانید، چون علامت اختلاف به آن وابسته است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_36(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    age_group = np.select([df.user_id.isin([1,3,7]),df.user_id.isin([2,5,8,10])],['young','middle'],default='older')
    df['age_group'] = age_group

    # BEGIN PRESERVED ANALYSIS: lesson-21
    from statsmodels.stats.multicomp import pairwise_tukeyhsd
    tukey = pairwise_tukeyhsd(df.hand_length_cm, df.age_group, alpha=.05)
    print(tukey.summary())

    fig,ax=plt.subplots(figsize=(8,4));
    for i,(delta,ci) in enumerate(zip(tukey.meandiffs,tukey.confint)):
        ax.plot(ci,[i,i],color=BLUE); ax.scatter(delta,i,color=GOLD)
    ax.axvline(0,color='0.4',ls='--'); ax.set(yticks=range(3),yticklabels=['middle → older','middle → young','older → young'],xlabel='group2 − group1 length (cm)',title='Tukey simultaneous 95% confidence intervals'); plt.tight_layout(); plt.show()

    say('در جدول Tukey ستون meandiff اختلاف group2−group1 است؛ p-adj برای خانوادهٔ مقایسه‌ها اصلاح شده است. بازه‌های همهٔ زوج‌ها صفر را شامل می‌شوند؛ هیچ زوجی در سطح ۰٫۰۵ رد نمی‌شود، اما برابری اثبات نشده است.')
    # END PRESERVED ANALYSIS: lesson-21
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-21'] = run_lesson_36()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

ستون reject=False به معنی نبودِ رد در سطح تعیین‌شده است. بازهٔ پهن نشان می‌دهد چه اختلاف‌های منفی و مثبتی هنوز با داده سازگارند. به جای تنها انتخاب کوچک‌ترین p، همهٔ مقایسه‌های از پیش تعریف‌شده و مقیاس اختلاف را ببینید.

#### کاربرد و محدودیت

وقتی واریانس‌ها متفاوت‌اند، پرسش تعقیبی مشابه را می‌توان با Games–Howell بررسی کرد. انتخاب آزمون با فرض‌ها و طرح پژوهش تعیین می‌شود، نه اینکه کدام روش نتیجهٔ معنادار تولید کند.

#### تمرین

برای پنج گروه چند مقایسهٔ جفتی داریم؟

#### پاسخ تشریحی

۵×۴÷۲=۱۰ مقایسه. هرچه خانواده بزرگ‌تر شود، لزوم کنترل چندمقایسه‌ای بیشتر به چشم می‌آید.

### درس 37. مقایسهٔ Games–Howell

**هدف یادگیری:** مقایسهٔ جفتیِ میانگین‌ها را با واریانس جدا انجام دهید.

#### مفهوم و تشریح کامل

Games–Howell برای هر جفت، خطای استاندارد و درجهٔ آزادی جدا می‌سازد و تعدیل چندمقایسه‌ای را با توزیع دامنهٔ دانشجوشده انجام می‌دهد. این روش پارامتری است و نام‌گذاریِ «تعقیبی غیرپارامتری» در فایل اولیه درست نبود.

#### فرمول و قرارداد

```text
SEij=√(si²/ni+sj²/nj); q=√2×|mean_i−mean_j|/SEij
```

#### مثال حل‌شدهٔ مرجع

اختلاف جوان−میانی −۲٫۵ با df≈۴٫۸۵ و p تعدیل‌شده≈۰٫۴۹۷۸ است. جوان−بالاتر −۱٫۶۶۷ و p≈۰٫۶۱۵۲؛ میانی−بالاتر +۰٫۸۳۳ و p≈۰٫۹۱۷۴ است. هیچ‌کدام در سطح۰٫۰۵ رد نمی‌شوند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_37(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    age_group = np.select([df.user_id.isin([1,3,7]),df.user_id.isin([2,5,8,10])],['young','middle'],default='older')
    df['age_group'] = age_group
    groups_age = [df.loc[df.age_group==g,'hand_length_cm'].to_numpy() for g in ['young','middle','older']]

    # BEGIN PRESERVED ANALYSIS: lesson-26
    from itertools import combinations
    rows = []
    group_names = ['young','middle','older']
    for i,j in combinations(range(3),2):
        a,b = groups_age[i],groups_age[j]
        va,vb = np.var(a,ddof=1),np.var(b,ddof=1)
        se = np.sqrt(va/len(a)+vb/len(b))
        dof = (va/len(a)+vb/len(b))**2 / ((va/len(a))**2/(len(a)-1)+(vb/len(b))**2/(len(b)-1))
        q = np.sqrt(2)*abs(a.mean()-b.mean())/se
        p = stats.studentized_range.sf(q, len(groups_age), dof)
        rows.append([group_names[i],group_names[j],a.mean()-b.mean(),dof,p])
    display(pd.DataFrame(rows,columns=['group A','group B','mean difference (cm)','df','adjusted p']).round(4))

    gh_table=pd.DataFrame(rows,columns=['group A','group B','mean difference (cm)','df','adjusted p'])
    fig,ax=plt.subplots(); ax.barh([r[0]+' − '+r[1] for r in rows],[r[2] for r in rows],color=BLUE); ax.axvline(0,color='0.3'); ax.set(xlabel='Mean difference (cm)',title='Games–Howell pairwise effect estimates (see adjusted p table)'); plt.tight_layout(); plt.show()

    say(f'در Games–Howell سه مقایسهٔ میانگین با واریانس جدا و p اصلاح‌شده انجام شد. کوچک‌ترین p={gh_table["adjusted p"].min():.4f} است؛ در سطح ۰٫۰۵ هیچ زوج رد نمی‌شود. ستون اختلاف این جدول group A−group B است.')
    # END PRESERVED ANALYSIS: lesson-26
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-26'] = run_lesson_37()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

هر سطر پرسش جدا با برآورد اختلافِ دارای جهت است. بزرگ بودن p به معنی احتمال برابری گروه‌ها نیست. با سه یا چهار نفر در گروه، واریانس تخمینی خودش ناپایدار است و نمی‌توان با عدم رد، محصول مشترک برای همهٔ گروه‌ها را اثبات کرد.

#### کاربرد و محدودیت

در صورت نگرانی دربارهٔ ناهمگنی، Welch و Games–Howell چارچوب سازگارتری برای پرسش میانگین‌ها فراهم می‌کنند. برای پرسش رتبه‌ای یا دادهٔ زوجی باید روش متناسب با همان ساختار انتخاب شود.

#### تمرین

چرا همهٔ جفت‌های Games–Howell لزوماً df یکسان ندارند؟

#### پاسخ تشریحی

چون n و s² هر دو گروه در تصحیح درجهٔ آزادی وارد می‌شوند و ترکیب این مقدارها برای هر جفت متفاوت است.

### درس 38. Kruskal–Wallis

**هدف یادگیری:** مقایسهٔ رتبه‌ایِ بیش از دو گروه را تفسیر کنید.

#### مفهوم و تشریح کامل

Kruskal–Wallis رتبه‌های کل مشاهده‌ها را بین گروه‌ها مقایسه می‌کند و ties را اصلاح می‌کند. فرض عمومی دربارهٔ یکسانی توزیع‌هاست. تنها با شکل‌های قابل مقایسه می‌توان آن را به جابه‌جایی مکان ربط داد؛ هر اختلاف پراکندگی هم ممکن است بر نتیجه اثر بگذارد.

#### فرمول و قرارداد

```text
H=12/[N(N+1)] × Σ(Ri²/ni) −3(N+1), با تصحیحِ ties
```

#### مثال حل‌شدهٔ مرجع

سه گروه سنی در این نمونه H≈۱٫۳۴۱ و p≈۰٫۵۱۱۴ دارند. مجموع رتبه‌ها از همان طول‌های ده نفر به دست آمده است. اعداد۲۰ و۲۱ تکرار دارند و باید رتبهٔ میانگین بگیرند؛ رتبه‌دادنِ متفاوت به دو مقدار برابر نتیجه را تغییر می‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_38(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    age_group = np.select([df.user_id.isin([1,3,7]),df.user_id.isin([2,5,8,10])],['young','middle'],default='older')
    df['age_group'] = age_group
    groups_age = [df.loc[df.age_group==g,'hand_length_cm'].to_numpy() for g in ['young','middle','older']]

    # BEGIN PRESERVED ANALYSIS: lesson-28
    kw = stats.kruskal(*groups_age)
    print(f'H={kw.statistic:.3f}, p={kw.pvalue:.4f}')

    rank_all=stats.rankdata(np.concatenate(groups_age)); starts=np.cumsum([0]+[len(g) for g in groups_age])
    fig,ax=plt.subplots(); ax.boxplot([rank_all[starts[i]:starts[i+1]] for i in range(3)],tick_labels=['young','middle','older']); ax.set(ylabel='Pooled average rank',title='Rank distributions used in Kruskal–Wallis'); plt.show()

    say(f'Kruskal–Wallis: H={kw.statistic:.3f} و p={kw.pvalue:.4f}. در سطح ۰٫۰۵ شواهد کافی برای اختلاف توزیع‌های رتبه‌ای نیست؛ این نتیجه اثبات برابر بودن همهٔ میانه‌های جامعه نیست.')
    # END PRESERVED ANALYSIS: lesson-28
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-28'] = run_lesson_38()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

نتیجهٔ کلی رد نشده است؛ این به معنی یکسان‌بودن همهٔ توزیع‌ها نیست. حتی در صورت رد کلی، آزمون تعیین نمی‌کند کدام جفت متفاوت است؛ پس‌آزمون رتبه‌ای و کنترل چندمقایسه‌ای لازم می‌شود. با گروه‌های بسیار کوچک، تقریب مجانبی را با احتیاط بخوانید.

#### کاربرد و محدودیت

استفاده از روش رتبه‌ای، سوگیری نمونه‌گیری یا وابستگی افراد را حل نمی‌کند. یک داده را با چند آزمون بررسی‌کردن به معنی چند منبع شاهد مستقل نیست.

#### تمرین

وقتی دو فرد مقدار۲۰ دارند، چگونه رتبه می‌گیرند؟

#### پاسخ تشریحی

اگر جایگاه‌های سوم وچهارم باشند، هر دو رتبه۳٫۵ می‌گیرند. مجموع رتبه‌ها همچنان باید N(N+۱)/۲ باشد.

### درس 39. ANOVA عاملی: کنترل شناسایی‌پذیری

**هدف یادگیری:** پیش از آزمون تعامل، امکان شناسایی آن را بررسی کنید.

#### مفهوم و تشریح کامل

در طرح دو عاملی، اثر عامل A، عامل B و تعامل یعنی تغییر اثر A در سطوح B را می‌سنجیم. برای جداسازی این اثرها باید ترکیب‌های عامل‌ها مشاهده شده باشند. خانهٔ خالی یا هم‌پوشانی کامل دو عامل می‌تواند ماتریس طرح را کم‌رتبه کند.

#### فرمول و قرارداد

```text
Y=μ+α(age)+β(experience)+αβ+ε; نخست جدول شمارشِ ترکیب عامل‌ها
```

#### مثال حل‌شدهٔ مرجع

با سن≤۲۴ در برابر سن>۲۴ و سابقه≤۵ در برابر>۵، جدول فقط دو خانهٔ پر دارد: جوان‌تر/کم‌سابقه۴ نفر و بالاتر/پرسابقه۶ نفر؛ دو خانهٔ دیگر صفرند. بنابراین اثر سن و سابقه در این تقسیم از هم جدا نمی‌شوند و جدول p تعامل معتبر تولید نشده است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_39(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    df['experience_group'] = np.where(df.gaming_years<=5,'low','high')

    # BEGIN PRESERVED ANALYSIS: lesson-27
    df['age_binary'] = np.where(df.age_years<=24,'younger','older')
    design = pd.crosstab(df.age_binary, df.experience_group)
    display(design)
    full_design = bool((design>0).all().all())
    print('Full 2×2 design:', full_design)
    if not full_design:
        print('Interaction ANOVA is not estimable from these data: at least one cell is empty.')
    else:
        import statsmodels.formula.api as smf
        from statsmodels.stats.anova import anova_lm
        fit = smf.ols('hand_length_cm ~ C(age_binary)*C(experience_group)', data=df).fit()
        display(anova_lm(fit, typ=2))

    from patsy import dmatrix
    design_matrix=dmatrix('C(age_binary)*C(experience_group)',data=df)
    design_rank=np.linalg.matrix_rank(design_matrix); design_columns=design_matrix.shape[1]
    fig,ax=plt.subplots(figsize=(5,4)); ax.imshow(design,cmap='Blues',vmin=0,vmax=design.to_numpy().max())
    ax.set(xticks=range(2),xticklabels=design.columns,yticks=range(2),yticklabels=design.index,title='Observed 2×2 cell counts (zeros are structural gaps)')
    for i in range(2):
        for j in range(2): ax.text(j,i,str(design.iloc[i,j]),ha='center',va='center',color='black',bbox=dict(facecolor='white',alpha=.8,edgecolor='none'))
    plt.show()

    say(f'طرح کامل ۲×۲؟ {full_design}. دو ترکیب صفر دارند؛ ماتریس طرح رتبهٔ {design_rank} از {design_columns} ستون دارد. در این داده اثر سن و تجربه از هم جدا نمی‌شود؛ خروجی عددیِ اثر متقابل گزارش نمی‌کنیم.')
    # END PRESERVED ANALYSIS: lesson-27
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-27'] = run_lesson_39()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

به جدول شمارش نگاه کنید، نه فقط یک interaction plot که نرم‌افزار ممکن است با خانه‌های حذف‌شده رسم کند. برای تصور تعامل در یک مثال مستقل، پاسخ روش A در مبتدی/حرفه‌ای۶ و۸ و روش B در همان گروه‌ها۷ و۷ باشد: اثر A نسبت به B از−۱ به+۱ عوض می‌شود.

#### کاربرد و محدودیت

برای پژوهش جدید از هر ترکیب عامل نمونه جمع کنید و تعداد کافی در سلول‌ها داشته باشید. اگر تخصیص مشاهده‌ای است، حتی طرحِ قابل برازش نیز علت را بدون کنترل لازم ثابت نمی‌کند.

#### تمرین

آیا نرم‌افزاری که با خانه‌های خالی یک جدول ANOVA می‌دهد، مشکل طراحی را حل کرده است؟

#### پاسخ تشریحی

خیر. ممکن است مدل کاهش‌یافته، قرارداد محدودیت یا نتیجهٔ تکین گزارش شود. اول رتبهٔ طرح و فرض‌های قابل شناسایی را بررسی و صریحاً بیان کنید.

### درس 40. اندازهٔ اثر، اهمیت عملی و آزمون هم‌ارزی

**پرسش شروع:** دو رابط تفاوت معناداری ندارند؛ پس می‌توانیم بگوییم تجربه‌شان یکسان است؟

#### مفهوم، دقیق اما ساده

عدم ردِ اختلاف صفر، اثبات هم‌ارزی نیست. مطالعهٔ کم‌توان ممکن است اختلاف مهمی را تشخیص ندهد. برای هم‌ارزی، حاشیهٔ عملی Δ را پیش از دیدن نتیجه تعریف می‌کنیم و می‌پرسیم آیا اثر با اطمینان کافی داخل (−Δ,+Δ) قرار دارد. در TOST دو فرض یک‌طرفه دربارهٔ مرزهای بد بودن بررسی می‌شوند. با سطح هر آزمون ۰٫۰۵، معادلِ متناظر بر مبنای مدل t این است که CI دوطرفهٔ ۹۰٪ کاملاً در حاشیه باشد؛ CI نوددرصدی برای این تصمیم به معنی شل‌کردن خودسرانهٔ استاندارد نیست. اندازهٔ اثر استانداردشده برای مقایسهٔ مقیاس‌ها مفید است اما واحد عملی، هزینه و دامنهٔ اثر را جایگزین نمی‌کند.

#### فرمول و قرارداد

```text
Equivalence: −Δ < effect < +Δ
TOST α=.05 ↔ model-based 90% two-sided CI fully inside the equivalence margin.
```

#### مثال حل‌شده

هشت اختلافِ امتیاز راحتیِ «جدید منهای قدیم» برابر ۰٫۱،۰٫۲،۰،−۰٫۱،۰٫۱،۰٫۲،−۰٫۲،۰٫۱ بسازید. حاشیهٔ مهم را از پیش ±۰٫۵ امتیاز قرار می‌دهیم. میانگین، CI نوددرصدی، دو p یک‌طرفه و d_z زوجی محاسبه می‌شوند. اگر حاشیه ±۰٫۰۵ انتخاب شده بود، نتیجه می‌توانست متفاوت شود؛ حاشیه تصمیم طراحی است نه چیزی که از p استخراج شود. این مثال اندازه‌گیری تکراریِ هشت فرد مستقل با فرض مدل t برای اختلاف‌هاست.

````python
def run_lesson_40(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: equivalence
    rng=np.random.default_rng(202611)
    diff=np.array([.1,.2,0,-.1,.1,.2,-.2,.1]); margin=.5; avg=diff.mean(); se=stats.sem(diff); ci=stats.t.interval(.9,len(diff)-1,loc=avg,scale=se)
    p_lower=stats.ttest_1samp(diff,-margin,alternative='greater').pvalue; p_upper=stats.ttest_1samp(diff,margin,alternative='less').pvalue
    fig,ax=plt.subplots(figsize=(9,3)); ax.axvspan(-margin,margin,color=GOLD,alpha=.18,label='Predeclared equivalence region'); ax.errorbar(avg,0,xerr=[[avg-ci[0]],[ci[1]-avg]],fmt='o',color=BLUE,capsize=8,label='90% t CI'); ax.axvline(0,color='0.4',ls='--'); ax.set(yticks=[],xlabel='Paired comfort-score difference (new − old)',title='Equivalence asks more than failure to reject zero'); ax.legend(); show()
    say(f'اثر میانگین={avg:.3f}، CI نوددرصدی=[{ci[0]:.3f},{ci[1]:.3f}]؛ pهای TOST={p_lower:.5f} و {p_upper:.5f}. هم‌ارزی در Δ=۰٫۵؟ {max(p_lower,p_upper)<.05}. d_z={avg/diff.std(ddof=1):.3f} واحد استاندارد اختلاف‌هاست، نه امتیاز راحتی.')
    # END PRESERVED ANALYSIS: equivalence
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['equivalence'] = run_lesson_40()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

در نمودار، ناحیهٔ طلایی محدودهٔ هم‌ارزی عملی و خط آبی CI نوددرصدی اثر میانگین است. داخل‌بودن کل بازه، نه فقط نقطهٔ مرکزی، اهمیت دارد. اگر بازه از مرز بیرون برود و آزمون اختلاف هم رد نشود، نتیجه «نامطمئن» است؛ نه برابر و نه حتماً متفاوت. برای تحلیل واقعی، حاشیه را با کاربر و معیار عملکرد مستند کنید.

#### دام رایج

کوچک‌کردن حاشیه بعد از دیدن داده یا تبدیل p>۰٫۰۵ به «دو روش یکسان‌اند» نتیجهٔ معتبر نمی‌دهد.

#### تمرین

اگر CI از −۱ تا۱ و Δ=۰٫۵ باشد، آیا عدم معناداری، هم‌ارزی را تأیید می‌کند؟

#### پاسخ تشریحی

خیر؛ بازه اثرهای مهم خارج حاشیه را هم مجاز می‌داند. داده برای ادعای هم‌ارزی کافی نیست و افزایش اطلاعات یا بازبینی طرح لازم است.

### درس 41. چندآزمونی، Bonferroni و نرخ کشف کاذب

**پرسش شروع:** از میان هزار ویژگی، چند مورد p کوچک دارند؛ آیا کشف تازه‌ای کرده‌ایم؟

#### مفهوم، دقیق اما ساده

هر بار آزمون‌کردن فرصتی برای رد اشتباه می‌سازد. کنترل خطای خانواده‌ای یا FWER احتمال حداقل یک رد کاذب در کل خانواده را محدود می‌کند؛ Bonferroni با آستانهٔ α/m بدون نیاز به استقلال، محافظه‌کارانه این هدف را دنبال می‌کند. FDR به امیدِ نسبتِ کشف‌های کاذب به کل کشف‌ها مربوط است، با قرارداد صفر در نبود کشف. روش Benjamini–Hochberg با فرض استقلال یا برخی وابستگی‌های مثبت، FDR را کنترل می‌کند؛ برای وابستگی عمومی روش‌هایی مثل BY محافظه‌کارترند. FDR=۵٪ تضمین نمی‌کند در هر اجرای خاص دقیقاً ۵٪ یافته‌ها غلط‌اند، و هر p تعدیل‌شده احتمال نادرستی فرضیه نیست.

#### فرمول و قرارداد

```text
Bonferroni: reject p≤α/m
BH: sort p_(i); find largest k with p_(k)≤(k/m)q; reject ranks 1..k.
```

#### مثال حل‌شده

هشت p آموزشی برابر ۰٫۰۰۱،۰٫۰۱۲،۰٫۰۱۸،۰٫۰۴،۰٫۲،۰٫۳،۰٫۶ و۰٫۹ داریم. بدون اصلاح چهار مورد زیر۰٫۰۵ هستند. Bonferroni فقط اولی و BH با q=۰٫۰۵ سه مورد اول را نگه می‌دارد. این pها فهرست ساخته‌شده‌اند؛ چون حقیقت هر فرضیه را در این فهرست تعیین نکرده‌ایم، تعداد واقعی خطاهای کاذب را از روی آن‌ها نمی‌دانیم. برای هزار سنجهٔ اکتشافی، تعریف خانواده باید از سؤال تحلیل بیاید، نه بعد از دیدن اینکه کدام اصلاح مطلوب‌تر است.

````python
def run_lesson_41(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: fdr
    rng=np.random.default_rng(202612)
    from statsmodels.stats.multitest import multipletests
    ps=np.array([.001,.012,.018,.04,.2,.3,.6,.9]); bonf=multipletests(ps,method='bonferroni'); bh=multipletests(ps,method='fdr_bh')
    display(pd.DataFrame({'raw p':ps,'Bonferroni adjusted':bonf[1],'BH adjusted':bh[1],'BH reject':bh[0]}))
    rank=np.arange(1,len(ps)+1); fig,ax=plt.subplots(); ax.plot(rank,ps,marker='o',color=BLUE,label='Sorted p-values'); ax.plot(rank,.05*rank/len(ps),color=GOLD,label='BH rank thresholds'); ax.axhline(.05/len(ps),color='0.4',ls='--',label='Bonferroni'); ax.set(xlabel='Ordered test rank',ylabel='p / threshold (log scale)',yscale='log',title='Different error-control objectives'); ax.legend(); show()
    say(f'کشف خام={sum(ps<.05)}، Bonferroni={bonf[0].sum()} و BH={bh[0].sum()}. بدون دانستن حقیقت فرضیه‌های این فهرستِ آموزشی، تعداد کشف کاذب مشاهده‌شده را نمی‌توان اعلام کرد.')
    # END PRESERVED ANALYSIS: fdr
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['fdr'] = run_lesson_41()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

در نمودار، نقاط pهای مرتب و خط BH آستانهٔ وابسته به رتبه است. فقط مقایسهٔ جداگانهٔ هر نقطه با خط کافی نیست؛ ابتدا بزرگ‌ترین رتبهٔ قابل‌قبول را بیابید و همهٔ رتبه‌های پیش از آن را در نظر بگیرید. جدول کنار شکل، p تعدیل‌شده و تصمیم هر روش را نشان می‌دهد. کشف آماری باید با اندازهٔ اثر، کیفیت داده و تکرار مستقل همراه شود.

#### دام رایج

فقط pهای معنادار را وارد اصلاح نکنید؛ این کار تعداد آزمون‌های واقعاً انجام‌شده را پنهان می‌کند.

#### تمرین

آیا با BH در q=۰٫۰۵ می‌توان گفت احتمال درست‌بودن هر یافته ۹۵٪ است؟

#### پاسخ تشریحی

خیر؛ این روش دربارهٔ نرخ مورد انتظارِ کشف‌های کاذب در یک فرایند تکرارشونده تحت فرض‌هاست، نه احتمال پسین یک فرضیهٔ مشخص.

### درس 42. آزمایش A/B، تصادفی‌سازی، زوج‌سازی و آزمون جایگشتی

**پرسش شروع:** هر شخص هر دو رابط را امتحان کرده؛ چرا دادهٔ زوجی را مثل دو جمعیت مستقل تحلیل نکنیم؟

#### مفهوم، دقیق اما ساده

تصادفی‌سازی کمک می‌کند تخصیص درمان یا طراحی، در انتظار با عوامل پیشین مستقل باشد؛ اما اجرای بد، ریزش متفاوت و سرایت اثر بین افراد همچنان خطرند. در طرح زوجی، اختلاف درون فرد اغلب بخشی از تنوع بین افراد را حذف می‌کند. ترتیب استفاده باید متوازن یا تصادفی شود تا اثر یادگیری با طراحی قاطی نشود. آزمون تصادفی‌سازی از سازوکار تخصیص استفاده می‌کند؛ آزمون جایگشتیِ برچسب‌های دلخواه بدون فرض تبادل‌پذیری همیشه معتبر نیست. برای زوج‌های تصادفی با فرض صفر تیز، تغییر علامت اختلاف هر زوج تخصیص ممکن دیگری را نشان می‌دهد. دادهٔ مشاهده‌ای صرف را با جابه‌جاکردن برچسب‌ها به آزمایش علّی تبدیل نمی‌کنیم.

#### فرمول و قرارداد

```text
Paired effect = mean(new_i−old_i)
Randomization null: compare observed statistic to allowed assignment statistics.
```

#### مثال حل‌شده

هشت اختلاف زمانی ۲،۱،۳،۲،۰،۱،−۱،۲ میلی‌ثانیه را برای «قدیم منهای جدید» بسازید؛ مثبت یعنی رابط جدید سریع‌تر است. تحت تخصیص زوجی تصادفی و فرض صفر مناسب، ۲^۸=۲۵۶ الگوی علامت ممکن را کامل می‌شماریم. p دوطرفه سهم الگوهایی است که قدر مطلق میانگینشان حداقل میانگین مشاهده‌شده است. چون همهٔ حالت‌ها شمرده شده‌اند، خطای شبیه‌سازی مونت‌کارلو نداریم؛ ولی انتخاب آماره و مفروضات طرح هنوز مهم‌اند.

````python
def run_lesson_42(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: randomization
    rng=np.random.default_rng(202613)
    from itertools import product
    delta=np.array([2,1,3,2,0,1,-1,2],dtype=float); signs=np.array(list(product([-1,1],repeat=len(delta)))); null_stats=(signs*delta).mean(axis=1); obs=delta.mean(); exact_p=np.mean(np.abs(null_stats)>=abs(obs)-1e-12)
    fig,ax=plt.subplots(); ax.hist(null_stats,bins=np.arange(-2.6,2.8,.25),color=BLUE,edgecolor='white'); ax.axvline(obs,color=GOLD,lw=2); ax.axvline(-obs,color=GOLD,ls='--'); ax.set(xlabel='Mean paired improvement under allowed sign flips (ms)',ylabel='Assignments out of 256',title='Exact paired randomization demonstration'); show()
    say(f'میانگین بهبود={obs:.3f} میلی‌ثانیه و p دقیق دوطرفه={exact_p:.5f} در {len(signs)} حالت است. تفسیر آزمون به تصادفی‌سازی زوج‌ها/فرض صفر متناسب وابسته است؛ این داده‌ها ساخته‌شده‌اند.')
    # END PRESERVED ANALYSIS: randomization
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['randomization'] = run_lesson_42()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

هیستوگرام، توزیع آماره تحت تخصیص‌های مجازِ فرض صفر است، نه توزیع اثر واقعی در جامعه. دو خط مرزهای ±قدر مطلق اثر مشاهده‌شده‌اند. ستون‌های بیرون این مرزها در صورت p شمرده می‌شوند. با n کم، p گسسته است؛ اضافه‌کردن رقم اعشار به معنای دقت علمی بیشتر نیست. در مطالعهٔ واقعی، CI و هزینهٔ عملی چند میلی‌ثانیه را هم گزارش کنید.

#### دام رایج

اگر یک شخص در هر دو گروه باشد، دو گروه مستقل نداریم. اگر همیشه طرح جدید دوم اجرا شود، اثر طراحی و یادگیری ممکن است تفکیک‌ناپذیر شوند.

#### تمرین

چرا ترتیب استفاده از دو رابط باید تصادفی یا متوازن باشد؟

#### پاسخ تشریحی

خستگی و تمرین می‌توانند در مرحلهٔ دوم نتیجه را عوض کنند. ترتیب متوازن این عامل را با طراحی یکی نمی‌کند و باید در طرح و تحلیل لحاظ شود.

### درس 43. نگاه‌کردن مکرر به p و توقف زودهنگام

**پرسش شروع:** آزمایش را هر روز نگاه می‌کنیم و اولین روز p<۰٫۰۵ شد متوقف می‌کنیم؛ چه چیزی خراب می‌شود؟

#### مفهوم، دقیق اما ساده

آزمون با حجم ثابت برای یک برنامهٔ نمونه‌گیری مشخص کالیبره شده است. اگر در هر نگاه فرصت تازه‌ای برای اعلام موفقیت داشته باشیم، خطای کل فرایند می‌تواند از پنج درصد بیشتر شود، حتی وقتی هر مقایسه ظاهراً با α=۰٫۰۵ انجام شده است. نگاه‌های تجمعی مستقل نیستند، پس فرمول ۱−۰٫۹۵^m را مستقیم برای آن‌ها به کار نمی‌بریم. در برنامهٔ از پیش مشخصِ m نگاه، Bonferroni یک اصلاح ساده و محافظه‌کارانه است؛ طراحی‌های گروه‌متوالی، تخصیص α و روش‌های always-valid ابزارهای دقیق‌ترند. قاعدهٔ توقف، معیار اصلی و تحلیل ریزش باید قبل از اجرای واقعی مشخص شوند.

#### فرمول و قرارداد

```text
Planned m looks: a conservative boundary is p≤α/m at each look.
Dependent cumulative looks do not satisfy 1−(1−α)^m.
```

#### مثال حل‌شده

در ۳٬۰۰۰ آزمایش ساختگی، هر دو گروه نرخ موفقیت واقعی ۰٫۵ دارند. از ۵۰ تا۱٬۰۰۰ نفر در هر گروه، هر۵۰ نفر یک بار z تقریبی اختلاف نرخ‌ها را نگاه می‌کنیم. سه قاعده را مقایسه می‌کنیم: فقط نگاه آخر، حداقل یکی از۲۰ نگاه با p<۰٫۰۵، و حداقل یکی با p<۰٫۰۰۲۵. این یک شبیه‌سازی آموزشی با آزمون نرمال تقریبی است؛ خطای نوع اولِ اصلاح‌شده نیز به معتبر بودن pهای هر نگاه وابسته است.

````python
def run_lesson_43(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: peeking
    rng=np.random.default_rng(202614)
    trials=3000; max_n=1000; looks=np.arange(50,max_n+1,50)
    aa=rng.binomial(1,.5,size=(trials,max_n)).cumsum(axis=1)[:,looks-1]/looks
    bb=rng.binomial(1,.5,size=(trials,max_n)).cumsum(axis=1)[:,looks-1]/looks
    z=(bb-aa)/np.sqrt(.5/looks); pv=2*stats.norm.sf(np.abs(z))
    false_rates=np.array([(pv[:,-1]<.05).mean(),(pv<.05).any(axis=1).mean(),(pv<.05/len(looks)).any(axis=1).mean()])
    fig,ax=plt.subplots(); ax.bar(['Final look only','Any of 20 looks','20-look Bonferroni'],100*false_rates,color=[BLUE,GOLD,TEAL]); ax.axhline(5,color='0.3',ls='--',label='Nominal 5%'); ax.set(ylabel='Simulated false-discovery experiments (%)',title='No treatment effect in any simulated experiment'); ax.legend(); show()
    say(f'نرخ رد کاذب تجربی: فقط پایان={false_rates[0]:.2%}؛ نگاه مکرر خام={false_rates[1]:.2%}؛ اصلاح ۲۰نگاهی={false_rates[2]:.2%}. این برآورد از ۳٬۰۰۰ تکرارِ سناریوی بدون اثر است، نه نرخ قطعی هر آزمایش واقعی.')
    # END PRESERVED ANALYSIS: peeking
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['peeking'] = run_lesson_43()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

میله‌ها سهم آزمایش‌های بدون اثرند که اشتباهاً «کشف» اعلام کرده‌اند؛ این سهم توان نیست. قاعدهٔ نگاه مکرر معمولاً میلهٔ بلندتری دارد. اعداد دقیق با بذر و تعداد تکرار تغییر می‌کنند. Bonferroni برای آموزش هزینهٔ چندفرصتی است، نه پیشنهاد خودکار برای همهٔ پلتفرم‌های A/B؛ طراحی متوالی حرفه‌ای باید متناسب با هزینه و اطلاعات هر مرحله باشد.

#### دام رایج

توقف پس از نخستین p کوچک و گزارش همان p به‌عنوان آزمونِ ازپیش‌طراحی‌شده، فرایند انتخاب را پنهان می‌کند.

#### تمرین

آیا چون نگاه‌ها به دادهٔ تجمعی وابسته‌اند، دیگر خطای چندبار نگاه‌کردن وجود ندارد؟

#### پاسخ تشریحی

نه. وابستگی فقط مقدار تورم خطا را عوض می‌کند و فرمول استقلال را نامناسب می‌سازد؛ کالیبراسیونِ قاعدهٔ توقف همچنان لازم است.

## فصل 07: از همبستگی تا پرسش علّی

رابطهٔ مشاهده‌شده را از اثر مداخله جدا می‌کنیم؛ سیمپسون، نمودار علّی و وزن‌دهی نشان می‌دهند چرا یک ضریب کافی نیست.

### درس 44. نقشهٔ همبستگی

**هدف یادگیری:** ماتریس همبستگی و حدود نتیجه‌گیری آن را بخوانید.

#### مفهوم و تشریح کامل

هر خانه رابطهٔ خطیِ دو ستون را خلاصه می‌کند. ماتریس متقارن است و قطر برای ستونِ دارای پراکندگی مقدار ۱ دارد. رنگ فقط راهنمای الگو است؛ مقدار عددی، حجم نمونه و نمودار پراکندگی باید همراهش خوانده شوند. همبستگی صفر نیز استقلال کامل متغیرها را ثابت نمی‌کند.

#### فرمول و قرارداد

```text
rXY=Σ(x−x̄)(y−ȳ)/√[Σ(x−x̄)²Σ(y−ȳ)²]
```

#### مثال حل‌شدهٔ مرجع

در خروجی گرد‌شده، طول و عرض r≈۰٫۹۹، سن و سابقه r≈۰٫۹۸، طول و سن r≈۰٫۴۳ و طول و سابقه r≈۰٫۵۸ دارند. بنابراین ادعای اولیهٔ استقلال ابعاد دست از سن از همین جدول به دست نمی‌آید. سن و سابقهٔ بسیار همبسته می‌توانند در مدل چندگانه هم‌خطی ایجاد کنند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_44(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()

    # BEGIN PRESERVED ANALYSIS: lesson-14
    numeric_cols = ['hand_length_cm','hand_width_cm','age_years','gaming_years']
    corr = df[numeric_cols].corr()
    display(corr.round(2))
    fig, ax = plt.subplots(figsize=(6,5))
    im = ax.imshow(corr, vmin=-1, vmax=1, cmap='BrBG')
    ax.set_xticks(range(4), numeric_cols, rotation=30, ha='right')
    ax.set_yticks(range(4), numeric_cols)
    for i in range(4):
        for j in range(4): ax.text(j,i,f'{corr.iloc[i,j]:.2f}',ha='center',va='center',bbox=dict(facecolor='white',alpha=.85,edgecolor='none'))
    fig.colorbar(im, ax=ax, label='Pearson r')
    ax.set_title('Correlation among measured attributes')
    plt.tight_layout(); plt.show()

    say(f'پیرسون طول–عرض={corr.iloc[0,1]:.3f} و سن–سابقه={corr.iloc[2,3]:.3f} است. وابستگی شدید پیش‌بین‌ها می‌تواند ضرایب رگرسیون چندگانه را ناپایدار کند؛ رابطهٔ عددی علّیت نیست.')
    # END PRESERVED ANALYSIS: lesson-14
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-14'] = run_lesson_44()
plt.close('all')
````

#### کاربرد و محدودیت

در پروژهٔ همبستگی، دما با اجاره روی ۱۷٬۳۷۹ ساعت بررسی می‌شود. وابستگی زمان و ترکیب فصل‌ها همچنان مهم است؛ زیادبودن تعداد سطر، همهٔ اشکال نمونه‌گیری را برطرف نمی‌کند.

#### تمرین

#### پاسخ تشریحی

### درس 45. ماتریس پراکندگی و بررسی رابطهٔ غیرخطی

**هدف یادگیری:** پشت هر خانهٔ همبستگی، نقاط واقعی را ببینید.

#### مفهوم و تشریح کامل

در ماتریس پراکندگی، هر خانه یک زوج ویژگی را رسم می‌کند؛ محورهای سطر و ستون تعیین می‌کنند چه چیزی مقایسه می‌شود. خانه‌های دو سوی قطر اطلاعات جفتی مشابه با محورهای جابه‌جا دارند. قطر را می‌توان با هیستوگرام همان متغیر پر کرد. این نمایش، رابطهٔ خمیده، تجمع، ties و نقاط اثرگذار را از یک عدد همبستگی بهتر آشکار می‌کند.

#### فرمول و قرارداد

```text
چهار ویژگی → ۴×۴ خانه؛ تعداد زوج‌های یکتا = 4×3/2 = 6
```

#### مثال حل‌شدهٔ مرجع

در دادهٔ نمونه، دو زوج طول–عرض و سن–سابقه همبستگی نزدیک۱ دارند. اما در زوج طول–سن r≈۰٫۴۳ است و تنها ده نقطه وجود دارد. در مثال جداگانه X=[−۲،−۱،۰،۱،۲] وY=[۴،۱،۰،۱،۴]، پیرسون صفر می‌شود ولی شکلU کاملاً مشخص است. بنابراین نقشهٔ رنگیِ نزدیک صفر ممکن است یک رابطهٔ مهم را پنهان کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_45(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    numeric_cols = ['hand_length_cm','hand_width_cm','age_years','gaming_years']

    # BEGIN PRESERVED ANALYSIS: scatter-matrix
    from pandas.plotting import scatter_matrix
    scatter_matrix(df[numeric_cols],diagonal='hist',figsize=(10,10),alpha=.8,color=BLUE)
    plt.suptitle('Four measured attributes: pairwise observations and histograms',y=1.01); plt.show()
    u=np.arange(-2,3); v=u**2
    fig,ax=plt.subplots(); ax.scatter(u,v,color=GOLD,s=70); ax.set(xlabel='X (constructed)',ylabel='Y = X²',title=f'Nonlinear dependence despite Pearson r={stats.pearsonr(u,v).statistic:.2f}'); plt.show()
    say(f'چهار ویژگی شش زوج یکتا می‌سازند. در مثال ساخته‌شدهٔ Uشکل، r={stats.pearsonr(u,v).statistic:.2f} ولی Y دقیقاً تابع X است؛ پس همبستگی صفر به معنی نبود هر رابطه نیست. مقیاس هر محور را جدا بخوانید.')
    # END PRESERVED ANALYSIS: scatter-matrix
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['scatter-matrix'] = run_lesson_45()
plt.close('all')
````

### درس 46. پارادوکس سیمپسون: وقتی جمع‌بندی جهت را برمی‌گرداند

**پرسش شروع:** روش B در کارهای آسان و سخت بهتر است، اما در کل بدتر دیده می‌شود. تناقض کجاست؟

#### مفهوم، دقیق اما ساده

نرخ کل میانگین وزنی نرخ‌های زیرگروه است؛ اگر وزن زیرگروه‌ها بین دو روش متفاوت باشد، رابطهٔ تجمیعی ممکن است جهت متفاوتی از رابطهٔ درون گروه‌ها داشته باشد. این پدیده تناقض حساب نیست؛ مخرج‌های متفاوت را مخلوط کرده‌ایم. اگر سختی کار هم بر انتخاب روش و هم بر موفقیت اثر بگذارد، یک مخدوش‌کنندهٔ محتمل است. اما «همیشه زیرگروه‌بندی کن» راه‌حل عمومی نیست؛ اگر متغیر انتخابی collider یا پیامدِ درمان باشد، شرطی‌کردن می‌تواند سوگیری ایجاد کند. برای تصمیم علّی، ساختار زمانی و فرض دربارهٔ روابط لازم است. برای توصیف نیز باید وزن‌های جامعهٔ هدف را روشن کرد.

#### فرمول و قرارداد

```text
Rate(A)=Σ P(group|A)×Rate(A,group)
Standardized rate uses the same declared group weights for both methods.
```

#### مثال حل‌شده

در کار آسان A برابر۸۱/۹۰=۹۰٪ و B برابر۱۹/۲۰=۹۵٪ موفق است. در کار سخت A برابر۱/۱۰=۱۰٪ و B برابر۱۶/۸۰=۲۰٪ است. با وجود برتری B در هر دو، کل A برابر۸۲٪ و کل B برابر۳۵٪ می‌شود، چون بیشتر کارهای B سخت‌اند. با وزن هدفِ برابرِ آسان/سخت، A=۵۰٪ و B=۵۷٫۵٪ است. این وزن۵۰/۵۰ صرفاً هدف فرضی ماست، نه ترکیب واقعی هر بازار.

````python
def run_lesson_46(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: simpson
    rng=np.random.default_rng(202617)
    success=np.array([[81,1],[19,16]]); trials=np.array([[90,10],[20,80]]); group_rates=success/trials; overall=success.sum(axis=1)/trials.sum(axis=1); standard=group_rates@np.array([.5,.5])
    display(pd.DataFrame({'method':['A','B'],'easy success':success[:,0],'easy n':trials[:,0],'hard success':success[:,1],'hard n':trials[:,1],'pooled rate':overall,'50:50 standardized':standard}))
    fig,ax=plt.subplots(); positions=np.arange(3); ax.bar(positions-.17,100*np.r_[group_rates[0],overall[0]],width=.34,color=BLUE,label='A'); ax.bar(positions+.17,100*np.r_[group_rates[1],overall[1]],width=.34,color=GOLD,label='B'); ax.set(xticks=positions,xticklabels=['Easy','Hard','Pooled actual mix'],ylabel='Success rate (%)',ylim=(0,100),title='B wins within groups; A appears better in the pooled mix'); ax.legend(); show()
    say(f'تجمیع خام: A={overall[0]:.0%} و B={overall[1]:.0%}. با وزن مشترک ۵۰/۵۰: A={standard[0]:.1%} و B={standard[1]:.1%}. تغییر جهت از وزن‌های متفاوت می‌آید؛ اثر علّی بدون طرح و فرض تعیین نمی‌شود.')
    # END PRESERVED ANALYSIS: simpson
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['simpson'] = run_lesson_46()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

سه خوشهٔ ستون را جدا بخوانید: آسان، سخت و تجمیع واقعی. برچسب n در جدول نشان می‌دهد چرا ستون آخر از نظر ترکیب همسان نیست. برآورد استانداردشده پاسخ یک جامعهٔ هدف مشترک است؛ بدون فرض عدم مخدوش‌سازیِ باقیمانده، آن را اثر علّی روش ننامید.

#### دام رایج

یک میانگین کل ممکن است تفاوت ترکیب را بازتاب دهد. از طرف دیگر، تنظیم برای هر ستون موجود هم الزاماً درست نیست؛ رابطهٔ علّی متغیر مهم است.

#### تمرین

اگر فقط نرخ کل ۸۲٪ و۳۵٪ را ببینیم، چه اطلاعاتی پیش از تصمیم لازم داریم؟

#### پاسخ تشریحی

تعداد و موفقیت در سطح سختی، منطق تخصیص روش، تعریف موفقیت و ترکیب هدف. سپس می‌توان توصیف همسان‌شده و محدودیت علّی آن را گزارش کرد.

### درس 47. نمودار علّی: مخدوش‌کننده، میانجی و Collider

**پرسش شروع:** چرا اضافه‌کردن یک متغیر کنترلی گاهی نتیجه را خراب می‌کند؟

#### مفهوم، دقیق اما ساده

در DAG پیکان، فرض جهت اثر علّی است و صرفاً از همبستگی یاد گرفته نمی‌شود. مخدوش‌کنندهٔ پیشین C می‌تواند علت مشترک X وY باشد؛ کنترل مناسب آن مسیر پشت‌در را می‌بندد. میانجی M روی مسیر X→M→Y بخشی از اثر کل را منتقل می‌کند؛ کنترل M سؤال را به اثر مستقیم و فرض‌های تازه تغییر می‌دهد. Collider در ساختار X→S←Y معلول مشترک است؛ شرطی‌کردن روی S یا برخی فرزندانش می‌تواند بین X وY رابطه بسازد، حتی اگر قبلاً مستقل باشند. انتخاب افراد پذیرفته‌شده یا سفارش‌های رسیدگی‌شده می‌تواند چنین سازوکاری داشته باشد. کنترل بیشتر، مترادف سوگیری کمتر نیست.

#### فرمول و قرارداد

```text
Confounder: X←C→Y
Mediator: X→M→Y
Collider: X→S←Y; conditioning on S can open a path.
```

#### مثال حل‌شده

دو توانایی مستقل X وY را نرمال بسازید. پذیرش به مجموع آن‌ها با نویز کم وابسته باشد: S=X+Y+ε. اگر فقط افراد با S نزدیک صفر را ببینیم، فردی با X بالا معمولاً Y پایین دارد تا جمع نزدیک صفر بماند. در کل جامعه همبستگی نزدیک صفر و در این انتخاب همبستگی منفی قوی ظاهر می‌شود. این یک سازوکار انتخاب ساخته‌شده است؛ ثابت نمی‌کند پذیرش واقعی دانشگاه یا استخدام خاصی همین ساختار را دارد.

````python
def run_lesson_47(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: dag
    rng=np.random.default_rng(202618)
    xx=rng.normal(size=8000); yy=rng.normal(size=8000); selection=xx+yy+rng.normal(scale=.15,size=8000); chosen=np.abs(selection)<.15
    fig,axes=plt.subplots(1,2,figsize=(11,4.5),layout='constrained'); axes[0].axis('off')
    for name,pos in [('X',(.15,.7)),('Y',(.85,.7)),('S',(.5,.25))]: axes[0].text(*pos,name,ha='center',va='center',fontsize=20,bbox=dict(boxstyle='circle',facecolor='#edf4f6',edgecolor=BLUE),transform=axes[0].transAxes)
    for start in [(.2,.65),(.8,.65)]: axes[0].annotate('',xy=(.5,.31),xytext=start,xycoords='axes fraction',arrowprops=dict(arrowstyle='->',color=BLUE,lw=2))
    axes[0].set_title('Constructed collider: select on S')
    axes[1].scatter(xx,yy,s=6,color='0.7',alpha=.2,label='All'); axes[1].scatter(xx[chosen],yy[chosen],s=12,color=BLUE,label='|S| < 0.15'); axes[1].set(xlabel='Independent X before selection',ylabel='Independent Y before selection',title='Selection induces an association'); axes[1].legend(); show()
    say(f'r در کل={np.corrcoef(xx,yy)[0,1]:.3f}؛ در {chosen.sum()} مورد انتخاب‌شده={np.corrcoef(xx[chosen],yy[chosen])[0,1]:.3f}. این مثال نشان می‌دهد چرا شرطی‌کردن بر collider می‌تواند رابطه بسازد.')
    # END PRESERVED ANALYSIS: dag
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['dag'] = run_lesson_47()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

در پنل چپ دو پیکان به S می‌رسند؛ این یک ادعای سازندهٔ سناریوست. در پنل راست نقاط خاکستری کل نمونه و نقاط آبی زیرگروه انتخاب‌شده‌اند. جهت منفی در انتخاب، رابطهٔ تازه‌ای است که از شرط بر معلول مشترک ایجاد شده؛ آن را به جامعهٔ کامل تعمیم ندهید. حذف اثر انتخاب نیز بدون دانستن سازوکار انتخاب همیشه ممکن نیست.

#### دام رایج

کنترل متغیرِ پس از درمان می‌تواند بخشی از اثر را حذف یا مسیر سوگیری باز کند. فهرست بلند کنترل‌ها جای دانش مسئله را نمی‌گیرد.

#### تمرین

برای اثر کل طراحی بر خستگی، آیا باید زمان استفاده‌ای را که خود طراحی تغییر داده همیشه کنترل کنیم؟

#### پاسخ تشریحی

نه لزوماً. اگر زمان استفاده میانجی باشد، کنترل آن اثر کل را حذف می‌کند و کمیت هدف عوض می‌شود. باید اثر کل/مستقیم، ترتیب زمانی و فرض‌ها پیش از تحلیل روشن شوند.

### درس 48. اثر علّی، پیامد بالقوه و وزن‌دهی Propensity

**پرسش شروع:** کاربران حرفه‌ای بیشتر طرح جدید را انتخاب کرده‌اند؛ اختلاف عملکرد واقعاً اثر طرح است؟

#### مفهوم، دقیق اما ساده

برای هر فرد، پیامد بالقوه Y(1) وY(0) را تصور می‌کنیم؛ اثر متوسط ATE میانگین تفاوت آن‌هاست، اما برای یک فرد معمولاً فقط یکی مشاهده می‌شود. آزمایش تصادفی کمک می‌کند گروه‌ها قابل‌مقایسه شوند. در دادهٔ مشاهده‌ای، شناسایی با وزن propensity به فرض‌هایی مانند سازگاریِ تعریف درمان، عدم مخدوش‌سازی اندازه‌گیری‌نشده، مثبت‌بودن احتمال هر درمان در سطح ویژگی‌ها و نبود تداخلِ نامتناسب متکی است. نمره propensity احتمال دریافت درمان با توجه به متغیرهای پیش از درمان است، نه احتمال موفقیت پیامد. وزن‌های معکوس گروه‌ها را به یک توزیع هدف نزدیک می‌کنند؛ اما فرض‌های شناسایی را از داده اثبات نمی‌کنند.

#### فرمول و قرارداد

```text
ATE=E[Y(1)−Y(0)]
e(X)=P(T=1|X); weights=T/e(X)+(1−T)/(1−e(X)).
```

#### مثال حل‌شده

C را ویژگیِ مهارت بسازید؛ احتمال انتخاب طرح جدید در C=۰ برابر۰٫۲ و در C=۱ برابر۰٫۸ است. پیامد از Y=۲T+۳C+ε تولید شود، پس اثر علّیِ معلومِ سازنده ۲ است. اختلاف خام، تفاوت مهارت را هم جذب می‌کند. با propensity معلومِ همین سناریو، میانگین‌های وزنیِ خودنرمال‌شده را محاسبه می‌کنیم. در پژوهش واقعی e(X) معلوم نیست و باید مناسب برآورد و تعادل بررسی شود؛ در اینجا مسئلهٔ برآورد propensity را عمداً جدا گذاشته‌ایم.

````python
def run_lesson_48(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: causal
    rng=np.random.default_rng(202619)
    cc=rng.binomial(1,.5,3000); propensity=.2+.6*cc; treatment=rng.binomial(1,propensity); outcome=2*treatment+3*cc+rng.normal(size=len(cc))
    naive=outcome[treatment==1].mean()-outcome[treatment==0].mean(); w1=treatment/propensity; w0=(1-treatment)/(1-propensity); ipw=(w1@outcome)/w1.sum()-(w0@outcome)/w0.sum(); weight=w1+w0
    fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained'); axes[0].bar(['Known simulated ATE','Raw difference','IPW Hajek difference'],[2,naive,ipw],color=[GOLD,BLUE,TEAL]); axes[0].set(ylabel='Outcome difference (constructed units)',title='Adjustment under explicit identifying assumptions'); axes[1].hist(weight,bins=np.linspace(1,5.5,10),color=BLUE,edgecolor='white'); axes[1].set(xlabel='Inverse-propensity weight',ylabel='People',title='Overlap controls weight stability'); show()
    say(f'اثر سازنده=۲، اختلاف خام={naive:.3f} و برآورد IPW خودنرمال‌شده={ipw:.3f}. propensity در این مثال از سازوکار معلوم آمده است؛ روی دادهٔ واقعی نبود مخدوش‌کنندهٔ پنهان را از این تطابق نمی‌توان نتیجه گرفت.')
    # END PRESERVED ANALYSIS: causal
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['causal'] = run_lesson_48()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

سه ستون اثر واقعیِ سناریو، اختلاف خام و اختلاف وزنی را نشان می‌دهند. وزن‌دهی ممکن است نزدیک‌تر شود، نه اینکه در هر نمونه دقیقاً ۲ بدهد. هیستوگرام وزن‌ها دربارهٔ ناپایداری سرنخ می‌دهد. وزن بسیار بزرگ معمولاً علامت هم‌پوشانی کم و دقت پایین است؛ حذف یا برش وزن نیز کمیت هدف و سوگیری را تغییر می‌دهد.

#### دام رایج

نتیجهٔ وزن‌دهی را بدون فرض «همهٔ مخدوش‌کننده‌های لازم اندازه‌گیری شده‌اند» علّی اعلام نکنید. ستون‌های پس از درمان را بی‌منطق وارد propensity نکنید.

#### تمرین

اگر گروهی از کاربران هرگز طرح قدیم را نمی‌گیرند، چه فرضی مشکل دارد؟

#### پاسخ تشریحی

مثبت‌بودن/هم‌پوشانی؛ در آن ناحیه پیامد ضدواقعِ طرح قدیم از داده شناسایی نمی‌شود. جمعیت هدف باید محدود یا اطلاعات و فرض‌های اضافی روشن شوند.

## فصل 08: رگرسیون، تعمیم و منظم‌سازی

رگرسیون ساده و بازهٔ شیب، سپس چندمتغیره، پاسخ دودویی و شمارشی؛ در پایان تعادل پیچیدگی با Ridge و اشتراک اطلاعات بین گروه‌ها.

### درس 49. رگرسیون خطی ساده و باقیمانده‌ها

**هدف یادگیری:** از خط برازش به پیش‌بینی و باقیمانده برسید.

#### مفهوم و تشریح کامل

کمترین مربعات خطی را انتخاب می‌کند که مجموع مربعِ فاصله‌های عمودی Y از خط کمینه شود. شیب در واحد پاسخ به ازای یک واحد X است. باقیمانده اختلاف مشاهده و برازش است؛ مثبت یعنی مدل Y را کمتر از مقدار واقعی پیش‌بینی کرده است.

#### فرمول و قرارداد

```text
ŷ=0.104+0.444x; e=y−ŷ; R²=1−SSE/SST
```

#### مثال حل‌شدهٔ مرجع

برای طول ۲۲ سانتی‌متر، عرض پیش‌بینی‌شده حدود ۹٫۸۷۲ است. عرض ثبت‌شدهٔ همان فرد ۱۰ است، پس با ضرایب گرد‌شده e≈۰٫۱۲۸ سانتی‌متر. R² آموزشی≈۰٫۹۸۴ است؛ یعنی خط نسبت به مدل میانگین بخش بزرگی از تغییر Yِ همین داده را بازسازی می‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_49(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    x = df.hand_length_cm.to_numpy(dtype=float)

    # BEGIN PRESERVED ANALYSIS: lesson-15
    import statsmodels.api as sm
    simple_fit = sm.OLS(df.hand_width_cm, sm.add_constant(df.hand_length_cm)).fit()
    print(pd.DataFrame({'term':simple_fit.params.index, 'estimate':simple_fit.params.values}).to_string(index=False))
    print(f'In-sample R² = {simple_fit.rsquared:.3f}')
    fig, axes = plt.subplots(1, 2, figsize=(11,4))
    axes[0].scatter(x, df.hand_width_cm, color=BLUE)
    grid = np.linspace(x.min(),x.max(),100)
    axes[0].plot(grid, simple_fit.params.iloc[0]+simple_fit.params.iloc[1]*grid, color=GOLD)
    axes[0].set(xlabel='Hand length (cm)', ylabel='Hand width (cm)', title='Linear fit')
    axes[1].scatter(simple_fit.fittedvalues, simple_fit.resid, color=TEAL)
    axes[1].axhline(0, color='black', lw=1)
    axes[1].set(xlabel='Fitted width (cm)', ylabel='Residual (cm)', title='Residual check')
    plt.tight_layout(); plt.show()

    say(f'عرضِ برازش‌شده={simple_fit.params.iloc[0]:.3f}+{simple_fit.params.iloc[1]:.3f}×طول؛ R² آموزشی={simple_fit.rsquared:.4f}. شیب واحد cm/cm دارد؛ باقیمانده‌ها باید جداگانه بررسی شوند و این R² ارزیابیِ فرد جدید نیست.')
    # END PRESERVED ANALYSIS: lesson-15
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-15'] = run_lesson_49()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

در نمودار اول محور X طول و Y عرض است. در نمودار باقیمانده، محور افقی برازش و عمودی e است؛ به خمیدگی، قیف، نقاط اثرگذار و وابستگی زمانی نگاه کنید. نبودِ الگوی آشکار در ده نقطه به معنی اثبات همهٔ فرض‌ها نیست. R² بالا هم خطای کاربران جدید را گزارش نمی‌کند.

#### کاربرد و محدودیت

عرض دستِ پیش‌بینی‌شده، عرض بدنهٔ ماوس نیست. برای تبدیل آن به طراحی باید یک رابطهٔ ارگونومیک مجزا را با آزمون کاربری برقرار کنید. بیرون بازهٔ ۱۸ تا ۲۷ نیز برون‌یابیِ همین خط قابل اتکای خودکار نیست.

#### تمرین

اگر y=۹ و ŷ=۹٫۴ باشد، باقیمانده چیست و جهت خطا کدام است؟

#### پاسخ تشریحی

e=−۰٫۴ سانتی‌متر؛ مدل مقدار عرض را ۰٫۴ بیش‌برآورد کرده است.

### درس 50. بازهٔ اطمینان ۹۵٪ شیب

**هدف یادگیری:** اندازه و عدم‌قطعیت شیب را هم‌زمان گزارش کنید.

#### مفهوم و تشریح کامل

برآورد شیب با تغییر نمونه نوسان می‌کند. خطای استاندارد شیب از پراکندگی باقیمانده و گستردگی X به دست می‌آید. با همان خطای مدل، گسترده‌تر بودن X اطلاعات بیشتری دربارهٔ شیب فراهم می‌کند. بازهٔ t به فرض‌های مدل و استقلال وابسته است.

#### فرمول و قرارداد

```text
CI(b)=b±t(0.975,n−2)×SE(b); SE(b)=√[SSE/(n−2)] / √Σ(x−x̄)²
```

#### مثال حل‌شدهٔ مرجع

شیب ۰٫۴۴۴ و بازهٔ ۹۵٪ آن تقریباً [۰٫۳۹۷،۰٫۴۹۱] است. واحد: سانتی‌متر عرض به ازای یک سانتی‌متر طول. برای اختلاف ۴ سانتی‌متری X، اختلاف متوسط برازش‌شده ۱٫۷۷۶ سانتی‌متر است؛ این محاسبه دربارهٔ تغییر میانگین مدل است و دامنهٔ Y یک فرد را نمی‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_50(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    simple_fit = sm.OLS(df.hand_width_cm,sm.add_constant(df.hand_length_cm)).fit()

    # BEGIN PRESERVED ANALYSIS: lesson-17
    slope_name = 'hand_length_cm'
    slope_ci = simple_fit.conf_int(alpha=.05).loc[slope_name]
    print(f'Slope = {simple_fit.params[slope_name]:.3f} cm width / cm length')
    print(f'95% CI = [{slope_ci.iloc[0]:.3f}, {slope_ci.iloc[1]:.3f}]')

    fig,ax=plt.subplots(figsize=(8,2.8)); ax.errorbar([simple_fit.params[slope_name]],[0],xerr=[[simple_fit.params[slope_name]-slope_ci.iloc[0]],[slope_ci.iloc[1]-simple_fit.params[slope_name]]],fmt='o',capsize=8,color=BLUE)
    ax.axvline(0,color='0.4',ls='--'); ax.set(xlabel='Slope (cm width / cm length)',yticks=[],title='Slope and model-based 95% confidence interval'); plt.show()

    say(f'شیب {simple_fit.params[slope_name]:.3f} و بازهٔ ۹۵٪ [{slope_ci.iloc[0]:.3f}, {slope_ci.iloc[1]:.3f}] است؛ صفر بیرون این بازه است، تحت فرض‌های مدل. این بازه برای شیب است، نه پیش‌بینی عرض یک فرد.')
    # END PRESERVED ANALYSIS: lesson-17
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-17'] = run_lesson_50()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

بازه صفر را شامل نمی‌کند و با جهت مثبت رابطه سازگار است. عرض بازه دقت برآوردِ مشروط به مدل را نشان می‌دهد. نباید آن را با بازهٔ پیش‌بینی عرضِ یک فرد اشتباه گرفت؛ برای پیش‌بینی فردی، خطای فرد نیز وارد می‌شود.

#### کاربرد و محدودیت

در گزارش طراحی جملهٔ «شیب مثبت با این بازه» را کنار نمودار، n و بازهٔ مشاهده‌شدهٔ X بنویسید. از آن برای اثبات رابطهٔ علّی میان ویژگی‌های فیزیکی یا تبدیل مستقیم به ابعاد ماوس استفاده نکنید.

#### تمرین

دو شیب برابرند ولی بازهٔ یکی بسیار پهن‌تر است. چه تفاوتی در دانش ما وجود دارد؟

#### پاسخ تشریحی

برآورد نقطه‌ای یکسان است، اما دقت برآورد دوم کمتر است. صرف مقایسهٔ شیب‌های گرد‌شده این عدم‌قطعیت را پنهان می‌کند.

### درس 51. رگرسیون چندگانه

**هدف یادگیری:** ضرایب شرطی، هم‌خطی و پیچیدگی مدل را درک کنید.

#### مفهوم و تشریح کامل

در رگرسیون چندگانه هر ضریب با ثابت نگه‌داشتن دیگر پیش‌بین‌های مدل تفسیر می‌شود. این یک توصیف شرطی است؛ در دادهٔ مشاهده‌ای الزاماً اثر مداخله نیست. پیش‌بین‌های شدیداً همبسته می‌توانند ضرایب را ناپایدار کنند، حتی زمانی که پیش‌بینی مجموع مدل تغییر کمی دارد.

#### فرمول و قرارداد

```text
width≈1.4355+0.4149×length−0.0516×age+0.0923×experience
```

#### مثال حل‌شدهٔ مرجع

مدل سه پیش‌بین و یک عرض از مبدأ روی ده نفر دارد؛ درجهٔ آزادی باقیمانده ۶ است. R²≈۰٫۹۸۶ و تعدیل‌شده≈۰٫۹۷۸ است. با ورودی طول ۲۲، سن ۲۶ و سابقه ۷، با ضرایب گرد‌شده ŷ≈۹٫۸۶۸ سانتی‌متر به دست می‌آید. p سن≈۰٫۶۷۶ و سابقه≈۰٫۶۱۱ در حضور بقیهٔ متغیرهاست.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_51(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()

    # BEGIN PRESERVED ANALYSIS: lesson-20
    multi_X = df[['hand_length_cm','age_years','gaming_years']]
    multi_fit = sm.OLS(df.hand_width_cm, sm.add_constant(multi_X)).fit()
    display(pd.DataFrame({'coefficient':multi_fit.params, 'p_value':multi_fit.pvalues}).round(4))
    print(f'R²={multi_fit.rsquared:.3f}; adjusted R²={multi_fit.rsquared_adj:.3f}; residual df={multi_fit.df_resid:.0f}')
    print('Design matrix condition number:', round(np.linalg.cond(sm.add_constant(multi_X)), 1))

    coef_ci=multi_fit.conf_int().drop('const'); coef_values=multi_fit.params.drop('const')
    fig,ax=plt.subplots(figsize=(9,4)); ax.errorbar(coef_values,np.arange(3),xerr=[coef_values-coef_ci[0],coef_ci[1]-coef_values],fmt='o',capsize=6,color=BLUE)
    ax.axvline(0,color='0.4',ls='--'); ax.set(yticks=np.arange(3),yticklabels=coef_values.index,xlabel='Conditional coefficient, 95% model-based CI (different input units)',title='Multiple-regression coefficients'); plt.tight_layout(); plt.show()

    say(f'R² آموزشی={multi_fit.rsquared:.4f} و تعدیل‌شده={multi_fit.rsquared_adj:.4f} با df باقیمانده={multi_fit.df_resid:.0f}. ضریب سن به‌شرط ثابت‌بودن بقیه است؛ هم‌خطی و نمونهٔ کم تفسیر مستقل و تعمیم را محدود می‌کنند.')
    # END PRESERVED ANALYSIS: lesson-20
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-20'] = run_lesson_51()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

بالارفتن جزئی R² نسبت به مدل ساده انتظار می‌رود؛ افزودن پیش‌بین معمولاً خطای آموزش را کم می‌کند. مقایسهٔ ضرایب خامِ سن و طول به‌عنوان اهمیت درست نیست، چون واحدها فرق دارند. عدد شرطِ ماتریسِ بدون مقیاس‌بندی نیز به مقیاس حساس است؛ هم‌خطی را فقط از آن تشخیص ندهید.

#### کاربرد و محدودیت

برای مقایسهٔ پیش‌بینی، یک pipeline با پیش‌پردازش درون هر fold و معیار مشترک بسازید. ضریب کوچک و p بزرگ، بی‌اهمیتی قطعیِ متغیر را ثابت نمی‌کند، به‌ویژه با ده سطر.

#### تمرین

آیا با افزودن ده پیش‌بین دیگر به ده فرد، کیفیت تصمیم بهتر می‌شود؟

#### پاسخ تشریحی

تعداد پارامترها نسبت به اطلاعات بسیار زیاد و مدل نامعین یا بیش‌برازش می‌شود. ابتدا هدف، دادهٔ بیشتر و ارزیابی خارج از نمونه لازم است.

### درس 52. رگرسیون لجستیک: احتمال، Odds و نسبت شانس

**پرسش شروع:** یک ساعت آموزش بیشتر، احتمال موفقیت را «دو برابر» می‌کند یا شانس موفقیت را؟

#### مفهوم، دقیق اما ساده

رگرسیون خطی برای پیامد صفر و یک ممکن است پیش‌بینی زیر صفر یا بالای یک بدهد. مدل لجستیک ترکیب خطی ویژگی‌ها را به log-odds پیوند می‌دهد و سپس آن را با تابع sigmoid به احتمال تبدیل می‌کند. Odds برابر p/(1−p) است؛ odds ratio با risk ratio یکی نیست. افزایش یک واحد X، با ثابت‌بودن بقیهٔ پیش‌بین‌ها، odds را در exp(β) ضرب می‌کند؛ تغییر احتمال به احتمال اولیه وابسته است. ضرایب مشاهده‌ای اثر علّی محسوب نمی‌شوند. نمونهٔ ناکافی، جدایی کامل کلاس‌ها و هم‌خطی می‌توانند برازش و عدم‌قطعیت را مشکل‌دار کنند؛ منظم‌سازی و طرح داده مهم‌اند.

#### فرمول و قرارداد

```text
log(p/(1−p))=β0+β1x
p=1/(1+exp(−η)); OR=exp(β1)
```

#### مثال حل‌شده

اگر p ابتدا ۰٫۲ باشد، odds برابر۰٫۲۵ است. دو برابرشدن odds آن را به۰٫۵ می‌رساند و احتمال تازه۰٫۳۳۳ می‌شود، نه۰٫۴. اگر p ابتدا۰٫۸ باشد، odds=۴ و odds تازه۸ است؛ احتمال۰٫۸۸۹ می‌شود. در شبیه‌سازی، ۵۰۰ مقدار ساعت تمرین و موفقیت ساخته و مدل GLM دوجمله‌ای برازش می‌کنیم؛ ضریب برآوردشده الزاماً دقیقاً ضریب سازنده نیست، چون داده تصادفی‌اند.

````python
def run_lesson_52(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: logistic
    rng=np.random.default_rng(202615)
    hours=rng.uniform(0,6,500); true_prob=1/(1+np.exp(-(-2+.7*hours))); outcome=rng.binomial(1,true_prob)
    logit_fit=sm.GLM(outcome,sm.add_constant(hours),family=sm.families.Binomial()).fit()
    grid=np.linspace(0,6,150); fitted_prob=logit_fit.predict(sm.add_constant(grid))
    fig,ax=plt.subplots(); ax.scatter(hours,outcome+rng.uniform(-.025,.025,len(outcome)),alpha=.15,s=12,color=BLUE); ax.plot(grid,fitted_prob,color=GOLD,lw=2); ax.set(xlabel='Constructed practice hours',ylabel='Outcome / fitted probability',ylim=(-.08,1.08),title='Logistic GLM on constructed binary outcomes'); show()
    display(pd.DataFrame({'coefficient':logit_fit.params,'exp(coefficient)':np.exp(logit_fit.params)},index=['Intercept','Hours']))
    say(f'ضریب ساعت={logit_fit.params[1]:.3f} و OR={np.exp(logit_fit.params[1]):.3f} است؛ تغییر احتمال به مقدار آغازین وابسته است. برای OR=۲، احتمال ۰٫۲ به {(.2/.8*2)/(1+.2/.8*2):.3f} می‌رسد، نه۰٫۴.')
    # END PRESERVED ANALYSIS: logistic
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['logistic'] = run_lesson_52()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

منحنی آبی احتمال برازش‌شده بین صفر ویک است. نقاطِ موفقیت و شکست کمی برای خوانایی در راستای عمودی جابه‌جا شده‌اند؛ این jitter پیامد جدیدی نساخته است. جدول ضریب و exp(ضریب) را با واحد ورودی بخوانید. برای کیفیت پیش‌بینیِ فرد تازه باید جداگانه از آموزش ارزیابی شود؛ این شکل برازش درون‌نمونه است.

#### دام رایج

عبارت «OR=۲ یعنی احتمال دو برابر» به‌خصوص برای پیامد رایج اشتباه است. همچنین ضریب با کنترل متغیر نامناسب می‌تواند سوگیر شود.

#### تمرین

با p=۰٫۵ و OR=۳، احتمال جدید چقدر است؟

#### پاسخ تشریحی

odds اولیه۱ و جدید۳ است؛ p=۳/(۱+۳)=۰٫۷۵. افزایش احتمال۲۵ واحد درصد و نسبت احتمال۱٫۵ است، نه۳.

### درس 53. پواسون، بیش‌پراکندگی و مدل شمارشی با Exposure

**پرسش شروع:** دو شعبه هر کدام ۱۰ شکایت دارند؛ یکی صد مشتری داشته و دیگری ده‌هزار مشتری.

#### مفهوم، دقیق اما ساده

شمارش را بدون اندازهٔ فرصت وقوع مقایسه نکنید. در مدل پواسون با exposure، لگاریتم فرصت مانند مدت یا تعداد مشتری می‌تواند offset با ضریب ثابت یک باشد؛ مدل، نرخ را با پاسخ شمارشی مرتبط می‌کند. مدل پواسون ساده واریانس شرطی برابر میانگین شرطی دارد. اگر واریانسِ شرطی خیلی بیشتر باشد، بیش‌پراکندگی ممکن است از ناهمگنی، وابستگی، صفر اضافی یا مدل میانگین نامناسب بیاید. توزیع نگاتیو بینوم با پارامتر پراکندگی یکی از راه‌هاست، نه تنها راه. برابری واریانس و میانگین در کل فایل جای بررسی شرطی بعد از کنترل ویژگی‌ها را نمی‌گیرد.

#### فرمول و قرارداد

```text
log(E[count|X,exposure])=Xβ+log(exposure)
Poisson Var(Y|X)=μ; NB2 Var(Y|X)=μ+αμ².
```

#### مثال حل‌شده

۱۰ شکایت در۱۰۰ مشتری نرخ۱۰٪ و در۱۰٬۰۰۰ مشتری نرخ۰٫۱٪ است؛ شمارش برابر، نرخ صدبرابر متفاوت دارد. در مثال کد، ۶۰۰ شمارش با میانگینِ وابسته به ساعات کار و exposure تصادفی از NB2 با α=۰٫۸ می‌سازیم. سپس پواسون و NB را برازش می‌کنیم. برای NB، α در این نمایش همان پارامترِ معلومِ سازنده است؛ آن را برآوردشده از داده جا نمی‌زنیم. در کار واقعی باید پراکندگی را با روش معتبر تخمین یا حساسیت‌سنجی کرد.

````python
def run_lesson_53(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: counts
    rng=np.random.default_rng(202616)
    cx=rng.uniform(0,3,600); exposure=rng.uniform(.5,2,600); mu=exposure*np.exp(.3+.4*cx); alpha=.8
    counts=rng.negative_binomial(1/alpha,(1/alpha)/(1/alpha+mu)); design_x=sm.add_constant(cx)
    poisson=sm.GLM(counts,design_x,family=sm.families.Poisson(),offset=np.log(exposure)).fit()
    nb_demo=sm.GLM(counts,design_x,family=sm.families.NegativeBinomial(alpha=alpha),offset=np.log(exposure)).fit()
    dispersion=poisson.pearson_chi2/poisson.df_resid
    fig,ax=plt.subplots(); ax.scatter(poisson.fittedvalues,counts,color=BLUE,alpha=.3,s=15); lim=max(counts.max(),poisson.fittedvalues.max()); ax.plot([0,lim],[0,lim],color='0.4',ls='--'); ax.set(xlabel='Poisson fitted conditional count',ylabel='Observed constructed count',title='NB-generated counts: Poisson conditional variance is too small'); show()
    say(f'نسبت تشخیصی χ²/df پواسون={dispersion:.2f} است. ضریب x در پواسون={poisson.params[1]:.3f} و NB با α معلوم ۰٫۸ برابر {nb_demo.params[1]:.3f} است؛ این سناریو دادهٔ واقعی شکایت نیست و ارزیابی پیش‌بینی مستقل انجام نشده است.')
    # END PRESERVED ANALYSIS: counts
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['counts'] = run_lesson_53()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

در ابر نقاط، محور افقی میانگین شرطیِ پیش‌بینی‌شدهٔ پواسون و عمودی شمارش دیده‌شده است؛ پراکندگی زیاد اطراف خط y=x با توضیح بیش‌پراکندگی سازگار است. نسبت Pearson χ²/df فقط یک تشخیص تقریبی با مدل و حجم نمونهٔ مناسب است. مقدار بزرگ هشدار است، ولی علت نهایی مشکل را تعیین نمی‌کند.

#### دام رایج

افزودن offset اشتباه، مانند مدتِ حاصل از خود پیامد، می‌تواند سوگیری بسازد. زیادبودن صفرها هم به‌تنهایی اثبات یک مدل zero-inflated نیست.

#### تمرین

اگر exposure دو برابر و نرخ ثابت باشد، شمارش مورد انتظار چه می‌شود؟

#### پاسخ تشریحی

دو برابر می‌شود؛ offset لگاریتم exposure این تناسب را به مدل تحمیل می‌کند. اگر تناسب در واقعیت درست نباشد، باید مدل فرصت و ساختار سؤال بازبینی شود.

### درس 54. بایاس–واریانس، منظم‌سازی و Ridge

**پرسش شروع:** مدلی که از همهٔ نقاط آموزش می‌گذرد چرا روی دادهٔ جدید بد عمل می‌کند؟

#### مفهوم، دقیق اما ساده

مدل بسیار انعطاف‌پذیر می‌تواند نویزِ نمونه را هم یاد بگیرد. بایاس–واریانس راهی برای فکرکردن به خطای مورد انتظار مدل در تکرار آموزش است، نه اینکه برای هر دادهٔ واقعی به سادگی دو عدد قابل مشاهده داشته باشیم. منظم‌سازی با جریمه‌کردن پیچیدگی، بخشی از انعطاف را کم می‌کند؛ معمولاً بایاس را بالا و واریانس را پایین می‌آورد، ولی مقدار نهایی خطا به مسئله وابسته است. Ridge ضرایب را با جریمهٔ L2 جمع می‌کند و برای هم‌خطی مفید است؛ مقدار α و مقیاس ویژگی‌ها مهم‌اند. استانداردسازی و انتخاب α باید فقط در آموزش و foldهای داخلی انجام شوند. آزمون نهایی برای انتخاب درجه و α دوباره مصرف نمی‌شود.

#### فرمول و قرارداد

```text
Ridge minimizes ||y−Xβ||² + α||β||²
The intercept is generally not penalized; scaling changes what a unit penalty means.
```

#### مثال حل‌شده

۳۰ نقطه از sin(x) با نویز بسازید. درجهٔ۱۲ِ چندجمله‌ای را یک بار بدون جریمه و یک بار با Ridge مقایسه کنید. جست‌وجوی شبکه‌ایِ Pipeline کامل مقدار α را در CV داخلیِ همین۳۰ نقطه انتخاب می‌کند. هزار نقطهٔ آزمون مستقلِ همان فرایند ساخته می‌شوند و فقط برای گزارش خطای نهایی استفاده می‌شوند. نتیجهٔ یک بذر تضمین نمی‌کند Ridge همیشه برنده باشد؛ طراحی مثال برای دیدن سازوکار نوسان و کنترل آن است، نه مسابقهٔ عمومی الگوریتم‌ها.

````python
def run_lesson_54(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: ridge
    rng=np.random.default_rng(202622)
    from sklearn.preprocessing import PolynomialFeatures,StandardScaler
    from sklearn.pipeline import make_pipeline
    from sklearn.linear_model import LinearRegression,RidgeCV
    train_x=rng.uniform(-2,2,(30,1)); train_y=np.sin(train_x[:,0])+rng.normal(scale=.3,size=30); test_x=rng.uniform(-2,2,(1000,1)); test_y=np.sin(test_x[:,0])+rng.normal(scale=.3,size=1000)
    ols=make_pipeline(PolynomialFeatures(12,include_bias=False),StandardScaler(),LinearRegression()).fit(train_x,train_y)
    # Scaling must be fit inside each CV fold: use GridSearchCV around the entire pipeline.
    from sklearn.model_selection import GridSearchCV,KFold
    from sklearn.linear_model import Ridge
    search=GridSearchCV(make_pipeline(PolynomialFeatures(12,include_bias=False),StandardScaler(),Ridge()),{'ridge__alpha':np.logspace(-3,3,13)},cv=KFold(5,shuffle=True,random_state=42),scoring='neg_mean_squared_error').fit(train_x,train_y)
    gg=np.linspace(-2,2,300)[:,None]; fig,ax=plt.subplots(); ax.scatter(train_x[:,0],train_y,color=BLUE,label='30 training points'); ax.plot(gg[:,0],np.sin(gg[:,0]),color='0.3',ls='--',label='Known simulation mean'); ax.plot(gg[:,0],ols.predict(gg),color=GOLD,label='Degree 12, no penalty'); ax.plot(gg[:,0],search.predict(gg),color=TEAL,label='Degree 12, CV Ridge'); ax.set(xlabel='Constructed x',ylabel='Constructed response',ylim=(-3,3),title='Same features, different regularization (view clipped to −3..3)'); ax.legend(); show()
    say(f'α منتخبِ CV داخلی={search.best_params_["ridge__alpha"]:.4g}. MSE آزمون مستقل: بدون جریمه={mean_squared_error(test_y,ols.predict(test_x)):.4f}، Ridge={mean_squared_error(test_y,search.predict(test_x)):.4f}. این مقایسه یک اجرای ساختگی است، نه تضمین برتری همیشگی.')
    # END PRESERVED ANALYSIS: ridge
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['ridge'] = run_lesson_54()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

ابر نقاط، آموزشِ نویزی است؛ خط تیره حقیقتِ معلومِ شبیه‌سازی و دو منحنی برازش‌اند. خطای کمِ آموزش با نزدیک‌بودن به حقیقت یا خطای آزمون یکی نیست. حاشیهٔ دامنه معمولاً محل نوسان شدیدتر چندجمله‌ای است. عنوان «بهترین α» فقط به شبکهٔ مقادیر و CV داخلیِ همین طرح مربوط است.

#### دام رایج

استفاده از آزمون برای انتخاب α، حتی اگر فقط یک پارامتر باشد، ارزیابی نهایی را خوش‌بینانه می‌کند. Pipeline همهٔ نشت‌ها را خودکار حل نمی‌کند؛ جداسازی فرد و زمان هم لازم است.

#### تمرین

اگر همهٔ ویژگی‌ها را در عدد۱۰۰ ضرب کنیم، همان α الزاماً همان مدل Ridge را می‌دهد؟

#### پاسخ تشریحی

خیر؛ جریمه به مقیاس ضرایب وابسته است. استانداردسازی مناسب و انتخاب α در فرایند آموزش لازم است.

### درس 55. تجمیع جزئی، Shrinkage و شهود مدل چندسطحی

**پرسش شروع:** یک شعبه از دو سفارش دو موفقیت دارد و شعبه‌ای از هزار سفارش۹۰۰ موفقیت؛ کدام مطمئن‌تر است؟

#### مفهوم، دقیق اما ساده

برآورد جداگانهٔ هر گروه در گروه‌های کم‌داده ناپایدار است؛ ادغام کامل هم اختلاف واقعی گروه‌ها را حذف می‌کند. تجمیع جزئی اطلاعات مشترک را با اطلاعات اختصاصی ترکیب می‌کند. در مدل چندسطحی واقعی، پارامترهای گروه‌ها از یک توزیع جمعیتی می‌آیند و پارامترهای آن نیز برآورد یا دارای پیشین‌اند. برای دیدن سازوکار، می‌توان از پیشین Beta مشترکِ ثابت استفاده کرد: گروه کوچک بیشتر و گروه بزرگ کمتر به میانگین پیشین جمع می‌شود. این نمایش، برازش کامل سلسله‌مراتبی نیست. اگر گروه‌ها واقعاً قابل‌تبادل نباشند یا تفاوت ساختاری مهم داشته باشند، جمع‌کردن مشترک می‌تواند مسئله را پنهان کند.

#### فرمول و قرارداد

```text
Posterior mean=(s+a)/(n+a+b)
= [n/(n+a+b)]×(s/n) + [(a+b)/(n+a+b)]×[a/(a+b)].
```

#### مثال حل‌شده

شعبه‌ها با (موفقیت،n) برابر (۲،۲)،(۸،۱۰)،(۷۰،۱۰۰)،(۹۰۰،۱۰۰۰) بسازید. پیشینِ آموزشیِ ثابت Beta(7,3) میانگین۰٫۷ و قدرت معادل۱۰ تلاش دارد. برآورد۲/۲ از۱ به۹/۱۲=۰٫۷۵ می‌رسد، ولی۹۰۰/۱۰۰۰ فقط اندکی به۰٫۷ نزدیک می‌شود. این «تنبیه شعبهٔ کوچک» نیست؛ عدم‌قطعیت بیشتر دادهٔ کمتر را منعکس می‌کند. اما انتخاب پیشین۷،۳ باید توجیه یا حساسیت‌سنجی شود؛ از خود این مثال، نرخ واقعی یک شرکت به دست نیامده است.

````python
def run_lesson_55(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: pooling
    rng=np.random.default_rng(202624)
    s=np.array([2,8,70,900]); n=np.array([2,10,100,1000]); a,b=7,3; raw=s/n; partial=(s+a)/(n+a+b); intervals=np.array([stats.beta.ppf([.025,.975],a+si,b+ni-si) for si,ni in zip(s,n)])
    display(pd.DataFrame({'n':n,'raw rate':raw,'fixed-prior posterior mean':partial,'95% lower':intervals[:,0],'95% upper':intervals[:,1]}))
    fig,ax=plt.subplots()
    for i in range(4): ax.plot([raw[i],partial[i]],[i,i],color='0.6')
    ax.scatter(raw,np.arange(4),color=BLUE,label='Raw'); ax.scatter(partial,np.arange(4),color=GOLD,marker='s',label='Beta(7,3) posterior mean'); ax.axvline(.7,color='0.3',ls='--',label='Fixed prior mean'); ax.set(yticks=np.arange(4),yticklabels=[f'Group n={v}' for v in n],xlabel='Success probability estimate',xlim=(0,1.05),title='Partial-pooling intuition, not a fitted hierarchical model'); ax.legend(); show()
    say(f'گروه n=۲ از {raw[0]:.2f} به {partial[0]:.2f} جمع شد؛ گروه n=۱۰۰۰ از {raw[-1]:.3f} به {partial[-1]:.3f}. بازهٔ اعتباری گروه کوچک [{intervals[0,0]:.3f},{intervals[0,1]:.3f}] است؛ عدم‌قطعیت رتبه‌بندی را نادیده نگیرید.')
    # END PRESERVED ANALYSIS: pooling
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['pooling'] = run_lesson_55()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

نقاط خام و جمع‌شده را با n بخوانید. خطوط بین دو نقطه مقدار جمع‌شدن را نشان می‌دهند؛ خط‌چین۰٫۷ میانگین پیشین است. این خطوط CI نیستند. بازهٔ اعتباری گروه کوچک معمولاً پهن‌تر است و رتبه‌بندی قطعی شعبه‌ها از یک برآورد نقطه‌ای توصیه نمی‌شود. برای مدل کامل، ناهمگنی و ارتباط گروه‌ها باید بررسی شود.

#### دام رایج

نام «چندسطحی» را روی هر میانگین وزنی نگذارید؛ اینجا پیشین مشترک ثابت است، نه مدل سلسله‌مراتبی با فراپارامترهای آموخته‌شده.

#### تمرین

اگر a+b را بیشتر کنیم، برای یک شعبه با n ثابت چه رخ می‌دهد؟

#### پاسخ تشریحی

قدرت پیشین بیشتر و جمع‌شدن به میانگین a/(a+b) قوی‌تر می‌شود، اگر آن میانگین ثابت نگه داشته شود. این باید با دانش و تحلیل حساسیت همراه باشد.

## فصل 09: کاهش بُعد و کشف ساختار

استانداردسازی، مؤلفه‌ها و بارگذاری‌ها را پیش از خوشه‌بندی و نمایش غیرخطی می‌خوانیم؛ ساختار تصویری به‌تنهایی حقیقت جمعیت نیست.

### درس 56. PCA: تحلیل مؤلفه‌های اصلی

**هدف یادگیری:** فشرده‌سازی خطی را از طبقه‌بندی و کشف علت جدا کنید.

#### مفهوم و تشریح کامل

PCA جهت‌هایی متعامد می‌سازد که واریانس دادهٔ مرکز‌شده را به‌ترتیب بیشینه کنند. چون سانتی‌متر و سال مقیاس‌های متفاوت دارند، در این مثال ابتدا هر ستون استاندارد می‌شود. مؤلفه‌ها از همهٔ ستون‌ها ترکیب می‌شوند؛ «مستقل بودن آماری» از متعامد یا ناهمبسته بودن نتیجه نمی‌شود.

#### فرمول و قرارداد

```text
Z=(X−mean)/SD; scores=ZV; explained ratio=λj/Σλ
```

#### مثال حل‌شدهٔ مرجع

سهم PC1≈۷۵٫۷٪ و PC2≈۲۴٫۰٪ است؛ مجموع حدود۹۹٫۷٪ واریانسِ همین ویژگی‌های استانداردشده را نگه می‌دارند. ضرایب PC1 تقریباً[۰٫۴۹۸،۰٫۵۰۷،۰٫۴۷۳،۰٫۵۲۲] و PC2≈[−۰٫۵۰۷،−۰٫۴۷۸،۰٫۵۷۸،۰٫۴۲۴] است. مؤلفهٔ دوم اندازه‌های دست را در برابر سن/سابقه قرار می‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_56(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    numeric_cols = ['hand_length_cm','hand_width_cm','age_years','gaming_years']

    # BEGIN PRESERVED ANALYSIS: lesson-24
    from sklearn.decomposition import PCA
    X_all = df[numeric_cols].to_numpy(dtype=float)
    X_scaled = StandardScaler().fit_transform(X_all)
    pca = PCA(n_components=2).fit(X_scaled)
    scores_pca = pca.transform(X_scaled)
    print('Explained variance ratio:', np.round(pca.explained_variance_ratio_,3))
    display(pd.DataFrame(pca.components_.T, index=numeric_cols, columns=['PC1','PC2']).round(3))
    fig, ax = plt.subplots()
    ax.scatter(scores_pca[:,0], scores_pca[:,1], color=BLUE)
    for i, (a,b) in enumerate(scores_pca): ax.annotate(str(i+1),(a,b),xytext=(4,4),textcoords='offset points')
    ax.set(xlabel='PC1 (standardized units)', ylabel='PC2 (standardized units)', title='PCA projection of 10 users')
    plt.show()

    say(f'PC1={100*pca.explained_variance_ratio_[0]:.2f}٪ و PC2={100*pca.explained_variance_ratio_[1]:.2f}٪، مجموع {100*pca.explained_variance_ratio_.sum():.2f}٪ واریانس استانداردشده را حفظ می‌کنند؛ محور‌ها واحد سانتی‌متر ندارند.')
    # END PRESERVED ANALYSIS: lesson-24
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-24'] = run_lesson_56()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

در شکل هر نقطه یک فرد با مختصات PC1 وPC2 است. این محورها سانتی‌متر نیستند و درصد واریانس «دقت پیش‌بینی» نیست. علامت هر مؤلفه قراردادی است؛ وارونگی هم‌زمان محور و ضرایب، همان هندسه را می‌دهد.

#### کاربرد و محدودیت

برای استفادهٔ پیش‌بینی، استانداردسازی و PCA را داخل fold آموزش برازش کنید. روی همهٔ افراد fit کردن و سپس ارزیابی روی افراد کنارگذاشته‌شده، اطلاعات آزمون را به تبدیل وارد می‌کند.

#### تمرین

آیا دو مؤلفه با ۹۹٫۷٪ واریانس، ۹۹٫۷٪ دقتِ تعیین سایز می‌دهند؟

#### پاسخ تشریحی

خیر. واریانسِ ویژگی‌ها با صحتِ یک هدف برچسب‌دار متفاوت است. برای دقت باید هدف واقعی و ارزیابی جدا تعریف شود.

### درس 57. Scree، بارگذاری، Biplot و بازسازی PCA

**هدف یادگیری:** سه خروجی متفاوت PCA را به یکدیگر وصل کنید.

#### مفهوم و تشریح کامل

Scree سهم یا مقدار ویژهٔ مؤلفه‌ها را به‌ترتیب نشان می‌دهد؛ افت شدید می‌تواند برای انتخاب بُعد سرنخ باشد. بارگذاری‌ها نشان می‌دهند هر ویژگی چگونه در مؤلفه شرکت می‌کند. score مختصاتِ افراد روی مؤلفه‌هاست. در Biplot افراد و بردار ویژگی‌ها با یک قرارداد مقیاس‌بندی کنار هم قرار می‌گیرند؛ بدون اعلام قرارداد، طول پیکان‌ها قابل مقایسهٔ دقیق نیست.

#### فرمول و قرارداد

```text
scores=ZV; Ẑ=scores Vᵀ; X̂=scaler.inverse_transform(Ẑ)
```

#### مثال حل‌شدهٔ مرجع

دو مؤلفهٔ اول حدود۷۵٫۷٪ و۲۴٫۰٪ را نگه می‌دارند. برای فردی با بردار استاندارد z، score اول با ضرب هر چهار مقدار در[۰٫۴۹۸،۰٫۵۰۷،۰٫۴۷۳،۰٫۵۲۲] و جمع آن‌ها محاسبه می‌شود. PC2 برای طول/عرض علامت منفی و برای سن/سابقه مثبت دارد. در این تعریف components_ ضرایبِ بردارهای مؤلفه‌اند؛ برخی کتاب‌ها loading را با ضرب در√λ تعریف می‌کنند و عددهایشان متفاوت است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_57(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    numeric_cols = ['hand_length_cm','hand_width_cm','age_years','gaming_years']

    # BEGIN PRESERVED ANALYSIS: pca-reading
    scaler_detail=StandardScaler().fit(df[numeric_cols]); Z_detail=scaler_detail.transform(df[numeric_cols])
    full_pca=PCA().fit(Z_detail); detail_pca=PCA(n_components=2).fit(Z_detail); detail_scores=detail_pca.transform(Z_detail)
    detail_reconstruction=detail_pca.inverse_transform(detail_scores)
    display(pd.DataFrame(full_pca.components_.T,index=numeric_cols,columns=['PC1','PC2','PC3','PC4']).round(4))
    fig,axes=plt.subplots(1,2,figsize=(12,5),layout='constrained')
    axes[0].bar(np.arange(1,5),100*full_pca.explained_variance_ratio_,color=BLUE); axes[0].plot(np.arange(1,5),100*full_pca.explained_variance_ratio_.cumsum(),color=GOLD,marker='o'); axes[0].set(xlabel='Component',ylabel='Explained variance (%)',xticks=[1,2,3,4],ylim=(0,105),title='Scree bars and cumulative proportion')
    axes[1].scatter(detail_scores[:,0],detail_scores[:,1],color=BLUE)
    for i,(a,b) in enumerate(detail_scores): axes[1].annotate(str(i+1),(a,b),xytext=(4,4),textcoords='offset points')
    short_names=['Length','Width','Age','Experience']
    label_offsets=[(-18,-24),(8,-5),(-25,18),(8,-12)]
    for label,(a,b),offset in zip(short_names,detail_pca.components_.T,label_offsets):
        axes[1].annotate('',xy=(2*a,2*b),xytext=(0,0),arrowprops=dict(arrowstyle='->',color=GOLD,lw=1.5)); axes[1].annotate(label,(2*a,2*b),xytext=offset,textcoords='offset points')
    axes[1].set(xlabel='PC1 score',ylabel='PC2 score',title='Scores + component vectors ×2 (declared scaling)',xlim=(-4,5),ylim=(-3,3)); plt.show()
    say(f'دو مؤلفه {detail_pca.explained_variance_ratio_.sum():.3%} واریانس را حفظ می‌کنند؛ MSE بازسازی آموزشی در فضای استاندارد {mean_squared_error(Z_detail,detail_reconstruction):.5f} است. پیکان‌ها components_×۲ هستند، نه همبستگی‌های دقیقِ بارگذاری. علامت محور دلخواه و خطای فوق آموزشی است.')
    # END PRESERVED ANALYSIS: pca-reading
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['pca-reading'] = run_lesson_57()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

در Scree جمع نسبت‌ها باید۱ شود. در نمودار score، دوربودن نقطه ازمرکز می‌تواند ترکیبِ اندازه‌های استانداردِ بزرگ باشد. زاویهٔ پیکان‌ها فقط با مقیاس‌بندی مناسب و بازنمایی خوبِ ویژگی‌ها سرنخ همبستگی است؛ در صفحهٔ دوبعدیِ ناقص آن را دقیق نخوانید.

#### کاربرد و محدودیت

اگر هدف فشرده‌سازی است، MSE بازسازیِ دادهٔ کنارگذاشته‌شده را هم بسنجید. اگر هدف پیش‌بینی است، تعداد مؤلفه را برای همان معیار پیش‌بینی انتخاب کنید؛ بیشترین واریانس الزاماً بیشترین اطلاعاتِ هدف نیست.

#### تمرین

با وارونه‌کردن علامت score اول و بردار اول، بازسازی تغییر می‌کند؟

#### پاسخ تشریحی

خیر. حاصل‌ضرب دو علامت منفی مثبت است و همان مؤلفهٔ بازسازی به دست می‌آید. تفسیر فیزیکی را به جهت ظاهری محور گره نزنید.

### درس 58. خوشه‌بندی K-means

**هدف یادگیری:** خوشه‌بندی را به‌عنوان سناریوی اکتشافی ارزیابی کنید.

#### مفهوم و تشریح کامل

K-means فاصلهٔ مربع‌شدهٔ نقاط تا مرکز خوشه‌شان را کمینه می‌کند. مقیاس ویژگی‌ها تعیین‌کننده است؛ اینجا طول و عرض استاندارد شده‌اند. الگوریتم همیشه برای k درخواستی خوشه می‌سازد، پس دیدن سه رنگ به معنی کشفِ قطعی سه گروه طبیعی نیست.

#### فرمول و قرارداد

```text
J=Σk Σi∈Ck ||zi−μk||²; silhouette compares within/between cluster distance
```

#### مثال حل‌شدهٔ مرجع

برای k=۲، SSE≈۶٫۷۶۹ و silhouette≈۰٫۵۵۹ است؛ k=۳:۲٫۷۴۸ و۰٫۴۶۵؛ k=۴:۱٫۳۹۲ و۰٫۴۱۳. SSE با افزایش k پایین می‌آید، اما در این سه سناریو silhouette برای۲ بهتر است. بنابراین ادعای خودکار «بهترین تعداد سایز=۳» با این جدول پشتیبانی نمی‌شود.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_58(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    age_group = np.select([df.user_id.isin([1,3,7]),df.user_id.isin([2,5,8,10])],['young','middle'],default='older')
    df['age_group'] = age_group

    # BEGIN PRESERVED ANALYSIS: lesson-30
    from sklearn.cluster import KMeans
    from sklearn.metrics import silhouette_score
    X_size = StandardScaler().fit_transform(df[['hand_length_cm','hand_width_cm']])
    cluster_checks = []
    for k in [2,3,4]:
        model = KMeans(n_clusters=k, random_state=SEED, n_init=10).fit(X_size)
        cluster_checks.append([k,model.inertia_,silhouette_score(X_size,model.labels_)])
    display(pd.DataFrame(cluster_checks,columns=['k','within-cluster SSE','silhouette']).round(3))
    k3 = KMeans(n_clusters=3,random_state=SEED,n_init=10).fit(X_size)
    fig, ax = plt.subplots()
    ax.scatter(df.hand_length_cm, df.hand_width_cm, c=np.array([BLUE,GOLD,TEAL])[k3.labels_], s=80)
    for i,row in df.iterrows(): ax.annotate(str(row.user_id),(row.hand_length_cm,row.hand_width_cm),xytext=(4,4),textcoords='offset points')
    ax.set(xlabel='Hand length (cm)',ylabel='Hand width (cm)',title='Illustrative k=3 size clusters')
    plt.show()

    say(f'برای k=2,3,4 امتیاز silhouette برابر {[round(row[2],3) for row in cluster_checks]} است. در این مقایسه k={max(cluster_checks,key=lambda row:row[2])[0]} بیشترین امتیاز را دارد؛ شکل k=3 فقط یک سناریوست، نه اثبات وجود سه سایز طبیعی.')
    # END PRESERVED ANALYSIS: lesson-30
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-30'] = run_lesson_58()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

در شکل، نقاط کاربر و رنگ‌ها انتساب خوشه‌اند. مرکز خوشه میانگین ویژگی‌هاست و الزاماً شکل یا ابعاد محصول قابل تولید نیست. فرد انتهایی می‌تواند یک خوشهٔ کوچک بسازد؛ پایداری با نمونه‌ها و شروع‌های مختلف را بسنجید.

#### کاربرد و محدودیت

k=۳ می‌تواند فرضیهٔ طراحی S/M/L باشد. انتخاب نهایی به هزینهٔ قالب، پوشش کاربران، موجودی و نتایج آزمون راحتی وابسته است. خوشه‌های ناپایدار را به بخش بازار تبدیل نکنید.

#### تمرین

چرا کم‌ترین SSE به‌تنهایی k مناسب را مشخص نمی‌کند؟

#### پاسخ تشریحی

اگر هر فرد خوشهٔ خودش باشد، SSE صفر می‌شود. باید پیچیدگی، جدایی، پایداری و هدف کاربردی را هم در نظر گرفت.

### درس 59. PPCA: تحلیل مؤلفه‌های اصلی احتمالاتی

**هدف یادگیری:** تفسیر احتمالاتیِ کاهش بُعد را یاد بگیرید.

#### مفهوم و تشریح کامل

PPCA فرض می‌کند داده از یک متغیر نهفتهٔ کم‌بُعد با توزیع گاوسی و نویز همسانگرد ساخته می‌شود. در کنار تصویر دوبعدی، یک مدل احتمالاتی برای کوواریانس و درست‌نمایی فراهم می‌کند. برآوردِ نویز کوچک فقط در چارچوب همین فرض‌ها معنا دارد.

#### فرمول و قرارداد

```text
x=Wz+μ+ε; z~N(0,I); ε~N(0,σ²I)
```

#### مثال حل‌شدهٔ مرجع

برای دو مؤلفه روی چهار ستونِ استانداردشده، noise_variance_≈۰٫۰۰۷۵ و میانگین log-likelihood آموزش≈−۱٫۲۱۷۴ ثبت شده است. این اعداد در واحدِ فضای استانداردند، نه cm² مستقیم. سهم واریانسِ دو مؤلفه همان حدود۷۵٫۷٪ و۲۴٫۰٪ است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_59(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    numeric_cols = ['hand_length_cm','hand_width_cm','age_years','gaming_years']
    X_scaled = StandardScaler().fit_transform(df[numeric_cols].to_numpy(dtype=float))

    # BEGIN PRESERVED ANALYSIS: lesson-31
    ppca = PCA(n_components=2, svd_solver='full').fit(X_scaled)
    print('Noise variance estimate:', round(ppca.noise_variance_,4))
    print('Training mean log-likelihood per sample:', round(ppca.score(X_scaled),4))
    print('Explained variance:', np.round(ppca.explained_variance_ratio_,3))

    display(pd.DataFrame({'user_id':df.user_id,'training_log_density':ppca.score_samples(X_scaled)}))
    fig,ax=plt.subplots(); ax.bar(df.user_id,ppca.score_samples(X_scaled),color=TEAL); ax.set(xlabel='User ID',ylabel='Training log density',title='PPCA density diagnostic (not held-out performance)'); plt.show()

    say(f'واریانس نویز PPCA={ppca.noise_variance_:.5f} و log-likelihood میانگین آموزش={ppca.score(X_scaled):.4f}. این کیفیت روی همان دادهٔ برازش‌شده است، نه احتمال موفقیت محصول یا نتیجهٔ اعتبارسنجی مستقل.')
    # END PRESERVED ANALYSIS: lesson-31
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-31'] = run_lesson_59()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

درست‌نمایی عددی برای مقایسهٔ مدل‌ها با تعریف یکسانِ داده و مقیاس به کار می‌رود؛ بزرگ‌بودن مقدارِ آموزش به‌تنهایی مدل بهترِ کاربران آینده را اثبات نمی‌کند. تغییر استانداردسازی یا تعداد ویژگی‌ها مقیاس آن را عوض می‌کند.

#### کاربرد و محدودیت

برای انتخاب تعداد مؤلفه، امتیاز دادهٔ اعتبارسنجی و هدف بازسازی/پیش‌بینی را بررسی کنید. PPCA خوشه‌بند نیست و محورها برچسب مستقیمِ اندازهٔ محصول ندارند.

#### تمرین

آیا σ² کوچک، نبود خطای اندازه‌گیری واقعی را ثابت می‌کند؟

#### پاسخ تشریحی

خیر. این پارامترِ نویزِ مدل گاوسی در فضای انتخابی است و با خطای دستگاه یا همهٔ انواع خطای داده برابر نیست.

### درس 60. t-SNE: نمایش غیرخطی

**هدف یادگیری:** نقشهٔ t-SNE را بدون داستان‌سازی بخوانید.

#### مفهوم و تشریح کامل

t-SNE شباهت همسایگی‌های فضای اصلی را با شباهت در نقشهٔ کم‌بُعد تقریب می‌زند. محورهای نقشه واحد فیزیکی ندارند. فاصلهٔ خوشه‌های دور، مساحت جزیره‌ها و جهت چرخش را نباید مثل فاصله‌های اصلی تفسیر کرد.

#### فرمول و قرارداد

```text
هدف: کمینه‌سازی KL(P||Q) برای شباهت‌های جفتی
```

#### مثال حل‌شدهٔ مرجع

در این اجرا چهار ویژگی استاندارد، n=۱۰، perplexity=۳، شروع PCA و بذر۴۲ استفاده شده است. perplexity باید با حجم داده سازگار باشد. اگر آن را به۲ یا۴ تغییر دهید، ممکن است چیدمان عوض شود؛ این آزمایشِ حساسیت است و دادهٔ جدید تولید نمی‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_60(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    numeric_cols = ['hand_length_cm','hand_width_cm','age_years','gaming_years']
    X_scaled = StandardScaler().fit_transform(df[numeric_cols].to_numpy(dtype=float))

    # BEGIN PRESERVED ANALYSIS: lesson-32
    from sklearn.manifold import TSNE
    tsne_xy = TSNE(n_components=2, perplexity=3, init='pca', learning_rate='auto',
                   random_state=SEED).fit_transform(X_scaled)
    fig, ax = plt.subplots()
    ax.scatter(tsne_xy[:,0], tsne_xy[:,1], color=TEAL)
    for i,(a,b) in enumerate(tsne_xy): ax.annotate(str(i+1),(a,b),xytext=(4,4),textcoords='offset points')
    ax.set(xlabel='t-SNE 1 (arbitrary)',ylabel='t-SNE 2 (arbitrary)',title='Exploratory t-SNE map, perplexity=3')
    plt.show()

    say('نقشهٔ t-SNE با perplexity=3 و بذر ثابت ساخته شد. همسایگی‌های محلی سرنخ اکتشافی‌اند؛ جهت، فاصلهٔ خوشه‌های دور و اندازهٔ فیزیکی محور قابل تفسیر نیست. گروه‌بندی تجاری از این شکل به‌تنهایی به دست نمی‌آید.')
    # END PRESERVED ANALYSIS: lesson-32
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-32'] = run_lesson_60()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

هر نقطه همان یک کاربر است. نزدیکی چند کاربر می‌تواند همسایگی محلی را منعکس کند، ولی فاصلهٔ زیاد دو جزیره دلیل دو گروه تجاری نیست. در نمودارهای بدون برچسب مستقل، نام «دست کوچک» و«بزرگ» را از روی جایگاه بصری تحمیل نکنید.

#### کاربرد و محدودیت

برای اکتشاف دادهٔ بزرگ‌تر، نقشه را کنار ویژگی‌های اصلی، برچسب‌های مستقل و اجرای چند تنظیم بخوانید. برای این ده نفر، جدول و پراکندگی طول–عرض اغلب از نقشهٔ پیچیده قابل تفسیرتر است.

#### تمرین

آیا کاربری که دو برابر دورتر روی t-SNE است، دو برابر تفاوت فیزیکی دارد؟

#### پاسخ تشریحی

خیر. محور و فاصلهٔ جهانیِ نقشه چنین مقیاس مستقیمی ندارند؛ هدف روش تقریبِ ساختار محلی است.

## فصل 10: پیش‌بینی قابل اعتماد و تصمیم

خط پایه و جلوگیری از نشت پیش از مدل‌ها می‌آیند؛ بعد از طبقه‌بندها، اطلاعات، کالیبراسیون، هزینه و بازهٔ Conformal را می‌خوانیم.

### درس 61. ارزیابی مدل، OOB، خط پایه و اهمیت جایگشتی

**هدف یادگیری:** کیفیت مدل‌ها را با سؤال و معیار مشترک مقایسه کنید.

#### مفهوم و تشریح کامل

آموزش برای یادگیری پارامترها، اعتبارسنجی برای انتخاب تنظیمات و آزمون برای برآورد نهاییِ تعمیم است. اگر یک فرد چند سطر دارد، سطرهای او باید در یک گروهِ جداسازی بمانند. برای زمان، گذشته باید قبل از آینده باشد. درpipeline، scaler وPCA و هر انتخاب ویژگی فقط روی بخش آموزش fit می‌شوند.

#### فرمول و قرارداد

```text
MAE=Σ|e|/n; RMSE=√(Σe²/n); R²=1−SSE/SST
```

#### مثال حل‌شدهٔ مرجع

برای خطاهای[−۱،۰،۲]، MAE=۱، MSE=۵/۳ وRMSE≈۱٫۲۹۱ است. RMSE به خطای بزرگ حساس‌تر است. در مدل‌های طولِ این دوره، جنگلMAE≈۱٫۲۰۱ وBoosting≈۱٫۲۳۵ دارند؛ خط پایهٔ میانگین≈۲٫۲۲۲. MLP باMAE≈۰٫۲۳۴ هدفِ دیگری یعنی عرض دارد و در همین رتبه‌بندی قرار نمی‌گیرد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_61(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()

    # BEGIN PRESERVED ANALYSIS: evaluation
    from sklearn.dummy import DummyRegressor
    from sklearn.ensemble import RandomForestRegressor
    from sklearn.inspection import permutation_importance
    eval_X=df[['age_years','gaming_years','hand_width_cm']].to_numpy(); eval_y=df.hand_length_cm.to_numpy()
    eval_cv=LeaveOneOut(); eval_models={'Mean baseline':DummyRegressor(),'Random forest':RandomForestRegressor(n_estimators=100,min_samples_leaf=2,random_state=SEED)}
    eval_rows=[]
    for name,estimator in eval_models.items():
        pred=cross_val_predict(estimator,eval_X,eval_y,cv=eval_cv)
        eval_rows.append([name,np.mean(abs(eval_y-pred)),np.sqrt(mean_squared_error(eval_y,pred))])
    eval_table=pd.DataFrame(eval_rows,columns=['model','LOOCV MAE (cm length)','LOOCV RMSE (cm length)']); display(eval_table.round(4))
    oob_demo=RandomForestRegressor(n_estimators=500,min_samples_leaf=2,oob_score=True,random_state=SEED).fit(eval_X,eval_y)
    print('Illustrative OOB R²:',oob_demo.oob_score_)
    # A fixed 7/3 random split, not a reliable benchmark or an extra independent population.
    from sklearn.model_selection import train_test_split
    perm_train,perm_test=train_test_split(np.arange(len(eval_y)),test_size=.3,random_state=SEED)
    perm_model=RandomForestRegressor(n_estimators=100,min_samples_leaf=2,random_state=SEED).fit(eval_X[perm_train],eval_y[perm_train])
    perm=permutation_importance(perm_model,eval_X[perm_test],eval_y[perm_test],scoring='neg_mean_absolute_error',n_repeats=50,random_state=SEED)
    print('Permutation test user IDs:',df.user_id.iloc[perm_test].tolist())
    fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained')
    axes[0].bar(eval_table.model,eval_table.iloc[:,1],color=BLUE); axes[0].set(ylabel='LOOCV MAE (cm length)',title='Same target, same held-out scheme')
    axes[1].barh(['Age','Experience','Width'],perm.importances_mean,xerr=perm.importances_std,color=GOLD,capsize=4); axes[1].axvline(0,color='0.4'); axes[1].set(xlabel='Increase in test MAE (cm length)',title='Permutation importance: only 3 test users'); plt.show()
    say(f'خط پایهٔ LOOCV برابر {eval_rows[0][1]:.3f} و RF برابر {eval_rows[1][1]:.3f} سانتی‌متر طول خطا دارد. OOB R²={oob_demo.oob_score_:.3f} یک برآورد داخلی جنگل است. اهمیت جایگشتی روی فقط سه فرد کنارگذاشته‌شده بسیار ناپایدار است؛ میلهٔ خطا SD تکرار جایگشت است، نه CI جمعیتی. انتخاب مدل با این ده فرد به اعتبارسنجی مستقل نیاز دارد.')
    # END PRESERVED ANALYSIS: evaluation
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['evaluation'] = run_lesson_61()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

دقت آموزش را از دقت کنارگذاشته‌شده جدا بخوانید. در bootstrapِ جنگل، احتمال کنارماندن یک مشاهده از یک بازنمونهٔnتایی (۱−۱/n)ⁿ و برایn بزرگ حدود۳۶٫۸٪ است. OOB از همین درخت‌ها برای آن مشاهده استفاده می‌کند. در اهمیت جایگشتی، یک ستونِ آزمون را جابه‌جا و افزایش خطا را می‌سنجیم؛ همبستگی ویژگی‌ها روی این تفسیر هم اثر دارد.

#### کاربرد و محدودیت

با جست‌وجوی مکرر تنظیمات روی یک مجموعهٔ آزمون، آن مجموعه دیگر آزمونِ دست‌نخورده نیست. با دادهٔ کافی از ارزیابی تودرتو یا آزمون مستقل استفاده کنید. R² آزمون حتی می‌تواند منفی شود؛ یعنی از میانگینِ مرجعِ آن ارزیابی ضعیف‌تر است.

#### تمرین

اگر هر فرد در پنج روز اندازه‌گیری شده، جداسازی تصادفیِ سطرها چه مشکلی دارد؟

#### پاسخ تشریحی

اطلاعات همان فرد می‌تواند هم درآموزش و هم درآزمون باشد و تعمیم به فرد جدید خوش‌بینانه شود. GroupKFold یا طرح متناسب با فرد و زمان را بررسی کنید.

### درس 62. نشت داده، Pipeline و اعتبارسنجی گروهی و تو‌در‌تو

**پرسش شروع:** مدلی روی دادهٔ کاملاً تصادفی دقت خوب دارد؛ احتمالاً کشف نکرده، نشت کرده است.

#### مفهوم، دقیق اما ساده

آزمون باید اطلاعاتی باشد که در زمان یادگیری و انتخاب مدل دیده نشده است. نشت فقط ورود مستقیم y نیست؛ انتخاب ویژگی با کل داده، جایگذاری قبل از جداسازی، استانداردسازی خارج fold، تکرار یک فرد در آموزش و آزمون و استفاده از آینده هم نمونه‌های آن‌اند. Pipeline کمک می‌کند تبدیل‌های قابل‌آموزش در fold مناسب fit شوند. برای انتخاب پارامتر، CV داخلی و برای ارزیابی فرایند انتخاب، CV بیرونی یا آزمون نهاییِ مستقل لازم است. GroupKFold برای افراد/دستگاه‌های تکراری و طرح زمانی برای پیش‌بینی آینده به‌کار می‌روند؛ انتخاب تصادفی سطرها همیشه کافی نیست. هیچ ابزار خودکاری بدون تعریف زمان دسترسی ویژگی‌ها، نشت مفهومی را حل نمی‌کند.

#### فرمول و قرارداد

```text
Train-only fitting applies to preprocessing, feature selection and tuning.
Nested CV: inner folds select; outer folds estimate the selected workflow.
```

#### مثال حل‌شده

۱۸۰ ردیف، ۲٬۰۰۰ ویژگیِ نویز و یک برچسب تصادفی می‌سازیم. در روش غلط، بهترین۲۰ ویژگی با استفاده از همهٔ yها انتخاب می‌شوند و سپس CV انجام می‌شود؛ برچسب آزمون قبلاً در انتخاب ویژگی مصرف شده است. در روش درست SelectKBest داخل Pipeline است و هر fold انتخاب خودش را فقط با آموزش انجام می‌دهد. اگر دقت روش غلط بالاتر شد، این شاهد قابل‌تعمیم از رابطهٔ واقعی نیست؛ خودِ سازوکار داده هیچ رابطه‌ای ندارد. روش درست هم در یک نمونه دقیقاً۰٫۵ نمی‌شود.

````python
def run_lesson_62(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: validation
    rng=np.random.default_rng(202625)
    from sklearn.feature_selection import SelectKBest,f_classif
    from sklearn.linear_model import LogisticRegression
    from sklearn.model_selection import StratifiedKFold,cross_val_score
    noise_x=rng.normal(size=(180,2000)); random_y=rng.integers(0,2,180); cv=StratifiedKFold(5,shuffle=True,random_state=42)
    leaked=SelectKBest(f_classif,k=20).fit_transform(noise_x,random_y)
    wrong=cross_val_score(LogisticRegression(max_iter=3000),leaked,random_y,cv=cv)
    proper=make_pipeline(SelectKBest(f_classif,k=20),StandardScaler(),LogisticRegression(max_iter=3000))
    right=cross_val_score(proper,noise_x,random_y,cv=cv)
    display(pd.DataFrame({'fold':np.arange(1,6),'leaked accuracy':wrong,'pipeline accuracy':right}))
    fig,ax=plt.subplots(); ax.bar(['Leaked selection (wrong)','Train-only pipeline'],[wrong.mean(),right.mean()],color=[GOLD,BLUE]); ax.axhline(.5,color='0.4',ls='--'); ax.set(ylabel='5-fold accuracy',ylim=(0,1),title='All 2,000 features and labels are independent noise'); show()
    say(f'دقت با نشت={wrong.mean():.3f}، با Pipeline={right.mean():.3f}. داده هیچ سیگنالِ سازنده‌ای ندارد؛ اختلاف، اثر فرایند انتخاب است و نه اعتبار یک مدل واقعی.')
    # END PRESERVED ANALYSIS: validation
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['validation'] = run_lesson_62()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

میله‌ها میانگین دقت پنج fold در یک اجرای ثابت‌اند؛ خط۰٫۵ مرجعِ کلاس متوازنِ تقریبی است، نه CI یا تضمین دقت تصادفی. توزیع foldها و خط پایهٔ اکثریت را هم در جدول ببینید. شکل تفاوتِ فرایند را نشان می‌دهد؛ جست‌وجوی بذر برای بیشینه‌کردن اختلاف، خودش نمونه‌ای از انتخاب پس از مشاهده است.

#### دام رایج

استفادهٔ تکراری از همان مجموعهٔ آزمون برای انتخاب مدل، آزمون را عملاً به validation تبدیل می‌کند. نام‌گذاری «test» جلوی این مشکل را نمی‌گیرد.

#### تمرین

آیا وجود Pipeline کافی است اگر خریدهای یک فرد در هر دو سمت تقسیم باشند؟

#### پاسخ تشریحی

خیر؛ Pipeline تبدیل‌ها را درون fold نگه می‌دارد، ولی واحد جداسازی را درست انتخاب نمی‌کند. گروه‌بندی فرد یا زمان باید در خود splitter لحاظ شود.

### درس 63. طبقه‌بندی LDA

**هدف یادگیری:** طبقه‌بندی، برچسب و ارزیابی کنارگذاشته‌شده را بشناسید.

#### مفهوم و تشریح کامل

LDA از میانگین کلاس‌ها و کوواریانس مشترک، تابع تصمیم خطی می‌سازد. در این تمرین برچسبِ B برابر طول دست>۲۱ است و پیش‌بین‌ها عرض و سن‌اند. این برچسب ترجیح واقعی سایز ماوس نیست. در LOOCV هر بار یک فرد کنار گذاشته و تبدیل و مدل روی نه فرد دیگر آموزش داده می‌شود.

#### فرمول و قرارداد

```text
δk(x)=xᵀΣ⁻¹μk−0.5μkᵀΣ⁻¹μk+log πk
```

#### مثال حل‌شدهٔ مرجع

ماتریس خطای کنارگذاشته‌شده برای ترتیب A/B برابر [[۶،۰]،[۲،۲]] است. ازده فرد۸ درست‌اند و دقت۸۰٪ می‌شود. برای B، حساسیت۲÷۴=۵۰٪ و برای A نرخ تشخیص۶÷۶=۱۰۰٪ است؛ پس دقت کلی بخشی از ضعف تشخیص B را پنهان می‌کند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_63(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()

    # BEGIN PRESERVED ANALYSIS: lesson-29
    from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
    y_class = (df.hand_length_cm>21).astype(int).to_numpy()
    X_class = df[['hand_width_cm','age_years']].to_numpy()
    lda_pipe = make_pipeline(StandardScaler(), LinearDiscriminantAnalysis())
    loo = LeaveOneOut()
    lda_cv_pred = cross_val_predict(lda_pipe, X_class, y_class, cv=loo, method='predict')
    print('LOOCV confusion matrix [A,B]:')
    display(pd.DataFrame(confusion_matrix(y_class, lda_cv_pred), index=['actual A','actual B'],
                         columns=['pred A','pred B']))
    print('LOOCV accuracy:', round((lda_cv_pred==y_class).mean(),3))

    cm=confusion_matrix(y_class,lda_cv_pred); class_sensitivity=cm[1,1]/cm[1].sum(); class_precision=cm[1,1]/cm[:,1].sum()
    fig,ax=plt.subplots(figsize=(5,4)); ax.imshow(cm,cmap='Blues',vmin=0)
    ax.set(xticks=[0,1],xticklabels=['≤21 cm','>21 cm'],yticks=[0,1],yticklabels=['≤21 cm','>21 cm'],xlabel='Predicted class',ylabel='Actual class',title='LDA LOOCV confusion counts')
    for i in range(2):
        for j in range(2): ax.text(j,i,str(cm[i,j]),ha='center',va='center',bbox=dict(facecolor='white',edgecolor='none',alpha=.8))
    plt.show()

    say(f'در LOOCV دقت={(lda_cv_pred==y_class).mean():.0%}، حساسیت کلاس مثبت={class_sensitivity:.0%} و precision={class_precision:.0%} است. کلاس مثبت طول بیشتر از ۲۱ سانتی‌متر است؛ تنها چهار مثبت واقعی داریم، پس دقت ۸۰٪ به‌تنهایی کافی نیست.')
    # END PRESERVED ANALYSIS: lesson-29
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-29'] = run_lesson_63()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

سطر ماتریس کلاس واقعی و ستون کلاس پیش‌بینی است. جای سطر و ستون را با قرارداد نمودار کنترل کنید. دو خطای B→A دربارهٔ از دست رفتن کاربران کلاس B است و هزینهٔ کاربردیِ آن می‌تواند از یک خطای نوع دیگر متفاوت باشد.

#### کاربرد و محدودیت

برای توصیهٔ اندازه، هدف واقعی را با آزمون تناسب و راحتی ثبت کنید. برچسب‌سازی از طول و سپس استفاده از همان طول به‌عنوان ورودی، مسئله را تقریباً به بازسازی یک آستانه تبدیل می‌کند.

#### تمرین

در این خروجی تعداد مثبت کاذب برای کلاسB چقدر است؟

#### پاسخ تشریحی

صفر؛ هیچ فردA بهB پیش‌بینی نشده است. منفی کاذب برایB برابر۲ است و با وجود دقت۸۰٪، حساسیتB فقط۵۰٪ است.

### درس 64. ROC/AUC برای LDA

**هدف یادگیری:** دقت، آستانه و توان رتبه‌بندی را از هم تشخیص دهید.

#### مفهوم و تشریح کامل

ROC با تغییر آستانهٔ امتیاز، TPR را مقابل FPR رسم می‌کند. AUC توان رتبه‌بندی یک مثبت در برابر یک منفی را خلاصه می‌کند؛ دقت در یک آستانه یا کالیبراسیون احتمال نیست. امتیاز باید از فردی به دست آمده باشد که در برازش آن مدل استفاده نشده است.

#### فرمول و قرارداد

```text
TPR=TP/(TP+FN); FPR=FP/(FP+TN)
```

#### مثال حل‌شدهٔ مرجع

برای آستانهٔ طبقه‌بندی LDAِ این نمونه، TP=۲، FN=۲، TN=۶ وFP=۰ است؛ بنابراین TPR=۰٫۵، FPR=۰، precision=۱ و accuracy=۰٫۸ است. precision یک‌بودن، از دست‌رفتن نصف مثبت‌ها را پنهان نمی‌کند. منحنی از امتیازهای احتمالِ LOOCV ساخته شده است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_64(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    X_class = df[['hand_width_cm','age_years']].to_numpy()
    y_class = (df.hand_length_cm>21).astype(int).to_numpy()
    lda_pipe = make_pipeline(StandardScaler(),LinearDiscriminantAnalysis())
    loo = LeaveOneOut()

    # BEGIN PRESERVED ANALYSIS: lesson-36
    lda_cv_prob = cross_val_predict(lda_pipe, X_class, y_class, cv=loo, method='predict_proba')[:,1]
    auc = roc_auc_score(y_class, lda_cv_prob)
    fpr,tpr,thresholds = roc_curve(y_class, lda_cv_prob)
    fig, ax = plt.subplots()
    ax.plot(fpr,tpr,marker='o',color=BLUE,label=f'LOOCV AUC = {auc:.2f}')
    ax.plot([0,1],[0,1],ls='--',color='gray',label='Chance line')
    ax.set(xlabel='False positive rate',ylabel='True positive rate',xlim=(0,1),ylim=(0,1.05),
           title='LDA ROC from held-out predictions')
    ax.legend(); plt.show()

    say(f'AUC کنارگذاشته‌شده={auc:.4f}؛ خط چین مرجع رتبه‌بندی تصادفی است. AUC توان رتبه‌بندی دو کلاس را توصیف می‌کند و دقت در یک آستانه یا کالیبراسیون احتمال نیست؛ n=10 عدم‌قطعیت زیادی دارد.')
    # END PRESERVED ANALYSIS: lesson-36
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-36'] = run_lesson_64()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

خط قطری مرجعِ رتبه‌بندی تصادفی است؛ بالا–چپ معمولاً مطلوب‌تر است، اما هزینهٔ خطاها و کاربرد انتخاب آستانه را تعیین می‌کنند. با فقط۴ مثبت و۶ منفی، پله‌های منحنی درشت‌اند و عدد AUC ناپایدار است. حتی تجمیع امتیازهای foldهای مختلف محدودیتِ مقایسهٔ امتیاز دارد.

#### کاربرد و محدودیت

اگر کلاس مثبت نادر باشد، precision–recall و تعداد واقعی خطاها را هم گزارش کنید. انتخاب آستانه روی دادهٔ آزمون نهایی، آزمون را به بخشی از تنظیم مدل تبدیل می‌کند.

#### تمرین

اگر پیش‌بینی همهٔ افراد را منفی کنیم، حساسیتB وFPR چیست؟

#### پاسخ تشریحی

هر دو صفر می‌شوند. FPR کم به‌تنهایی کافی نیست، چون هیچ مثبت واقعی کشف نشده است.

### درس 65. جنگل تصادفی و اهمیت متغیرها

**هدف یادگیری:** جنگل تصادفی و اهمیت ویژگی را در چارچوب پیش‌بینی بخوانید.

#### مفهوم و تشریح کامل

جنگل از درخت‌های متعدد روی بازنمونه‌ها و زیرمجموعه‌های تصادفی ویژگی‌ها استفاده می‌کند. میانگین‌گیری پیش‌بینی‌ها نوسان درخت منفرد را کاهش می‌دهد، اما دادهٔ کم را پرحجم نمی‌کند. اهمیت ناخالصی سهم کاهش خطای تقسیم‌های آموزشی است و علت یا ارزش طراحی را نمی‌سنجد.

#### فرمول و قرارداد

```text
ŷ(x)=(1/B)Σb tree_b(x); MAE=mean(|y−ŷ|)
```

#### مثال حل‌شدهٔ مرجع

پاسخ این مدل طول دست است و ورودی‌ها سن، سابقه و عرض‌اند. خطای آموزش≈۰٫۶۸۷ سانتی‌متر و LOOCV≈۱٫۲۰۱ است. اهمیت آموزشی عرض≈۰٫۸۶۱، سابقه≈۰٫۰۸۲ و سن≈۰٫۰۵۷ است. افزایش خطا در فرد کنارگذاشته‌شده، تفاوت یادگیری و حفظ‌کردن را نشان می‌دهد.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_65(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    loo = LeaveOneOut()

    # BEGIN PRESERVED ANALYSIS: lesson-33
    from sklearn.ensemble import RandomForestRegressor
    X_reg = df[['age_years','gaming_years','hand_width_cm']].to_numpy()
    y_reg = df.hand_length_cm.to_numpy()
    rf = RandomForestRegressor(n_estimators=100, min_samples_leaf=2, random_state=SEED)
    rf_cv = cross_val_predict(rf, X_reg, y_reg, cv=loo)
    rf.fit(X_reg,y_reg)
    print(f'Train MAE={np.mean(abs(y_reg-rf.predict(X_reg))):.3f} cm; LOOCV MAE={np.mean(abs(y_reg-rf_cv)):.3f} cm')
    display(pd.DataFrame({'feature':['age','experience','width'], 'training impurity importance':rf.feature_importances_}).round(3))

    prediction_figure(y_reg,rf_cv,'RF LOOCV: hand length','Hand length (cm)')

    say(f'RF: MAE آموزش={np.mean(abs(y_reg-rf.predict(X_reg))):.3f} و LOOCV={np.mean(abs(y_reg-rf_cv)):.3f} سانتی‌متر طول. اختلاف دو خطا خوش‌بینیِ آموزش را نشان می‌دهد؛ اهمیتِ ناخالصی سنجهٔ اثر علّی نیست.')
    # END PRESERVED ANALYSIS: lesson-33
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-33'] = run_lesson_65()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

اهمیت‌ها در این تعریف جمعاً۱ می‌شوند؛ مقدار۰٫۸۶۱ به معنی «۸۶٫۱٪ اثر علّی عرض» نیست. همبستگی ویژگی‌ها می‌تواند اهمیت را بین آن‌ها تقسیم کند. مقایسهٔ مدل‌ها باید با پاسخ، افراد آزمون و واحد خطای یکسان انجام شود.

#### کاربرد و محدودیت

برای دادهٔ بزرگ‌تر، اهمیت جایگشتی را روی دادهٔ کنارگذاشته‌شده بسنجید. OOB نیز برای هر نمونه از درخت‌هایی استفاده می‌کند که آن نمونه در بازنمونهٔ آموزشی‌شان نبوده است؛ در نمونهٔ خیلی کوچک خودِ این برآورد ناپایدار است.

#### تمرین

آیا MAE≈۱٫۲۰۱ یعنی خطای هر فرد دقیقاً۱٫۲۰۱ است؟

#### پاسخ تشریحی

خیر؛ این میانگین قدرمطلق خطاهای افراد است. خطای فردی می‌تواند کوچک‌تر یا بزرگ‌تر باشد و باید توزیع خطا را هم دید.

### درس 66. شبکهٔ عصبی کوچک MLP

**هدف یادگیری:** شبکهٔ عصبی را با داده و خطای مناسب ارزیابی کنید.

#### مفهوم و تشریح کامل

MLP ترکیب‌های خطی و تبدیل غیرخطی را پشت سر هم می‌گذارد. تعداد پارامترها می‌تواند نسبت به ده مشاهده زیاد شود، بنابراین استانداردسازی، منظم‌سازی و ارزیابیِ درست مهم‌اند. در این اجرا یک لایهٔ پنهان چهارواحدی و solver=lbfgs به کار رفته است.

#### فرمول و قرارداد

```text
h=f(W1z+b1); ŷ=W2h+b2; سه ورودی → چهار واحد پنهان → یک پاسخ
```

#### مثال حل‌شدهٔ مرجع

هدف این مدل عرض دست است. MAE کنارگذاشته‌شده≈۰٫۲۳۴ سانتی‌متر شده است. برای کاربر۱، مشاهده۸ و پیش‌بینیِ کنارگذاشته‌شده≈۸٫۲۶ است، یعنی خطا≈+۰٫۲۶؛ برای فرد۹، مشاهده۱۱ و پیش‌بینی≈۱۰٫۶۳ است، یعنی خطا≈−۰٫۳۷.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_66(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    loo = LeaveOneOut()

    # BEGIN PRESERVED ANALYSIS: lesson-34
    from sklearn.neural_network import MLPRegressor
    X_mlp = df[['hand_length_cm','age_years','gaming_years']].to_numpy()
    y_mlp = df.hand_width_cm.to_numpy()
    mlp = make_pipeline(StandardScaler(), MLPRegressor(hidden_layer_sizes=(4,), solver='lbfgs',
                            alpha=1.0, max_iter=1000, random_state=SEED))
    mlp_cv = cross_val_predict(mlp, X_mlp, y_mlp, cv=loo)
    print(f'MLP LOOCV MAE={np.mean(abs(y_mlp-mlp_cv)):.3f} cm')
    display(pd.DataFrame({'observed_width_cm':y_mlp,'held_out_prediction_cm':mlp_cv}).round(2))

    prediction_figure(y_mlp,mlp_cv,'MLP LOOCV: hand width','Hand width (cm)')

    say(f'MLP: MAE کنارگذاشته‌شده={np.mean(abs(y_mlp-mlp_cv)):.3f} سانتی‌متر عرض دست. هدف این مدل با RF که طول را پیش‌بینی می‌کند فرق دارد؛ مقایسهٔ مستقیم دو MAE برای انتخاب الگوریتم معتبر نیست.')
    # END PRESERVED ANALYSIS: lesson-34
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-34'] = run_lesson_66()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

این خطا را با MAE جنگلِ درس قبل مستقیم رتبه‌بندی نکنید؛ آن مدل طول را پیش‌بینی می‌کرد و این یکی عرض را. کاهش loss آموزش نیز با خطای افراد آینده یکسان نیست. lbfgs الزاماً یک منحنی loss در هر epoch مانند SGD ارائه نمی‌کند.

#### کاربرد و محدودیت

برای مقایسهٔ عادلانه، یک رگرسیون ساده و یک شبکه را روی همان پاسخ، همان foldها و همان معیار بسنجید. تنظیم مدل با دیدن نتایج همهٔ foldها می‌تواند برآورد را خوش‌بینانه کند؛ برای تنظیم جدی ارزیابیِ تودرتو یا آزمون مستقل لازم است.

#### تمرین

چرا scaler باید داخل pipeline و داخل fold fit شود؟

#### پاسخ تشریحی

میانگین وSD افراد آزمون نباید در پیش‌پردازش آموزشی دخالت کند. pipeline از ورود این اطلاعات به مدلِ آن fold جلوگیری می‌کند.

### درس 67. رگرسیون Gradient Boosting

**هدف یادگیری:** یادگیری ترتیبی و مقایسه با خط پایه را ببینید.

#### مفهوم و تشریح کامل

Gradient Boosting با مدل ساده شروع و در هر گام یک یادگیرنده برای اصلاح خطا اضافه می‌کند. در زیان مربع، هدفِ گام بعدی باقیمانده‌های فعلی است. نرخ یادگیری کوچک، سهم هر درخت را محدود می‌کند؛ تعداد گام و عمق باید با ارزیابی مناسب تنظیم شوند.

#### فرمول و قرارداد

```text
Fm(x)=Fm−1(x)+νhm(x); برای square loss: residual=y−Fm−1(x)
```

#### مثال حل‌شدهٔ مرجع

مدل۳۰ درختِ عمق۱ با نرخ۰٫۰۵ برای پیش‌بینی طول دارد. LOOCV MAE≈۱٫۲۳۵ سانتی‌متر است؛ خط پایهٔ میانگینِ نه فرد آموزش≈۲٫۲۲۲ است. نسبت کاهشِ این خطا در همین ارزیابی حدود۴۴٫۴٪ است؛ این به معنی افزایش۴۴٫۴٪ فروش یا اثر طراحی نیست.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_67(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    X_reg = df[['age_years','gaming_years','hand_width_cm']].to_numpy()
    y_reg = df.hand_length_cm.to_numpy()
    loo = LeaveOneOut()

    # BEGIN PRESERVED ANALYSIS: lesson-37
    from sklearn.ensemble import GradientBoostingRegressor
    gb = GradientBoostingRegressor(n_estimators=30, learning_rate=.05, max_depth=1,
                                   min_samples_leaf=2, random_state=SEED)
    gb_cv = cross_val_predict(gb, X_reg, y_reg, cv=loo)
    baseline_cv = np.array([y_reg[np.arange(len(y_reg))!=i].mean() for i in range(len(y_reg))])
    print(f'Boosting LOOCV MAE={np.mean(abs(y_reg-gb_cv)):.3f} cm')
    print(f'Mean baseline LOOCV MAE={np.mean(abs(y_reg-baseline_cv)):.3f} cm')
    gb.fit(X_reg,y_reg)
    fig, ax = plt.subplots()
    ax.plot(gb.train_score_,color=BLUE)
    ax.set(xlabel='Boosting stage',ylabel='Training loss',title='Training loss (not a validation curve)')
    plt.show()

    say(f'Boosting LOOCV MAE={np.mean(abs(y_reg-gb_cv)):.3f} در برابر خط پایه={np.mean(abs(y_reg-baseline_cv)):.3f} سانتی‌متر طول. نمودار کاهش زیان فقط آموزش را نشان می‌دهد؛ مقایسهٔ مناسب با RF روی همین هدف ممکن است، با MLP عرض خیر.')
    # END PRESERVED ANALYSIS: lesson-37
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-37'] = run_lesson_67()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور افقی منحنی loss تعداد مراحل boosting و عمودی خطای آموزش است. نزول آن نشان می‌دهد مدل دادهٔ آموزش را بهتر بازسازی می‌کند. برای تعیین نقطهٔ توقف باید منحنیِ اعتبارسنجی را داشته باشیم؛ کمینهٔ آموزش معمولاً مدل مناسبِ آینده را تضمین نمی‌کند.

#### کاربرد و محدودیت

جنگل وBoosting در این مثال پاسخ وfoldهای مشترک دارند و MAE آن‌ها نزدیک است؛ با این n اختلاف کوچک را برتری پایدار اعلام نکنید. پیچیدگی، پایداری و قابلیت توضیح را هم مقایسه کنید.

#### تمرین

اگر loss آموزش کم شود ولی خطای اعتبارسنجی زیاد شود، نشانهٔ چه وضعیتی است؟

#### پاسخ تشریحی

بیش‌برازش محتمل است؛ مدل جزئیات خاص آموزش را یاد می‌گیرد. توقف، منظم‌سازی یا مدل ساده‌تر را با طرح ارزیابی مناسب بررسی کنید.

### درس 68. Autoencoder و فضای نهفته

**هدف یادگیری:** بازسازیِ خوب را از کشفِ ساختار واقعی جدا کنید.

#### مفهوم و تشریح کامل

Autoencoder ورودی را به گلوگاه کم‌بُعد می‌برد و دوباره بازسازی می‌کند. در این مثال ورودی و هدف هر دو چهار ویژگی استانداردشده‌اند؛ پس مسئله نظارت‌شدهٔ پیش‌بینی عرض از بقیه نیست. مدل کم‌ظرفیت می‌تواند ساختار مشترک را فشرده کند، ولی بازسازی آموزشی به‌تنهایی تعمیم را نمی‌سنجد.

#### فرمول و قرارداد

```text
z=tanh(Xscaled We+be); X̂=decoder(z); loss=mean((Xscaled−X̂)²)
```

#### مثال حل‌شدهٔ مرجع

معماری۴→۲→۴ است و MSE بازسازیِ آموزش≈۰٫۰۶۰۹ در واحد استانداردشده ثبت شده است. اجرای ذخیره‌شده هشدار همگراییِ lbfgs دارد؛ بنابراین همین خروجی را با قیدِ همگرا نشدن گزارش می‌کنیم و از آن ادعای مدل نهایی نمی‌سازیم.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_68(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()
    numeric_cols = ['hand_length_cm','hand_width_cm','age_years','gaming_years']
    X_scaled = StandardScaler().fit_transform(df[numeric_cols].to_numpy(dtype=float))

    # BEGIN PRESERVED ANALYSIS: lesson-38
    auto = MLPRegressor(hidden_layer_sizes=(2,), activation='tanh', solver='adam',
                        alpha=.01, max_iter=20000, tol=1e-7, n_iter_no_change=100, learning_rate_init=.01, random_state=SEED).fit(X_scaled, X_scaled)
    latent = np.tanh(X_scaled @ auto.coefs_[0] + auto.intercepts_[0])
    reconstruction = auto.predict(X_scaled)
    print(f'Training reconstruction MSE in standardized units={mean_squared_error(X_scaled,reconstruction):.4f}')
    fig, ax = plt.subplots()
    ax.scatter(latent[:,0],latent[:,1],color=TEAL)
    for i,(a,b) in enumerate(latent): ax.annotate(str(i+1),(a,b),xytext=(4,4),textcoords='offset points')
    ax.set(xlabel='Latent unit 1',ylabel='Latent unit 2',title='Two-neuron autoencoder bottleneck (training data)')
    plt.show()

    say(f'MSE بازسازی آموزشی={mean_squared_error(X_scaled,reconstruction):.5f} در واحد استانداردشده است. دو نورون گلوگاه کد فشرده‌اند، نه دو ویژگی فیزیکی. نسخهٔ حاضر با Adam، alpha=0.01، نرخ یادگیری 0.01 و توقف پس از {auto.n_iter_} تکرار اجرا شد؛ عدد اجرای قدیمیِ lbfgs در مثال مرجع حفظ شده ولی نتیجهٔ این اجرا نیست. توقف بهینه‌ساز و اعتبار خارج از نمونه دو موضوع جدا هستند.')
    # END PRESERVED ANALYSIS: lesson-38
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-38'] = run_lesson_68()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

هر نقطهٔ فضای نهفته یک فرد است؛ مختصات آن طول یا سن نیست. جهت محور و فاصله‌ها به وزن‌های آموزش وابسته‌اند. برای بازسازی در واحد اصلی باید تبدیل معکوس scaler انجام شود؛ خطای عددیِ استاندارد را مستقیم سانتی‌متر نخوانید.

#### کاربرد و محدودیت

مدل را باPCA، دادهٔ کنارگذاشته‌شده و چند بذر مقایسه کنید. با ده نفر، یک شبکهٔ پیچیده می‌تواند تقریباً حفظ کند؛ وجود جزیره در فضای نهفته دلیل تعیین سه سایز نیست.

#### تمرین

چرا MSE=۰٫۰۶۰۹ را نباید «خطای۰٫۰۶۰۹ سانتی‌متر» نوشت؟

#### پاسخ تشریحی

داده‌ها استاندارد شده‌اند و خروجی میانگین مربع خطا روی چهار ویژگی با واحدهای اولیهٔ متفاوت است. برای تفسیر فیزیکی باید خطای هر ستون را پس از تبدیل معکوس سنجید.

### درس 69. آنتروپی، Cross-entropy و اطلاعات متقابل

**پرسش شروع:** پیش‌بینیِ غلط با اطمینان۹۹٫۹٪ چرا از غلط با احتمال۵۱٪ پرهزینه‌تر است؟

#### مفهوم، دقیق اما ساده

آنتروپی میزان عدم‌قطعیت یک توزیع را با واحد تعریف‌شده می‌سنجد؛ برای سکهٔ منصفانه با لگاریتم پایه۲، یک بیت است. آنتروپی برنولی در p=۰٫۵ بیشینه و در دو انتهای قطعی صفر است. Cross-entropy یا log loss، توزیع پیش‌بینی را با پیامد می‌سنجد و به پیش‌بینیِ بسیار مطمئنِ نادرست جریمهٔ بزرگ می‌دهد. این را با «درصد خطا» یکی ندانید؛ واحد log طبیعی nat و پایه۲ bit است. اطلاعات متقابل می‌تواند وابستگی غیرخطیِ فراتر از پیرسون را ببیند، ولی برآورد آن برای پیوسته‌ها، تعداد داده و شیوهٔ binning حساس است و اثبات علّیت نیست. اطلاعات بیشتر دربارهٔ یک برچسب لزوماً اطلاعات اخلاقاً مجاز برای تصمیم‌گیری نیست.

#### فرمول و قرارداد

```text
H(p)=−p log₂p−(1−p)log₂(1−p)
Binary log loss=−[y ln(q)+(1−y)ln(1−q)]
MI(X;Y)=Σp(x,y)log[p(x,y)/(p(x)p(y))].
```

#### مثال حل‌شده

اگر نتیجه واقعاً۱ شود، پیش‌بینی q=۰٫۹۹ زیان−ln(۰٫۹۹)≈۰٫۰۱ دارد، q=۰٫۵ حدود۰٫۶۹۳ و q=۰٫۰۰۱ حدود۶٫۹۰۸. با این حال، قضاوت کالیبراسیون از یک مشاهده ممکن نیست؛ زیان باید روی مجموعهٔ ارزیابی مناسب گزارش شود. در مثال Uشکل، پیرسون ممکن است صفر باشد ولی وابستگی واقعی وجود داشته باشد؛ MI مناسب می‌تواند آن را نشان دهد، با توجه به محدودیت تخمین. منحنی آنتروپی را هم به‌عنوان کیفیت پیش‌بینی یک مدل اشتباه نخوانید؛ توزیع و مدل پیش‌بینی دو نقش متفاوت دارند.

````python
def run_lesson_69(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: information
    rng=np.random.default_rng(202631)
    p=np.linspace(.001,.999,500); entropy=-(p*np.log2(p)+(1-p)*np.log2(1-p)); loss=-np.log(p)
    fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained'); axes[0].plot(p,entropy,color=BLUE); axes[0].set(xlabel='Bernoulli event probability p',ylabel='Entropy (bits)',ylim=(0,1.05),title='Uncertainty is greatest at p=0.5'); axes[1].plot(p,loss,color=GOLD); axes[1].set(xlabel='Predicted probability assigned to y=1',ylabel='Log loss (natural-log nats), observed y=1',title='Confident errors receive a large penalty'); show()
    say(f'برای یک نتیجهٔ درستِ۱: log loss با q=.99 برابر {-np.log(.99):.3f} و با q=.001 برابر {-np.log(.001):.3f} nat است. این مقایسهٔ یک پیامد است؛ کالیبراسیون و کیفیت کل مدل از یک مشاهده نتیجه نمی‌شود.')
    # END PRESERVED ANALYSIS: information
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['information'] = run_lesson_69()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

پنل آنتروپی حول۰٫۵ متقارن است و نقطه‌های قطعی کم‌عدم‌قطعیت‌اند؛ این به معنای خوب‌بودن هر پیش‌بینیِ مطمئن نیست. پنل log loss برای یک y=۱ ثابت، هرچه احتمالِ داده‌شده به نتیجهٔ درست کمتر شود زیان بیشتر می‌شود. پایهٔ لگاریتم‌ها در دو پنل متفاوت و روی محور اعلام شده است.

#### دام رایج

حداقل‌کردن entropy خروجی، بدون توجه به درست‌بودن پیش‌بینی، می‌تواند فقط مدل را بیش‌اطمینان کند. برای log loss، clipکردن احتمال برای پایداری عددی باید روشن باشد.

#### تمرین

آیا مدلی که همیشه۰ یا۱ می‌گوید به دلیل آنتروپی کم حتماً بهتر است؟

#### پاسخ تشریحی

خیر. اگر با اطمینان غلط باشد، log loss بسیار بزرگ می‌شود. کالیبراسیون، قدرت تفکیک و هزینهٔ تصمیم باید با دادهٔ آزمون مناسب سنجیده شوند.

### درس 70. عدم‌توازن کلاس، کالیبراسیون و تصمیمِ هزینه‌محور

**پرسش شروع:** دو مدل افراد را دقیقاً یکسان رتبه‌بندی می‌کنند، اما یکی با اطمینان کاذب عدد۰٫۹۹ می‌دهد.

#### مفهوم، دقیق اما ساده

رتبه‌بندی و احتمال‌دادن دو مهارت متفاوت‌اند. ROC/AUC از ترتیب امتیازها استفاده می‌کند؛ کالیبراسیون می‌پرسد بین مواردی که احتمال۰٫۲ گرفته‌اند تقریباً۲۰٪ مثبت‌اند یا نه. Brier میانگین مربعِ خطای احتمال است و هم اجزای کالیبراسیون و تفکیک را منعکس می‌کند. با کلاس نادر، precision/recall و حجم بررسی معمولاً کنار ROC لازم‌اند؛ PR نیز به نرخ پایه وابسته است. آستانهٔ۰٫۵ قراردادی است، نه همیشه بهینه. برای احتمالِ کالیبره، با هزینهٔ ثابت FP وFN و اقدام دودوییِ ساده، آستانهٔ نظری C_FP/(C_FP+C_FN) است؛ هزینه‌ها، محدودیت ظرفیت و تغییر نرخ پایه می‌توانند تصمیم را عوض کنند.

#### فرمول و قرارداد

```text
Brier=mean((p−y)²)
With calibrated p and simple fixed costs: predict positive if p>C_FP/(C_FP+C_FN).
```

#### مثال حل‌شده

۴٬۰۰۰ احتمال حقیقی از Beta(2,8) و برچسب برنولی می‌سازیم. پیش‌بینی اول همان احتمال سازنده است؛ دومی log-odds را سه برابر می‌کند و بیش‌اطمینان می‌شود. چون تبدیل یکنواخت است، AUC و رتبه‌بندی یکسان می‌مانند، اما Brier و نمودار کالیبراسیون فرق می‌کنند. اگر FN پنج واحد وFP یک واحد هزینه داشته باشد، آستانهٔ نظریِ احتمال کالیبره۱/۶ است. این هزینه قراردادی است، نه ارزش‌گذاری اخلاقی یا مالیِ واقعیِ افراد.

````python
def run_lesson_70(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: calibration
    rng=np.random.default_rng(202626)
    from sklearn.calibration import calibration_curve
    from sklearn.metrics import brier_score_loss,roc_auc_score,precision_score,recall_score
    prob=rng.beta(2,8,4000); truth=rng.binomial(1,prob); over=1/(1+np.exp(-3*np.log(prob/(1-prob))))
    fig,ax=plt.subplots()
    for name,pred,style in [('Generating probabilities',prob,'-'),('Overconfident monotone transform',over,'--')]:
        fraction,mean_pred=calibration_curve(truth,pred,n_bins=8,strategy='quantile'); ax.plot(mean_pred,fraction,marker='o',ls=style,label=name)
    ax.plot([0,1],[0,1],color='0.4',ls=':'); ax.set(xlabel='Mean predicted probability in each bin',ylabel='Observed positive fraction in same bin',xlim=(0,1),ylim=(0,1),title='Same ranking, different calibration'); ax.legend(); show()
    rows=[]
    for threshold in [.5,1/6]:
        pred=prob>threshold; fp=np.sum(pred&(truth==0)); fn=np.sum((~pred)&(truth==1)); rows.append([threshold,precision_score(truth,pred,zero_division=0),recall_score(truth,pred),fp,fn,fp+5*fn])
    display(pd.DataFrame(rows,columns=['threshold','precision','recall','FP','FN','cost FP+5FN']))
    say(f'AUC دو مدل={roc_auc_score(truth,prob):.4f} و {roc_auc_score(truth,over):.4f}؛ Brier={brier_score_loss(truth,prob):.4f} و {brier_score_loss(truth,over):.4f}. یکسانی AUC، کالیبراسیون یکسان را تضمین نکرد. هزینهٔ۱ و۵ صرفاً سناریوی آموزشی است.')
    # END PRESERVED ANALYSIS: calibration
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['calibration'] = run_lesson_70()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

محور افقیِ reliability میانگین احتمال پیش‌بینی‌شدهٔ هر بازه و عمودی سهم مثبت واقعی همان بازه است. خط قطر، کالیبراسیون ایدئال است؛ تعداد محدود هر بازه نوسان ایجاد می‌کند. جدول هزینه برای دو آستانه با همان احتمالِ کالیبره محاسبه شده است، نه با احتمالات بیش‌اطمینان. این یک نمایش با مدلِ سازندهٔ معلوم است؛ در کار واقعی کالیبراسیون باید با دادهٔ جدا از آموزش تنظیم و آزمون شود.

#### دام رایج

AUC بالا به معنی احتمال درست یا بهترین آستانه نیست. یک عدد accuracy نیز خطاهای پرهزینهٔ کلاس نادر را پنهان می‌کند.

#### تمرین

اگر تبدیل یکنواخت امتیاز، AUC را ثابت نگه دارد، آیا می‌توان هر امتیازی را احتمال نامید؟

#### پاسخ تشریحی

خیر. رتبه ممکن است درست بماند، ولی معنای احتمالات و تصمیم هزینه‌محور خراب شود. باید کالیبراسیون، عدم‌قطعیت و نرخ پایه را بررسی کرد.

### درس 71. پیش‌بینی Conformal و ضمانتی که باید درست خوانده شود

**پرسش شروع:** مدل یک عدد پیش‌بینی می‌کند؛ چطور یک بازهٔ همراه با پوششِ قابل‌تعریف بسازیم؟

#### مفهوم، دقیق اما ساده

در split conformal، داده را به آموزش، کالیبراسیون و آزمون تقسیم می‌کنیم. مدل فقط با آموزش fit می‌شود؛ خطاهای مطلقِ کالیبراسیون امتیاز عدم‌انطباق‌اند. یک صدک با تصحیح نمونهٔ محدود انتخاب می‌کنیم و به پیش‌بینی نقطه‌ای کم و زیاد می‌کنیم. با تبادل‌پذیری دادهٔ کالیبراسیون و مشاهدهٔ آینده و رعایت استقلال نقش‌ها، پوشش حاشیه‌ای حداقل۱−α به دست می‌آید؛ این به معنی پوششِ شرطیِ هر x یا هر گروه نیست. مدل لازم نیست درست باشد، اما بازهٔ مدل بد معمولاً پهن و کم‌فایده می‌شود. تغییر توزیع، وابستگی زمانی و مصرف آزمون در تنظیم، ضمانت ساده را مخدوش می‌کنند.

#### فرمول و قرارداد

```text
k=ceil((n_cal+1)(1−α)); q=the k-th smallest calibration score
Prediction set: [f(x)−q,f(x)+q]; requires k≤n_cal or an infinite-bound convention.
```

#### مثال حل‌شده

۶۰۰ نقطه با رابطهٔ y=۲x+نویز و نویزی که با |x| بزرگ‌تر می‌شود بسازید. ۳۰۰ آموزش،۱۵۰ کالیبراسیون و۱۵۰ آزمون مستقلِ همان فرایند کنار می‌گذاریم. α=۰٫۱ یعنی هدف پوشش حاشیه‌ای۹۰٪. k=ceil(۱۵۱×۰٫۹)=۱۳۶ و q خطای۱۳۶ام مرتب‌شده است. بازهٔ ساده پهنای ثابت دارد، درحالی‌که نویز در انتهای محور بیشتر است؛ پس پوشش در بخش مرکزی و انتهایی می‌تواند متفاوت باشد، حتی اگر پوشش کل مناسب باشد.

````python
def run_lesson_71(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: conformal
    rng=np.random.default_rng(202627)
    from sklearn.linear_model import LinearRegression
    cx=rng.uniform(-3,3,(600,1)); cy=2*cx[:,0]+rng.normal(size=600)*(.3+.8*np.abs(cx[:,0])); model=LinearRegression().fit(cx[:300],cy[:300]); alpha=.1
    scores=np.abs(cy[300:450]-model.predict(cx[300:450])); rank=int(np.ceil((len(scores)+1)*(1-alpha))); assert rank<=len(scores); q=np.sort(scores)[rank-1]
    test_x=cx[450:]; test_y=cy[450:]; pred=model.predict(test_x); hits=np.abs(test_y-pred)<=q; order=np.argsort(test_x[:,0]); central=np.abs(test_x[:,0])<1.5
    fig,ax=plt.subplots(); ax.scatter(test_x[:,0],test_y,color=BLUE,s=18,alpha=.65,label='150 untouched test points'); ax.plot(test_x[order,0],pred[order],color=GOLD); ax.fill_between(test_x[order,0],pred[order]-q,pred[order]+q,color=GOLD,alpha=.18,label='90% marginal conformal target'); ax.set(xlabel='Constructed x',ylabel='Constructed response',title='Constant-width split conformal with heteroskedastic noise'); ax.legend(); show()
    say(f'n کالیبراسیون={len(scores)}، رتبهٔ تصحیح‌شده={rank}، نیم‌عرض q={q:.3f}. پوشش آزمون این اجرا={hits.mean():.1%}؛ مرکز={hits[central].mean():.1%} و انتها={hits[~central].mean():.1%}. ضمانت حاشیه‌ای، پوشش برابر همهٔ xها یا گروه‌ها نیست.')
    # END PRESERVED ANALYSIS: conformal
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['conformal'] = run_lesson_71()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

نقاط آزمون و نوار بازه را ببینید؛ برای مرتب‌سازی نمودار فقط x مرتب می‌شود و هیچ نقطه‌ای از محاسبهٔ پوشش حذف نشده است. نرخ پوشش مشاهده‌شدهٔ۱۵۰ آزمون لزوماً دقیقاً۹۰٪ یا بالاتر نیست؛ ضمانت دربارهٔ احتمال حاشیه‌ای تحت فرض‌هاست، نه حد قطعیِ درصد هر مجموعهٔ محدود. دو زیرگروه را هم گزارش می‌کنیم تا پنهان‌شدن نابرابریِ پوشش معلوم شود.

#### دام رایج

عبارت «distribution-free» به معنی «بی‌نیاز از فرض و معتبر تحت هر تغییر توزیع» نیست. تبادل‌پذیری و تعریف نوع پوشش ضروری‌اند.

#### تمرین

اگر بازه در مجموع۹۰٪ پوشش دهد، آیا برای کاربران یک گروه کوچک هم خودکار۹۰٪ است؟

#### پاسخ تشریحی

خیر؛ پوشش حاشیه‌ای گروه‌ها را تضمین نمی‌کند. روش‌های گروه‌شرطی یا کالیبراسیون مناسب، حجم داده و مصالحهٔ پهنا/پوشش را باید جدا بررسی کرد.

## فصل 11: زمان تا رخداد و سانسور

منحنی بقا را با معنای رخداد و سانسور می‌خوانیم و بعد تفاوت گروه‌ها را با Log-rank می‌سنجیم.

### درس 72. تحلیل بقا Kaplan–Meier

**هدف یادگیری:** خرابی و سانسور را در منحنی بقا تفکیک کنید.

#### مفهوم و تشریح کامل

تابع بقا S(t) احتمالِ باقی‌ماندن بدون رخداد تا بعد از t را توصیف می‌کند. Kaplan–Meier در زمان رخداد با ضریب۱−d/nr افت می‌کند. سانسور راست یعنی می‌دانیم محصول تا زمان پیگیری کار کرده ولی زمان خرابی بعدی‌اش را ندیده‌ایم؛ سانسور، خرابی نیست.

#### فرمول و قرارداد

```text
Ŝ(t)=∏(1−di/ni), برای زمان‌های رخداد ti≤t
```

#### مثال حل‌شدهٔ مرجع

مثال موجود ساختگی است و سه گروه۵تاییِ بدون سانسور دارد. در گروه سبک، خرابی‌ها۲۴،۲۶،۲۸،۳۰،۳۲ ماه‌اند؛ بعد از نخستین خرابی S=۴/۵=۰٫۸ و بعد از دومین S=۰٫۶ است. میانهٔ بقا نخستین زمان S≤۰٫۵ است: سبک۲۸، متوسط۲۲ و سنگین۱۴ ماه.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_72(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # BEGIN PRESERVED ANALYSIS: lesson-35
    survival = {'light':[24,26,28,30,32], 'medium':[18,20,22,24,26],
                'heavy':[10,12,14,16,18]}
    def km_all_events(times):
        values, counts = np.unique(times, return_counts=True)
        at_risk, probability, steps = len(times), 1.0, [(0,1.0)]
        for t,d in zip(values, counts):
            probability *= 1-d/at_risk
            steps.append((t,probability))
            at_risk -= d
        return np.asarray(steps)
    fig, ax = plt.subplots()
    for name, times in survival.items():
        steps = km_all_events(times)
        ax.step(steps[:,0],steps[:,1],where='post',label=f'{name} (median {np.median(times):.0f} mo)')
    ax.set(xlabel='Months to simulated failure',ylabel='Survival probability',ylim=(0,1.05),
           title='Kaplan–Meier curves: simulated failures only')
    ax.legend(); plt.show()

    say(f'میانه‌های بقای ساختگی: سبک={np.median(survival["light"]):.0f}، متوسط={np.median(survival["medium"]):.0f}، سنگین={np.median(survival["heavy"]):.0f} ماه. هر ۱۵ مشاهده رخداد دارد؛ این مثال شاهد گارانتی واقعی نیست.')
    # END PRESERVED ANALYSIS: lesson-35
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-35'] = run_lesson_72()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

محور افقی ماه و عمودی احتمال بقا ازصفر تا۱ است. پلهٔ افقی به معنی نبودِ رخداد ثبت‌شده در آن فاصله است. اگر سانسور وجود داشته باشد، تعداد در معرض خطر کاهش می‌یابد ولی خود سانسور افتِ احتمال ایجاد نمی‌کند. نقاط انتهایی با افراد در معرض خطرِ کم، عدم‌قطعیت بیشتری دارند.

#### کاربرد و محدودیت

برای گارانتی باید دادهٔ واقعیِ تست، وضعیت رخداد، شرایط استفاده، علت خروج و هزینهٔ خدمات ثبت شود. میانه‌های دادهٔ ساختگی هیچ ضمانتی دربارهٔ عمر محصول تولیدی ایجاد نمی‌کنند.

#### تمرین

محصولی تا ماه۱۸ کار کرده و سپس مطالعه تمام شده است. status آن چیست؟

#### پاسخ تشریحی

در تعریف رایج، status=۰ یعنی سانسور در۱۸؛ status=۱ برای رخداد خرابی است. ثبت آن به‌عنوان خرابی۱۸ماهه منحنی را غلط می‌کند.

### درس 73. سانسور و آزمون Log-rank، قدم‌به‌قدم

**هدف یادگیری:** افت منحنی بقا و مقایسهٔ دو منحنی را با شمارشِ در معرض خطر یاد بگیرید.

#### مفهوم و تشریح کامل

برای مقایسهٔ دو منحنی، Log-rank در هر زمان رخداد تعداد رخدادِ مشاهده‌شدهٔ یک گروه را با تعداد مورد انتظار تحت فرض صفر مقایسه می‌کند. تعداد در معرض خطر باید شامل افرادِ زنده و هنوز تحت پیگیری باشد. سانسور مستقل از خطر رخداد، مشروط به اطلاعات مناسب، از فرض‌های مهم تفسیر است.

#### فرمول و قرارداد

```text
E1j=dj×n1j/nj; U=Σ(O1j−E1j); χ²=U²/Var(U), df=1
```

#### مثال حل‌شدهٔ مرجع

یک مثال مستقلِ چهارتایی در زمان‌های ۲، ۳، ۴، ۵ داریم: در ۲ یک خرابی، در ۳ یک سانسور، در ۴ یک خرابی و در ۵ یک سانسور. S پس از ۲ برابر ۳/۴=۰٫۷۵ است؛ سانسور ۳ آن را کم نمی‌کند. قبل از ۴ دو محصول در معرض خطرند، پس S پس از ۴ برابر ۰٫۷۵×(۱−۱/۲)=۰٫۳۷۵ است. این همان میانهٔ بقای برآوردی ۴ است. در مقایسهٔ جداگانهٔ گروه‌های ساختگیِ سبک و سنگین، χ²≈۹٫۷۰۰۷ با یک درجهٔ آزادی و p≈۰٫۰۰۱۸۴۲ به دست می‌آید؛ این عدد توصیفِ سناریوی ساختگی است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_73(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # BEGIN PRESERVED ANALYSIS: logrank
    from statsmodels.duration.survfunc import SurvfuncRight, survdiff
    light=np.array([24,26,28,30,32]); heavy=np.array([10,12,14,16,18])
    lr_time=np.r_[light,heavy]; lr_status=np.ones(10,dtype=int); lr_group=np.r_[np.zeros(5,dtype=int),np.ones(5,dtype=int)]
    lr_chi,lr_p=survdiff(lr_time,lr_status,lr_group)
    example=SurvfuncRight(np.array([2,3,4,5]),np.array([1,0,1,0]))
    display(pd.DataFrame({'event_time':example.surv_times,'estimated_survival':example.surv_prob}))
    lr_rows=[]
    for t in np.unique(lr_time):
        n1=(light>=t).sum(); n2=(heavy>=t).sum(); d1=(light==t).sum(); d2=(heavy==t).sum(); n=n1+n2; d=d1+d2
        variance=n1*n2*d*(n-d)/(n*n*(n-1)) if n>1 else 0
        lr_rows.append([t,n1,n2,d1,d2,d*n1/n,variance])
    lr_table=pd.DataFrame(lr_rows,columns=['time','risk light','risk heavy','events light','events heavy','expected light','variance'])
    display(lr_table.round(4)); lr_u=(lr_table['events light']-lr_table['expected light']).sum(); lr_manual=lr_u**2/lr_table.variance.sum()
    assert np.isclose(lr_manual,lr_chi)
    fig,ax=plt.subplots(figsize=(8,4)); ax.step(np.r_[0,example.surv_times,5],np.r_[1,example.surv_prob,example.surv_prob[-1]],where='post',color=BLUE)
    ax.scatter([3,5],[.75,.375],marker='+',s=140,color=GOLD,label='Censoring (no drop)'); ax.axhline(.5,color='0.4',ls='--'); ax.set(xlabel='Time (constructed units)',ylabel='Estimated survival probability',ylim=(0,1.05),title='Censoring removes risk time, not survival probability'); ax.legend(); plt.show()
    say(f'مثال سانسور: S(2)=0.75 و S(4)=0.375؛ میانهٔ بقا ۴ است. مقایسهٔ جداگانهٔ سبک/سنگینِ ساختگی χ²={lr_chi:.4f} و p={lr_p:.6f} می‌دهد؛ محاسبهٔ دستی جدول دقیقاً با کتابخانه تطبیق دارد. سانسور ۳ و ۵ منحنی را پایین نمی‌آورد. این شواهد از دادهٔ خرابی صنعتی نیست.')
    # END PRESERVED ANALYSIS: logrank
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['logrank'] = run_lesson_73()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

در مثال ساختگیِ سه گروه قبلی، هر۱۵ محصول خرابی دارند؛ status همه۱ است. اگر دو گروه سبک و سنگین را باکد زیر مقایسه کنیم، p صرفاً خروجیِ همان سناریوی ساختگی است. در دادهٔ واقعی، هم‌پوشانی و عبور منحنی‌ها مهم است و یکp همهٔ شکل اختلاف را خلاصه نمی‌کند.

#### کاربرد و محدودیت

میانهٔ بقا، نرخ خرابی و هزینهٔ گارانتی پرسش‌های متفاوت‌اند. برای اعتبار صنعتی، شرایط آزمایش و پیگیریِ خروج‌کرده‌ها باید ثبت شوند. متغیرهای استفاده یا گروه‌های تصادفی‌نشده، استنباط علّی را محدود می‌کنند.

#### تمرین

چرا حذفِ کامل محصولاتِ هنوز سالم از جدول، منحنی را عوض می‌کند؟

#### پاسخ تشریحی

آن‌ها تا زمان سانسور در معرض خطر بوده‌اند. حذفشان مخرجِ رخدادهای قبلی را کاهش می‌دهد و معمولاً بقا را کمتر از واقع نشان می‌دهد.

## فصل 12: وابستگی زمانی و پیش‌بینی آینده

ابتدا خودهمبستگی و تغییر توزیع؛ سپس تفکیک زمانی، فروش آموزشی و ابزارهای هموارسازی، بدون استفاده از آینده.

### درس 74. خودهمبستگی، اعتبارسنجی زمانی و تغییر توزیع

**پرسش شروع:** صد مقدار پشت‌سرهم که تقریباً شبیه هم‌اند، چقدر اطلاعات تازه دارند؟

#### مفهوم، دقیق اما ساده

در سری زمانی، ترتیب داده بخشی از مسئله است. خودهمبستگی یعنی مقدارها با گذشتهٔ خود رابطه دارند؛ در AR(1) پایای ساده، اثر یک شوک با توان‌های φ کم می‌شود. وقتی φ مثبت و بزرگ است، میانگین صد روز به اندازهٔ صد مشاهدهٔ مستقل دقیق نیست. تقریب n_eff≈n(1−φ)/(1+φ) برای شرایط خاصِ AR(1) و طول کافی است؛ برای سری فصلی، غیرپایا یا وابستگی پیچیده قانون عمومی نیست. تقسیم تصادفی سطرها می‌تواند آینده را به آموزش بیاورد. اعتبارسنجی مبدأهای متحرک، gap مناسب و ویژگی‌هایی که فقط در لحظهٔ پیش‌بینی در دسترس‌اند مهم‌اند. Drift ممکن است توزیع X، شیوع Y یا خود رابطهٔ P(Y|X) را عوض کند.

#### فرمول و قرارداد

```text
AR(1): y_t=φy_(t−1)+ε_t, |φ|<1
Under suitable stationary AR(1) conditions: n_eff≈n(1−φ)/(1+φ).
```

#### مثال حل‌شده

با φ=۰٫۸، n=۴۰۰ و پس از burn-in، سری ساخته می‌شود. تقریب n مؤثر فقط۴۴٫۴ است. می‌توان هر۴۰۰ سطر را داشت اما اطلاعات تازهٔ میانگین کمتر باشد. در مثال پیش‌بینی، میانگین متحرکِ centered از آیندهٔ نقطه استفاده می‌کند؛ trailing فقط از گذشته و جاری استفاده می‌کند. حتی trailing اگر هدف همان y جاری باشد، ممکن است خود y را داخل ویژگی گذاشته و نشت کند؛ برای پیش‌بینی گام بعد باید جابه‌جایی و زمان دسترسی را دقیق تعریف کرد.

````python
def run_lesson_74(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: time-dependence
    rng=np.random.default_rng(202628)
    from statsmodels.tsa.stattools import acf
    phi=.8; eps=rng.normal(size=700); series=np.zeros(700)
    for i in range(1,len(series)): series[i]=phi*series[i-1]+eps[i]
    series=series[300:]; coefficients=acf(series,nlags=20,fft=True)
    fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained'); axes[0].plot(np.arange(len(series)),series,color=BLUE); axes[0].set(xlabel='Ordered time index after burn-in',ylabel='Constructed value',title='Stationary AR(1), φ=0.8'); axes[1].bar(np.arange(21),coefficients,color=TEAL); axes[1].set(xlabel='Lag',ylabel='Sample autocorrelation',title='ACF: adjacent rows are not independent'); show()
    say(f'n={len(series)} ولی تقریب n مؤثرِ AR(1) با φ معلوم={len(series)*(1-phi)/(1+phi):.1f} است؛ ACF نمونه در lag۱={coefficients[1]:.3f}. این فرمول برای وابستگی یا drift دلخواه معتبر نیست.')
    # END PRESERVED ANALYSIS: time-dependence
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['time-dependence'] = run_lesson_74()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

پنل اول ترتیب نوسان‌ها و پنل دوم ACF را نشان می‌دهد. میلهٔ lag=۰ تعریفاً۱ است و رابطهٔ علمی تازه‌ای نیست. ACF نمونه از φ^lag دقیقاً پیروی نمی‌کند. ناحیه‌های اطمینان تقریبیِ ACF در مدل‌های مختلف فرض‌های متفاوت دارند؛ اینجا عمداً فقط ضرایب بدون نوار قطعیت رسم می‌شوند. روند یا فصل قوی هم می‌تواند همبستگی ظاهری بسازد.

#### دام رایج

ارزیابی تصادفی زمانی را فقط به دلیل راحتی اجرا انتخاب نکنید. زنجیرهٔ ویژگی→زمان دسترسی→افق پیش‌بینی باید قابل ممیزی باشد.

#### تمرین

چرا مدلی که در۲۰۱۱ عالی بوده ممکن است در۲۰۱۲ افت کند، حتی با کد بدون اشکال؟

#### پاسخ تشریحی

تغییر کاربران، نرخ پایه، سیاست، سنجه‌گیری یا رابطهٔ ورودی و پاسخ ممکن است رخ داده باشد. باید دادهٔ تازه، خط پایه و انواع drift را بررسی کرد؛ افت خودکار علت خاصی را اثبات نمی‌کند.

### درس 75. سری زمانی و پیش‌بینی فروش

**هدف یادگیری:** پیش‌بینی زمانی را با گذشتهٔ مجاز ارزیابی کنید.

#### مفهوم و تشریح کامل

در سری زمانی ترتیب داده بخشی از مسئله است. ماه‌های آینده نباید در برازش یا انتخاب تنظیماتِ پیش‌بینیِ گذشته استفاده شوند. خط پایهٔ فصلی مقدار همان ماهِ سال قبل را پیش‌بینی می‌کند؛ خط روند فقط تغییر خطی متوسط را می‌بیند. دادهٔ۲۴ماههٔ این درس نمایشی است.

#### فرمول و قرارداد

```text
seasonal naive: ŷt=yt−12; MAEtest=mean(|yt−ŷt|), t=19,…,24
```

#### مثال حل‌شدهٔ مرجع

ماه‌های۱ تا۱۸ آموزش و۱۹ تا۲۴ آزمون‌اند. برای ماه۱۹ مقدار واقعی۲۴هزار و فصلیِ ساده۲۲هزار است؛ خطای مطلق۲هزار. MAE شش ماهِ آزمون برای فصلی۲ و برای روند≈۲٫۴۷۸هزار است. پیش‌بینی فصلیِ ماه‌های۲۵ تا۳۰ برابر[۱۴،۱۶،۱۷،۱۹،۲۰،۲۲]هزار است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_75(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # BEGIN PRESERVED ANALYSIS: lesson-39
    sales = np.array([12,14,15,17,18,20,22,21,19,18,16,15,
                      14,16,17,19,20,22,24,23,21,20,18,17], dtype=float)
    months = np.arange(1,25)
    train_end = 18
    seasonal_test = sales[np.arange(train_end,24)-12]
    trend_coef = np.polyfit(months[:train_end],sales[:train_end],deg=1)
    trend_test = np.polyval(trend_coef,months[train_end:])
    test_actual = sales[train_end:]
    comparison = pd.DataFrame({'month':months[train_end:], 'actual_demo_sales_k':test_actual,
                               'seasonal_naive_k':seasonal_test, 'linear_trend_k':trend_test})
    display(comparison.round(2))
    for name,pred in [('seasonal naive',seasonal_test),('linear trend',trend_test)]:
        print(name, 'holdout MAE (thousand units)=',round(np.mean(abs(test_actual-pred)),3))
    future_months = np.arange(25,31)
    future_forecast = sales[future_months-13]
    display(pd.DataFrame({'future_month':future_months,'seasonal_naive_forecast_k':future_forecast}))
    fig, ax = plt.subplots()
    ax.plot(months,sales,marker='o',color=BLUE,label='Illustrative history')
    ax.plot(future_months,future_forecast,marker='o',ls='--',color=GOLD,label='Seasonal-naive forecast')
    ax.axvline(train_end+.5,color='gray',ls=':',label='Holdout begins')
    ax.set(xlabel='Month index',ylabel='Illustrative sales (thousand units)',
           title='24-month demo series and six-month seasonal baseline')
    ax.legend(); plt.show()

    say(f'MAE شش ماه آخر: فصلی ساده={np.mean(abs(test_actual-seasonal_test)):.3f} و روند خطی={np.mean(abs(test_actual-trend_test)):.3f} هزار واحد. پیش‌بینی آیندهٔ ماه‌های ۲۵–۳۰={future_forecast.tolist()}. سری ساختگی است و بازهٔ عدم‌قطعیت آینده در این خط پایه برآورد نشده است.')
    # END PRESERVED ANALYSIS: lesson-39
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['lesson-39'] = run_lesson_75()
plt.close('all')
````

### درس 76. میانگین متحرک، تجزیه و Holt–Winters

**هدف یادگیری:** هموارسازیِ تاریخ، تجزیهٔ الگو و پیش‌بینیِ آینده را جدا بخوانید.

#### مفهوم و تشریح کامل

میانگین متحرک trailing از زمان جاری و گذشته استفاده می‌کند؛ میانگین centered برای توصیف می‌تواند از آیندهٔ همان نقطه هم استفاده کند و برای پیش‌بینیِ آنلاین بدون تغییر مناسب نیست. تجزیهٔ جمعی، سری را به روند، فصل و باقیمانده می‌شکند. Holt–Winters سطح و روند و الگوی فصلی را به‌صورت بازگشتی به‌روزرسانی می‌کند.

#### فرمول و قرارداد

```text
MA3(t)=(y[t−2]+y[t−1]+y[t])/3; y[t]=T[t]+S[t]+R[t]
ℓ[t]=α(y[t]−s[t−m])+(1−α)(ℓ[t−1]+b[t−1])
b[t]=β(ℓ[t]−ℓ[t−1])+(1−β)b[t−1]
s[t]=γ(y[t]−ℓ[t])+(1−γ)s[t−m]
forecast at t+h = ℓ[t]+h×b[t]+appropriate seasonal index
```

#### مثال حل‌شدهٔ مرجع

سه فروش آخرِ سری نمایشی ۲۰، ۱۸، ۱۷ هزار است، پس MA3 در ماه ۲۴ برابر ۱۸٫۳۳۳ هزار می‌شود. این هموارسازیِ تاریخ است. برای یک گامِ نمونهٔ Holt–Winters جمعی، سطح قبلی ۱۸، روند قبلی ۱، فصلِ متناظرِ قبلی ۲ و مشاهدهٔ تازه ۲۳ را بگیرید. با α=۰٫۳، β=۰٫۲ و γ=۰٫۴، سطح جدید ۰٫۳×(۲۳−۲)+۰٫۷×(۱۸+۱)=۱۹٫۶ است؛ روند جدید ۰٫۲×(۱۹٫۶−۱۸)+۰٫۸×۱=۱٫۱۲ و جزء فصلی جدید ۰٫۴×(۲۳−۱۹٫۶)+۰٫۶×۲=۲٫۵۶ است. این مثالِ یک گام، مستقل از برازش خودکارِ سری ۲۴ماهه است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_76(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    sales = np.array([12,14,15,17,18,20,22,21,19,18,16,15,14,16,17,19,20,22,24,23,21,20,18,17],dtype=float)

    # BEGIN PRESERVED ANALYSIS: time-tools
    from statsmodels.tsa.holtwinters import ExponentialSmoothing
    from statsmodels.tsa.seasonal import seasonal_decompose
    series=pd.Series(sales,index=pd.date_range('2020-01-01',periods=24,freq='MS'))
    ma3=series.rolling(3).mean(); decomp=seasonal_decompose(series,model='additive',period=12)
    fig,ax=plt.subplots(figsize=(10,4)); ax.plot(series,marker='o',color=BLUE,label='Constructed history'); ax.plot(ma3,color=GOLD,ls='--',label='Trailing MA3'); ax.set(ylabel='Demo sales (thousand units)',xlabel='Month',title='Trailing moving average: smoothing, not validated forecasting'); ax.legend(); plt.show()
    fig=decomp.plot(); fig.set_size_inches(10,8); fig.suptitle('Additive decomposition: only two constructed seasonal cycles',y=1.01); fig.tight_layout(); plt.show()
    hw=ExponentialSmoothing(series,trend='add',seasonal='add',seasonal_periods=12,initialization_method='estimated').fit()
    hw_future=hw.forecast(6); display(pd.DataFrame({'forecast_demo_sales_k':hw_future}).round(3))
    fig,ax=plt.subplots(figsize=(10,4)); ax.plot(series,color=BLUE,label='History used to fit'); ax.plot(hw.fittedvalues,color=TEAL,ls=':',label='In-sample fit'); ax.plot(hw_future,color=GOLD,marker='o',ls='--',label='Illustrative forecast (no interval)'); ax.set(xlabel='Month',ylabel='Demo sales (thousand units)',title='Holt–Winters fitted on all 24 months: not a holdout evaluation'); ax.legend(); plt.show()
    level=.3*(23-2)+.7*(18+1); trend=.2*(level-18)+.8*1; seasonal=.4*(23-level)+.6*2
    say(f'آخرین MA3={ma3.iloc[-1]:.3f} هزار واحد است. مثال یک گام مستقل: سطح={level:.2f}، روند={trend:.2f}، فصل={seasonal:.2f}. پیش‌بینی خودکار HW شش ماه بعد {np.round(hw_future.to_numpy(),3).tolist()} است. همهٔ ۲۴ ماه برای برازش مصرف شده‌اند؛ این نمودار ارزیابی آزمون یا بازهٔ تضمین‌شده نیست.')
    # END PRESERVED ANALYSIS: time-tools
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['time-tools'] = run_lesson_76()
plt.close('all')
````

### درس 77. مطالعهٔ واقعی ۱: چارک‌های ANSUR II و پوشش اندازه

**هدف یادگیری:** روش چارک و تفسیر را روی هزاران اندازه‌گیری واقعی اجرا کنید.

#### مفهوم و تشریح کامل

نسخهٔ منتخب ANSUR II شامل ۶٬۰۶۸ فرد، ۴٬۰۸۲ مرد و ۱٬۹۸۶ زن از نظامیان آمریکاست. طول، عرض و طول کف دست در فایل اصلی میلی‌مترند. این جامعه با کاربران ایرانیِ ماوس یکی نیست؛ تعداد زیاد سطر به‌تنهایی نمایندگی بازار هدف را ایجاد نمی‌کند. گروه و متغیر را پیش از مقایسه مشخص کنید.

#### فرمول و قرارداد

```text
type 7 quantile: h=(n−1)p; cm=mm/10
```

#### مثال حل‌شدهٔ مرجع

برای طول دست مردان Q1=۱۸۷، میانه=۱۹۳ و Q3=۲۰۰ میلی‌متر است؛ IQR=۱۳ و حصارها ۱۶۷٫۵ و ۲۱۹٫۵ میلی‌مترند. برای زنان Q1=۱۷۴، میانه=۱۸۰ و Q3=۱۸۷ است؛ IQR نیز ۱۳ میلی‌متر است. میانه‌ها ۱۳ میلی‌متر فرق دارند؛ برابر بودن IQR به معنی توزیع یا بازهٔ پوششِ برابر نیست.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_77(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # BEGIN PRESERVED ANALYSIS: real-quartiles
    # Editable analysis parameters; run this cell again after changing them.
    ANSUR_SEX='male'  # 'male', 'female', or 'all'; do not infer a target-market sex mix.
    ANSUR_VARIABLE='handlength'  # handlength / handbreadth / palmlength (mm)
    ansur_view=ansur if ANSUR_SEX=='all' else ansur.loc[ansur.sex.eq(ANSUR_SEX)]
    av=ansur_view[ANSUR_VARIABLE].to_numpy(float); aq=np.quantile(av,[.05,.25,.5,.75,.95],method='linear'); aiqr=aq[3]-aq[1]; afences=(aq[1]-1.5*aiqr,aq[3]+1.5*aiqr)
    summary_ansur=ansur.groupby('sex')[ANSUR_VARIABLE].agg(['count','min','mean','median','max'])
    for p in [.05,.25,.75,.95]: summary_ansur[f'P{100*p:g}']=ansur.groupby('sex')[ANSUR_VARIABLE].quantile(p,interpolation='linear')
    display(summary_ansur.round(3))
    fig,axes=plt.subplots(1,2,figsize=(12,4.5),layout='constrained')
    axes[0].hist(av,bins=35,color=BLUE,edgecolor='white');
    for q,label,ls in zip(aq[1:4],['Q1','Median','Q3'],['--','-',':']): axes[0].axvline(q,color=GOLD,ls=ls,label=f'{label}={q:g} mm')
    axes[0].set(xlabel=f'{ANSUR_VARIABLE} (mm)',ylabel='People',title=f'ANSUR II: {ANSUR_SEX}, n={len(av):,}'); axes[0].legend()
    axes[1].boxplot([ansur.loc[ansur.sex.eq(s),ANSUR_VARIABLE] for s in ['female','male']],tick_labels=['Female (n=1,986)','Male (n=4,082)'],showfliers=True); axes[1].set(ylabel=f'{ANSUR_VARIABLE} (mm)',title='Separate populations, same measurement scale'); plt.show()
    az=np.sort(av); fig,ax=plt.subplots(); ax.step(az,np.arange(1,len(az)+1)/len(az),where='post',color=BLUE); ax.axhline(.95,color=GOLD,ls='--'); ax.axvline(aq[-1],color=GOLD,ls=':'); ax.set(xlabel=f'{ANSUR_VARIABLE} (mm)',ylabel='Empirical cumulative share',ylim=(0,1.03),title='Percentile coverage in the selected ANSUR population'); plt.show()
    say(f'انتخاب فعلی: {ANSUR_SEX}، متغیر {ANSUR_VARIABLE}، n={len(av):,}. چارک‌ها {aq[1]:.2f}، {aq[2]:.2f} و {aq[3]:.2f} میلی‌متر؛ IQR={aiqr:.2f} و حصارها [{afences[0]:.2f},{afences[1]:.2f}] هستند. صدک ۹۵={aq[-1]:.2f} میلی‌متر توصیف همین گروه است؛ طول بدنهٔ ماوس از آن به‌طور مستقیم به دست نمی‌آید. نمایندگی کاربران ایرانی یا بازار هدف ادعا نمی‌شود.')
    # END PRESERVED ANALYSIS: real-quartiles
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['real-quartiles'] = run_lesson_77()
plt.close('all')
````

در آزمایشگاه چارک‌ها ابتدا «طول دست/مردان» را انتخاب کنید؛ جای جعبه و میانه را با همین اعداد تطبیق دهید. سپس گروه را عوض و حرکت جعبه را ببینید. تغییر متغیر به عرض دست، واحد را ثابت نگه می‌دارد اما معنای اندازه و دامنه را تغییر می‌دهد. علامت‌های بیرون حصار، خطای قطعی نیستند.

برای فرضیهٔ سایزبندی، ابتدا جامعهٔ هدفِ واقعی و پوشش مطلوب را تعریف کنید، سپس نسبت‌های جمعیتی را با طرح نمونه‌گیری و وزن‌دهی مناسب وارد کنید. تقسیم دلخواهِ طول به سه بخش، تناسب چندبعدی یا تعداد مدل تجاری را تضمین نمی‌کند.

#### تمرین

اگر Q3=۲۰۰ میلی‌متر باشد، آیا طول بدنهٔ ماوس باید ۲۰ سانتی‌متر شود؟

خیر. این صدکِ اندازهٔ دست است و رابطهٔ مستقیم با طول بدنه از آن استخراج نشده است. ابعاد محصول باید با هندسهٔ گرفتن و آزمون تناسب تعیین شوند.

### درس 78. مطالعهٔ واقعی ۲: گشتاورهای تقاضای ساعتی

**هدف یادگیری:** توان‌های فاصله و حساسیت به انتهاها را در ۱۷٬۳۷۹ ساعت ببینید.

#### مفهوم و تشریح کامل

در Bike Sharing هر سطر یک تاریخ×ساعت است؛ cnt تعداد اجاره و نه تعداد افراد یکتاست. این فایل همهٔ ساعت‌های تقویمی را ندارد و ساعت غایب را نباید خودکار صفر کرد. دادهٔ دو سال، فصل و زمان روز دارد؛ مخلوط این وضعیت‌ها بر شکل توزیع اثر می‌گذارد.

#### فرمول و قرارداد

```text
m2=Σ(x−x̄)²/n; g1=m3/m2^(3/2); β2=m4/m2²
```

#### مثال حل‌شدهٔ مرجع

برای کل فایل، میانگین≈۱۸۹٫۴۶۳، میانه ۱۴۲، SD توصیفی≈۱۸۱٫۳۸۲ و بیشینه ۹۷۷ اجاره در ساعت است. چولگی≈۱٫۲۷۷، کشیدگی≈۴٫۴۱۶ و اضافی≈۱٫۴۱۶ است. تعداد ۹۳۸ ساعت، یعنی حدود ۵٫۴٪، حداقل ۲SD از میانگین دورند و حدود ۸۱٫۷٪ مجموع توان چهارم فاصله‌ها را می‌سازند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_78(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # BEGIN PRESERVED ANALYSIS: real-moments
    BIKE_VARIABLE='cnt'  # cnt / casual / registered
    BIKE_YEAR='all'  # 'all', 2011, 2012
    bv_frame=bike if BIKE_YEAR=='all' else bike.loc[bike.dteday.dt.year.eq(int(BIKE_YEAR))]
    bv=bv_frame[BIKE_VARIABLE].to_numpy(float); bm=bv.mean(); bc=bv-bm; bsd=bv.std(ddof=0); bz=bc/bsd
    braw=np.array([np.mean(bv**o) for o in range(1,5)]); bcentral=np.array([np.mean(bc**o) for o in range(1,5)])
    bskew=np.mean(bz**3); bkurt=np.mean(bz**4); tail=np.abs(bz)>=2; tailshare=np.sum(bz[tail]**4)/np.sum(bz**4)
    display(pd.DataFrame({'order':[1,2,3,4],'raw moment (rental^order)':braw,'central moment (rental^order)':bcentral}))
    display(pd.DataFrame({'n':[len(bv)],'mean':[bm],'median':[np.median(bv)],'SD descriptive':[bsd],'skew moment':[bskew],'excess moment':[bkurt-3],'skew adjusted':[stats.skew(bv,bias=False)],'excess adjusted':[stats.kurtosis(bv,bias=False)]}).round(5))
    fig,axes=plt.subplots(1,2,figsize=(12,4.5),layout='constrained'); axes[0].hist(bv,bins=45,color=BLUE,edgecolor='white'); axes[0].axvline(bm,color=GOLD,label=f'Mean={bm:.1f}'); axes[0].axvline(np.median(bv),color='0.3',ls='--',label=f'Median={np.median(bv):.1f}'); axes[0].set(xlabel=f'{BIKE_VARIABLE}: rentals per observed hour',ylabel='Observed hours',title=f'Bike Sharing {BIKE_YEAR}: n={len(bv):,}'); axes[0].legend()
    axes[1].bar([0,1],[100*tail.mean(),100*tailshare],color=[BLUE,GOLD]); axes[1].set(xticks=[0,1],xticklabels=['Share of hours','Share of total z⁴'],ylabel='Percent (different denominators)',ylim=(0,100),title='Hours with |z| ≥ 2: frequency vs fourth-moment influence');
    for i,v in enumerate([100*tail.mean(),100*tailshare]): axes[1].text(i,v+2,f'{v:.1f}%',ha='center')
    plt.show()
    fig,axes=plt.subplots(1,2,figsize=(12,4),layout='constrained'); axes[0].hist(bz,bins=50,density=True,color=BLUE,alpha=.65,label='Standardized observed rentals'); ng=np.linspace(-3,5,250); axes[0].plot(ng,stats.norm.pdf(ng),color=GOLD,label='Standard normal reference'); axes[0].set(xlabel='z = (rentals − mean)/descriptive SD',ylabel='Density',title='Shape comparison, not a normality conclusion'); axes[0].legend(); stats.probplot(bv,dist='norm',plot=axes[1]); axes[1].set(title='Normal Q–Q of observed hourly rentals',ylabel='Rentals per observed hour'); plt.show()
    say(f'فیلتر فعلی {BIKE_VARIABLE}/{BIKE_YEAR}: n={len(bv):,}، میانگین={bm:.3f}، میانه={np.median(bv):.1f} و SD={bsd:.3f}. چولگی گشتاوری={bskew:.3f} و کشیدگی={bkurt:.3f} (اضافی={bkurt-3:.3f}) است. {tail.sum():,} ساعت ({tail.mean():.2%}) با |z|≥۲، سهم {tailshare:.2%} از Σz⁴ دارند. میلهٔ دوم تعداد ساعت نیست. این فایل ساعت‌های ثبت‌شدهٔ ۲۰۱۱–۲۰۱۲ است، نه افراد یکتا یا همهٔ ساعات تقویم.')
    # END PRESERVED ANALYSIS: real-moments
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['real-moments'] = run_lesson_78()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

در هیستوگرام، بیشتر ساعت‌ها نسبت به بیشینه کم‌تقاضاترند و انتهای راست گسترده است. در نمودار سهمِ چهارم، درصد ساعت‌ها و درصد Σz⁴ مخرج‌های متفاوت دارند. ستون طلایی بزرگِ گروه انتهایی به معنی فراوانی بیشترِ آن گروه نیست؛ به معنی وزن بسیار زیادِ فاصله‌های آن در توان چهار است.

#### کاربرد و محدودیت

متغیر را به casual یا registered و سال را به ۲۰۱۲ تغییر دهید. در cnt سال ۲۰۱۲ چولگی≈۰٫۹۸۷ و کشیدگی≈۳٫۳۷ است؛ پس تفسیر کل داده را برای هر زیرگروه کپی نکنید. برازش توزیع تقاضا نیازمند توجه به ساعت روز و فصل است.

#### تمرین

#### پاسخ تشریحی

یکی بر تعداد ساعت‌ها تقسیم شده، دیگری بر مجموع توان چهارم فاصله‌ها. چند فاصلهٔ بزرگ می‌توانند سهم بیشترِ گشتاور چهارم را بسازند.

### درس 79. مطالعهٔ واقعی ۳: همبستگی، رتبه و ترکیب زیرگروه‌ها

**هدف یادگیری:** یک رابطه را با جهت، شدت، شکل و محدودهٔ معتبر گزارش کنید.

#### مفهوم و تشریح کامل

آزمایشگاه همبستگی دما، رطوبت یا شاخص باد را مقابل تعداد کل، گذری یا عضو قرار می‌دهد. تمام ضرایب از همهٔ زوج‌های همان فیلتر حساب می‌شوند. برای دما و نام فصل بین README بسته و صفحهٔ فعلی UCI اختلاف وجود دارد؛ بنابراین دما را به‌صورت شاخص نرمال‌شدهٔ فایل و فصل‌ها را با کد ۱ تا ۴ گزارش می‌کنیم. رطوبت ۱۰۰×hum درصد است.

#### فرمول و قرارداد

```text
Pearson: linear association; Spearman: Pearson of average-tie ranks
```

#### مثال حل‌شدهٔ مرجع

در کل ساعت‌ها، شاخص دما با cnt پیرسون≈۰٫۴۰۵ و اسپیرمن≈۰٫۴۲۳ دارد و r²≈۰٫۱۶۴ است. برای رطوبت با cnt، پیرسون≈−۰٫۳۲۳ و اسپیرمن≈−۰٫۳۶۰ است. در فصل کد ۱ سال ۲۰۱۲، پیرسونِ رطوبت≈−۰٫۲۵۸ می‌شود. تبدیل خطیِ مثبتِ شاخص دما ضرایب را عوض نمی‌کند. تغییر با فیلتر، توصیفِ جامعهٔ انتخاب‌شده است، نه اثبات اینکه فیلتر علت رابطه بوده است.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_79(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # BEGIN PRESERVED ANALYSIS: real-correlation
    CORR_X='temp'  # normalized temp / hum / windspeed
    CORR_Y='cnt'  # cnt / casual / registered
    CORR_YEAR='all'  # all / 2011 / 2012
    CORR_SEASON='all'  # all / 1 / 2 / 3 / 4 (codes only: conflicting metadata)
    cf=bike.copy()
    if CORR_YEAR!='all': cf=cf.loc[cf.dteday.dt.year.eq(int(CORR_YEAR))]
    if CORR_SEASON!='all': cf=cf.loc[cf.season.eq(int(CORR_SEASON))]
    assert len(cf)>3
    cx=cf[CORR_X].to_numpy(float)*(100 if CORR_X=='hum' else 1); cy=cf[CORR_Y].to_numpy(float)
    cr=stats.pearsonr(cx,cy).statistic; crho=stats.spearmanr(cx,cy).statistic; cfit=np.polyfit(cx,cy,1)
    cx_label={'temp':'Normalized temperature index (original file)','hum':'Relative humidity (%)','windspeed':'Normalized windspeed index'}[CORR_X]
    fig,axes=plt.subplots(1,2,figsize=(12,4.5),layout='constrained'); axes[0].scatter(cx,cy,s=9,alpha=.12,color=BLUE,rasterized=True); cgrid=np.linspace(cx.min(),cx.max(),100); axes[0].plot(cgrid,np.polyval(cfit,cgrid),color=GOLD,lw=2); axes[0].set(xlabel=cx_label,ylabel=f'{CORR_Y}: rentals per hour',title=f'All {len(cf):,} selected hours; descriptive linear fit')
    bins=pd.cut(cx,bins=10); cbin=pd.DataFrame({'band':bins,'x':cx,'rentals':cy}).groupby('band',observed=True).agg(x=('x','mean'),mean_rentals=('rentals','mean'),n=('rentals','size'))
    axes[1].plot(cbin.x,cbin.mean_rentals,marker='o',color=BLUE); axes[1].set(xlabel=cx_label,ylabel='Mean rentals within band',title='Binned averages hide within-band spread')
    for row in cbin.itertuples(): axes[1].annotate(f'n={row.n}',(row.x,row.mean_rentals),xytext=(0,8),textcoords='offset points',ha='center',fontsize=8)
    plt.show(); display(cbin.round(3))
    correlation_rows=[]
    for (year,season_code),part in cf.groupby([cf.dteday.dt.year,'season']):
        correlation_rows.append([year,season_code,len(part),part[CORR_X].corr(part[CORR_Y]),part[CORR_X].corr(part[CORR_Y],method='spearman')])
    display(pd.DataFrame(correlation_rows,columns=['year','season code','observed hours','Pearson r','Spearman rho']).round(4))
    check_mask=(bike.dteday.dt.year.eq(2012)&bike.season.eq(1)); humidity_sub_r=bike.loc[check_mask,'hum'].corr(bike.loc[check_mask,'cnt'])
    say(f'انتخاب {CORR_X} در برابر {CORR_Y}، سال={CORR_YEAR}، فصل کد={CORR_SEASON}: n={len(cf):,}، r={cr:.4f}، ρ={crho:.4f} و r²={cr**2:.3f}. r² سهم توضیح توصیفیِ خط ساده با عرض از مبدأ است، نه سهم اثر علّی. مثال کنترل رطوبت/کل در فصل کد ۱ سال ۲۰۱۲: r={humidity_sub_r:.4f}. همهٔ زوج‌های انتخاب‌شده برای ضرایب و نقاط مصرف شده‌اند؛ ساعت روز و روند رشد می‌توانند مخدوش‌کننده باشند.')
    # END PRESERVED ANALYSIS: real-correlation
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['real-correlation'] = run_lesson_79()
plt.close('all')
````

#### راهنمای کامل خواندن خروجی و نمودار

در ابر نقاط هر نقطه یک ساعت و شفافیّت نشان‌دهندهٔ تراکم است. خط طلایی برازش خطیِ توصیفی است. در نمودار بازه‌ای، ارتفاع میانگین اجاره و n تعداد ساعت در بازه است؛ میانگین‌گیری می‌تواند پراکندگیِ درون بازه را پنهان کند. r²=۰٫۱۶۴ را «۱۶٫۴٪ اثر علّی دما» ننویسید.

#### کاربرد و محدودیت

سال، فصل و نوع روز را تغییر دهید و جدول گروه‌ها را بخوانید. هوا و تقاضا با زمان روز، تعطیلی، فصل، اندازهٔ شبکه و رشد سیستم درهم‌آمیخته‌اند. cnt از جمع casual و registered ساخته شده، پس همبستگی اجزا با کل بخشی از وابستگیِ تعریف‌شدهٔ جمع را منعکس می‌کند.

#### تمرین

اگر ρ بزرگ‌تر از r باشد، آیا حتماً رابطه علّی است؟

#### پاسخ تشریحی

خیر. ممکن است همراهی رتبه‌ای از نزدیکی به خط قوی‌تر باشد یا مقدارهای انتهایی بر پیرسون اثر گذاشته باشند. شکل نمودار و طرح داده‌برداری را بررسی کنید.

### درس 80. آمار مسئولانه: زیرگروه، حریم خصوصی و دفتر تصمیم

**پرسش شروع:** دقت کل۹۰٪ است؛ آیا مدل برای همه خوب است و می‌توان دادهٔ ریز را منتشر کرد؟

#### مفهوم، دقیق اما ساده

میانگین خوب می‌تواند عملکرد بدِ یک زیرگروه کوچک را پنهان کند. ارزیابی مسئولانه باید نرخ پایه، معیارهای متناسب با خطا، حجم هر گروه، عدم‌قطعیت و زمینهٔ استفاده را کنار هم بگذارد. معیارهای عدالت مانند برابری نرخ مثبت یا برابری خطای شرطی یک مفهوم واحد نیستند و گاهی با نرخ‌های پایهٔ متفاوت در تعارض‌اند؛ انتخاب باید مسئله‌محور و با مشارکت ذی‌نفعان باشد. حذف نام افراد به‌تنهایی ناشناس‌سازی نیست؛ ترکیب چند ویژگی یا اتصال به فایل دیگر می‌تواند بازشناسایی کند. تجمیع، حداقل‌سازی داده، کنترل دسترسی و روش‌های رسمیِ حریم خصوصی نقش‌های متفاوت دارند. هیچ‌کدام از یک مثال کوچک مجوز تصمیم پرخطر دربارهٔ فرد نمی‌سازد.

#### فرمول و قرارداد

```text
Aggregate performance = weighted subgroup performance
Responsible record: intended use + evidence + uncertainty + harm + privacy + monitoring + owner.
```

#### مثال حل‌شده

۹۰۰ مورد گروهA با۹۵٪ صحت و۱۰۰ مورد گروهB با۴۵٪ صحت، دقت کلی۹۰٪ می‌سازند: (۸۵۵+۴۵)/۱۰۰۰. اعلام فقط۹۰٪ شکست در گروهB را پنهان می‌کند. این دو گروه، برچسب مصنوعی‌اند و ویژگی حساس واقعی افراد نیستند. مثال دوم: جدول بدون نامی که سن دقیق، محل دقیق و زمان دقیق دارد ممکن است هنوز قابل اتصال به افراد باشد؛ «اسم را پاک کردم» تضمین محرمانگی نمی‌دهد. پیش از انتشار باید ضرورت هر فیلد و خطر ترکیب را بررسی کرد.

````python
def run_lesson_80(BLUE='#277f9f', GOLD='#c08b2c', TEAL='#278b80'):
    # BEGIN PRESERVED ANALYSIS: responsible
    rng=np.random.default_rng(202632)
    correct=np.array([855,45]); totals=np.array([900,100]); rates=correct/totals; pooled=correct.sum()/totals.sum()
    fig,ax=plt.subplots(); ax.bar(['Group A n=900','Group B n=100','Aggregate n=1000'],100*np.r_[rates,pooled],color=[BLUE,GOLD,TEAL]); ax.set(ylabel='Constructed accuracy (%)',ylim=(0,100),title='A good aggregate can hide a poorly served subgroup'); show()
    checklist=pd.DataFrame({'review':['Target & intended use','Sampling & measurement','Held-out and subgroup evidence','Privacy & data minimization','Monitoring & accountable owner'],'required evidence':['Population, unit, estimand','Source, missingness, grain, units','Metric, n, uncertainty, error costs','Necessity, access, re-identification review','Drift checks, failure threshold, review date']}); display(checklist)
    say(f'دقت کل={pooled:.0%} از گروه‌های {rates[0]:.0%} و {rates[1]:.0%} به دست آمد. میانگین مناسب، عملکرد برابر یا محرمانگی داده را تضمین نمی‌کند؛ این مثال صرفاً ساختگی و آموزشی است.')
    # END PRESERVED ANALYSIS: responsible
    return locals()

with plt.rc_context(STYLE_SIM):
    lesson_results['responsible'] = run_lesson_80()
plt.close('all')
````

#### راهنمای خواندن نمودار و نتیجه

هر میله نرخ صحت را با مخرج همان گروه نشان می‌دهد. گروه کوچک‌تر سهم کمتری در عدد کل دارد، ولی اهمیت تجربهٔ افرادش کمتر نیست. برای مقایسهٔ علمی واقعی به CI، تعریف برچسب و ماتریس خطا نیاز داریم؛ accuracy نمونهٔ این درس فقط برای آشکارکردن پنهان‌سازی در تجمیع است. از تفاوت این نرخ‌ها به‌تنهایی علت، تبعیض عمدی یا راه‌حل حقوقی نتیجه نمی‌گیریم.

#### دام رایج

عدد کلی، تعهد اخلاقی یا حقوقی را ثابت نمی‌کند. یک روش fairness یا چند نویز تصادفی نیز به‌تنهایی تضمین عدالت یا حریم خصوصی رسمی نیست.

#### تمرین

پیش از انتشار تحلیل برای تصمیم واقعی، چه پنج پرسشی می‌پرسید؟

#### پاسخ تشریحی

کمیت هدف و جمعیت چیست؟ منبع و کیفیت داده چگونه کنترل شد؟ عدم‌قطعیت و عملکرد زیرگروه‌ها چیست؟ چه آسیب یا بازشناسایی ممکن است؟ چه کسی پایش تغییر توزیع و اصلاح تصمیم را بر عهده دارد؟ پاسخ‌ها را همراه معیار توقف و بازبینی ثبت کنید.

### درس 81. از همهٔ تحلیل‌ها تا یک تصمیم طراحیِ قابل آزمون

**هدف یادگیری:** یک آموزش آماری را به برنامهٔ پژوهش و پروتوتایپ تبدیل کنید.

#### مفهوم و تشریح کامل

زنجیرهٔ تصمیم با هدف کاربر شروع می‌شود: راحتی، دقت، خستگی یا سرعت. ابعاد بدن فقط بخشی از توضیح عملکردند. سپس جامعهٔ هدف و تنوع گرفتن را تعریف، اندازه‌گیری را کالیبره و چند پروتوتایپ می‌سازیم. روش آماری باید به پرسشِ از پیش تعریف‌شده پاسخ دهد؛ زیادکردن تعداد مدل جای طرح پژوهش را نمی‌گیرد.

#### فرمول و قرارداد

```text
پرسش → معیار موفقیت → دادهٔ مناسب → برآورد و عدم‌قطعیت → پروتوتایپ → آزمون مستقل
```

#### مثال حل‌شدهٔ مرجع

برای ده فرد آموزشی، تقسیم ≤۲۰، بیش از ۲۰ تا ۲۳ و >۲۳ به‌ترتیب ۴، ۴ و ۲ نفر می‌دهد. این یک سناریوی S/M/L است. برای آزمون، سه پروتوتایپ را با ترتیب متوازن روی افراد جدید امتحان کنید، دقت وظیفه و امتیاز راحتی را ثبت و اثر تکرارِ هر شخص را در تحلیل لحاظ کنید. آستانه‌های قبول باید از معنای عملی بیایند.

#### اجرای محاسبه و نمودار

کد زیر را اجرا کنید؛ تفسیر اجرای حاضر پس از خروجی از خود محاسبات ساخته می‌شود.

````python
def run_lesson_81(BLUE='#2475a4', GOLD='#d09331', TEAL='#168a7c'):
    # Explicit inputs; never read a previous lesson namespace.
    df = make_hand_data()

    # BEGIN PRESERVED ANALYSIS: design
    size_scenario=pd.cut(df.hand_length_cm,bins=[-np.inf,20,23,np.inf],labels=['S (≤20)','M (>20 to 23)','L (>23)'],right=True)
    size_counts=size_scenario.value_counts(sort=False); display(size_counts.rename('people').to_frame())
    # Explicitly constructed engineering tolerance example; these are NOT measurements of manufactured parts.
    demo_parts=np.array([113.8,114.2,114.7,114.9,115.0,115.1,115.3,115.8,116.0,116.2]); specification=(114,116)
    in_spec=(demo_parts>=specification[0])&(demo_parts<=specification[1])
    fig,axes=plt.subplots(1,2,figsize=(11,4),layout='constrained'); axes[0].bar(size_counts.index.astype(str),size_counts,color=BLUE); axes[0].set(ylabel='People in the 10-person teaching sample',title='An arbitrary size scenario: not market segmentation')
    axes[1].scatter(np.arange(1,11),demo_parts,color=BLUE); axes[1].axhspan(*specification,color=GOLD,alpha=.18,label='Engineering specification 114–116 mm'); axes[1].axhline(115,color='0.3',ls='--'); axes[1].set(xlabel='Illustrative part ID',ylabel='Constructed part length (mm)',title='Specification limits are not statistical control limits'); axes[1].legend(); plt.show()
    say(f'سناریوی قراردادی سایزها {size_counts.tolist()} فرد است. در مثال مستقلِ کاملاً ساختگیِ قطعه، {in_spec.sum()} از {len(demo_parts)} مقدار در محدودهٔ ۱۱۴–۱۱۶ میلی‌مترند؛ این عدد کیفیت خط تولید واقعی یا حدود کنترل را بیان نمی‌کند. تصمیم نهایی باید با افراد جدید، معیار راحتی/دقت/خستگی، ترتیب متوازن و تحلیل اندازه‌گیری تکراری ارزیابی شود.')
    # END PRESERVED ANALYSIS: design
    return locals()

with plt.rc_context(STYLE_HAND):
    lesson_results['design'] = run_lesson_81()
plt.close('all')
````

## کنترل‌های نهایی و جمع‌بندی

سلول زیر اعداد کلیدی را با محاسبات مستقل کنترل می‌کند: مجموع و واریانس دستی، همبستگی حاصل از حاصل‌ضرب مرکزی، PCA از تجزیهٔ SVD، U رتبه‌ای، Log-rank دستی، تعداد سطر داده‌های واقعی و قرارداد چارک‌ها. این‌ها کنترل نسخهٔ مرجع‌اند؛ اگر عمداً داده‌ها را تغییر داده‌اید، انتظار تطابق با اعداد مرجع نداشته باشید.


````python
# Bind the reference checks to the exact lesson-local results.
x = lesson_results['lesson-3']['x']
df = make_hand_data()
corr = lesson_results['lesson-14']['corr']
X_scaled = lesson_results['lesson-24']['X_scaled']
pca = lesson_results['lesson-24']['pca']
low_exp = lesson_results['lesson-19']['low_exp']
high_exp = lesson_results['lesson-19']['high_exp']
mw = lesson_results['lesson-23']['mw']
lr_chi = lesson_results['logrank']['lr_chi']
lr_manual = lesson_results['logrank']['lr_manual']
full_design = lesson_results['lesson-27']['full_design']
design_rank = lesson_results['lesson-27']['design_rank']
design_columns = lesson_results['lesson-27']['design_columns']
future_forecast = lesson_results['lesson-39']['future_forecast']
assert np.isclose(sum(x)/len(x),21.5)
assert np.isclose(sum((value-21.5)**2 for value in x)/len(x),6.25)
assert np.allclose(np.quantile(x,[.25,.5,.75],method='linear'),[20,21,22.75])
direct_r=np.sum((x-x.mean())*(df.hand_width_cm-df.hand_width_cm.mean()))/np.sqrt(np.sum((x-x.mean())**2)*np.sum((df.hand_width_cm-df.hand_width_cm.mean())**2))
assert np.isclose(direct_r,corr.loc['hand_length_cm','hand_width_cm'])
_,sv,_=np.linalg.svd(X_scaled,full_matrices=False)
assert np.allclose(sv[:2]**2/np.sum(sv**2),pca.explained_variance_ratio_)
manual_u=sum((a>b)+.5*(a==b) for a in low_exp for b in high_exp)
assert np.isclose(manual_u,mw.statistic)
assert np.isclose(lr_chi,lr_manual)
assert not full_design and design_rank<design_columns
assert np.allclose(ansur.loc[ansur.sex.eq('male'),'handlength'].quantile([.25,.5,.75]),[187,193,200])
assert len(bike)==17379 and np.isclose(bike.cnt.mean(),189.46308763450142)
assert np.isclose(bike.temp.corr(bike.cnt),.404772275778658)
assert len(future_forecast)==6 and np.allclose(future_forecast,[14,16,17,19,20,22])
say('همهٔ کنترل‌های نسخهٔ مرجع گذشتند؛ هر ۸۱ درس با ترتیب سایت اجرا شد. داده، واحد، روش، عدم‌قطعیت و مرز تعمیم باید همراه هر نتیجه گزارش شوند. مدل پیچیده جای نمونه‌گیری و آزمایش مستقل را نمی‌گیرد.')
````

## واژه‌نامهٔ کامل دوره

**جامعه و نمونه:** جامعه هدفِ تعمیم است؛ نمونه افراد یا واحدهای واقعاً دیده‌شده‌اند.

**پارامتر و آماره:** پارامتر ویژگی جامعه و آماره تابعی از نمونه است.

**برآورد نقطه‌ای و بازه‌ای:** یک عدد برای پارامتر، در برابر مجموعه‌ای از مقادیر با روش و پوشش مشخص.

**SD و SE:** SD پراکندگی مشاهده‌ها و SE نوسان برآوردگر در تکرار نمونه‌گیری است.

**چگالی و احتمال:** احتمالِ یک بازه از مساحت چگالی به دست می‌آید؛ ارتفاع یک نقطه احتمال آن نقطه نیست.

**چارک و صدک:** آستانه‌های موقعیتیِ توزیع؛ تعریف نرم‌افزاری در نمونهٔ محدود باید مشخص شود.

**مقدار انتهایی و خطای ثبت:** انتهایی‌بودن یک مشاهده لزوماً خطا نیست؛ اصالت با منبع و اندازه‌گیری بررسی می‌شود.

**p-value و α:** p از داده و مدل محاسبه می‌شود؛ α آستانهٔ تصمیمِ از پیش تعیین‌شده است.

**اندازهٔ اثر:** مقدارِ اختلاف یا رابطه با واحد و تعریفِ متناسب با پرسش.

**عدم رد و هم‌ارزی:** عدم رد، نبودِ شواهد کافی است؛ هم‌ارزی نیازمند حاشیه و آزمون مشخص است.

**باقیمانده:** اختلاف مشاهده و مقدار برازش‌شده؛ نشانه‌ای برای بررسیِ ساختار خطای مدل.

**هم‌خطی:** وابستگی شدید پیش‌بین‌ها که می‌تواند ضرایبِ شرطی را ناپایدار کند.

**بارگذاری و score:** ضرایبِ ویژگی‌ها در مؤلفه، در برابر مختصاتِ هر مشاهده روی مؤلفه.

**آموزش و آزمون:** دادهٔ یادگیری مدل، در برابر دادهٔ کنارگذاشته‌شده برای سنجش تعمیم.

**نشت داده:** ورود اطلاعاتی به فرایند آموزش/انتخاب که در لحظهٔ استفاده در دسترس نیست.

**سانسور:** اطلاع ناقص اما مفید دربارهٔ زمان رخداد؛ از جمله سالم‌بودن تا پایان پیگیری.

**اثر متقابل:** تغییر رابطهٔ یک عامل با پاسخ در سطح‌های عامل دیگر.

**خط پایه:** پیش‌بینی ساده‌ای که مدل پیچیده باید ارزش افزوده‌اش را نسبت به آن نشان دهد.

## گام‌های بعدی و پروژهٔ پایانی

۱. فیلترهای سه کارگاه واقعی را تغییر دهید و برای هر انتخاب، n، واحد، برآورد و محدودیت را دوباره گزارش کنید. تفسیر خودکار کنار خروجی تغییر می‌کند؛ مثال مرجع متن ثابت می‌ماند.

۲. یک نتیجهٔ نامعتبر رایج را اصلاح کنید: SD به‌جای CI، p بزرگ به‌جای هم‌ارزی، r² به‌جای اثر علّی، یا خطای آموزش به‌جای آزمون. پاسخ باید به عدد و طراحی مطالعه ارجاع بدهد.

۳. برای تصمیم بین دو و سه سایز ماوس، جمعیت هدف، روش اندازه‌گیری، معیار حداقل اثر مهم، نمونهٔ مستقل و ثبت خستگی/دقت/راحتی را از پیش تعریف کنید. ANSUR و ده فرد آموزشی جای آن آزمایش را نمی‌گیرند.
